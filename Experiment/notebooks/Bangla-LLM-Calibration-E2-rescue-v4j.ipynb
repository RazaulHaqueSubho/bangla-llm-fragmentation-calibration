# Does Tokenization Fragmentation Predict Miscalibration in Bangla LLMs? (v4j E2 Rescue: Cyclic4 + LLaMA-3.2-3B (Final Run-All))
## A Unified Cross-Lingual Evaluation & Econometric Analysis Framework [v4j]

### Research Objective & Hypotheses
This self-contained research notebook investigates the empirical relationship between **tokenization fragmentation** and **model miscalibration** in Large Language Models (LLMs) evaluated on Bengali (Bangla) tasks, using **BnMMLU** as the primary multilingual benchmark (~134K multiple-choice questions across 41 subjects) with an **English control**: the English translation of the *same* BnMMLU questions (item-parallel; `CONFIG.en_source="bnmmlu_translation"`), or optionally a subject-matched English MMLU sample (`en_source="mmlu"`).

#### Core Research Hypotheses:
1. **$H_1$ (Vocabulary Coverage & Calibration Error):** Tokenizers with lower vocabulary coverage for Bangla (e.g., standard Latin-centric Llama BPE tokenizers relying on byte fallbacks) produce severe subword fragmentation, which statistically predicts increased model miscalibration (higher Expected Calibration Error, Brier Score, and Maximum Calibration Error).
2. **$H_2$ (Item-Level Residual Prediction):** At the question level, tokenization fragmentation metrics (Tokens-per-Word, Single-Token Retention Rate, Tokens-per-Character) positively predict the magnitude of individual model calibration residuals ($e_i = |\hat{p}_i - \mathbb{I}(\hat{y}_i = y_i)|$) after controlling for correctness, subject fixed effects, and prompt length.
3. **$H_3$ (Cross-Lingual Divergence):** The miscalibration-fragmentation relationship exhibits significant divergence across languages, reflecting the distinct morphological and script characteristics of Bengali (an inflectional abugida with complex conjuncts) compared to English (Latin alphabetic).

> [!IMPORTANT]
> **Epistemic & Non-Causal Framing:** All statistical models in this study estimate **observational associations**, never causal effects. Observational evaluations cannot eliminate confounding from pretraining corpus composition, alignment regimes, or parameter scale. All findings are strictly framed as "predicts" or "is associated with."

---

### Mathematical Formulations

#### 1. Tokenization Fragmentation Metrics
* **Tokens-per-Word (TPW / Fertility):**
  $$\text{TPW} = \frac{N_{\text{tokens}}}{N_{\text{words}}}$$
  Whitespace-delimited word baseline directly comparable to prior literature (e.g., TituLLMs, AdaptMix).
* **Single-Token Retention Rate (STRR):**
  $$\text{STRR} = \frac{1}{|W|} \sum_{w \in W} \mathbb{I}(\text{len}(\text{tokenize}(w)) = 1)$$
  Reported in both `isolated` mode (tokenizing words in isolation) and `in_context` mode (tokenizing `" " + word` to account for BPE space-prefix tokens).
* **Tokens-per-Character (TPC / Character Fertility):**
  $$\text{TPC} = \frac{N_{\text{tokens}}}{N_{\text{non-whitespace characters}}}$$
  Normalizes token count against non-whitespace character count, providing a script-agnostic cross-lingual baseline.

#### 2. Option Extraction, Uncertainty Quantification & Calibration Metrics
* **Candidate Option Probability Extraction (Logsumexp & Softmax):**
  For candidate option letters $k \in \{A, B, C, D\}$, all single-token variants (e.g., `"A"` and `" A"`) are resolved to token IDs $V_k$. Log-probabilities over the full vocabulary are pooled via logsumexp and renormalized over $A-D$:
  $$s_k = \log \sum_{v \in V_k} \exp(\log P(v \mid \text{prompt})), \quad P(k) = \frac{\exp(s_k)}{\sum_{j \in \{A, B, C, D\}} \exp(s_j)}$$
  **Confidence** is defined as the Max Softmax Probability (MSP): $\hat{p} = \max_{k \in \{A, B, C, D\}} P(k)$.
* **Option Mass ($\text{Mass}_{A-D}$):**
  Total unnormalized full-vocabulary probability mass allocated to the candidate option tokens:
  $$\text{Option Mass} = \sum_{k \in \{A, B, C, D\}} \sum_{v \in V_k} P(v \mid \text{prompt})$$
* **Expected Calibration Error (ECE):**
  $$\text{ECE} = \sum_{m=1}^M \frac{|B_m|}{N} |\text{acc}(B_m) - \text{conf}(B_m)|$$
* **Brier Score (BS) & Multiclass Brier Score:**
  $$\text{BS}_{\text{top1}} = \frac{1}{N} \sum_{i=1}^N (\hat{p}_i - \mathbb{I}(\hat{y}_i = y_i))^2, \quad \text{BS}_{\text{multi}} = \frac{1}{N} \sum_{i=1}^N \sum_{k=1}^4 (P_{i,k} - y_{i,k})^2$$
* **Negative Log-Likelihood (NLL):**
  $$\text{NLL} = -\frac{1}{N} \sum_{i=1}^N \log(\max(P_{i, y_i}, \epsilon))$$
* **Item-Level Calibration Residual:**
  $$e_i = |\hat{p}_i - \mathbb{I}(\hat{y}_i = y_i)|$$

### Prompt Variants & Position Debiasing Strategy
Small and base models frequently suffer from single-letter logit collapse or low option mass. We provide four orthogonal degrees of adaptation:
1. **`plain` (default):** Prompt ends with the task instruction followed by `:\n`, expecting an immediate bare letter (`A`, `B`, `C`, or `D`).
2. **`answer_cue`:** Appends a language-appropriate completion cue (`উত্তর:` for Bangla, `Answer:` for English) directly after the instruction. Many instruction-tuned models follow completion cues more reliably.
3. **`few_shot` (4-shot or 8-shot):** Prepends balanced, fixed in-context exemplars drawn from the held-out exemplar pool (`fewshot_exemplars.json`).
4. **`cyclic4` position debiasing:** Computes predictions across 4 cyclic permutations of option order (positions shifted by $s \in \{0, 1, 2, 3\}$) and averages mapped probabilities back to canonical options. Items with self-referential / cross-referencing options (e.g. "both A and B", "all of the above", "উপরের সবগুলো") are automatically exempted from permutation.

> **Methodology & Validity Selection Rules:**
> - Settings (`prompt_variant`, `n_shots`, `use_chat_template`, `debias`) are selected **per model** using **validity gates only** (`mean_mass >= MIN_OPTION_MASS_THRESHOLD` and `top_letter_share <= MAX_COLLAPSE_THRESHOLD`).
> - **Never select settings by accuracy**: the diagnostic subset is held **disjoint** from the evaluation pool (verified in Section 5); selecting variants by accuracy would still create data leakage via the shared source pool and invalidate calibration findings.
> - **Calibration interpretation under cyclic4:** `cyclic4` averages across permutations, changing the empirical definition of model "confidence". Both raw (`pred_raw`, `confidence_raw`) and debiased calibration metrics must be reported.
> - **Reporting:** Every chosen per-model setting must be explicitly reported in the paper's methodology and limitations sections.

```python
# ==========================================
# GLOBAL STRICT REAL-ONLY CONFIGURATION
# ==========================================
STRICT_REAL_MODE: bool = True
os.environ["STRICT_REAL_MODE"] = "True"

# ==========================================
# FIX 1: Kaggle input paths (single config cell)
# ==========================================
ANALYSIS_PLAN_DIR = "/kaggle/input/datasets/asfiazaman/analysis-plan"
GATE_CONFIG_DIR   = "/kaggle/input/datasets/asfiazaman/bangla-calibration-config"
BNMMLU_DIR        = "/kaggle/input/datasets/asfiazaman/bnmmlu-full-en"

import hashlib as _hashlib

def _resolve_in_dir(resource_dir, fname, label):
    # Resolve fname by searching resource_dir recursively.
    # Log resolved full path + sha256. Source files are only read (hashed), never modified.
    # NOTE: resource_dir encodes the EXPECTED Kaggle dataset owner (e.g. asfiazaman).
    # If that owner/dataset name differs in the user's Kaggle input, fall back to an
    # owner-agnostic recursive search under /kaggle/input so the notebook still works.
    found = None
    if os.path.isdir(resource_dir):
        for _r, _d, _files in os.walk(resource_dir):
            if fname in _files:
                found = os.path.join(_r, fname)
                break
    if found is None and os.path.isdir("/kaggle/input"):
        # Owner-agnostic fallback: find the file anywhere under /kaggle/input.
        for _r, _d, _files in os.walk("/kaggle/input"):
            if fname in _files:
                found = os.path.join(_r, fname)
                break
    if found is None:
        existing = []
        if os.path.isdir(resource_dir):
            for _r, _d, _files in os.walk(resource_dir):
                existing.extend(os.path.join(_r, f) for f in _files)
        raise FileNotFoundError(
            f"{label} '{fname}' not found under {resource_dir} (and not anywhere under /kaggle/input).\n"
            f"  Files present in {resource_dir}:\n    "
            + "\n    ".join(existing)
            + "\n  Confirm the Kaggle dataset owner/name attached in Kaggle input (constants in cell 6)."
        )
    digest = _hashlib.sha256(open(found, "rb").read()).hexdigest()
    logger.info(f"[FIX 1] {label}: {found} | sha256={digest}")
    return found


MODEL_TAG = "tigerllm_1b"

# ==========================================
# FIX 2 / FIX 3: stage enable flags
# ==========================================
RUN_E2_EXISTING = False   # existing three models already evaluated; skip on Run All
RUN_E3 = True             # run E3 extension for the four new approved models

# E3 completion / exploratory flags
E3_COMPLETED = ["qwen25_3b", "bloomz_3b", "llama32_3b_base", "phi35_mini"]   # finished in an earlier session, results downloaded
E3_FORCE_RERUN = False
RUN_PHI_EXPLORATORY = True

# E4: new approved tokenizer models (added after the preregistered six-model set)
E4_NEW_MODELS = ["mgpt_1_3b", "olmo2_1b", "smollm2_1_7b", "sarvam1_2b"]
E4_EXPLORATORY_MODELS = []  # frozen before any inference; filled manually if needed

# Compute the final E3 execute list (skips completed models unless force rerun;
# phi35_mini runs only when RUN_PHI_EXPLORATORY=True).
REGISTRY_ORDER = ["qwen25_3b", "bloomz_3b", "phi35_mini", "llama32_3b_base", "mgpt_1_3b", "olmo2_1b", "smollm2_1_7b", "sarvam1_2b"]
if E3_FORCE_RERUN:
    TARGET_E3_MODELS = list(REGISTRY_ORDER)
else:
    TARGET_E3_MODELS = [
        m for m in REGISTRY_ORDER
        if m not in E3_COMPLETED and (m != "phi35_mini" or RUN_PHI_EXPLORATORY)
    ]
print(f"E3_COMPLETED={E3_COMPLETED} E3_FORCE_RERUN={E3_FORCE_RERUN} RUN_PHI_EXPLORATORY={RUN_PHI_EXPLORATORY}")
print(f"📋 RUN ALL will execute models: {TARGET_E3_MODELS}")
assert set(TARGET_E3_MODELS) == set(E4_NEW_MODELS), f"TARGET_E3_MODELS {TARGET_E3_MODELS} != E4_NEW_MODELS {E4_NEW_MODELS}"
_seven_completed = {"titullm_1b", "tigerllm_1b", "llama32_3b", "qwen25_3b", "bloomz_3b", "llama32_3b_base", "phi35_mini"}
assert not (set(TARGET_E3_MODELS) & _seven_completed), f"Completed models in TARGET_E3_MODELS: {set(TARGET_E3_MODELS) & _seven_completed}"



# ==========================================
# MODEL REGISTRY  (all IDs verified on the Hugging Face Hub)
# ==========================================
MODEL_REGISTRY: Dict[str, Dict[str, Any]] = {
    "llama32_3b": {
        "hf_id": "meta-llama/Llama-3.2-3B-Instruct",
        "use_chat_template": True,   # NOTE: registry default; effective Run A used use_chat_template=False (overridden by model_settings.json via apply_uniform_setting)
        "dtype": "float16",
        "needs_hf_token": True,                 # GATED: accept the Llama 3.2 license on the model page first
        "tokenizer_family": "llama3_bpe_128k",
        "expected_model_type": "llama",
        "notes": "Anchor model: original Llama-3.2 128K BPE tokenizer (byte-level, no Bangla-specific vocabulary). use_chat_template=True required for instruction-tuned variant: plain next-token prediction yields option_mass < 0.50.",
    },
    "titullm_1b": {
        "hf_id": "hishab/titulm-llama-3.2-1b-v1.1",
        "use_chat_template": False,
        "prompt_variant": "answer_cue",    # NOTE: registry default; effective Run A used prompt_variant=few_shot (overridden by model_settings.json via apply_uniform_setting)
        "collapse_threshold_override": None, # None = use global MAX_COLLAPSE_THRESHOLD; set float to relax
        "dtype": "float16",
        "needs_hf_token": False,
        "tokenizer_family": "llama3_continued_pretrain",   # Audited in Section 14b
        "expected_model_type": "llama",
        "notes": ("Bangla continued-pretraining of Llama-3.2-1B (base model, no chat tuning). "
                  "prompt_variant=answer_cue applied to prevent logit collapse on base models. "
                  "Section 14b tokenizer audit tests whether vocabulary/tokenization differs from Llama-3.2."),
    },
    "tigerllm_1b": {
        "hf_id": "md-nishat-008/TigerLLM-1B-it",
        "use_chat_template": False,
        "dtype": "float32",                      # Gemma-3 is unstable in fp16 and NVIDIA T4 has no native bf16
        "needs_hf_token": False,
        "tokenizer_family": "gemma3",           # Hub repo tagged gemma3_text; Gemma-3 262K vocabulary
        "expected_model_type": "gemma3_text",
        "notes": ("Out-of-family replication model. Hub config is Gemma-3 (262K vocab), NOT Llama, "
                  "providing a distinct tokenizer family comparison."),
    },
    # ============================================================
    # S9a: Approved New Models Configuration (Uniform debias=cyclic4, chat=off)
    # ============================================================
    "phi35_mini": {
        "hf_id": "microsoft/Phi-3.5-mini-instruct",
        "tokenizer_family": "phi_family",
        "model_size": "3.82B",
        "dtype": "float16",
        "use_chat_template": False,
        "max_seq_len": 2048,
        "estimated_gpu_memory": "8.1 GB",
        "gated_status": "ungated",
        "needs_hf_token": False,
        "model_loading_config": {
            "torch_dtype": "float16",
            "device_map": "auto",
            "low_cpu_mem_usage": True,
            "trust_remote_code": True,
        },
        "expected_model_type": "phi3",
        "notes": "Phi-3.5 Mini Instruct (Microsoft). Ungated, float16 on T4, use_chat_template=False.",
    },
    "qwen25_3b": {
        "hf_id": "Qwen/Qwen2.5-3B-Instruct",
        "tokenizer_family": "qwen_family",
        "model_size": "3.09B",
        "dtype": "float16",
        "use_chat_template": False,
        "max_seq_len": 2048,
        "estimated_gpu_memory": "6.7 GB",
        "gated_status": "ungated",
        "needs_hf_token": False,
        "model_loading_config": {
            "torch_dtype": "float16",
            "device_map": "auto",
            "low_cpu_mem_usage": True,
            "trust_remote_code": True,
        },
        "expected_model_type": "qwen2",
        "notes": "Qwen 2.5 3B Instruct (Alibaba). Ungated, float16 on T4, use_chat_template=False.",
    },
    "bloomz_3b": {
        "hf_id": "bigscience/bloomz-3b",
        "tokenizer_family": "bloom_family",
        "model_size": "3.00B",
        "dtype": "float16",
        "use_chat_template": False,
        "max_seq_len": 2048,
        "estimated_gpu_memory": "6.5 GB",
        "gated_status": "ungated",
        "needs_hf_token": False,
        "model_loading_config": {
            "torch_dtype": "float16",
            "device_map": "auto",
            "low_cpu_mem_usage": True,
            "trust_remote_code": True,
        },
        "expected_model_type": "bloom",
        "notes": "BLOOMZ 3B (BigScience multilingual). Ungated, float16 on T4, use_chat_template=False.",
    },
    "llama32_3b_base": {
        "hf_id": "meta-llama/Llama-3.2-3B",
        "tokenizer_family": "llama_family",
        "model_size": "3.21B",
        "dtype": "float16",
        "use_chat_template": False,
        "max_seq_len": 2048,
        "estimated_gpu_memory": "6.9 GB",
        "gated_status": "gated",
        "needs_hf_token": True,
        "model_loading_config": {
            "torch_dtype": "float16",
            "device_map": "auto",
            "low_cpu_mem_usage": True,
            "trust_remote_code": True,
        },
        "expected_model_type": "llama",
        "notes": "Llama 3.2 3B Base (Meta unaligned base model). Gated (requires HF_TOKEN), float16 on T4, use_chat_template=False.",
    },
    # ============================================================
    # S9b: E4 new approved tokenizer models (added after preregistered six-model set)
    # ============================================================
    "mgpt_1_3b": {
        "hf_id": "ai-forever/mGPT",
        "tokenizer_family": "mGPT_GPT2_multilingual",
        "model_size": "UNVERIFIED",
        "dtype": "float16",
        "use_chat_template": False,
        "max_seq_len": 2048,
        "estimated_gpu_memory": "UNVERIFIED",
        "gated_status": "ungated",
        "needs_hf_token": False,
        "model_loading_config": {
            "torch_dtype": "float16",
            "device_map": "auto",
            "low_cpu_mem_usage": True,
            "trust_remote_code": False,
        },
        "expected_model_type": "gpt2",
        "notes": "mGPT (ai-forever). model_size UNVERIFIED: HuggingFace Hub reports no safetensors weight file for this repo (legacy PyTorch bin), so HfApi().model_info().safetensors.total is None; estimated_gpu_memory cannot be computed from safetensors. Ungated, float16 on T4, use_chat_template=False, native GPT2LMHeadModel (no trust_remote_code). max_position_embeddings=2048 verified via AutoConfig; step-10 context check passed (bn_max 1013 < 2048). Registry max_seq_len=2048 matches completed-model convention (registry value is not read by any inference/cap code; cap is manifest-only at 2048).",
    },
    "olmo2_1b": {
        "hf_id": "allenai/OLMo-2-0425-1B",
        "tokenizer_family": "OLMo2_GPTNeoX",
        "model_size": "1484916736",
        "dtype": "float16",
        "use_chat_template": False,
        "max_seq_len": 2048,
        "estimated_gpu_memory": "3.5 GB",
        "gated_status": "ungated",
        "needs_hf_token": False,
        "model_loading_config": {
            "torch_dtype": "float16",
            "device_map": "auto",
            "low_cpu_mem_usage": True,
            "trust_remote_code": False,
        },
        "expected_model_type": "olmo2",
        "notes": "OLMo-2-0425-1B (allenai). model_size=1,484,916,736 params from HfApi().model_info().safetensors.total (verified). estimated_gpu_memory=params*2 bytes(fp16)+~0.5 GB overhead ≈ 3.5 GB. Ungated, float16 on T4, use_chat_template=False, native Olmo2ForCausalLM (no trust_remote_code). max_position_embeddings=4096 verified via AutoConfig; step-10 context check passed (bn_max 1499 < 4096). Registry max_seq_len=2048 matches completed-model convention (registry value is not read by any inference/cap code; cap is manifest-only at 2048).",
    },
    "smollm2_1_7b": {
        "hf_id": "HuggingFaceTB/SmolLM2-1.7B",
        "tokenizer_family": "Llama3_BPE_SmolLM2",
        "model_size": "1711376384",
        "dtype": "float16",
        "use_chat_template": False,
        "max_seq_len": 2048,
        "estimated_gpu_memory": "3.9 GB",
        "gated_status": "ungated",
        "needs_hf_token": False,
        "model_loading_config": {
            "torch_dtype": "float16",
            "device_map": "auto",
            "low_cpu_mem_usage": True,
            "trust_remote_code": False,
        },
        "expected_model_type": "llama",
        "notes": "SmolLM2-1.7B (HuggingFaceTB). model_size=1,711,376,384 params from HfApi().model_info().safetensors.total (verified). estimated_gpu_memory=params*2 bytes(fp16)+~0.5 GB overhead ≈ 3.9 GB. Ungated, float16 on T4, use_chat_template=False, native LlamaForCausalLM (no trust_remote_code). max_position_embeddings=8192 verified via AutoConfig; step-10 context check passed (bn_max 2228 < 8192). Registry max_seq_len=2048 matches completed-model convention (registry value is not read by any inference/cap code; cap is manifest-only at 2048). NOTE: 45.73% of BN 4-shot prompts exceed 2048 tokens (max 3057) — if a 2048 cap were enforced these would be right-truncated, cutting the answer cue; no cap is currently enforced.",
    },
    "sarvam1_2b": {
        "hf_id": "sarvamai/sarvam-1",
        "tokenizer_family": "Llama3_BPE_Sarvam_Indic",
        "model_size": "2525087744",
        "dtype": "float16",
        "use_chat_template": False,
        "max_seq_len": 2048,
        "estimated_gpu_memory": "5.6 GB",
        "gated_status": "ungated",
        "needs_hf_token": False,
        "model_loading_config": {
            "torch_dtype": "float16",
            "device_map": "auto",
            "low_cpu_mem_usage": True,
            "trust_remote_code": False,
        },
        "expected_model_type": "llama",
        "notes": "Sarvam-1 2B (sarvamai, Indic-focused). model_size=2,525,087,744 params from HfApi().model_info().safetensors.total (verified). estimated_gpu_memory=params*2 bytes(fp16)+~0.5 GB overhead ≈ 5.6 GB. Ungated, float16 on T4, use_chat_template=False, native LlamaForCausalLM (no trust_remote_code). max_position_embeddings=8192 verified via AutoConfig; step-10 context check passed (bn_max 608 < 8192). Registry max_seq_len=2048 matches completed-model convention (registry value is not read by any inference/cap code; cap is manifest-only at 2048).",
    },
}


@dataclass
class ExperimentConfig:
    # Model selector: "llama32_3b", "titullm_1b", or "tigerllm_1b"
    model_key: str = "llama32_3b"

    # Backend engine: STRICT REAL evaluation uses "transformers"
    backend: str = "transformers"

    # Evaluation mode: "extended" (primary) or "baseline"
    mode: str = "extended"

    # Dataset file path (optional local override; real runs download from Hub)
    dataset_path: Optional[str] = None

    # Number of items PER LANGUAGE: Exactly 1500 samples per language
    n_samples: int = 1500

    # Batch size (only 1 is supported; batched inference is NOT implemented)
    batch_size: int = 1

    # Optional path to an attached Kaggle dataset directory containing previous session outputs
    previous_results_dir: Optional[str] = None

    # Reproducibility seed
    seed: int = DEFAULT_SEED

    # Calibration binning
    n_bins: int = 10
    bin_strategy: str = "uniform"

    # Run legacy baseline text generation (opt-in; default False)
    run_baseline: bool = False

    # Output directory for results: Default /kaggle/working/results/ on Kaggle
    output_dir: str = OUTPUT_DIR

    # Hugging Face access token (retrieved securely from Kaggle Secrets or env)
    hf_token: Optional[str] = None

    # ---------------- Real-data settings ----------------
    # BnMMLU is a GATED dataset with two configs ("full" and "hard"); config 'full' is required.
    bn_dataset_id: str = "samanjoy2/BnMMLU"
    bn_hf_config: str = "full"
    # Optional local parquet/csv copy of BnMMLU (overrides the Hub download when set)
    bn_local_path: Optional[str] = None
    # Only needed if a local copy stores answers as bare digits; the Hub version uses letters a-d
    bn_index_base: Optional[int] = None
    # English control: cais/mmlu, configuration all, split test
    en_dataset_id: str = "cais/mmlu"
    en_hf_config: str = "all"
    en_split: str = "test"

    # English-side source.
    #   "mmlu"               = ORIGINAL design: subject-matched (not item-parallel) control sampled from cais/mmlu.
    #   "bnmmlu_translation" = ITEM-PARALLEL: the English translation of the SAME BnMMLU questions, read from one
    #                          wide CSV (question_bn/question_en, option_a_bn..option_d_en, subject_name, correct_answer).
    en_source: str = "mmlu"
    # Path of the wide bn/en CSV (translation mode). None = auto-find bnmmlu_full_en*.csv under /kaggle/input.
    en_translation_path: Optional[str] = None
    # Subjects dropped in translation mode: their English translation is no longer the same task
    # (questions about Bangla grammar / literature / poetry), so it is not a valid control for tokenization.
    translation_exclude_subjects: Tuple[str, ...] = ("Bengali Language & Syntax", "Bengali Literature", "Bengali Poetry")

    # BnMMLU keeps MathML tags inside question text. Those tags inflate token counts and would be
    # measured as "fragmentation" of the language. True = drop items containing MathML.
    exclude_markup: bool = True

    # "intersection": only BnMMLU subjects that map to an MMLU subject; equal per-subject counts in both languages.
    # "all_bn": stratified sample over ALL BnMMLU subjects; English matched where possible, topped up otherwise.
    subject_match_mode: str = "intersection"

    # Diagnostic / dev set items per language
    dev_n: int = 200

    # Smoke test items
    smoke_n: int = 30
    # Strict smoke test abort on collapse
    strict_smoke: bool = False

    # Prompt variant: "plain" (original) or "answer_cue" (appends language-specific answer cue)
    prompt_variant: str = "answer_cue"  # CONFIG default; effective Run A used few_shot (set by apply_uniform_setting in Section 14c)

    # Few-shot count: 4 = 4-shot (default), 8 = 8-shot. Allowed values: {4, 8}
    n_shots: int = 4

    # Debiasing strategy: "none" (default) or "cyclic4" (v3b+)
    debias: str = "none"

    # Allow few-shot English candidates to fall back to unused test pool if dev split fails
    allow_fewshot_test_pool_fallback: bool = False

    # Per-run collapse threshold override (None = use global MAX_COLLAPSE_THRESHOLD = 0.80).
    # Set to 1.0 to fully suppress the collapse validity gate for base-model analysis runs.
    collapse_threshold_override: Optional[float] = None

    # Analysis validity filter: if True, filter out runs with validity_flag != 'VALID'
    analysis_valid_only: bool = True


ANALYSIS_PLAN_PATH = _resolve_in_dir(ANALYSIS_PLAN_DIR, "analysis_plan.yaml", "Analysis plan")
GATE_CONFIG_PATH = _resolve_in_dir(GATE_CONFIG_DIR, "gate.yaml", "Gate config")
EN_DATA_PATH = _resolve_in_dir(BNMMLU_DIR, "bnmmlu_full_en.csv", "English CSV")

def validate_config(config: "ExperimentConfig") -> None:
    if getattr(config, "dev_n", 200) < 0:
        raise ValueError(f"CONFIG.dev_n must be non-negative, got {config.dev_n}")
    """Validate global experiment configuration parameters."""
    if STRICT_REAL_MODE and config.backend != "transformers":
        raise ValueError(f"STRICT_REAL_MODE=True requires CONFIG.backend='transformers', but got '{config.backend}'. Mock modes are prohibited.")

    if config.batch_size != 1:
        raise NotImplementedError("Batched inference is not implemented in this notebook. Keep CONFIG.batch_size = 1.")

    if config.prompt_variant not in VALID_PROMPT_VARIANTS:
        raise ValueError(f"Invalid CONFIG.prompt_variant='{config.prompt_variant}'. Allowed: {VALID_PROMPT_VARIANTS}")

    if config.prompt_variant == "few_shot":
        if config.n_shots not in {4, 8}:
            raise ValueError(f"Invalid CONFIG.n_shots={config.n_shots} with prompt_variant='few_shot'. Allowed: {{4, 8}}")
    else:
        if config.n_shots not in {0, 4, 8}:
            raise ValueError(f"Invalid CONFIG.n_shots={config.n_shots} with prompt_variant='{config.prompt_variant}'. Allowed: {{0, 4, 8}}")

    if config.en_source not in ("mmlu", "bnmmlu_translation"):
        raise ValueError(f"Invalid CONFIG.en_source='{config.en_source}'. Allowed: 'mmlu', 'bnmmlu_translation'.")

    _ALLOWED_DEBIAS = ("none", "cyclic4", "cloze", "prior_debias")
    if config.debias not in _ALLOWED_DEBIAS:
        raise ValueError(f"Invalid CONFIG.debias='{config.debias}'. Allowed: {_ALLOWED_DEBIAS}")


CONFIG = ExperimentConfig(
    model_key=MODEL_TAG,
    backend="transformers",
    mode="extended",
    n_samples=1500,
    dev_n=200,
    batch_size=1,
    run_baseline=False,
    output_dir=OUTPUT_DIR,
    # English = translation of the SAME BnMMLU items (item-parallel). Set en_source="mmlu" for the original MMLU control.
    en_source="bnmmlu_translation",
    en_translation_path=None,   # e.g. "/kaggle/input/<your-dataset>/bnmmlu_full_en.csv"
)

# ---- en_source must be bnmmlu_translation (item-parallel translation mode) ----
assert CONFIG.en_source == "bnmmlu_translation", (
    "CONFIG.en_source must be 'bnmmlu_translation' for item-parallel evaluation, got "
    + repr(CONFIG.en_source)
)


# ============================================================
# S1: Raw and Cyclic4 Logging Configuration
# ============================================================
LOG_RAW_AND_CYCLIC4: bool = True
VERIFY_OPTION_TOKENS_IN_CONTEXT: bool = True
OPTION_TOKEN_STRICT_ATOMIC: bool = True
SAVE_RAW_PROBABILITIES: bool = True


# ============================================================
# S2: Provenance, Dtype and Resume Configuration
# ============================================================
RESUME_ENABLED: bool = True
SHIFT_CHECKPOINT_ENABLED: bool = True
DTYPE_SENSITIVITY_ACTIVE: bool = True
DTYPE_SENSITIVITY_RUN: bool = False   # dtype sensitivity: False => skipped on Run All

# E3 completion / exploratory flags
E3_COMPLETED = ["qwen25_3b", "bloomz_3b", "llama32_3b_base", "phi35_mini"]   # finished in an earlier session, results downloaded
E3_FORCE_RERUN = False
RUN_PHI_EXPLORATORY = True
CLEAN_HF_CACHE_PER_MODEL: bool = True
MAX_SEQ_LENGTH_OBSERVED: int = 0
ASSERTION_STATUS_RECORD_ALL: bool = True

validate_config(CONFIG)
os.makedirs(CONFIG.output_dir, exist_ok=True)

# Securely retrieve Hugging Face token without exposing it
if not CONFIG.hf_token:
    CONFIG.hf_token = os.environ.get("HF_TOKEN") or os.environ.get("HUGGINGFACE_TOKEN")
    if not CONFIG.hf_token and IS_KAGGLE:
        try:
            from kaggle_secrets import UserSecretsClient
            user_secrets = UserSecretsClient()
            CONFIG.hf_token = user_secrets.get_secret("HF_TOKEN")
            logger.info("Successfully retrieved HF_TOKEN from Kaggle Secrets.")
        except Exception as e:
            logger.warning(f"Could not retrieve HF_TOKEN from Kaggle Secrets: {e}")
    if not CONFIG.hf_token and IS_COLAB:
        try:
            from google.colab import userdata
            CONFIG.hf_token = userdata.get("HF_TOKEN")
            logger.info("Successfully retrieved HF_TOKEN from Colab secrets.")
        except Exception:
            pass

# Verify HF_TOKEN is present for real gated resources
_active_needs_token = MODEL_REGISTRY.get(CONFIG.model_key, {}).get("needs_hf_token", False) or (CONFIG.en_source != "bnmmlu_translation" and not CONFIG.bn_local_path)
if not CONFIG.hf_token and _active_needs_token:
    raise RuntimeError(
        f"No Hugging Face token found (HF_TOKEN). Model '{CONFIG.model_key}' requires authenticated access.\n"
        "  1. Accept license on the Hugging Face model / dataset page\n"
        "  2. Add HF_TOKEN under Kaggle Secrets or environment variables.\n"
        "STRICT_REAL_MODE is active: mock execution and synthetic data fallbacks are disabled."
    )
elif not CONFIG.hf_token:
    logger.info(f"HF_TOKEN not set; '{CONFIG.model_key}' does not require authentication and translations are loaded from CSV.")

# Persist the retrieved token in os.environ so downstream cells (E3 gated check,
# tokenizer/model loading for gated repos) read the same value without re-querying
# Kaggle Secrets. Never print the token value.
if CONFIG.hf_token:
    os.environ["HF_TOKEN"] = CONFIG.hf_token
    os.environ["HUGGING_FACE_HUB_TOKEN"] = CONFIG.hf_token

logger.info(f"STRICT_REAL_MODE: {STRICT_REAL_MODE}")
logger.info(f"Active Model Key: {CONFIG.model_key}")
logger.info(f"Backend Engine:   {CONFIG.backend}")
logger.info(f"Sample Count:     {CONFIG.n_samples} per language")
logger.info(f"Dev Set Count:    {CONFIG.dev_n} per language")
# Resolve translation CSV path + sha256 for logging (cell 11's _file_sha256 not yet defined here)
if CONFIG.en_source == "bnmmlu_translation":
    _csv_log_path = None
    _etp_log = getattr(CONFIG, "en_translation_path", None)
    if _etp_log and os.path.exists(_etp_log):
        _csv_log_path = _etp_log
    else:
        import glob as _glob_log
        _found_log = sorted(set(
            _glob_log.glob("/kaggle/input/*/bnmmlu_full_en*.csv")
            + _glob_log.glob("/kaggle/input/*/*/bnmmlu_full_en*.csv")
            + [p for p in ("bnmmlu_full_en.csv", "data/bnmmlu_full_en.csv") if os.path.exists(p)]
        ))
        if _found_log:
            _csv_log_path = _found_log[0]
    if _csv_log_path:
        _h_log = _hashlib.sha256()
        with open(_csv_log_path, "rb") as _f:
            for _chunk in iter(lambda: _f.read(1 << 20), b""):
                _h_log.update(_chunk)
        _csv_log_sha = _h_log.hexdigest()
        logger.info(f"BnMMLU source: {_csv_log_path} (sha256={_csv_log_sha})")
        globals()["_E3_CSV_PATH"] = _csv_log_path
        globals()["_E3_CSV_SHA256"] = _csv_log_sha
    else:
        logger.info(f"BnMMLU source: {CONFIG.bn_local_path or (CONFIG.bn_dataset_id + ' [' + CONFIG.bn_hf_config + ']')}")
else:
    logger.info(f"BnMMLU source: {CONFIG.bn_local_path or (CONFIG.bn_dataset_id + ' [' + CONFIG.bn_hf_config + ']')}")
if CONFIG.en_source == "bnmmlu_translation":
    logger.info(f"English source:   BnMMLU translation (item-parallel) | csv={CONFIG.en_translation_path or 'auto-find under /kaggle/input'} | excluded subjects={list(CONFIG.translation_exclude_subjects)}")
else:
    logger.info(f"MMLU source:      {CONFIG.en_dataset_id} ['{CONFIG.en_hf_config}', split='{CONFIG.en_split}']")
logger.info(f"Markup excluded:  {CONFIG.exclude_markup} | Subject matching: {CONFIG.subject_match_mode}")
logger.info(f"Output dir:       {CONFIG.output_dir}")
logger.info(f"Prompt variant:   {CONFIG.prompt_variant}")
# ==========================================
# GATE CONFIGURATION (notebook defaults + optional gate.yaml override)
# ==========================================
# Defaults (GATE_DEFAULTS) are defined in Section 2, so GATE_CFG always exists. If a gate.yaml is found
# (GATE_YAML_PATH env var, GATE_CONFIG_PATH, config/, code/config/, /kaggle/working/config/, or a Kaggle input dataset),
# its values override the defaults after strict validation. A missing file is NOT an error outside Kaggle; the source is logged either way.
import glob as _glob
import math as _math
import yaml as _yaml

def validate_gate_dict(loaded: Any, source: str = "gate.yaml") -> Dict[str, float]:
    """Strictly validate GATE_CFG mapping per pre-registered research protocol."""
    if not isinstance(loaded, dict):
        raise TypeError(f"{source} must be a dictionary/mapping, got {type(loaded).__name__}")
    req_keys = ("top_share_max", "min_letter_share", "alpha")
    missing = [k for k in req_keys if k not in loaded]
    if missing:
        raise ValueError(f"{source} is missing required gate keys: {missing}")
    out = {}
    for k in req_keys:
        val = loaded[k]
        if isinstance(val, bool):
            raise TypeError(f"Invalid key '{k}' in {source}: boolean values are rejected (got {val!r})")
        if not isinstance(val, (int, float)):
            raise TypeError(f"Invalid key '{k}' in {source}: value must be numeric, got {type(val).__name__} ({val!r})")
        fval = float(val)
        if not _math.isfinite(fval):
            raise ValueError(f"Invalid key '{k}' in {source}: value must be finite, got {fval}")
        out[k] = fval
    t_val = out["top_share_max"]
    if not (0.0 < t_val <= 1.0):
        raise ValueError(f"Invalid key 'top_share_max' in {source}: must be in (0.0, 1.0], got {t_val}")
    m_val = out["min_letter_share"]
    if not (0.0 <= m_val < 0.25):
        raise ValueError(f"Invalid key 'min_letter_share' in {source}: must be in [0.0, 0.25), got {m_val}")
    a_val = out["alpha"]
    if not (0.0 < a_val < 1.0):
        raise ValueError(f"Invalid key 'alpha' in {source}: must be in (0.0, 1.0), got {a_val}")
    return out

_gate_candidates = []
if os.environ.get("GATE_YAML_PATH"):
    _gate_candidates.append(os.environ["GATE_YAML_PATH"])
_gate_candidates.append(GATE_CONFIG_PATH)
_gate_candidates.extend([
    "config/gate.yaml",
    "gate.yaml",
    "../gate.yaml",
    "code/config/gate.yaml",
    "/kaggle/working/config/gate.yaml",
])
if os.path.isdir("/kaggle/input"):
    for _r, _d, _files in os.walk("/kaggle/input"):
        if "gate.yaml" in _files:
            _gate_candidates.append(os.path.join(_r, "gate.yaml"))
_gate_candidates.extend(_glob.glob("/kaggle/input/*/gate.yaml"))
_gate_candidates.extend(_glob.glob("/kaggle/input/*/*/gate.yaml"))
_gate_candidates.extend(_glob.glob("/kaggle/input/*/*/*/gate.yaml"))

_gate_path = next((p for p in _gate_candidates if os.path.exists(p)), None)
GATE_CFG = dict(GATE_DEFAULTS, _source="notebook defaults")
if _gate_path:
    with open(_gate_path, "r", encoding="utf-8") as _gf:
        _loaded = _yaml.safe_load(_gf)
    _validated = validate_gate_dict(_loaded, _gate_path)
    GATE_CFG = dict(_validated, _source=_gate_path)
else:
    logger.warning("gate.yaml not found; using notebook gate defaults. Set GATE_YAML_PATH or attach Kaggle dataset.")

logger.info(f"Gate config source: {GATE_CFG['_source']} | top_share_max={GATE_CFG['top_share_max']}, min_letter_share={GATE_CFG['min_letter_share']}, alpha={GATE_CFG['alpha']}")
```

## Section 4: Dataset Discovery, Parsing, Label Validation & Stratified Sampling
Core data handling functions:
1. **`find_dataset_path`:** Ensures English searches never resolve to a Bangla file; raises a clear error in real mode if `dataset_path` is None (disallowing auto-guessing in production).
2. **`parse_answer_label`:** Disambiguates 0-indexed vs. 1-indexed numeric labels via explicit `index_base` parameter (raising `ValueError` if a bare digit is encountered without an explicit base).
3. **`validate_items`:** Drops items with `len(options) != 4` or invalid gold answer indices. Logs dropped counts and gold distributions. Never scores an invalid item as "wrong."
4. **`sample_stratified_subset`:** Proportional subject sampling with locked seeds.


```python
def find_dataset_path(preferred_path: Optional[str] = None, language: str = "bn", is_mock: bool = False) -> str:
    """Discover dataset file or raise error in real mode to avoid auto-guessing."""
    if preferred_path and os.path.exists(preferred_path):
        return preferred_path

    if STRICT_REAL_MODE or not is_mock:
        raise ValueError(
            f"Explicit 'dataset_path' must be provided for real evaluation (language='{language}'). "
            "Auto-discovery and synthetic fallback are strictly prohibited."
        )

    # Search candidates (Legacy offline development only)
    search_dirs = [
        "/kaggle/input",
        "/kaggle/input/bnmmlu",
        "/kaggle/input/bnmmlu-full",
        "/kaggle/input/bangla-mmlu",
        ".",
        "data"
    ]

    extensions = [".parquet", ".csv", ".jsonl", ".json"]
    for d in search_dirs:
        if not os.path.exists(d):
            continue
        for root, _, files in os.walk(d):
            for f in files:
                f_lower = f.lower()
                if any(f_lower.endswith(ext) for ext in extensions):
                    if language == "en":
                        # CRITICAL FIX: English search must NEVER resolve to a Bangla BnMMLU file
                        if "bnmmlu" in f_lower or "bangla" in f_lower:
                            continue
                        if "mmlu" in f_lower or "english" in f_lower:
                            candidate = os.path.join(root, f)
                            logger.info(f"Discovered candidate English dataset: {candidate}")
                            return candidate
                    else:
                        if any(kw in f_lower for kw in ["bnmmlu", "bangla"]):
                            candidate = os.path.join(root, f)
                            logger.info(f"Discovered candidate Bangla dataset: {candidate}")
                            return candidate

    return "FALLBACK_SYNTHETIC"


def parse_options(options_raw: Any) -> List[str]:
    """Robustly parse options into a list of 4 strings."""
    if options_raw is None:
        return []

    if isinstance(options_raw, (list, tuple, np.ndarray)):
        return [str(x).strip() for x in options_raw]

    if isinstance(options_raw, float) and pd.isna(options_raw):
        return []

    s = str(options_raw).strip()

    # 1. Try JSON
    try:
        obj = json.loads(s)
        if isinstance(obj, list):
            return [str(x).strip() for x in obj]
    except Exception:
        pass

    # 2. Try Python literal eval
    try:
        obj = ast.literal_eval(s)
        if isinstance(obj, list):
            return [str(x).strip() for x in obj]
    except Exception:
        pass

    # 3. Extract quoted strings using regex
    try:
        parts = []
        for m in re.finditer(r"'([^']*)'", s):
            parts.append(m.group(1).strip())
        if not parts:
            for m in re.finditer(r'"([^"]*)"', s):
                parts.append(m.group(1).strip())
        if parts:
            return parts
    except Exception:
        pass

    # 4. Fallback comma split
    try:
        return [p.strip().strip("'").strip('"') for p in s.strip("[]").split(",") if p.strip()]
    except Exception:
        return []


def parse_answer_label(ans_raw: Any, index_base: Optional[int] = None) -> Tuple[str, int]:
    """Parse raw answer into (letter, index).
    
    Supports:
    - 'A', 'B', 'C', 'D' / 'a', 'b', 'c', 'd'
    - Bengali numerals ১, ২, ৩, ৪
    - Digits: Requires explicit index_base (0 for 0-3, 1 for 1-4).
    """
    if ans_raw is None:
        return "", -1

    s = str(ans_raw).strip().upper()

    # Bengali numeral mapping
    bn_map = {"১": "A", "২": "B", "৩": "C", "৪": "D"}
    if s in bn_map:
        letter = bn_map[s]
        return letter, OPTION_LETTERS.index(letter)

    if s in OPTION_LETTERS:
        return s, OPTION_LETTERS.index(s)

    # Integer digits: Must strictly adhere to index_base
    if s in ["0", "1", "2", "3", "4"]:
        if index_base is None:
            raise ValueError(
                f"Ambiguous numeric answer label '{s}'. Must explicitly specify index_base=0 (0-3) or index_base=1 (1-4)."
            )
        val = int(s)
        if index_base == 0:
            if 0 <= val <= 3:
                return OPTION_LETTERS[val], val
            else:
                raise ValueError(f"Value '{s}' out of range for index_base=0 (expected 0-3).")
        elif index_base == 1:
            if 1 <= val <= 4:
                idx = val - 1
                return OPTION_LETTERS[idx], idx
            else:
                raise ValueError(f"Value '{s}' out of range for index_base=1 (expected 1-4).")
        else:
            raise ValueError(f"Invalid index_base={index_base}. Must be 0 or 1.")

    return s, -1


def validate_items(items: List[Dict[str, Any]]) -> List[Dict[str, Any]]:
    """Validate MCQ items: drop items with len(options) != 4 or answer_idx not in {0,1,2,3}.
    Prints dropped item counts with reasons and the gold label distribution.
    Never scores an invalid item as 'wrong'.
    """
    valid_items = []
    dropped_reasons: Dict[str, int] = {}
    gold_distribution: Dict[str, int] = {letter: 0 for letter in OPTION_LETTERS}

    for item in items:
        options = item.get("options", [])
        ans_idx = item.get("answer_idx", -1)
        gold_letter = item.get("correct_answer", "")
        question = str(item.get("question", "")).strip()

        if not question:
            dropped_reasons["empty_question"] = dropped_reasons.get("empty_question", 0) + 1
            continue

        if not isinstance(options, list) or len(options) != 4:
            count_key = f"invalid_options_count_{len(options) if isinstance(options, list) else 'non_list'}"
            dropped_reasons[count_key] = dropped_reasons.get(count_key, 0) + 1
            continue

        if any(not str(opt).strip() for opt in options):
            dropped_reasons["empty_option_string"] = dropped_reasons.get("empty_option_string", 0) + 1
            continue

        if ans_idx not in [0, 1, 2, 3] or gold_letter not in OPTION_LETTERS:
            dropped_reasons["invalid_answer_label"] = dropped_reasons.get("invalid_answer_label", 0) + 1
            continue

        valid_items.append(item)
        gold_distribution[gold_letter] += 1

    total_input = len(items)
    total_valid = len(valid_items)
    total_dropped = total_input - total_valid

    logger.info(f"Data Validation: {total_valid:,} valid items retained ({total_dropped:,} dropped).")
    if dropped_reasons:
        logger.info(f"Dropped items breakdown: {dropped_reasons}")
    logger.info(f"Gold label distribution: {gold_distribution}")

    return valid_items


def create_synthetic_benchmark_dataset() -> List[Dict[str, Any]]:
    """Create synthetic MCQ dataset in Bangla and English for testing without external files."""
    synthetic_items = [
        # Bangla items across varied subjects
        {"question_id": "SYN_BN_001", "subject": "Financial Accounting", "question": "প্রকৃত জাবেদায় নিচের কোনটি লিপিবদ্ধ করা হয়?", "options": ["ক্রয় জাবেদা", "বিক্রয় জাবেদা", "সমন্বয় জাবেদা", "বিশেষ জাবেদা"], "correct_answer": "C", "answer_idx": 2, "language": "bn", "benchmark": "BnMMLU"},
        {"question_id": "SYN_BN_002", "subject": "Physics", "question": "কোন মাধ্যমের মধ্য দিয়ে শব্দ দ্রুততম গতিতে সঞ্চালিত হয়?", "options": ["বায়বীয় মাধ্যম", "কঠিন মাধ্যম", "তরল মাধ্যম", "শূন্যস্থান"], "correct_answer": "B", "answer_idx": 1, "language": "bn", "benchmark": "BnMMLU"},
        {"question_id": "SYN_BN_003", "subject": "History", "question": "ঐতিহাসিক ভাষা আন্দোলনের চূড়ান্ত পর্যায় কোন সালে সংঘটিত হয়েছিল?", "options": ["১৯৪৭", "১৯৫২", "১৯৬৯", "১৯৭১"], "correct_answer": "B", "answer_idx": 1, "language": "bn", "benchmark": "BnMMLU"},
        {"question_id": "SYN_BN_004", "subject": "Computer Science", "question": "বাইনারি সার্চ ট্রিতে উপাদান অনুসন্ধানের গড় সময় জটিলতা কোনটি?", "options": ["O(1)", "O(n)", "O(log n)", "O(n log n)"], "correct_answer": "C", "answer_idx": 2, "language": "bn", "benchmark": "BnMMLU"},
        {"question_id": "SYN_BN_005", "subject": "General Science", "question": "উদ্ভিদ কোষে সালোকসংশ্লেষণ প্রক্রিয়ার জন্য প্রধানত কোন অঙ্গাণু দায়ী?", "options": ["মাইটোকন্ড্রিয়া", "রাইবোসোম", "প্লাস্টিড বা ক্লোরোপ্লাস্ট", "গলগি বডি"], "correct_answer": "C", "answer_idx": 2, "language": "bn", "benchmark": "BnMMLU"},
        {"question_id": "SYN_BN_006", "subject": "Bengali Grammar", "question": "'সন্ধি' ব্যাকরণের কোন অংশে আলোচিত হয়?", "options": ["রূপতত্ত্বে", "ধ্বনিযোগে ধ্বনিতত্ত্বে", "বাক্যতত্ত্বে", "অর্থতত্ত্বে"], "correct_answer": "B", "answer_idx": 1, "language": "bn", "benchmark": "BnMMLU"},
        {"question_id": "SYN_BN_007", "subject": "Geography", "question": "পৃথিবীর সর্ববৃহৎ ব-দ্বীপ অঞ্চল কোনটি?", "options": ["নীলনদের ব-দ্বীপ", "আমাজন ব-দ্বীপ", "গঙ্গা-ব্রহ্মপুত্র ব-দ্বীপ", "মিসিসিপি ব-দ্বীপ"], "correct_answer": "C", "answer_idx": 2, "language": "bn", "benchmark": "BnMMLU"},
        {"question_id": "SYN_BN_008", "subject": "Chemistry", "question": "সাধারণ তাপমাত্রায় নিচের কোন ধাতুটি তরল অবস্থায় থাকে?", "options": ["সোডিয়াম", "পারদ", "লোহা", "অ্যালুমিনিয়াম"], "correct_answer": "B", "answer_idx": 1, "language": "bn", "benchmark": "BnMMLU"},
        # English matched control items
        {"question_id": "SYN_EN_001", "subject": "Physics", "question": "Through which medium does sound travel at the highest speed?", "options": ["Gas", "Solid", "Liquid", "Vacuum"], "correct_answer": "B", "answer_idx": 1, "language": "en", "benchmark": "MMLU"},
        {"question_id": "SYN_EN_002", "subject": "Computer Science", "question": "What is the average time complexity of searching in a balanced BST?", "options": ["O(1)", "O(n)", "O(log n)", "O(n log n)"], "correct_answer": "C", "answer_idx": 2, "language": "en", "benchmark": "MMLU"},
        {"question_id": "SYN_EN_003", "subject": "General Science", "question": "Which organelle is responsible for photosynthesis in plants?", "options": ["Mitochondria", "Ribosome", "Chloroplast", "Golgi apparatus"], "correct_answer": "C", "answer_idx": 2, "language": "en", "benchmark": "MMLU"},
        {"question_id": "SYN_EN_004", "subject": "Chemistry", "question": "Which metal remains liquid at room temperature?", "options": ["Sodium", "Mercury", "Iron", "Aluminum"], "correct_answer": "B", "answer_idx": 1, "language": "en", "benchmark": "MMLU"},
    ]
    for idx, item in enumerate(synthetic_items):
        item["raw_index"] = idx
    return synthetic_items


# ---------------------------------------------------------------------------
# MathML / markup detection
# BnMMLU stores equations as MathML (outer <math> tags removed, inner tags kept). Those tags are not Bangla,
# but a tokenizer still splits them into many pieces, which would be mis-measured as "fragmentation".
# ---------------------------------------------------------------------------
MATHML_TAG_RE = re.compile(
    r"</?(?:math|mi|mn|mo|mrow|msup|msub|msubsup|mfrac|msqrt|mroot|mtext|mstyle|mover|munder|munderover|"
    r"mtable|mtr|mtd|mspace|mfenced|menclose|mpadded|mphantom|semantics|annotation)(?:\s[^<>]*)?/?>"
    r"|&lt;/?m(?:i|n|o|row|sup|sub|frac|sqrt|text)\b",
    re.IGNORECASE,
)


def contains_markup(question: str, options: List[str]) -> bool:
    """True if the question or any option contains MathML tags."""
    blob = str(question or "") + " " + " ".join(str(o or "") for o in (options or []))
    return bool(MATHML_TAG_RE.search(blob))


# ---------------------------------------------------------------------------
# Subject normalisation and the BnMMLU -> MMLU subject map
# ---------------------------------------------------------------------------
def normalize_subject(s: Any) -> str:
    """Lower-case, underscores to spaces, strip punctuation. Non-Latin names fall back to the stripped original."""
    raw = str(s or "").strip()
    norm = re.sub(r"[^a-z0-9 ]+", "", raw.lower().replace("_", " ").replace("-", " "))
    norm = re.sub(r"\s+", " ", norm).strip()
    return norm if norm else raw.lower()


# RESEARCH-DESIGN CHOICE (review and justify this in the paper): BnMMLU is built from Bangladeshi school/college and
# competitive-exam material, MMLU from US-oriented exams, so "subject matching" is approximate. Keys are normalised
# BnMMLU subject names (Table 5 of the BnMMLU paper); values are exact MMLU subject ids. Subjects with no MMLU
# analogue (Bengali, Agricultural Studies, General Knowledge, Finance, Science, ...) are deliberately unmapped.
SUBJECT_MAP_BN_TO_MMLU: Dict[str, List[str]] = {
    "accounting": ["professional_accounting"],
    "advanced mathematics": ["college_mathematics", "high_school_mathematics"],
    "general mathematics": ["elementary_mathematics", "high_school_mathematics"],
    "mathematics": ["elementary_mathematics", "high_school_mathematics"],
    "advanced physics": ["college_physics", "high_school_physics"],
    "general physics": ["conceptual_physics", "high_school_physics"],
    "physics": ["conceptual_physics", "high_school_physics"],
    "biology": ["high_school_biology", "college_biology"],
    "chemistry": ["high_school_chemistry", "college_chemistry"],
    "statistics": ["high_school_statistics"],
    "ict": ["high_school_computer_science", "college_computer_science"],
    "information and communication technology": ["high_school_computer_science", "college_computer_science"],
    "economics": ["high_school_macroeconomics", "high_school_microeconomics", "econometrics"],
    "geography": ["high_school_geography"],
    "psychology": ["high_school_psychology", "professional_psychology"],
    "logic": ["formal_logic", "logical_fallacies"],
    "civics and citizenship": ["high_school_government_and_politics"],
    "religion and moral education": ["world_religions", "philosophy"],
    "religious studies": ["world_religions", "philosophy"],
    "business entrepreneurship": ["management", "marketing", "business_ethics"],
    "production management": ["management"],
    "social work": ["sociology"],
}

# RESEARCH-DESIGN CHOICE: approximate subject matching between Bangladeshi
# curriculum (BnMMLU) and US-oriented MMLU; justify in the paper. Subjects with no MMLU analogue
# (Bengali language/literature/poetry, agriculture, history/culture, banking, entrepreneurship,
# production, general knowledge, general science, miscellaneous GK) are deliberately unmapped.
SUBJECT_MAP_EXTRA: Dict[str, List[str]] = {
    "cell biology genetics": ["high_school_biology", "college_biology", "medical_genetics"],
    "human biology anatomy": ["anatomy", "human_aging", "clinical_knowledge"],
    "ecology environmental biology": ["high_school_biology"],
    "organic chemistry": ["high_school_chemistry", "college_chemistry"],
    "inorganic chemistry": ["high_school_chemistry", "college_chemistry"],
    "physical analytical chemistry": ["high_school_chemistry", "college_chemistry"],
    "conceptual physics basic laws": ["conceptual_physics", "high_school_physics"],
    "mechanics": ["high_school_physics", "college_physics"],
    "thermodynamics electromagnetism": ["college_physics", "high_school_physics"],
    "relativity modern physics": ["astronomy", "college_physics"],
    "algebra number theory": ["abstract_algebra", "high_school_mathematics"],
    "calculus analysis": ["college_mathematics", "high_school_mathematics"],
    "statistics probability inference": ["high_school_statistics"],
    "programming algorithms": ["high_school_computer_science", "college_computer_science"],
    "networking security": ["computer_security"],
    "ai data science basics": ["machine_learning"],
    "cognitive psychology": ["high_school_psychology", "professional_psychology"],
    "behavioral psychology": ["high_school_psychology", "professional_psychology"],
    "civics governance": ["high_school_government_and_politics"],
    "financial accounting": ["professional_accounting"],
    "corporate finance": ["high_school_macroeconomics", "high_school_microeconomics"],
    "business strategy management": ["management", "marketing"],
    "comparative religion": ["world_religions"],
    "moral ethical studies": ["moral_disputes", "philosophy"],
    "critical thinking": ["logical_fallacies", "formal_logic"],
    "social work welfare": ["sociology"],
}


def build_subject_map() -> Dict[str, List[str]]:
    """Merged map with normalised BnMMLU keys and normalised MMLU values."""
    merged: Dict[str, List[str]] = {}
    for src in (SUBJECT_MAP_BN_TO_MMLU, SUBJECT_MAP_EXTRA):
        for k, v in src.items():
            merged[normalize_subject(k)] = [normalize_subject(x) for x in v]
    return merged


def items_from_dataframe(
    df: pd.DataFrame,
    language: str = "bn",
    benchmark: str = "BnMMLU",
    index_base: Optional[int] = None,
) -> List[Dict[str, Any]]:
    """Harmonise a DataFrame (from the Hub, parquet, csv or json) into MCQ item dicts."""
    Q_KEYS = ["question", "question_bn", "question_en"]
    O_KEYS = ["options", "options_bn", "options_en", "choices"]
    A_KEYS = ["correct_answer", "answer", "answer_idx", "correct"]
    for name, keys in (("question", Q_KEYS), ("options", O_KEYS), ("answer", A_KEYS)):
        if not any(k in df.columns for k in keys):
            raise ValueError(f"No {name} column found (looked for {keys}). Columns present: {list(df.columns)}")

    def _get_row_field(row, keys, default=None):
        for k in keys:
            if k in row:
                val = row[k]
                if val is not None:
                    if isinstance(val, float) and pd.isna(val):
                        continue
                    return val
        return default

    items = []
    for idx, row in df.iterrows():
        qid = _get_row_field(row, ["Unique_Serial", "question_id", "id", "item_id"], str(idx))
        subject = _get_row_field(row, ["subject_name", "subject", "category"], "general")
        question = _get_row_field(row, Q_KEYS, "")
        raw_options = _get_row_field(row, O_KEYS, [])
        raw_ans = _get_row_field(row, A_KEYS, "")
        subject_group = _get_row_field(row, ["subject_group"], None)

        options = parse_options(raw_options)
        correct_letter, answer_idx = parse_answer_label(raw_ans, index_base=index_base)

        items.append({
            "question_id": str(qid),
            "subject": str(subject),
            "subject_group": (str(subject_group) if subject_group is not None else normalize_subject(subject)),
            "question": str(question),
            "options": options,
            "correct_answer": correct_letter,
            "answer_idx": answer_idx,
            "language": language,
            "benchmark": benchmark,
            "has_markup": contains_markup(str(question), options),
            "raw_index": int(idx) if isinstance(idx, (int, np.integer)) else len(items),
        })

    # Duplicate question_ids would silently break resume logic and merges: make them unique
    ids = [it["question_id"] for it in items]
    if len(set(ids)) != len(ids):
        n_dup = len(ids) - len(set(ids))
        logger.warning(f"{n_dup:,} duplicate question_ids detected. Appending raw_index to make IDs unique.")
        for it in items:
            it["question_id"] = f"{it['question_id']}_{it['raw_index']}"

    logger.info(f"Harmonised {len(items):,} {benchmark} items ({sum(it['has_markup'] for it in items):,} contain MathML markup).")
    return items


def load_dataset_from_file(
    file_path: str,
    language: str = "bn",
    benchmark: str = "BnMMLU",
    index_base: Optional[int] = None,
    is_mock: bool = False,
) -> List[Dict[str, Any]]:
    """Load and harmonize dataset from Parquet, CSV, or JSONL. Synthetic data fallback is strictly prohibited."""
    if STRICT_REAL_MODE or not is_mock:
        if file_path == "FALLBACK_SYNTHETIC" or not os.path.exists(file_path):
            raise FileNotFoundError(
                f"Dataset file '{file_path}' not found. STRICT_REAL_MODE is active and authentic benchmark data is required. "
                "Silent fallback to synthetic or mock data is strictly prohibited."
            )
    else:
        if file_path == "FALLBACK_SYNTHETIC" or not os.path.exists(file_path):
            logger.info(f"Isolated unit test: Using synthetic fixtures (language='{language}').")
            items = [dict(it) for it in create_synthetic_benchmark_dataset() if it["language"] == language]
            for it in items:
                it.setdefault("subject_group", normalize_subject(it["subject"]))
                it.setdefault("has_markup", False)
            return items

    logger.info(f"Loading dataset from disk: {file_path}")
    if file_path.endswith(".parquet"):
        df = pd.read_parquet(file_path)
    elif file_path.endswith(".csv"):
        df = pd.read_csv(file_path, encoding="utf-8-sig")  # utf-8-sig strips the UTF-8 BOM present in bnmmlu_full_en.csv
    elif file_path.endswith(".json") or file_path.endswith(".jsonl"):
        df = pd.read_json(file_path, lines=file_path.endswith(".jsonl"))
    else:
        raise ValueError(f"Unsupported file format for: {file_path}")
    return items_from_dataframe(df, language=language, benchmark=benchmark, index_base=index_base)


def sample_stratified_subset(
    items: List[Dict[str, Any]],
    n_samples: int,
    seed: int = DEFAULT_SEED,
) -> List[Dict[str, Any]]:
    """Reproducibly extract a stratified random sample across benchmark subjects."""
    if n_samples <= 0 or n_samples >= len(items):
        return items

    rng = np.random.RandomState(seed)
    by_subject: Dict[str, List[int]] = {}
    for i, it in enumerate(items):
        subj = it.get("subject", "general")
        by_subject.setdefault(subj, []).append(i)

    total_len = len(items)
    selected_indices: List[int] = []

    for subj, indices in by_subject.items():
        prop = len(indices) / total_len
        k = int(round(prop * n_samples))
        if k > 0 and indices:
            chosen = rng.choice(indices, size=min(k, len(indices)), replace=False).tolist()
            selected_indices.extend(chosen)

    # Adjust rounding differences
    if len(selected_indices) < n_samples:
        remaining = list(set(range(len(items))) - set(selected_indices))
        if remaining:
            needed = min(n_samples - len(selected_indices), len(remaining))
            extra = rng.choice(remaining, size=needed, replace=False).tolist()
            selected_indices.extend(extra)
    elif len(selected_indices) > n_samples:
        selected_indices = rng.choice(selected_indices, size=n_samples, replace=False).tolist()

    selected_indices.sort()
    return [items[i] for i in selected_indices]
```

## Section 5: Data Preparation (BnMMLU + English control: item-parallel translation or subject-matched MMLU)
Prepares the standardized subsets (`bn_subset.csv`, `en_subset.csv`) of up to $N=1500$ items per language:
1. **BnMMLU:** `load_dataset("samanjoy2/BnMMLU", "full", token=...)` (gated). Items are validated (4 options, valid answer), optionally stripped of MathML items, and stratified-sampled with seed 42.
2. **MMLU:** `cais/mmlu`, config `all`, split `test`. English subjects are matched to BnMMLU subjects through the explicit `SUBJECT_MAP_BN_TO_MMLU` (Section 4); the matched BnMMLU label is stored as `subject_group` in **both** files so the paired Bangla-vs-English bootstrap works.
3. **Provenance:** `subset_meta.json` records the Hub commit SHAs of both datasets, `subject_mapping.json` records the per-subject mapping and which BnMMLU subjects have no MMLU analogue, and `sampled_question_ids.json` locks the exact items for every model.

> [!NOTE]
> **Two English-control modes (`CONFIG.en_source`).**
> - `"bnmmlu_translation"` (**current default of `CONFIG`**): item-parallel. Both languages are read from one wide CSV (`bnmmlu_full_en.csv`: `question_bn/question_en`, `option_*_bn/_en`). Every evaluation, dev and few-shot item exists in both languages with the **same `question_id`**. Filters (all counted in `subset_meta.json`): language-specific subjects (`CONFIG.translation_exclude_subjects`), MathML in either language, Bangla script left in the English side, untranslated (identical) questions, exact Bangla duplicates. Sampling is balanced across subjects (cap 1.5x equal share). The English text is a **translation (machine or otherwise) of Bangla-origin content**; its quality is not verified here. Report that as a limitation and spot-check a sample by hand.
> - `"mmlu"`: the original subject-matched control, described next. It is not item-parallel.
>
> **Original control framing (`en_source="mmlu"`):** the English subset is a **subject-matched control**, not an item-parallel translation. Subject matching is approximate (Bangladeshi school/exam content vs. US-oriented MMLU) and is a design choice to justify in the paper.
>
> The Hub card marks this BnMMLU release as *preliminary / non-final*. Report the recorded commit SHA.


```python
import hashlib


def _hub_revision(repo_id: str, repo_type: str, token: Optional[str]) -> str:
    """Commit SHA of a Hub repo, recorded so the exact data version can be reported in the paper."""
    try:
        from huggingface_hub import HfApi
        api = HfApi(token=token)
        info = api.dataset_info(repo_id) if repo_type == "dataset" else api.model_info(repo_id)
        return str(info.sha)
    except Exception as e:
        return f"unavailable ({type(e).__name__})"


def _explain_hub_failure(repo_id: str, err: Exception) -> RuntimeError:
    msg = str(err)
    hint = ""
    low = msg.lower()
    if "gated" in low or "401" in low or "403" in low or "access" in low or "restricted" in low or "authenticat" in low:
        hint = (f"\n=> '{repo_id}' is gated. While logged in to Hugging Face, open https://huggingface.co/datasets/{repo_id} "
                "and accept the access conditions, then make sure HF_TOKEN is a valid READ token for the SAME account.")
    elif "config" in low:
        hint = f"\n=> '{repo_id}' needs an explicit config name. Set CONFIG.bn_hf_config to 'full' or 'hard'."
    return RuntimeError(f"Could not load '{repo_id}' from the Hugging Face Hub: {type(err).__name__}: {msg}{hint}")


def load_bnmmlu_items(
    config: "ExperimentConfig",
    hf_token: Optional[str],
    is_mock: bool,
    bn_index_base: Optional[int],
) -> Tuple[List[Dict[str, Any]], Dict[str, Any]]:
    """Load the REAL BnMMLU (Hub config 'full' by default) or a local copy. Strict real execution enforced."""
    if STRICT_REAL_MODE or not is_mock:
        if is_mock:
            raise RuntimeError("Mock mode is disabled when STRICT_REAL_MODE=True. Authentic BnMMLU dataset is required.")
    else:
        items = load_dataset_from_file("FALLBACK_SYNTHETIC", "bn", "BnMMLU", is_mock=True)
        return items, {"source": "synthetic_mock"}

    if config.bn_local_path:
        if not os.path.exists(config.bn_local_path):
            raise FileNotFoundError(f"CONFIG.bn_local_path does not exist: {config.bn_local_path}")
        items = load_dataset_from_file(config.bn_local_path, "bn", "BnMMLU", index_base=bn_index_base, is_mock=False)
        return items, {"source": f"local:{config.bn_local_path}"}

    try:
        from datasets import load_dataset
        tok = {"token": hf_token} if hf_token else {}
        logger.info(f"Downloading BnMMLU from the Hub: {config.bn_dataset_id} (config='{config.bn_hf_config}')...")
        dd = load_dataset(config.bn_dataset_id, config.bn_hf_config, **tok)
    except Exception as e:
        raise _explain_hub_failure(config.bn_dataset_id, e)

    splits = list(dd.keys())
    if "test" in splits:
        use = ["test"]
    elif len(splits) == 1:
        use = splits
    else:
        use = splits  # several splits and no 'test': use all rows (zero-shot evaluation needs no train/test separation)
    logger.info(f"BnMMLU splits available: {splits} -> using {use}")
    df = pd.concat([dd[s].to_pandas() for s in use], ignore_index=True)
    logger.info(f"BnMMLU columns: {list(df.columns)} | rows: {len(df):,}")
    logger.info(f"BnMMLU distinct subjects: {df['subject_name'].nunique() if 'subject_name' in df.columns else 'n/a'}")
    items = items_from_dataframe(df, language="bn", benchmark="BnMMLU", index_base=bn_index_base)
    return items, {"source": f"hub:{config.bn_dataset_id}:{config.bn_hf_config}", "splits_used": use, "n_rows": int(len(df))}


def load_mmlu_items(config: "ExperimentConfig", hf_token: Optional[str], is_mock: bool, n_items: int) -> List[Dict[str, Any]]:
    """Load the REAL English MMLU (cais/mmlu, config 'all', split 'test'). Strict real execution enforced."""
    if STRICT_REAL_MODE or not is_mock:
        if is_mock:
            raise RuntimeError("Mock mode is disabled when STRICT_REAL_MODE=True. Authentic MMLU dataset is required.")
    else:
        synth_en = load_dataset_from_file("FALLBACK_SYNTHETIC", "en", "MMLU", is_mock=True)
        return [dict(it, question_id=f"{it['question_id']}_{k}") for k in range(n_items // 4 + 1) for it in synth_en]

    try:
        from datasets import load_dataset
        tok = {"token": hf_token} if hf_token else {}
        logger.info(f"Downloading English MMLU from the Hub: {config.en_dataset_id} ('{config.en_hf_config}', split='{config.en_split}')...")
        hf_en = load_dataset(config.en_dataset_id, config.en_hf_config, split=config.en_split, **tok)
    except Exception as e:
        raise _explain_hub_failure(config.en_dataset_id, e)

    items = []
    for idx, row in enumerate(hf_en):
        choices = [str(c) for c in row["choices"]]
        ans = int(row["answer"])
        subj = str(row.get("subject", "general"))
        q = str(row["question"])
        items.append({
            "question_id": f"MMLU_EN_{idx:05d}",
            "subject": subj,
            "subject_group": normalize_subject(subj),
            "question": q,
            "options": choices,
            "correct_answer": OPTION_LETTERS[ans] if 0 <= ans < 4 else "",
            "answer_idx": ans,
            "language": "en",
            "benchmark": "MMLU",
            "has_markup": contains_markup(q, choices),
            "raw_index": idx,
        })
    logger.info(f"Loaded {len(items):,} English MMLU items across {len({it['subject'] for it in items})} subjects.")
    return items



EXEMPLAR_VERSION = "v3_1_fewshot_v2"

def select_fewshot_exemplars(
    candidates: List[Dict[str, Any]],
    eval_ids: Union[set, List[str]],
    language: str,
    seed: int = DEFAULT_SEED,
) -> List[Dict[str, Any]]:
    """Select exactly 8 balanced few-shot exemplars disjoint from evaluation items.

    Each letter A-D appears exactly twice across the 8 exemplars, and any prefix of 4
    contains each letter A-D exactly once.
    Constraints on each exemplar:
      - 4 options
      - Valid gold label in {'A', 'B', 'C', 'D'}
      - has_markup is False (no MathML)
      - total character length (question + joined options) <= 350
      - question_id not in eval_ids
      - distinct subjects where possible
      - deterministic selection using fixed seed
    """
    eval_set = set(str(qid) for qid in eval_ids)

    # Filter eligible items
    eligible: List[Dict[str, Any]] = []
    for it in candidates:
        qid = str(it.get("question_id", ""))
        if qid in eval_set:
            continue
        gold = str(it.get("correct_answer", "")).upper()
        if gold not in OPTION_LETTERS:
            continue
        opts = it.get("options", [])
        if not isinstance(opts, list) or len(opts) != 4:
            continue
        if it.get("has_markup", False):
            continue
        q_text = str(it.get("question", ""))
        opts_text = "".join(str(o) for o in opts)
        if len(q_text) + len(opts_text) > 350:
            continue
        eligible.append(it)

    rng = np.random.RandomState(seed)
    shuffled_idx = rng.permutation(len(eligible))
    shuffled = [eligible[i] for i in shuffled_idx]

    # Pool by gold letter
    by_letter: Dict[str, List[Dict[str, Any]]] = {letter: [] for letter in OPTION_LETTERS}
    for it in shuffled:
        by_letter[str(it.get("correct_answer", "")).upper()].append(it)

    # Group 1: 1 of each letter A-D
    # Group 2: 1 of each letter A-D
    # Select with subject diversity where possible
    chosen_1: List[Dict[str, Any]] = []
    chosen_2: List[Dict[str, Any]] = []
    used_ids: set = set()
    used_subjects: set = set()

    for letter in OPTION_LETTERS:
        pool = by_letter[letter]
        pick = None
        # Try distinct subject
        for it in pool:
            subj = it.get("subject", "")
            if subj not in used_subjects:
                pick = it
                break
        if pick is None and pool:
            pick = pool[0]
        if pick is None:
            raise RuntimeError(f"Cannot build balanced few-shot set for {language}: insufficient items for letter {letter}")
        chosen_1.append(pick)
        used_ids.add(str(pick.get("question_id", "")))
        if pick.get("subject"):
            used_subjects.add(pick.get("subject"))

    for letter in OPTION_LETTERS:
        pool = [it for it in by_letter[letter] if str(it.get("question_id", "")) not in used_ids]
        pick = None
        for it in pool:
            subj = it.get("subject", "")
            if subj not in used_subjects:
                pick = it
                break
        if pick is None and pool:
            pick = pool[0]
        if pick is None:
            raise RuntimeError(f"Cannot build balanced few-shot set for {language}: insufficient items for second pass of letter {letter}")
        chosen_2.append(pick)
        used_ids.add(str(pick.get("question_id", "")))
        if pick.get("subject"):
            used_subjects.add(pick.get("subject"))

    # Verify no eval_ids overlap
    overlap = used_ids.intersection(eval_set)
    if overlap:
        raise RuntimeError(f"Few-shot exemplars overlap with evaluation items: {overlap}")

    # Build standardized exemplar dicts
    def _standardize(it):
        return {
            "question_id": str(it.get("question_id", "")),
            "subject": it.get("subject", ""),
            "question": str(it.get("question", "")),
            "options": [str(o) for o in it.get("options", [])],
            "correct_answer": str(it.get("correct_answer", "")).upper(),
            "source": it.get("benchmark", language),
        }

    return [_standardize(it) for it in chosen_1] + [_standardize(it) for it in chosen_2]

# ==========================================================================================
# TRANSLATION-PARALLEL MODE  (CONFIG.en_source == "bnmmlu_translation")
# The English side is the English translation of the SAME BnMMLU questions (one wide CSV with
# *_bn and *_en columns), instead of a subject-matched sample from cais/mmlu. Every evaluation,
# dev and few-shot item therefore exists in both languages with the same question_id.
# ==========================================================================================
TRANSLATION_FILTER_VERSION = "v1_translation_parallel"
TRANSLATION_SAMPLING_VERSION = "v4e_translation_parallel"
_BN_SCRIPT_RE = re.compile(r"[\u0980-\u09FF]")
_TRANSLATION_REQUIRED_COLS = (
    ["Unique_Serial", "subject_name", "correct_answer", "question_en", "question_bn"]
    + [f"option_{c}_{lang}" for c in "abcd" for lang in ("en", "bn")]
)


def en_benchmark_label(config: Optional["ExperimentConfig"] = None) -> str:
    """Benchmark label stored with English rows: 'BnMMLU-en' in translation mode, else 'MMLU'."""
    config = config or CONFIG
    return "BnMMLU-en" if getattr(config, "en_source", "mmlu") == "bnmmlu_translation" else "MMLU"


def _file_sha256(path: str) -> str:
    h = hashlib.sha256()
    with open(path, "rb") as f:
        for chunk in iter(lambda: f.read(1 << 20), b""):
            h.update(chunk)
    return h.hexdigest()


def resolve_translation_csv_path(config: "ExperimentConfig") -> str:
    """Find the wide BnMMLU bn/en CSV. An explicit CONFIG.en_translation_path always wins."""
    import glob as _glob
    explicit = getattr(config, "en_translation_path", None)
    if explicit:
        if not os.path.exists(explicit):
            raise FileNotFoundError(f"CONFIG.en_translation_path does not exist: {explicit}")
        return explicit
    found = sorted(set(
        _glob.glob("/kaggle/input/*/bnmmlu_full_en*.csv")
        + _glob.glob("/kaggle/input/*/*/bnmmlu_full_en*.csv")
        + [p for p in ("bnmmlu_full_en.csv", "data/bnmmlu_full_en.csv") if os.path.exists(p)]
    ))
    if not found:
        raise FileNotFoundError(
            "CONFIG.en_source='bnmmlu_translation' needs the wide BnMMLU bn/en CSV (columns question_bn, question_en, "
            "option_a_bn ... option_d_en). Attach it as a Kaggle dataset and set CONFIG.en_translation_path "
            "(e.g. '/kaggle/input/<your-dataset>/bnmmlu_full_en.csv')."
        )
    if len(found) > 1 and len({_file_sha256(p) for p in found}) > 1:
        raise ValueError(f"Several different candidate translation CSVs found: {found}. Set CONFIG.en_translation_path explicitly.")
    logger.info(f"Translation CSV auto-resolved to: {found[0]} (candidates: {found})")
    return found[0]


def load_translation_parallel_pairs(
    csv_path: str, config: "ExperimentConfig"
) -> Tuple[List[Dict[str, Any]], List[Dict[str, Any]], Dict[str, Any]]:
    """Read the wide CSV, apply quality filters, and return index-aligned (bn_items, en_items, report)."""
    df = pd.read_csv(csv_path, dtype=str, keep_default_na=False, encoding="utf-8-sig")  # utf-8-sig strips the UTF-8 BOM; keep_default_na: option text "None" must stay text
    missing = [c for c in _TRANSLATION_REQUIRED_COLS if c not in df.columns]
    if missing:
        raise ValueError(f"Translation CSV {csv_path} is missing required columns: {missing}")

    text_cols = [c for c in _TRANSLATION_REQUIRED_COLS if c.startswith(("question_", "option_"))]
    en_cols = ["question_en"] + [f"option_{c}_en" for c in "abcd"]
    report: Dict[str, Any] = {"n_raw": int(len(df)), "dropped": {}, "excluded_subjects_found": {}}

    def _drop(mask: "pd.Series", reason: str) -> None:
        nonlocal df
        n = int(mask.sum())
        if n:
            report["dropped"][reason] = n
            df = df[~mask]

    _drop((df[text_cols].apply(lambda s: s.str.strip()) == "").any(axis=1), "empty_text_field")
    _drop(~df["correct_answer"].str.strip().str.lower().isin(list("abcd")), "invalid_answer_label")
    _drop(df["Unique_Serial"].str.strip().duplicated(), "duplicate_unique_serial")

    excl = {normalize_subject(s) for s in getattr(config, "translation_exclude_subjects", ())}
    subj_norm = df["subject_name"].map(normalize_subject)
    if excl:
        report["excluded_subjects_found"] = {k: int(v) for k, v in df.loc[subj_norm.isin(excl), "subject_name"].value_counts().items()}
    _drop(subj_norm.isin(excl), "excluded_language_specific_subject")

    if config.exclude_markup:
        blob = df[text_cols].agg(" ".join, axis=1)
        _drop(blob.map(lambda s: bool(MATHML_TAG_RE.search(s))), "mathml_markup_either_language")

    _drop(df["question_bn"].str.strip() == df["question_en"].str.strip(), "untranslated_identical_question")
    en_blob = df[en_cols].agg(" ".join, axis=1)
    _drop(en_blob.map(lambda s: bool(_BN_SCRIPT_RE.search(s))), "bangla_script_left_in_english_side")
    bn_key = df[["question_bn"] + [f"option_{c}_bn" for c in "abcd"] + ["correct_answer"]].apply(lambda s: s.str.strip()).agg("\x1f".join, axis=1)
    _drop(bn_key.duplicated(), "duplicate_bn_item")

    report["n_after_filters"] = int(len(df))

    def _mk(r: Dict[str, str], lang: str, idx: int) -> Dict[str, Any]:
        q = r[f"question_{lang}"].strip()
        opts = [r[f"option_{c}_{lang}"].strip() for c in "abcd"]
        letter = r["correct_answer"].strip().upper()
        return {
            "question_id": r["Unique_Serial"].strip(),      # SAME id in both languages = natural pairing key
            "subject": r["subject_name"].strip(),
            "subject_group": normalize_subject(r["subject_name"]),
            "question": q, "options": opts,
            "correct_answer": letter, "answer_idx": OPTION_LETTERS.index(letter),
            "language": lang,
            "benchmark": "BnMMLU" if lang == "bn" else "BnMMLU-en",
            "has_markup": contains_markup(q, opts),
            "raw_index": idx,
        }

    records = df.to_dict("records")
    bn_items = validate_items([_mk(r, "bn", i) for i, r in enumerate(records)])
    en_items = validate_items([_mk(r, "en", i) for i, r in enumerate(records)])
    keep = {it["question_id"] for it in bn_items} & {it["question_id"] for it in en_items}
    bn_items = [it for it in bn_items if it["question_id"] in keep]
    en_items = [it for it in en_items if it["question_id"] in keep]
    assert [it["question_id"] for it in bn_items] == [it["question_id"] for it in en_items], "bn/en items not index-aligned"
    report["n_pairs"] = len(bn_items)
    return bn_items, en_items, report


def _balanced_subject_allocation(avail: Dict[str, int], n_items: int, rng: "np.random.RandomState") -> Tuple[Dict[str, int], int]:
    """Near-equal per-subject quotas with a cap (1.5x the equal share); leftovers go to subjects with spare items."""
    groups = sorted(avail)
    if sum(avail.values()) < n_items:
        raise RuntimeError(
            f"Only {sum(avail.values())} valid translation-parallel items available but n_items={n_items}. "
            f"Per-subject availability: {avail}. STRICT_REAL_MODE prohibits silently reducing the sample size."
        )
    cap = math.ceil(n_items / len(groups) * 1.5)
    alloc = {g: 0 for g in groups}
    remaining = n_items
    while remaining > 0:
        eligible = [g for g in groups if alloc[g] < min(avail[g], cap)]
        if not eligible:
            break
        step = max(1, remaining // len(eligible))
        for g in eligible:
            add = min(min(avail[g], cap) - alloc[g], step, remaining)
            alloc[g] += add
            remaining -= add
            if remaining == 0:
                break
    while remaining > 0:
        eligible = [g for g in groups if alloc[g] < avail[g]]
        if not eligible:
            break
        for j in rng.permutation(len(eligible)):
            g = eligible[int(j)]
            alloc[g] += 1
            remaining -= 1
            if remaining == 0:
                break
    return alloc, cap


def _save_items_csv(items: List[Dict[str, Any]], path: str, empty_cols: Optional[List[str]] = None) -> pd.DataFrame:
    out = pd.DataFrame(items) if items else pd.DataFrame(columns=empty_cols or [])
    if len(out):
        out["options"] = out["options"].apply(lambda opts: json.dumps(opts, ensure_ascii=False))
    out.to_csv(path, index=False, encoding="utf-8")
    return out


def _assert_no_stale_model_results(output_dir: str, bn_ids: set, en_ids: set) -> None:
    """Refuse to rebuild subsets underneath results that were produced on a different item set."""
    stale = []
    for m_key in MODEL_REGISTRY.keys():
        for lang, ids in (("bn", bn_ids), ("en", en_ids)):
            p = os.path.join(output_dir, m_key, f"{lang}_results.csv")
            if not os.path.exists(p):
                continue
            try:
                done = set(pd.read_csv(p, usecols=["question_id"], dtype=str)["question_id"])
            except Exception:
                done = set()
            if done and not done.issubset(ids):
                stale.append(p)
    if stale:
        raise RuntimeError(
            "The new translation-parallel item set differs from items already evaluated in:\n  " + "\n  ".join(stale) +
            "\nResuming would mix two different item sets. Archive or delete those model folders "
            "(e.g. rm -rf results/<model_key>) or point CONFIG.output_dir to a fresh directory, then re-run Section 5."
        )


def prepare_translation_parallel_subsets(
    output_dir: str, n_items: int, seed: int, config: "ExperimentConfig", is_mock: bool = False,
) -> Tuple[str, str]:
    """Build bn_subset.csv / en_subset.csv (+ dev, few-shot, provenance) from ONE wide bn/en CSV."""
    if is_mock:
        raise RuntimeError("Mock mode is disabled for translation-parallel preparation (STRICT_REAL_MODE).")
    os.makedirs(output_dir, exist_ok=True)
    bn_subset_path = os.path.join(output_dir, "bn_subset.csv")
    en_subset_path = os.path.join(output_dir, "en_subset.csv")
    dev_bn_path = os.path.join(output_dir, "dev_bn.csv")
    dev_en_path = os.path.join(output_dir, "dev_en.csv")
    sampled_ids_path = os.path.join(output_dir, "sampled_question_ids.json")
    mapping_path = os.path.join(output_dir, "subject_mapping.json")
    meta_path = os.path.join(output_dir, "subset_meta.json")
    fewshot_file = os.path.join(output_dir, "fewshot_exemplars.json")
    dev_n = int(getattr(config, "dev_n", 200))

    if config.bn_local_path:
        logger.warning("CONFIG.bn_local_path is ignored in translation-parallel mode: BOTH languages come from the same wide CSV.")

    csv_path = resolve_translation_csv_path(config)
    csv_sha = _file_sha256(csv_path)
    expected_meta = {
        "n_items": int(n_items), "seed": int(seed), "is_mock": False,
        "en_source": "bnmmlu_translation", "translation_csv_sha256": csv_sha,
        "exclude_markup": bool(config.exclude_markup),
        "translation_exclude_subjects": sorted(normalize_subject(s) for s in getattr(config, "translation_exclude_subjects", ())),
        "filter_version": TRANSLATION_FILTER_VERSION, "sampling_version": TRANSLATION_SAMPLING_VERSION,
        "n_shots_max": 8, "exemplar_version": EXEMPLAR_VERSION, "dev_n": dev_n, "dev_version": "v4e_dev",
    }
    if all(os.path.exists(p) for p in (bn_subset_path, en_subset_path, dev_bn_path, dev_en_path, meta_path, fewshot_file)):
        try:
            with open(meta_path, "r", encoding="utf-8") as f:
                saved_meta = json.load(f)
        except Exception:
            saved_meta = {}
        if {k: saved_meta.get(k) for k in expected_meta} == expected_meta:
            logger.info(f"Found existing translation-parallel subsets (settings + CSV hash match): {bn_subset_path}, {en_subset_path}")
            return bn_subset_path, en_subset_path
        logger.warning("Existing subsets were built with different settings. Rebuilding translation-parallel subsets.")

    logger.info(f"Preparing {n_items}-pair translation-parallel subsets from {csv_path} (sha256 {csv_sha[:12]}...)")
    bn_all, en_all, report = load_translation_parallel_pairs(csv_path, config)
    logger.info(f"Translation CSV filtering: {report['n_raw']:,} rows -> {report['n_pairs']:,} pairs | dropped: {report['dropped']}")
    if report["excluded_subjects_found"]:
        logger.info(f"Language-specific subjects excluded (their English translation is not a meaningful control): {report['excluded_subjects_found']}")

    rng = np.random.RandomState(seed)
    by_group: Dict[str, List[int]] = {}
    for i, it in enumerate(bn_all):
        by_group.setdefault(it["subject_group"], []).append(i)
    avail = {g: len(v) for g, v in by_group.items()}
    alloc, cap = _balanced_subject_allocation(avail, n_items, rng)
    chosen: List[int] = []
    for g in sorted(by_group):
        if alloc[g] > 0:
            chosen.extend(sorted(rng.choice(by_group[g], size=alloc[g], replace=False).tolist()))
    chosen = [chosen[j] for j in rng.permutation(len(chosen))]   # mixed subject order (smoke tests read the first items)
    bn_final = [bn_all[i] for i in chosen]
    en_final = [en_all[i] for i in chosen]
    if len(bn_final) != n_items:
        raise RuntimeError(f"Expected exactly {n_items} pairs, built {len(bn_final)}.")
    logger.info(f"Balanced subject sampling: {len(bn_final)} pairs across {sum(1 for v in alloc.values() if v)} subjects "
                f"(cap={cap}). Allocation: {alloc}")

    bn_eval_ids = {it["question_id"] for it in bn_final}
    en_eval_ids = {it["question_id"] for it in en_final}
    eval_texts = {it["question"] for it in bn_final} | {it["question"] for it in en_final}
    rest = [i for i in range(len(bn_all)) if bn_all[i]["question_id"] not in bn_eval_ids
            and bn_all[i]["question"] not in eval_texts and en_all[i]["question"] not in eval_texts]

    # Few-shot exemplars: chosen on the Bangla side, English = translation of the SAME items (parallel)
    bn_pool = [bn_all[i] for i in rest]
    en_by_id = {en_all[i]["question_id"]: en_all[i] for i in rest}
    bn_exemplars = select_fewshot_exemplars(bn_pool, bn_eval_ids, language="bn", seed=seed)
    en_exemplars = []
    for ex in bn_exemplars:
        e = en_by_id[ex["question_id"]]
        en_exemplars.append({"question_id": e["question_id"], "subject": e["subject"], "question": e["question"],
                             "options": list(e["options"]), "correct_answer": e["correct_answer"], "source": "BnMMLU-en"})
    ex_ids = {ex["question_id"] for ex in bn_exemplars}
    ex_texts = {ex["question"] for ex in bn_exemplars} | {ex["question"] for ex in en_exemplars}

    # Dev pool (same items in both languages), disjoint from evaluation items and exemplars
    dev_idx = [i for i in rest if bn_all[i]["question_id"] not in ex_ids
               and bn_all[i]["question"] not in ex_texts and en_all[i]["question"] not in ex_texts]
    if dev_n > 0:
        if len(dev_idx) < dev_n:
            raise RuntimeError(f"Fewer than {dev_n} valid items remain for the dev pool (found {len(dev_idx)}).")
        dev_rng = np.random.RandomState(seed + 7)
        pick = sorted(dev_rng.choice(len(dev_idx), size=dev_n, replace=False).tolist())
        bn_dev = [bn_all[dev_idx[j]] for j in pick]
        en_dev = [en_all[dev_idx[j]] for j in pick]
    else:
        bn_dev, en_dev = [], []
    bn_dev_ids = {it["question_id"] for it in bn_dev}
    en_dev_ids = {it["question_id"] for it in en_dev}
    assert not (bn_eval_ids & ex_ids) and not (bn_eval_ids & bn_dev_ids) and not (ex_ids & bn_dev_ids), "eval/exemplar/dev overlap"
    assert not (en_eval_ids & ex_ids) and not (en_eval_ids & en_dev_ids), "English eval/exemplar/dev overlap"

    # Never rebuild underneath results from a different item set
    _assert_no_stale_model_results(output_dir, bn_eval_ids, en_eval_ids)

    # ---------------- SAVE ----------------
    bn_df = _save_items_csv(bn_final, bn_subset_path)
    en_df = _save_items_csv(en_final, en_subset_path)
    empty_cols = ["question_id", "subject", "subject_group", "question", "options", "correct_answer", "answer_idx", "language", "benchmark", "has_markup", "raw_index"]
    dev_bn_df = _save_items_csv(bn_dev, dev_bn_path, empty_cols)
    dev_en_df = _save_items_csv(en_dev, dev_en_path, empty_cols)
    logger.info(f"Saved: {bn_subset_path} ({len(bn_df):,}), {en_subset_path} ({len(en_df):,}), dev {len(dev_bn_df):,}/{len(dev_en_df):,}")
    if not bn_df["question_id"].is_unique or not en_df["question_id"].is_unique:
        raise RuntimeError("Prepared subsets contain duplicate question_ids.")

    with open(fewshot_file, "w", encoding="utf-8") as f:
        json.dump({"bn": bn_exemplars, "en": en_exemplars, "version": EXEMPLAR_VERSION, "n_shots_max": 8,
                   "english_exemplar_source": "bnmmlu_translation_parallel"}, f, indent=2, ensure_ascii=False)
    with open(sampled_ids_path, "w", encoding="utf-8") as f:
        json.dump({"bn_question_ids": bn_df["question_id"].tolist(), "en_question_ids": en_df["question_id"].tolist()}, f, indent=2)
    dev_by_subj = {str(k): int(v) for k, v in pd.Series([it["subject_group"] for it in bn_dev]).value_counts().items()}
    with open(mapping_path, "w", encoding="utf-8") as f:
        json.dump({"mode": "bnmmlu_translation", "per_subject": {g: {"n": int(alloc[g]), "n_available": int(avail[g])} for g in sorted(avail)},
                   "excluded_subjects": report["excluded_subjects_found"], "filter_report": report,
                   "dev_counts": {"dev_n": dev_n, "n_bn": len(bn_dev), "n_en": len(en_dev), "bn_by_subject": dev_by_subj, "en_by_subject": dev_by_subj}},
                  f, indent=2, ensure_ascii=False)
    meta_out = dict(expected_meta)
    meta_out["data_provenance"] = {"translation_csv": csv_path, "translation_csv_sha256": csv_sha, "filter_report": report,
                                   "n_bn": int(len(bn_df)), "n_en": int(len(en_df)),
                                   "note": "English side = translation of the same BnMMLU items; both languages read from one CSV."}
    with open(meta_path, "w", encoding="utf-8") as f:
        json.dump(meta_out, f, indent=2, ensure_ascii=False)
    return bn_subset_path, en_subset_path


def prepare_research_subsets(
    output_dir: str = OUTPUT_DIR,
    n_items: int = 1500,
    seed: int = DEFAULT_SEED,
    hf_token: Optional[str] = None,
    is_mock: bool = False,
    bn_index_base: Optional[int] = None,
    config: Optional["ExperimentConfig"] = None,
) -> Tuple[str, str]:
    """Prepare standardised, subject-matched Bangla (BnMMLU) and English (MMLU) subsets."""
    config = config or CONFIG
    if getattr(config, "en_source", "mmlu") == "bnmmlu_translation":
        # Item-parallel mode: both languages come from ONE wide CSV (no MMLU download, no subject map).
        return prepare_translation_parallel_subsets(output_dir, n_items, seed, config, is_mock=is_mock)
    bn_subset_path = os.path.join(output_dir, "bn_subset.csv")
    en_subset_path = os.path.join(output_dir, "en_subset.csv")
    dev_bn_path = os.path.join(output_dir, "dev_bn.csv")
    dev_en_path = os.path.join(output_dir, "dev_en.csv")
    sampled_ids_path = os.path.join(output_dir, "sampled_question_ids.json")
    mapping_path = os.path.join(output_dir, "subject_mapping.json")
    meta_path = os.path.join(output_dir, "subset_meta.json")

    subj_map = build_subject_map()
    map_hash = hashlib.md5(json.dumps(subj_map, sort_keys=True).encode("utf-8")).hexdigest()[:12]
    dev_n = getattr(config, "dev_n", 200)
    expected_meta = {
        "n_items": int(n_items), "seed": int(seed), "is_mock": bool(is_mock),
        "bn_source": config.bn_local_path or f"{config.bn_dataset_id}:{config.bn_hf_config}",
        "en_source": f"{config.en_dataset_id}:{config.en_hf_config}:{config.en_split}",
        "exclude_markup": bool(config.exclude_markup),
        "subject_match_mode": config.subject_match_mode,
        "subject_map_hash": map_hash,
        "n_shots_max": 8,
        "exemplar_version": EXEMPLAR_VERSION,
        "sampling_version": "v3_3_disjoint_english",
        "dev_n": int(dev_n),
        "dev_version": "v4a_dev",
    }

    # Reuse prepared subsets ONLY if they were built with identical settings (never reuse stale/mock subsets)
    if all(os.path.exists(p) for p in (bn_subset_path, en_subset_path, dev_bn_path, dev_en_path, meta_path)):
        try:
            with open(meta_path, "r", encoding="utf-8") as f:
                saved_meta = json.load(f)
        except Exception:
            saved_meta = {}
        if {k: saved_meta.get(k) for k in expected_meta} == expected_meta:
            logger.info(f"Found existing prepared subsets (settings match):\n   - {bn_subset_path}\n   - {en_subset_path}")
            return bn_subset_path, en_subset_path
        logger.warning("Existing subsets were built with different settings. Rebuilding.")

    if config.subject_match_mode not in ("intersection", "all_bn"):
        raise ValueError("CONFIG.subject_match_mode must be 'intersection' or 'all_bn'.")

    logger.info(f"Preparing {n_items}-item research subsets for BnMMLU and MMLU...")

    # ---------------- 1. LOAD REAL DATA ----------------
    bn_raw, bn_info = load_bnmmlu_items(config, hf_token, is_mock, bn_index_base)
    en_raw = load_mmlu_items(config, hf_token, is_mock, n_items)

    bn_valid = validate_items(bn_raw)
    en_valid = validate_items(en_raw)

    if config.exclude_markup:
        n_bn_before, n_en_before = len(bn_valid), len(en_valid)
        bn_valid = [it for it in bn_valid if not it.get("has_markup")]
        en_valid = [it for it in en_valid if not it.get("has_markup")]
        logger.info(f"Markup filter: BnMMLU {n_bn_before:,} -> {len(bn_valid):,} items "
                    f"({n_bn_before - len(bn_valid):,} MathML items removed); MMLU {n_en_before:,} -> {len(en_valid):,}.")

    # Make the real BnMMLU subject list visible so the subject map can be checked/extended
    bn_subj_counts = pd.Series([it["subject_group"] for it in bn_valid]).value_counts()
    logger.info(f"BnMMLU subjects after filtering ({len(bn_subj_counts)}): {bn_subj_counts.to_dict()}")

    # MMLU pools keyed by normalised subject
    en_pool: Dict[str, List[Dict[str, Any]]] = {}
    for it in en_valid:
        en_pool.setdefault(it["subject_group"], []).append(it)
    if not is_mock:
        bad = sorted({m for v in subj_map.values() for m in v} - set(en_pool))
        if bad:
            logger.warning(f"Subject map references MMLU subjects that were not found: {bad}")

    def mmlu_names_for(group: str) -> List[str]:
        names = list(subj_map.get(group, []))
        if group in en_pool and group not in names:   # identical names also match (used by mock data / custom names)
            names.append(group)
        return [n for n in names if n in en_pool]

    rng = np.random.RandomState(seed)

    # ---------------- 2. SAMPLE ----------------
    mapped_groups = {g for g in bn_subj_counts.index if mmlu_names_for(g)}
    unmapped = {g: int(c) for g, c in bn_subj_counts.items() if g not in mapped_groups}
    if unmapped:
        logger.info(f"BnMMLU subjects with NO MMLU analogue ({len(unmapped)}): {unmapped}")

    mapped_pool_counts = {
        g: {"n_bn": int(bn_subj_counts.get(g, 0)), "n_en": sum(len(en_pool[n]) for n in mmlu_names_for(g))}
        for g in sorted(mapped_groups)
    }
    logger.info(f"Mapped subject groups available for sampling ({len(mapped_groups)}): {mapped_pool_counts}")

    if config.subject_match_mode == "intersection":
        candidates = [it for it in bn_valid if it["subject_group"] in mapped_groups]
        if not candidates:
            raise RuntimeError(
                "No BnMMLU subject could be mapped to an MMLU subject. Compare the BnMMLU subject list logged above with "
                "SUBJECT_MAP_BN_TO_MMLU (Section 4) and add entries to SUBJECT_MAP_EXTRA."
            )

        # Pre-group available valid Bangla candidates
        bn_cand_by_group: Dict[str, List[Dict[str, Any]]] = {}
        for it in candidates:
            bn_cand_by_group.setdefault(it["subject_group"], []).append(it)

        # Available pairs per mapped group
        available_pairs = {
            g: min(len(bn_cand_by_group.get(g, [])), len([it for n in mmlu_names_for(g) for it in en_pool[n]]))
            for g in sorted(mapped_groups)
        }
        total_available_pairs = sum(available_pairs.values())

        if total_available_pairs < n_items:
            pairs_per_subj = {g: f"{available_pairs[g]}/{available_pairs[g]}" for g in sorted(mapped_groups)}
            raise RuntimeError(
                f"Fewer than {n_items} valid, subject-matched Bangla-English samples available (found {total_available_pairs}). "
                f"Subject-matched pairs per mapped subject (n_bn/n_en): {pairs_per_subj}. "
                f"STRICT_REAL_MODE prohibits silently reducing the sample count below exactly {n_items} items."
            )

        n_mapped_groups = len(mapped_groups)
        per_subject_cap = math.ceil(n_items / n_mapped_groups * 1.5)

        # Balanced capped allocation with leftover redistribution
        alloc: Dict[str, int] = {g: 0 for g in sorted(mapped_groups)}
        target_remaining = n_items

        # Pass 1: allocate up to min(available_pairs, per_subject_cap) near-equally
        while target_remaining > 0:
            eligible = [g for g in sorted(mapped_groups) if alloc[g] < min(available_pairs[g], per_subject_cap)]
            if not eligible:
                break
            step = max(1, target_remaining // len(eligible))
            for g in sorted(eligible):
                room = min(available_pairs[g], per_subject_cap) - alloc[g]
                add = min(room, step, target_remaining)
                alloc[g] += add
                target_remaining -= add
                if target_remaining == 0:
                    break

        # Pass 2: redistribute leftover quota to groups that still have spare pairs
        while target_remaining > 0:
            eligible = [g for g in sorted(mapped_groups) if alloc[g] < available_pairs[g]]
            if not eligible:
                break
            perm = rng.permutation(len(eligible))
            for idx in perm:
                g = eligible[idx]
                alloc[g] += 1
                target_remaining -= 1
                if target_remaining == 0:
                    break

        # ---- DISJOINT English sampling ----------------------------------------------------------------------
        # Several BnMMLU subjects share MMLU subjects (e.g. high_school_biology serves 'cell biology genetics' AND
        # 'ecology environmental biology'). Sampling each group independently would pick the SAME English question for
        # several groups, duplicating items. Every English item is therefore used AT MOST ONCE across all groups:
        # scarcest pools are filled first and any shortfall is redistributed to groups that still have unused
        # Bangla AND English items.
        used_en_ids: set = set()
        full_pool: Dict[str, List[Dict[str, Any]]] = {
            g: [it for n in mmlu_names_for(g) for it in en_pool[n]] for g in sorted(mapped_groups)
        }
        en_pick_by_group: Dict[str, List[Dict[str, Any]]] = {g: [] for g in sorted(mapped_groups)}

        def _unused_en(g: str) -> List[Dict[str, Any]]:
            return [it for it in full_pool[g] if it["question_id"] not in used_en_ids]

        for g in sorted(mapped_groups, key=lambda x: (len(full_pool[x]), x)):
            avail = _unused_en(g)
            take = min(alloc[g], len(avail))
            if take > 0:
                for j in sorted(rng.choice(len(avail), size=take, replace=False).tolist()):
                    en_pick_by_group[g].append(avail[j])
                    used_en_ids.add(avail[j]["question_id"])

        shortfall = n_items - sum(len(v) for v in en_pick_by_group.values())
        while shortfall > 0:
            spare = [
                g for g in sorted(mapped_groups)
                if len(en_pick_by_group[g]) < len(bn_cand_by_group.get(g, [])) and _unused_en(g)
            ]
            if not spare:
                break
            g = min(spare, key=lambda x: (len(en_pick_by_group[x]), x))
            avail = _unused_en(g)
            j = int(rng.randint(len(avail)))
            en_pick_by_group[g].append(avail[j])
            used_en_ids.add(avail[j]["question_id"])
            shortfall -= 1

        if shortfall > 0:
            pairs_per_subj = {
                g: f"{len(bn_cand_by_group.get(g, []))}/{len(full_pool[g])}" for g in sorted(mapped_groups)
            }
            raise RuntimeError(
                f"Fewer than {n_items} valid, subject-matched Bangla-English samples available "
                f"(found {n_items - shortfall} with every English item used at most once). "
                f"Subject-matched pairs per mapped subject (n_bn/n_en, English pools are shared between groups): {pairs_per_subj}. "
                f"STRICT_REAL_MODE prohibits silently reducing the sample count below exactly {n_items} items."
            )

        bn_final: List[Dict[str, Any]] = []
        en_final: List[Dict[str, Any]] = []
        mapping_report: Dict[str, Any] = {}
        for group in sorted(mapped_groups):
            names = mmlu_names_for(group)
            en_pick = [dict(it, subject_group=group) for it in en_pick_by_group[group]]
            k = len(en_pick)
            if k == 0:
                mapping_report[group] = {"mmlu_subjects": names, "n_bn": 0, "n_en": 0, "n_bn_available": len(bn_cand_by_group.get(group, [])),
                                         "n_en_available": len(full_pool[group]), "quota": int(alloc[group])}
                continue
            bn_items_g = bn_cand_by_group[group]
            bn_idx = sorted(rng.choice(len(bn_items_g), size=k, replace=False).tolist())
            bn_final.extend(bn_items_g[j] for j in bn_idx)
            en_final.extend(en_pick)
            mapping_report[group] = {"mmlu_subjects": names, "n_bn": k, "n_en": k, "n_bn_available": len(bn_items_g),
                                     "n_en_available": len(full_pool[group]), "quota": int(alloc[group])}

        used_groups_count = sum(1 for g in mapped_groups if en_pick_by_group[g])
        logger.info(
            f"Balanced subject sampling complete: {len(bn_final)} paired items across {used_groups_count} subject groups "
            f"(per-subject cap={per_subject_cap}, every English item used at most once). "
            f"Allocation: { {g: len(v) for g, v in en_pick_by_group.items()} }"
        )
    else:
        bn_sampled = sample_stratified_subset(bn_valid, n_samples=n_items, seed=seed)

        # ---------------- 3. MATCH ENGLISH ITEMS PER SUBJECT ----------------
        bn_by_group: Dict[str, List[Dict[str, Any]]] = {}
        for it in bn_sampled:
            bn_by_group.setdefault(it["subject_group"], []).append(it)

        bn_final: List[Dict[str, Any]] = []
        en_final: List[Dict[str, Any]] = []
        mapping_report: Dict[str, Any] = {}
        used_en_ids: set = set()

        for group, bn_items_g in sorted(bn_by_group.items()):
            names = mmlu_names_for(group)
            pool = [it for n in names for it in en_pool[n] if it["question_id"] not in used_en_ids]
            if not pool:
                bn_final.extend(bn_items_g)
                mapping_report[group] = {"mmlu_subjects": [], "n_bn": len(bn_items_g), "n_en": 0}
                continue
            k = min(len(bn_items_g), len(pool))
            bn_keep = bn_items_g
            en_idx = rng.choice(len(pool), size=k, replace=False).tolist()
            en_pick = [dict(pool[j], subject_group=group) for j in en_idx]
            used_en_ids.update(it["question_id"] for it in en_pick)
            bn_final.extend(bn_items_g)
            en_final.extend(en_pick)
            mapping_report[group] = {"mmlu_subjects": names, "n_bn": len(bn_items_g), "n_en": k}

    if config.subject_match_mode == "all_bn":
        target_n = len(bn_final)
        if len(en_final) < target_n:
            remaining = [it for it in en_valid if it["question_id"] not in used_en_ids]
            needed = min(target_n - len(en_final), len(remaining))
            if needed > 0:
                extra = rng.choice(len(remaining), size=needed, replace=False).tolist()
                en_final.extend(dict(remaining[j], subject_group="unmatched") for j in extra)
                logger.info(f"Topped up English subset with {needed} subject-UNMATCHED items (subject_group='unmatched').")

    if len(bn_final) != len(en_final):
        raise RuntimeError(f"Subset size mismatch: Bangla has {len(bn_final)} items, English has {len(en_final)} items.")
    if len(bn_final) < n_items:
        pairs_per_subj = {g: f"{rep['n_bn']}/{rep['n_en']}" for g, rep in sorted(mapping_report.items())}
        raise RuntimeError(
            f"Fewer than {n_items} valid, subject-matched Bangla-English samples available (found {len(bn_final)}). "
            f"Subject-matched pairs per mapped subject (n_bn/n_en): {pairs_per_subj}. "
            f"STRICT_REAL_MODE prohibits silently reducing the sample count below exactly {n_items} items."
        )
    # Ensure exactly n_items in final paired sets
    if len(bn_final) > n_items:
        bn_final = bn_final[:n_items]
        en_final = en_final[:n_items]

    # ---------------- 4. SAVE ----------------
    def _save(items: List[Dict[str, Any]], path: str) -> pd.DataFrame:
        out = pd.DataFrame(items)
        out["options"] = out["options"].apply(lambda opts: json.dumps(opts, ensure_ascii=False))
        out.to_csv(path, index=False, encoding="utf-8")
        return out

    bn_df = _save(bn_final, bn_subset_path)
    en_df = _save(en_final, en_subset_path)
    logger.info(f"Saved Bangla subset: {bn_subset_path} ({len(bn_df):,} items)")
    logger.info(f"Saved English subset: {en_subset_path} ({len(en_df):,} items)")
    if not en_df["question_id"].is_unique or not bn_df["question_id"].is_unique:
        raise RuntimeError("Prepared subsets contain duplicate question_ids (an item was selected more than once).")

    with open(mapping_path, "w", encoding="utf-8") as f:
        json.dump({"mode": config.subject_match_mode, "per_subject": mapping_report, "unmapped_bn_subjects": unmapped,
                   "coverage_of_bn_pool": round(sum(int(bn_subj_counts.get(g, 0)) for g in mapped_groups) / max(1, len(bn_valid)), 4),
                   "n_unique_en_ids": int(en_df["question_id"].nunique()), "n_unique_bn_ids": int(bn_df["question_id"].nunique())},
                  f, indent=2, ensure_ascii=False)

    meta_out = dict(expected_meta)
    meta_out["data_provenance"] = {
        "bn": bn_info,
        "bn_hub_revision": "n/a (local/mock)" if (is_mock or config.bn_local_path) else _hub_revision(config.bn_dataset_id, "dataset", hf_token),
        "en_hub_revision": "n/a (mock)" if is_mock else _hub_revision(config.en_dataset_id, "dataset", hf_token),
        "n_bn": int(len(bn_df)), "n_en": int(len(en_df)),
    }
    with open(meta_path, "w", encoding="utf-8") as f:
        json.dump(meta_out, f, indent=2, ensure_ascii=False)

    with open(sampled_ids_path, "w", encoding="utf-8") as f:
        json.dump({"bn_question_ids": bn_df["question_id"].tolist(), "en_question_ids": en_df["question_id"].tolist()}, f, indent=2)
    logger.info(f"Sampled question IDs locked to: {sampled_ids_path} | subject mapping: {mapping_path}")

    # Build and persist fewshot_exemplars.json
    fewshot_file = os.path.join(output_dir, "fewshot_exemplars.json")
    bn_eval_set = set(bn_df["question_id"].astype(str).tolist())
    en_eval_set = set(en_df["question_id"].astype(str).tolist())

    # Bangla candidates: valid BnMMLU items outside evaluation subset
    bn_candidates = [it for it in bn_valid if str(it.get("question_id", "")) not in bn_eval_set]

    # English candidates: dev split of cais/mmlu (or unused valid en items)
    en_candidates = []
    english_exemplar_source = "test_pool_fallback"
    allow_fallback = getattr(config, "allow_fewshot_test_pool_fallback", False)

    if not is_mock and hf_token:
        try:
            logger.info("Loading English MMLU dev split for few-shot exemplars...")
            from datasets import load_dataset
            en_dev_ds = load_dataset(config.en_dataset_id, "all", split="dev", token=hf_token)
            en_dev_raw = []
            for row_idx, r in enumerate(en_dev_ds):
                sub = str(r.get("subject", "general"))
                choices = list(r.get("choices", []))
                raw_ans = r.get("answer", 0)
                ans_idx = int(raw_ans)
                q_text = str(r.get("question", ""))
                has_markup = contains_markup(q_text, choices)

                en_dev_raw.append({
                    "question_id": f"en_dev_{row_idx}",
                    "subject": sub,
                    "subject_group": normalize_subject(sub),
                    "question": q_text,
                    "options": choices,
                    "correct_answer": OPTION_LETTERS[ans_idx],
                    "answer_idx": ans_idx,
                    "language": "en",
                    "benchmark": "MMLU",
                    "has_markup": has_markup,
                    "raw_index": row_idx,
                })
            en_candidates = validate_items(en_dev_raw)
            english_exemplar_source = "mmlu_dev"
        except Exception as e:
            if STRICT_REAL_MODE and not allow_fallback:
                logger.error(f"Failed to load English MMLU dev split under STRICT_REAL_MODE: {e}")
                raise RuntimeError(f"Could not load English MMLU dev split under STRICT_REAL_MODE (fallback disabled): {e}") from e
            logger.warning(f"Could not load MMLU dev split ({e}); falling back to unused test pool.")
            en_candidates = [it for it in en_valid if str(it.get("question_id", "")) not in en_eval_set]
            english_exemplar_source = "test_pool_fallback"
    else:
        en_candidates = [it for it in en_valid if str(it.get("question_id", "")) not in en_eval_set]
        english_exemplar_source = "test_pool_fallback"

    bn_exemplars = select_fewshot_exemplars(bn_candidates, bn_eval_set, language="bn", seed=seed)
    en_exemplars = select_fewshot_exemplars(en_candidates, en_eval_set, language="en", seed=seed)

    exemplars_payload = {
        "bn": bn_exemplars,
        "en": en_exemplars,
        "version": EXEMPLAR_VERSION,
        "n_shots_max": 8,
        "english_exemplar_source": english_exemplar_source,
    }
    with open(fewshot_file, "w", encoding="utf-8") as f:
        json.dump(exemplars_payload, f, indent=2, ensure_ascii=False)
    logger.info(f"Few-shot exemplars persisted to: {fewshot_file} (bn: {len(bn_exemplars)}, en: {len(en_exemplars)})")

    # ---------------- 5. DEV POOL CREATION (dev_bn.csv, dev_en.csv) ----------------
    bn_ex_ids = {str(it.get("question_id", "")) for it in bn_exemplars}
    en_ex_ids = {str(it.get("question_id", "")) for it in en_exemplars}

    if dev_n > 0:
        bn_dev_candidates = [it for it in bn_valid if str(it.get("question_id", "")) not in bn_eval_set and str(it.get("question_id", "")) not in bn_ex_ids]
        en_dev_candidates = [it for it in en_valid if str(it.get("question_id", "")) not in en_eval_set and str(it.get("question_id", "")) not in en_ex_ids]

        if len(bn_dev_candidates) < dev_n:
            raise RuntimeError(f"Fewer than {dev_n} valid items remain for Bangla dev pool (found {len(bn_dev_candidates)}).")
        if len(en_dev_candidates) < dev_n:
            raise RuntimeError(f"Fewer than {dev_n} valid items remain for English dev pool (found {len(en_dev_candidates)}).")

        # Separate RandomState(seed + 7) for dev pool sampling
        dev_rng = np.random.RandomState(seed + 7)

        bn_dev_idx = sorted(dev_rng.choice(len(bn_dev_candidates), size=dev_n, replace=False).tolist())
        en_dev_idx = sorted(dev_rng.choice(len(en_dev_candidates), size=dev_n, replace=False).tolist())

        bn_dev_items = [bn_dev_candidates[i] for i in bn_dev_idx]
        en_dev_items = [en_dev_candidates[i] for i in en_dev_idx]

        bn_dev_ids = {str(it.get("question_id", "")) for it in bn_dev_items}
        en_dev_ids = {str(it.get("question_id", "")) for it in en_dev_items}

        # Assert eval, exemplars, and dev are pairwise disjoint
        assert len(bn_eval_set & bn_ex_ids) == 0, "Bangla eval and exemplars overlap!"
        assert len(bn_eval_set & bn_dev_ids) == 0, "Bangla eval and dev overlap!"
        assert len(bn_ex_ids & bn_dev_ids) == 0, "Bangla exemplars and dev overlap!"

        assert len(en_eval_set & en_ex_ids) == 0, "English eval and exemplars overlap!"
        assert len(en_eval_set & en_dev_ids) == 0, "English eval and dev overlap!"
        assert len(en_ex_ids & en_dev_ids) == 0, "English exemplars and dev overlap!"

        # Save dev_bn.csv and dev_en.csv
        dev_bn_df = _save(bn_dev_items, dev_bn_path)
        dev_en_df = _save(en_dev_items, dev_en_path)
        logger.info(f"Saved Bangla dev set:  {dev_bn_path} ({len(dev_bn_df):,} items)")
        logger.info(f"Saved English dev set: {dev_en_path} ({len(dev_en_df):,} items)")

        bn_dev_subj_counts = {str(k): int(v) for k, v in pd.Series([it.get("subject_group", "") for it in bn_dev_items]).value_counts().items()}
        en_dev_subj_counts = {str(k): int(v) for k, v in pd.Series([it.get("subject_group", "") for it in en_dev_items]).value_counts().items()}
        n_bn_dev = len(dev_bn_df)
        n_en_dev = len(dev_en_df)
    else:
        # dev_n == 0: create empty dev files, mapping records 0
        empty_cols = ["question_id", "subject", "subject_group", "question", "options", "correct_answer", "answer_idx", "is_mock"]
        dev_bn_df = pd.DataFrame(columns=empty_cols)
        dev_en_df = pd.DataFrame(columns=empty_cols)
        dev_bn_df.to_csv(dev_bn_path, index=False, encoding="utf-8")
        dev_en_df.to_csv(dev_en_path, index=False, encoding="utf-8")
        bn_dev_subj_counts = {}
        en_dev_subj_counts = {}
        n_bn_dev = 0
        n_en_dev = 0

    # Add dev counts to subject_mapping.json
    with open(mapping_path, "r", encoding="utf-8") as f:
        mapping_data = json.load(f)
    mapping_data["dev_counts"] = {
        "dev_n": dev_n,
        "n_bn": n_bn_dev,
        "n_en": n_en_dev,
        "bn_by_subject": bn_dev_subj_counts,
        "en_by_subject": en_dev_subj_counts,
    }
    with open(mapping_path, "w", encoding="utf-8") as f:
        json.dump(mapping_data, f, indent=2, ensure_ascii=False)

    return bn_subset_path, en_subset_path


# Execute subset preparation
bn_subset_file, en_subset_file = prepare_research_subsets(
    output_dir=CONFIG.output_dir,
    n_items=CONFIG.n_samples,
    seed=CONFIG.seed,
    hf_token=CONFIG.hf_token,
    is_mock=(CONFIG.backend == "mock"),
    bn_index_base=CONFIG.bn_index_base,
    config=CONFIG,
)
```

## Section 6: Tokenization Fragmentation Metrics (TPW, STRR Isolated/In-Context, TPC)
Quantifies subword fragmentation:
1. **Tokens-per-Word (TPW / Fertility):** Whitespace word baseline.
2. **STRR (`isolated` and `in_context`):** Evaluates atomic retention with and without space prefixes.
3. **Tokens-per-Character (TPC):** Script-normalized character fertility.
4. **`compute_dataset_fragmentation`:** Lightweight profiling loading *only* tokenizers without model weights.


```python
def _tokenize_text(tokenizer: Any, text: str) -> List[Union[int, str]]:
    """Safely tokenize text across diverse tokenizer interfaces."""
    if not text:
        return []
    if hasattr(tokenizer, "encode"):
        try:
            return tokenizer.encode(text, add_special_tokens=False)
        except TypeError:
            return tokenizer.encode(text)
    elif hasattr(tokenizer, "tokenize"):
        return tokenizer.tokenize(text)
    elif callable(tokenizer):
        return tokenizer(text)
    else:
        return text.strip().split()


def extract_words(text: str) -> List[str]:
    """Extract words using whitespace splitting."""
    if not text:
        return []
    return text.strip().split()


def compute_fertility(tokenizer: Any, text: str) -> float:
    """Compute Tokens-per-Word (TPW / Fertility)."""
    words = extract_words(text)
    if not words:
        return 0.0
    tokens = _tokenize_text(tokenizer, text)
    return float(len(tokens) / len(words))


def compute_strr(tokenizer: Any, words_or_text: Union[List[str], str], strr_mode: str = "isolated") -> float:
    """Compute Single-Token Retention Rate (STRR).
    
    Modes:
    - 'isolated': tokenizes words in isolation.
    - 'in_context': tokenizes ' ' + word (accounting for BPE leading space prefixes).
    """
    words = extract_words(words_or_text) if isinstance(words_or_text, str) else words_or_text
    if not words:
        return 0.0

    single_token_count = 0
    for w in words:
        cand = f" {w}" if strr_mode == "in_context" else w
        toks = _tokenize_text(tokenizer, cand)
        if len(toks) == 1:
            single_token_count += 1

    return float(single_token_count / len(words))


def compute_char_fertility(tokenizer: Any, text: str, exclude_spaces: bool = True) -> float:
    """Compute Tokens-per-Character (TPC / Character Fertility)."""
    if not text:
        return 0.0
    chars = text.replace(" ", "") if exclude_spaces else text
    if not chars:
        return 0.0
    tokens = _tokenize_text(tokenizer, text)
    return float(len(tokens) / len(chars))


def compute_fragmentation_metrics(
    tokenizer: Any,
    question: str,
    options: List[str],
) -> Dict[str, Any]:
    """Calculate tokenization fragmentation metrics for an MCQ item (both isolated & in_context)."""
    q_text = str(question or "").strip()
    clean_options = [str(opt or "").strip() for opt in options]
    opts_text = " ".join(clean_options)
    full_text = (q_text + " " + opts_text).strip()

    # Question
    q_words = extract_words(q_text)
    q_tokens = _tokenize_text(tokenizer, q_text)
    q_char_count = len(q_text.replace(" ", ""))
    q_tpw = float(len(q_tokens) / len(q_words)) if q_words else 0.0
    q_strr = compute_strr(tokenizer, q_words, strr_mode="isolated")
    q_strr_in_context = compute_strr(tokenizer, q_words, strr_mode="in_context")
    q_tpc = float(len(q_tokens) / q_char_count) if q_char_count else 0.0

    # Options
    opt_tokens_per_choice = [_tokenize_text(tokenizer, opt) for opt in clean_options]
    opts_tokens_total = sum(len(toks) for toks in opt_tokens_per_choice)
    opts_words = extract_words(opts_text)
    opts_char_count = len(opts_text.replace(" ", ""))
    opts_tpw = float(opts_tokens_total / len(opts_words)) if opts_words else 0.0
    opts_strr = compute_strr(tokenizer, opts_words, strr_mode="isolated")
    opts_strr_in_context = compute_strr(tokenizer, opts_words, strr_mode="in_context")
    opts_tpc = float(opts_tokens_total / opts_char_count) if opts_char_count else 0.0

    # Total Prompt
    full_tokens = _tokenize_text(tokenizer, full_text)
    full_words = extract_words(full_text)
    full_char_count = len(full_text.replace(" ", ""))
    full_tpw = float(len(full_tokens) / len(full_words)) if full_words else 0.0
    full_strr = compute_strr(tokenizer, full_words, strr_mode="isolated")
    full_strr_in_context = compute_strr(tokenizer, full_words, strr_mode="in_context")
    full_tpc = float(len(full_tokens) / full_char_count) if full_char_count else 0.0

    return {
        "question_token_count": len(q_tokens),
        "question_word_count": len(q_words),
        "question_char_count": q_char_count,
        "question_tpw": q_tpw,
        "question_strr": q_strr,
        "question_strr_in_context": q_strr_in_context,
        "question_tpc": q_tpc,
        "options_token_count": opts_tokens_total,
        "options_word_count": len(opts_words),
        "options_tpw": opts_tpw,
        "options_strr": opts_strr,
        "options_strr_in_context": opts_strr_in_context,
        "options_tpc": opts_tpc,
        "total_token_count": len(full_tokens),
        "total_word_count": len(full_words),
        "overall_tpw": full_tpw,
        "overall_strr": full_strr,
        "overall_strr_in_context": full_strr_in_context,
        "overall_tpc": full_tpc,
    }


def compute_dataset_fragmentation(
    tokenizer_id: str,
    items: List[Dict[str, Any]],
    hf_token: Optional[str] = None,
) -> Tuple[pd.DataFrame, Dict[str, float]]:
    """Tokenizer-only computation: Computes fragmentation across dataset without downloading model weights."""
    logger.info(f"Loading tokenizer '{tokenizer_id}' for dataset fragmentation profiling...")
    if tokenizer_id == "mock-model":
        class MockTokenizer:
            def __init__(self): self.name_or_path = "mock-model"
            def encode(self, t, add_special_tokens=False): return t.split()
            def __len__(self): return 32000
        tok = MockTokenizer()
        vocab_size = 32000
        tok_class = "MockTokenizer"
    else:
        from transformers import AutoTokenizer
        tok = AutoTokenizer.from_pretrained(tokenizer_id, token=hf_token)
        vocab_size = len(tok)
        tok_class = type(tok).__name__

    logger.info(f"Tokenizer: {tok_class} | Vocab Size: {vocab_size:,}")

    records = []
    for it in items:
        frag = compute_fragmentation_metrics(tok, it.get("question", ""), it.get("options", []))
        records.append({
            "question_id": it.get("question_id", ""),
            "subject": it.get("subject", ""),
            "overall_tpw": frag["overall_tpw"],
            "overall_strr_isolated": frag["overall_strr"],
            "overall_strr_in_context": frag["overall_strr_in_context"],
            "overall_tpc": frag["overall_tpc"],
        })

    frag_df = pd.DataFrame(records)
    summary = {
        "tokenizer_id": tokenizer_id,
        "vocab_size": vocab_size,
        "tokenizer_class": tok_class,
        "mean_tpw": float(frag_df["overall_tpw"].mean()),
        "mean_strr_isolated": float(frag_df["overall_strr_isolated"].mean()),
        "mean_strr_in_context": float(frag_df["overall_strr_in_context"].mean()),
        "mean_tpc": float(frag_df["overall_tpc"].mean()),
    }
    logger.info(f"Summary: Mean TPW={summary['mean_tpw']:.2f} | Mean STRR={summary['mean_strr_isolated']:.3f} | Mean TPC={summary['mean_tpc']:.3f}")
    return frag_df, summary
```

## Section 7: Uncertainty Quantification & Calibration Metrics
Implements exact mathematical formulations for calibration:
- **Expected Calibration Error (ECE):** Uniform equal-width and quantile adaptive binning.
- **Maximum Calibration Error (MCE):** Worst-case bin calibration gap.
- **Top-1 Brier Score & Full Multiclass Brier Score:** Evaluates mean squared probability deviation.
- **Negative Log-Likelihood (NLL):** True-class cross-entropy with $\epsilon$-stabilization.


```python
def _filter_valid_pairs(
    confidences: Union[List[float], np.ndarray],
    corrects: Union[List[Union[bool, int, float]], np.ndarray],
) -> Tuple[np.ndarray, np.ndarray]:
    """Filter out NaN, None, or infinite values from confidence and correctness arrays."""
    conf = np.asarray(confidences, dtype=np.float64)
    corr = np.asarray(corrects, dtype=np.float64)

    if len(conf) != len(corr):
        raise ValueError(f"Length mismatch: confidences={len(conf)}, corrects={len(corr)}")

    valid_mask = np.isfinite(conf) & np.isfinite(corr)
    conf_clean = np.clip(conf[valid_mask], 0.0, 1.0)
    corr_clean = (corr[valid_mask] > 0.5).astype(np.float64)

    return conf_clean, corr_clean


def expected_calibration_error(
    confidences: Union[List[float], np.ndarray],
    corrects: Union[List[Union[bool, int, float]], np.ndarray],
    n_bins: int = 10,
    strategy: str = "uniform",
) -> float:
    """Calculate Expected Calibration Error (ECE)."""
    conf, corr = _filter_valid_pairs(confidences, corrects)
    n = len(conf)
    if n == 0:
        return float("nan")

    if strategy == "quantile":
        quantiles = np.linspace(0, 1, n_bins + 1)
        bins = np.percentile(conf, quantiles * 100)
        bins = np.unique(bins)
        if len(bins) < 2:
            bins = np.array([0.0, 1.0])
    else:
        bins = np.linspace(0.0, 1.0, n_bins + 1)

    actual_n_bins = len(bins) - 1
    ece = 0.0

    for i in range(actual_n_bins):
        lo, hi = bins[i], bins[i + 1]
        mask = (conf >= lo) & (conf <= hi) if i == 0 else (conf > lo) & (conf <= hi)
        bin_count = np.sum(mask)
        if bin_count == 0:
            continue

        bin_conf = np.mean(conf[mask])
        bin_acc = np.mean(corr[mask])
        ece += (bin_count / n) * abs(bin_conf - bin_acc)

    return float(ece)


def max_calibration_error(
    confidences: Union[List[float], np.ndarray],
    corrects: Union[List[Union[bool, int, float]], np.ndarray],
    n_bins: int = 10,
) -> float:
    """Calculate Maximum Calibration Error (MCE)."""
    conf, corr = _filter_valid_pairs(confidences, corrects)
    if len(conf) == 0:
        return float("nan")

    bins = np.linspace(0.0, 1.0, n_bins + 1)
    worst_error = 0.0
    has_bin = False

    for i in range(n_bins):
        lo, hi = bins[i], bins[i + 1]
        mask = (conf >= lo) & (conf <= hi) if i == 0 else (conf > lo) & (conf <= hi)
        bin_count = np.sum(mask)
        if bin_count == 0:
            continue

        has_bin = True
        bin_conf = np.mean(conf[mask])
        bin_acc = np.mean(corr[mask])
        worst_error = max(worst_error, abs(bin_conf - bin_acc))

    return float(worst_error) if has_bin else float("nan")


def brier_score(
    confidences: Union[List[float], np.ndarray],
    corrects: Union[List[Union[bool, int, float]], np.ndarray],
) -> float:
    """Calculate Top-1 / MSP-based Brier Score."""
    conf, corr = _filter_valid_pairs(confidences, corrects)
    if len(conf) == 0:
        return float("nan")
    return float(np.mean((conf - corr) ** 2))


def multiclass_brier_score(
    probabilities: Union[List[List[float]], np.ndarray],
    gold_indices: Union[List[int], np.ndarray],
    num_classes: int = 4,
) -> float:
    """Calculate Full Multiclass Brier Score across probability vectors."""
    probs = np.asarray(probabilities, dtype=np.float64)
    golds = np.asarray(gold_indices, dtype=np.int64)

    if len(probs) != len(golds) or len(probs) == 0:
        return float("nan")

    valid_mask = [
        row is not None and len(row) == num_classes and np.all(np.isfinite(row))
        for row in probs
    ]
    if not any(valid_mask):
        return float("nan")

    valid_probs = probs[valid_mask]
    valid_golds = golds[valid_mask]

    n = len(valid_probs)
    one_hot = np.zeros((n, num_classes), dtype=np.float64)
    for i, g in enumerate(valid_golds):
        if 0 <= g < num_classes:
            one_hot[i, g] = 1.0

    return float(np.mean(np.sum((valid_probs - one_hot) ** 2, axis=1)))


def negative_log_likelihood(
    probabilities: Union[List[List[float]], np.ndarray],
    gold_indices: Union[List[int], np.ndarray],
    eps: float = 1e-12,
) -> float:
    """Calculate Negative Log-Likelihood (NLL) on the true class."""
    probs = np.asarray(probabilities, dtype=np.float64)
    golds = np.asarray(gold_indices, dtype=np.int64)

    if len(probs) != len(golds) or len(probs) == 0:
        return float("nan")

    nll_vals = []
    for p_vec, g in zip(probs, golds):
        if p_vec is None or not (0 <= g < len(p_vec)):
            continue
        prob_gold = float(p_vec[g])
        if not np.isfinite(prob_gold):
            continue
        prob_gold = np.clip(prob_gold, eps, 1.0)
        nll_vals.append(-np.log(prob_gold))

    if not nll_vals:
        return float("nan")
    return float(np.mean(nll_vals))


def compute_calibration_bins(
    confidences: Union[List[float], np.ndarray],
    corrects: Union[List[Union[bool, int, float]], np.ndarray],
    n_bins: int = 10,
) -> List[Dict[str, Any]]:
    """Compute detailed bin statistics for calibration reliability diagrams."""
    conf, corr = _filter_valid_pairs(confidences, corrects)
    n = len(conf)
    bins = np.linspace(0.0, 1.0, n_bins + 1)
    bin_records = []

    for i in range(n_bins):
        lo, hi = float(bins[i]), float(bins[i + 1])
        mask = (conf >= lo) & (conf <= hi) if i == 0 else (conf > lo) & (conf <= hi)
        count = int(np.sum(mask))
        if count > 0:
            mean_conf = float(np.mean(conf[mask]))
            accuracy = float(np.mean(corr[mask]))
            gap = float(abs(mean_conf - accuracy))
        else:
            mean_conf = float((lo + hi) / 2.0)
            accuracy = 0.0
            gap = 0.0

        bin_records.append({
            "bin_index": i,
            "bin_lower": lo,
            "bin_upper": hi,
            "sample_count": count,
            "sample_fraction": float(count / n) if n > 0 else 0.0,
            "mean_confidence": mean_conf,
            "accuracy": accuracy,
            "calibration_gap": gap,
        })

    return bin_records


def compute_all_calibration_metrics(
    confidences: Union[List[float], np.ndarray],
    corrects: Union[List[Union[bool, int, float]], np.ndarray],
    probabilities: Optional[Union[List[List[float]], np.ndarray]] = None,
    gold_indices: Optional[Union[List[int], np.ndarray]] = None,
    n_bins: int = 10,
) -> Dict[str, Any]:
    """Compute a consolidated dictionary of all calibration metrics."""
    conf, corr = _filter_valid_pairs(confidences, corrects)
    total_valid = len(conf)

    if total_valid == 0:
        return {
            "sample_size": 0,
            "accuracy": float("nan"),
            "mean_confidence": float("nan"),
            f"ece_uniform_{n_bins}": float("nan"),
            f"ece_quantile_{n_bins}": float("nan"),
            "mce": float("nan"),
            "brier_score": float("nan"),
            "multiclass_brier_score": float("nan"),
            "nll": float("nan"),
            "bins": [],
        }

    acc = float(np.mean(corr))
    mean_conf = float(np.mean(conf))
    ece_u = expected_calibration_error(conf, corr, n_bins=n_bins, strategy="uniform")
    ece_q = expected_calibration_error(conf, corr, n_bins=n_bins, strategy="quantile")
    mce = max_calibration_error(conf, corr, n_bins=n_bins)
    bs = brier_score(conf, corr)

    mc_bs = float("nan")
    nll = float("nan")
    if probabilities is not None and gold_indices is not None:
        mc_bs = multiclass_brier_score(probabilities, gold_indices)
        nll = negative_log_likelihood(probabilities, gold_indices)

    bins = compute_calibration_bins(conf, corr, n_bins=n_bins)

    return {
        "sample_size": total_valid,
        "accuracy": acc,
        "mean_confidence": mean_conf,
        f"ece_uniform_{n_bins}": ece_u,
        f"ece_quantile_{n_bins}": ece_q,
        "mce": mce,
        "brier_score": bs,
        "multiclass_brier_score": mc_bs,
        "nll": nll,
        "bins": bins,
    }
```

## Section 8: Standardized Prompt Templates & Natural Token Continuation
Ensures prompt standardization across all models and languages. Built via a single unified builder (`_build_prompt`), both Bangla and English templates end with `:\n` in the `"plain"` variant, ensuring that candidate option letters ($A, B, C, D$) are the natural next tokens.

The `"answer_cue"` variant appends a language-appropriate cue (`উত্তর:` for Bangla, `Answer:` for English) after the instruction line, which can steer instruction-tuned models to concentrate more probability mass on the option tokens.

```python
def _build_prompt(
    question_prefix: str,
    question: str,
    options_prefix: str,
    options: List[str],
    instruction: str,
) -> str:
    """Single shared builder for multiple-choice prompts ensuring structural equivalence."""
    opts = list(options)
    if len(opts) < 4:
        opts.extend([""] * (4 - len(opts)))

    opt_block = "\n".join([f"{letter}) {opts[i]}" for i, letter in enumerate(OPTION_LETTERS)])
    return (
        f"{question_prefix}: {question}\n\n"
        f"{options_prefix}:\n"
        f"{opt_block}\n\n"
        f"{instruction}:\n"
    )


def format_bnmmlu_baseline_prompt(question: str, options: List[str]) -> str:
    """Standardized Bangla BnMMLU prompt template."""
    return _build_prompt(
        question_prefix="\u09aa\u09cd\u09b0\u09b6\u09cd\u09a8",
        question=question,
        options_prefix="\u09ac\u09bf\u0995\u09b2\u09cd\u09aa\u09b8\u09ae\u09c2\u09b9",
        options=options,
        instruction="\u09b8\u09b0\u09be\u09b8\u09b0\u09bf \u09b8\u09a0\u09bf\u0995 \u0989\u09a4\u09cd\u09a4\u09b0 \u09a6\u09bf\u09a8\u0964 \u09b6\u09c1\u09a7\u09c1 A, B, C, \u09ac\u09be D \u09b2\u09bf\u0996\u09c1\u09a8\u0964 \u0985\u09a8\u09cd\u09af \u0995\u09bf\u099b\u09c1 \u09b2\u09bf\u0996\u09ac\u09c7\u09a8 \u09a8\u09be",
    )


def format_english_mmlu_prompt(question: str, options: List[str]) -> str:
    """Matched English MMLU prompt template."""
    return _build_prompt(
        question_prefix="Question",
        question=question,
        options_prefix="Options",
        options=options,
        instruction="Answer directly with only the letter (A, B, C, or D). Do not write anything else",
    )



_FEWSHOT_EXEMPLARS_CACHE: Dict[str, Tuple[Optional[float], Dict[str, Any]]] = {}  # path -> (mtime, data)

def load_fewshot_exemplars(path: Optional[str] = None) -> Dict[str, Any]:
    """Load and cache fewshot_exemplars.json.

    Raises RuntimeError if file does not exist or cannot be parsed.
    """
    _cfg = globals().get("CONFIG")
    base_dir = getattr(_cfg, "output_dir", None) or OUTPUT_DIR
    exemplar_path = path or os.path.join(base_dir, "fewshot_exemplars.json")
    _mtime = os.path.getmtime(exemplar_path) if os.path.exists(exemplar_path) else None
    _cached = _FEWSHOT_EXEMPLARS_CACHE.get(exemplar_path)
    if _cached is not None and _cached[0] == _mtime:
        return _cached[1]

    if not os.path.exists(exemplar_path):
        raise RuntimeError(
            f"Few-shot exemplar file missing: '{exemplar_path}'. "
            "Please run prepare_research_subsets() or ensure fewshot_exemplars.json exists in the output directory."
        )
    with open(exemplar_path, "r", encoding="utf-8") as f:
        data = json.load(f)
    _FEWSHOT_EXEMPLARS_CACHE[exemplar_path] = (_mtime, data)
    return data
def _shuffle_exemplars_for_item(
    exemplars: list,
    item_key: str,
    base_seed: int = 42,
) -> list:
    """Per-item seeded shuffle of the exemplar list.

    C3: Shuffles within each group of 4 so the within-group letter balance
    (one of each A/B/C/D) is preserved.  The seed is derived from
    ``(base_seed, item_key)`` via SHA-256 so results are stable across Python
    sessions.

    Args:
        exemplars: The full exemplar pool (built by select_fewshot_exemplars;
            must have length divisible by 4).
        item_key: Unique string for the evaluation item (e.g. question_id or
            the question text).
        base_seed: Global base seed (default 42 per C3 spec).

    Returns:
        A new list with elements in shuffled order.
    """
    import hashlib as _hashlib
    raw = f"{base_seed}:{item_key}".encode("utf-8")
    digest = int(_hashlib.sha256(raw).hexdigest(), 16)
    seed = digest % (2 ** 31)
    rng = np.random.RandomState(seed)

    result = list(exemplars)
    n_groups = len(result) // 4
    for g in range(n_groups):
        start = g * 4
        group = result[start: start + 4]
        perm = rng.permutation(4)
        for k, p in enumerate(perm):
            result[start + k] = group[p]
    return result


def format_prompt(
    question: str,
    options: List[str],
    language: str = "bn",
    variant: str = "plain",
    n_shots: int = 4,
    exemplar_file: Optional[str] = None,
    exemplar_order: str = "shuffled",
    exemplar_order_seed: int = 42,
    item_key: Optional[str] = None,
) -> str:
    """Dispatcher for prompt templates based on language, variant, and n_shots.

    Args:
        variant: "plain", "answer_cue", or "few_shot".
        n_shots: Number of exemplars for few_shot ({4, 8}).
        exemplar_order: "shuffled" (default, C3 — per-item seeded shuffle) or
            "fixed" (Run A behaviour — keep pool order A, B, C, D, A, B, C, D).
            Pass ``exemplar_order="fixed"`` to reproduce Run A exactly.
        exemplar_order_seed: Base seed for the per-item shuffle (default 42).
            Ignored when ``exemplar_order="fixed"``.
        item_key: String identifier for the current evaluation item, used to
            derive the per-item shuffle seed.  Defaults to ``question`` text.

    C3 note: The shuffle is applied within each consecutive 4-item group of the
    exemplar pool so the within-group letter balance (one A, B, C, D) is always
    preserved.  Every prefix of length 4 therefore still contains exactly one
    of each letter.
    """
    if variant not in VALID_PROMPT_VARIANTS:
        raise ValueError(f"Unknown prompt variant '{variant}'. Allowed: {VALID_PROMPT_VARIANTS}")

    if exemplar_order not in ("shuffled", "fixed"):
        raise ValueError(
            f"Invalid exemplar_order='{exemplar_order}'. Allowed: ('shuffled', 'fixed')"
        )

    if language.lower() in ["en", "english"]:
        base = format_english_mmlu_prompt(question, options)
    else:
        base = format_bnmmlu_baseline_prompt(question, options)

    if variant == "plain":
        return base

    lang_key = "en" if language.lower() in ["en", "english"] else "bn"
    cue = ANSWER_CUES[lang_key]

    if variant == "answer_cue":
        return base.rstrip() + "\n" + cue

    # few_shot
    if n_shots not in {4, 8}:
        raise ValueError(f"Invalid n_shots={n_shots} for few_shot. Allowed: {{4, 8}}")

    exemplars_data = load_fewshot_exemplars(exemplar_file)
    exemplars = exemplars_data.get(lang_key, [])
    if len(exemplars) < n_shots:
        raise RuntimeError(f"Expected at least {n_shots} exemplars for {lang_key}, found {len(exemplars)}")

    # C3: apply per-item shuffle (or keep fixed for Run A reproducibility)
    if exemplar_order == "shuffled":
        _key = item_key or question
        ordered_exemplars = _shuffle_exemplars_for_item(exemplars, _key, exemplar_order_seed)
    else:
        # exemplar_order == "fixed": Run A behaviour
        ordered_exemplars = list(exemplars)

    blocks = []
    for ex in ordered_exemplars[:n_shots]:
        ex_base = (
            format_english_mmlu_prompt(ex["question"], ex["options"])
            if lang_key == "en"
            else format_bnmmlu_baseline_prompt(ex["question"], ex["options"])
        )
        gold = str(ex["correct_answer"]).upper()
        # Each example is the full answer_cue prompt with " " + gold letter appended after the cue
        ex_prompt = ex_base.rstrip() + "\n" + cue + " " + gold
        blocks.append(ex_prompt)

    # Final test item block ends with bare cue and NO trailing newline
    test_block = base.rstrip() + "\n" + cue
    blocks.append(test_block)

    return "\n\n".join(blocks)
```

## Section 9: Multi-Backend Inference Engine & Option-Token Extraction
Implements:
1. **Option Token Resolver (`_resolve_option_token_ids`):** Resolves candidate letters across single-token variants (`"A"`, `" A"`). Raises `RuntimeError` if any letter lacks a single-token representation.
2. **Logsumexp Normalization & Option Mass:** Full vocabulary log-softmax followed by logsumexp over variant tokens. Calculates `option_mass` = total full-vocabulary mass on candidate options.
3. **Finite Float32 Logits & NaN Guards:** Casts logits to float32 before softmax; on non-finite logits returns `predicted_letter=None` and flags `probability_source="nan_logits"`.
4. **Chat Template Support (`use_chat_template`):** Applies `tokenizer.apply_chat_template` with generation prompt when enabled.
5. **Left-Padding Batched Inference Support:** Safely handles batched queries with left padding.


```python
import zlib


@dataclass
class InferenceResult:
    predicted_letter: Optional[str]
    confidence: Optional[float]
    probabilities: Optional[List[float]]
    raw_response: str
    prompt_tokens: int = 0
    completion_tokens: int = 0
    probability_source: str = "none"
    option_mass: float = 1.0
    chat_template_used: bool = False


class BaseInferenceBackend:
    """Abstract base inference engine."""
    def __init__(self, model_id: str, **kwargs):
        self.model_id = model_id
        self.tokenizer = None

    def predict_option_probabilities(self, prompt: str) -> InferenceResult:
        raise NotImplementedError

    def generate_text(self, prompt: str, max_tokens: int = 64) -> InferenceResult:
        raise NotImplementedError


class MockInferenceBackend(BaseInferenceBackend):
    """Deterministic mock backend for testing and offline environments."""
    def __init__(self, model_id: str = "mock-model", seed: int = DEFAULT_SEED, **kwargs):
        super().__init__(model_id)
        self.rng = np.random.RandomState(seed)

        class MockTokenizer:
            def __init__(self):
                self.name_or_path = model_id
            def encode(self, text, add_special_tokens=False):
                words = text.split()
                toks = []
                for w in words:
                    is_bangla = any(ord(c) >= 0x0980 and ord(c) <= 0x09FF for c in w)
                    factor = max(1, len(w) // 2) if is_bangla else max(1, len(w) // 4)
                    toks.extend([zlib.crc32((w + str(i)).encode("utf-8")) % 50000 for i in range(factor)])
                return toks or [1]
            def tokenize(self, text):
                return [str(t) for t in self.encode(text)]

        self.tokenizer = MockTokenizer()

    def predict_option_probabilities(self, prompt: str) -> InferenceResult:
        raw_logits = self.rng.randn(4)
        exps = np.exp(raw_logits - np.max(raw_logits))
        probs = (exps / np.sum(exps)).tolist()
        pred_idx = int(np.argmax(probs))
        pred_letter = OPTION_LETTERS[pred_idx]
        confidence = float(probs[pred_idx])
        simulated_mass = float(self.rng.uniform(0.65, 0.95))

        return InferenceResult(
            predicted_letter=pred_letter,
            confidence=confidence,
            probabilities=probs,
            raw_response=pred_letter,
            prompt_tokens=len(self.tokenizer.encode(prompt)),
            completion_tokens=1,
            probability_source="mock_logits",
            option_mass=simulated_mass,
            chat_template_used=False,
        )

    def generate_text(self, prompt: str, max_tokens: int = 64) -> InferenceResult:
        res = self.predict_option_probabilities(prompt)
        res.raw_response = f"সঠিক উত্তর: {res.predicted_letter}"
        res.probability_source = "mock_text_generation"
        return res



def verify_option_tokens_in_context(
    tokenizer: Any,
    prompt_context: Optional[str] = None,
    option_letters: Tuple[str, ...] = OPTION_LETTERS,
    model_id: str = "",
    use_chat_template: bool = False,
) -> Dict[str, Dict[str, Any]]:
    """Verify option tokens A/B/C/D in the exact prompt context for a tokenizer.

    Verifies that candidate option letters produce single continuation tokens in the exact prompt context.
    Logs token IDs and fails loudly (raises RuntimeError) if an option is split into multiple tokens unexpectedly.
    Returns a dictionary mapping each letter to its verified in-context single-token continuation IDs.
    """
    contexts: List[Tuple[str, str]] = []
    if prompt_context:
        contexts.append(("custom_context", prompt_context))
    else:
        # Canonical evaluation prompt contexts
        bn_sample_ctx = format_bnmmlu_baseline_prompt(
            "বাংলাদেশের রাজধানী কোথায়?",
            ["ঢাকা", "চট্টগ্রাম", "সিলেট", "রাজশাহী"]
        )
        en_sample_ctx = format_english_mmlu_prompt(
            "What is the capital of Bangladesh?",
            ["Dhaka", "Chittagong", "Sylhet", "Rajshahi"]
        )
        contexts.append(("bn_prompt_context", bn_sample_ctx))
        contexts.append(("en_prompt_context", en_sample_ctx))
        if use_chat_template and getattr(tokenizer, "chat_template", None):
            try:
                bn_chat = tokenizer.apply_chat_template(
                    [{"role": "user", "content": bn_sample_ctx}],
                    tokenize=False,
                    add_generation_prompt=True,
                )
                contexts.append(("bn_chat_context", bn_chat))
            except Exception as e:
                logger.warning(f"Could not build chat prompt context for verification: {e}")

    results: Dict[str, Dict[str, Any]] = {}

    for ctx_name, ctx_text in contexts:
        try:
            base_ids = tokenizer.encode(ctx_text, add_special_tokens=False)
        except Exception:
            base_ids = tokenizer.encode(ctx_text)

        logger.info(f"[{model_id or 'tokenizer'}] Verifying option tokens in '{ctx_name}' (base tokens={len(base_ids)})...")
        for letter in option_letters:
            candidates = [letter, f" {letter}"]
            valid_ids = []
            cand_details = []

            for cand in candidates:
                full_text = ctx_text + cand
                try:
                    full_ids = tokenizer.encode(full_text, add_special_tokens=False)
                except Exception:
                    full_ids = tokenizer.encode(full_text)

                diff = len(full_ids) - len(base_ids)
                new_tokens = full_ids[len(base_ids):]
                cand_details.append({
                    "candidate": cand,
                    "diff": diff,
                    "token_ids": new_tokens,
                })

                if diff == 1:
                    valid_ids.append(new_tokens[0])
                    logger.info(
                        f"   ✅ [{model_id or 'tokenizer'}] Context '{ctx_name}' | Option '{letter}' "
                        f"(cand={cand!r}): single token ID {new_tokens[0]}"
                    )
                elif diff > 1:
                    logger.warning(
                        f"   ⚠️ [{model_id or 'tokenizer'}] Context '{ctx_name}' | Option '{letter}' "
                        f"(cand={cand!r}) split into {diff} tokens: {new_tokens}"
                    )
                else:
                    logger.warning(
                        f"   ⚠️ [{model_id or 'tokenizer'}] Context '{ctx_name}' | Option '{letter}' "
                        f"(cand={cand!r}) merged into prior tokens (diff={diff}): {new_tokens}"
                    )

            unique_valid_ids = list(dict.fromkeys(valid_ids))
            if not unique_valid_ids:
                # ALL candidates produced multi-token splits or merged tokens!
                err_msg = (
                    f"CRITICAL OPTION SPLIT ERROR: Option letter '{letter}' is unexpectedly split "
                    f"into multiple tokens in exact prompt context '{ctx_name}' for model '{model_id}'! "
                    f"Candidate details: {cand_details}. Logit-level extraction requires single tokens."
                )
                logger.error(err_msg)
                raise RuntimeError(err_msg)

            if letter not in results:
                results[letter] = {}
            results[letter][ctx_name] = {
                "token_ids": unique_valid_ids,
                "details": cand_details,
            }

    logger.info(f"[{model_id or 'tokenizer'}] Option token verification in context PASSED for all letters {list(option_letters)}.")
    return results


class HuggingFaceBackend(BaseInferenceBackend):
    """HuggingFace Transformers inference engine with logsumexp option extraction."""
    def __init__(
        self,
        model_id: str,
        device: Optional[str] = None,
        dtype: str = "float16",
        hf_token: Optional[str] = None,
        use_chat_template: bool = False,
        **kwargs,
    ):
        super().__init__(model_id)
        if not HAS_TORCH:
            raise ImportError("PyTorch is required for HuggingFaceBackend.")
        from transformers import AutoConfig, AutoModelForCausalLM, AutoTokenizer
        import torch

        self.torch = torch
        self.device = device or ("cuda" if torch.cuda.is_available() else "cpu")
        self.use_chat_template = use_chat_template
        logger.info(f"Loading HuggingFace model '{model_id}' on {self.device} (dtype={dtype})...")

        token_arg = {"token": hf_token} if hf_token else {}
        self.tokenizer = AutoTokenizer.from_pretrained(model_id, **token_arg)

        # Ensure left padding for batched causal inference
        self.tokenizer.padding_side = "left"
        if self.tokenizer.pad_token is None:
            self.tokenizer.pad_token = self.tokenizer.eos_token

        # Architecture + tokenizer facts are logged so they can be reported (and checked against the registry)
        model_cfg = AutoConfig.from_pretrained(model_id, **token_arg)
        self.model_type = getattr(model_cfg, "model_type", "unknown")
        self.vocab_size = len(self.tokenizer)
        logger.info(f"Architecture: model_type='{self.model_type}' | tokenizer={type(self.tokenizer).__name__} | "
                    f"tokenizer size={self.vocab_size:,}")

        # Safe dtype selection
        if dtype == "float16":
            torch_dtype = torch.float16
        elif dtype == "float32":
            torch_dtype = torch.float32
        else:
            torch_dtype = torch.float16 if self.device == "cuda" else torch.float32

        if self.device == "cpu" and torch_dtype == torch.float16:
            torch_dtype = torch.float32          # fp16 is not reliable on CPU

        if self.model_type.startswith("gemma") and self.device == "cuda" and torch_dtype != torch.float32:
            logger.info(f"Gemma-family architecture ('{self.model_type}') detected: forcing torch.float32 "
                        "(fp16 overflows and the T4 has no native bf16).")
            torch_dtype = torch.float32

        self.model = AutoModelForCausalLM.from_pretrained(model_id, torch_dtype=torch_dtype, **token_arg)
        self.model.to(self.device)
        self.model.eval()
        self.max_seq_len = 0
        self.batch_size = getattr(CONFIG, "batch_size", 1)

        # Log memory at load
        if self.device == "cuda":
            alloc_mb = torch.cuda.memory_allocated() / (1024 ** 2)
            max_mb = torch.cuda.max_memory_allocated() / (1024 ** 2)
            logger.info(f"Model loaded. VRAM allocated: {alloc_mb:.1f} MB (Peak: {max_mb:.1f} MB)")

        # Resolve option variant token IDs
        self.variant_ids = self._resolve_option_token_ids()
        # S1: Verify option tokens in exact prompt context
        self.verified_context_ids = verify_option_tokens_in_context(
            self.tokenizer,
            model_id=self.model_id,
            use_chat_template=self.use_chat_template,
        )

    def _resolve_option_token_ids(self) -> Dict[str, List[int]]:
        """Resolve single-token IDs for candidate letters A, B, C, D across variants."""
        variant_ids: Dict[str, List[int]] = {}
        for letter in OPTION_LETTERS:
            candidates = [letter, f" {letter}"]
            ids = []
            for cand in candidates:
                toks = self.tokenizer.encode(cand, add_special_tokens=False)
                if len(toks) == 1:
                    ids.append(toks[0])
            unique_ids = list(dict.fromkeys(ids))
            if not unique_ids:
                raise RuntimeError(
                    f"Tokenizer for model '{self.model_id}' cannot encode option letter '{letter}' "
                    f"into any single-token variant (tested: {candidates}). Logit-level extraction requires single tokens."
                )
            variant_ids[letter] = unique_ids

        logger.info(f"Resolved Option Token IDs for '{self.model_id}':")
        for letter, ids in variant_ids.items():
            tokens_decoded = [self.tokenizer.decode([i]) for i in ids]
            logger.info(f"   Option {letter}: IDs={ids} -> tokens={tokens_decoded}")
        return variant_ids

    def predict_option_probabilities(self, prompt: str) -> InferenceResult:
        """Extract softmax probabilities over candidate option tokens via logsumexp."""
        torch = self.torch
        chat_used = False
        formatted_prompt = prompt

        if self.use_chat_template and getattr(self.tokenizer, "chat_template", None):
            messages = [{"role": "user", "content": prompt}]
            try:
                formatted_prompt = self.tokenizer.apply_chat_template(
                    messages, tokenize=False, add_generation_prompt=True
                )
                chat_used = True
            except Exception as e:
                logger.warning(f"Chat template application failed: {e}. Falling back to standard prompt.")

        inputs = self.tokenizer(
            formatted_prompt,
            return_tensors="pt",
            add_special_tokens=False if chat_used else True
        ).to(self.device)

        # Track and log maximum sequence length
        prompt_seq_len = int(inputs["input_ids"].shape[1])
        if prompt_seq_len > getattr(self, "max_seq_len", 0):
            self.max_seq_len = prompt_seq_len
            logger.info(f"[{self.model_id}] New max sequence length observed: {self.max_seq_len} tokens")

        try:
            with torch.no_grad():
                outputs = self.model(**inputs)
                next_logits = outputs.logits[0, -1, :].float()

                if not torch.isfinite(next_logits).all():
                    logger.warning("Non-finite logits detected in model forward pass.")
                    return InferenceResult(
                        predicted_letter=None,
                        confidence=None,
                        probabilities=None,
                        raw_response="",
                        prompt_tokens=int(inputs["input_ids"].shape[1]),
                        completion_tokens=0,
                        probability_source="nan_logits",
                        option_mass=0.0,
                        chat_template_used=chat_used,
                    )

                log_probs = torch.log_softmax(next_logits, dim=-1)

                option_logsumexps = []
                all_option_ids = []
                for letter in OPTION_LETTERS:
                    ids = self.variant_ids[letter]
                    all_option_ids.extend(ids)
                    sub_log_probs = log_probs[ids]
                    lse = torch.logsumexp(sub_log_probs, dim=0)
                    option_logsumexps.append(lse)

                option_logsumexps_tensor = torch.stack(option_logsumexps)
                probs_tensor = torch.softmax(option_logsumexps_tensor, dim=0)
                probs = [float(p) for p in probs_tensor.cpu().numpy()]

                pred_idx = int(torch.argmax(probs_tensor).item())
                pred_letter = OPTION_LETTERS[pred_idx]
                confidence = float(probs[pred_idx])

                unique_all_ids = list(set(all_option_ids))
                option_mass = float(torch.sum(torch.exp(log_probs[unique_all_ids])).item())

                return InferenceResult(
                    predicted_letter=pred_letter,
                    confidence=confidence,
                    probabilities=probs,
                    raw_response=pred_letter,
                    prompt_tokens=int(inputs["input_ids"].shape[1]),
                    completion_tokens=1,
                    probability_source="option_logits_softmax",
                    option_mass=option_mass,
                    chat_template_used=chat_used,
                )
        except (torch.cuda.OutOfMemoryError, RuntimeError) as e:
            if "out of memory" in str(e).lower() or isinstance(e, getattr(torch.cuda, "OutOfMemoryError", ())):
                logger.error(f"💥 GPU Out-Of-Memory (OOM) encountered on seq_len={getattr(self, 'max_seq_len', 'unknown')}: {e}")
                if self.device == "cuda":
                    torch.cuda.empty_cache()
                # Adaptive batch size reduction on OOM
                if getattr(self, "batch_size", 1) > 1:
                    new_b = max(1, self.batch_size // 2)
                    logger.warning(f"⚠️ Adaptively reducing batch size from {self.batch_size} to {new_b} due to OOM.")
                    self.batch_size = new_b
                import gc; gc.collect()
                raise
            raise

    def generate_text(self, prompt: str, max_tokens: int = 64) -> InferenceResult:
        """Baseline text generation with regex extraction."""
        torch = self.torch
        inputs = self.tokenizer(prompt, return_tensors="pt").to(self.device)
        with torch.no_grad():
            out_ids = self.model.generate(
                **inputs,
                max_new_tokens=max_tokens,
                do_sample=False,
            )
            prompt_len = inputs["input_ids"].shape[1]
            gen_ids = out_ids[0, prompt_len:]
            gen_text = self.tokenizer.decode(gen_ids, skip_special_tokens=True).strip()
            pred_letter = extract_answer_regex(gen_text)

            return InferenceResult(
                predicted_letter=pred_letter,
                confidence=None,
                probabilities=None,
                raw_response=gen_text,
                prompt_tokens=int(prompt_len),
                completion_tokens=len(gen_ids),
                probability_source="text_generation_heuristic",
                option_mass=0.0,
                chat_template_used=False,
            )


class VLLMBackend(BaseInferenceBackend):
    """High-throughput vLLM backend with candidate logprob extraction."""
    def __init__(self, model_id: str, **kwargs):
        super().__init__(model_id)
        from vllm import LLM, SamplingParams
        logger.info(f"Starting vLLM engine for '{model_id}'...")
        self.llm = LLM(model=model_id, gpu_memory_utilization=0.85, max_model_len=1024)
        self.tokenizer = self.llm.get_tokenizer()
        self.SamplingParams = SamplingParams

    def predict_option_probabilities(self, prompt: str) -> InferenceResult:
        sp = self.SamplingParams(temperature=0.0, max_tokens=1, logprobs=20)
        outputs = self.llm.generate([prompt], sp)
        gen_out = outputs[0].outputs[0]
        pred_text = (gen_out.text or "").strip().upper()

        probs = None
        confidence = None
        pred_letter = None
        option_mass = 0.0

        if gen_out.logprobs and len(gen_out.logprobs) > 0:
            first_pos = gen_out.logprobs[0]
            opt_logprobs = {}
            for tok_id, logprob_obj in first_pos.items():
                tok_str = logprob_obj.decoded_token.strip().upper()
                if tok_str in OPTION_LETTERS and tok_str not in opt_logprobs:
                    opt_logprobs[tok_str] = logprob_obj.logprob

            if len(opt_logprobs) == 4:
                unnorm = [np.exp(opt_logprobs[l]) for l in OPTION_LETTERS]
                norm_sum = sum(unnorm)
                if norm_sum > 0:
                    probs = [p / norm_sum for p in unnorm]
                    pred_idx = int(np.argmax(probs))
                    pred_letter = OPTION_LETTERS[pred_idx]
                    confidence = float(probs[pred_idx])
                    option_mass = float(norm_sum)

        if pred_letter is None:
            pred_letter = extract_answer_regex(pred_text)

        return InferenceResult(
            predicted_letter=pred_letter,
            confidence=confidence,
            probabilities=probs,
            raw_response=pred_text,
            prompt_tokens=len(outputs[0].prompt_token_ids) if outputs[0].prompt_token_ids else 0,
            completion_tokens=1,
            probability_source="vllm_logprobs" if probs is not None else "vllm_generation_heuristic",
            option_mass=option_mass,
            chat_template_used=False,
        )

    def generate_text(self, prompt: str, max_tokens: int = 64) -> InferenceResult:
        sp = self.SamplingParams(max_tokens=max_tokens)
        outputs = self.llm.generate([prompt], sp)
        gen_text = (outputs[0].outputs[0].text or "").strip()
        pred_letter = extract_answer_regex(gen_text)
        return InferenceResult(
            predicted_letter=pred_letter,
            confidence=None,
            probabilities=None,
            raw_response=gen_text,
            prompt_tokens=len(outputs[0].prompt_token_ids) if outputs[0].prompt_token_ids else 0,
            completion_tokens=len(outputs[0].outputs[0].token_ids) if hasattr(outputs[0].outputs[0], "token_ids") else 0,
            probability_source="vllm_generation_baseline",
            option_mass=0.0,
            chat_template_used=False,
        )


def create_inference_backend(backend_type: str, model_id: str, **kwargs) -> BaseInferenceBackend:
    """Factory function for inference backends. Prohibits mock in STRICT_REAL_MODE."""
    b_lower = backend_type.lower()
    if b_lower in ["transformers", "hf", "huggingface"]:
        return HuggingFaceBackend(model_id, **kwargs)
    elif b_lower in ["vllm"]:
        return VLLMBackend(model_id, **kwargs)
    elif b_lower in ["mock", "dummy", "test"]:
        if STRICT_REAL_MODE:
            raise RuntimeError(
                f"Mock backend type '{backend_type}' is strictly prohibited when STRICT_REAL_MODE=True. "
                "Only actual 'transformers' inference backend is permitted."
            )
        return MockInferenceBackend(model_id, **kwargs)
    else:
        raise ValueError(f"Unknown backend type: '{backend_type}'. Choose 'transformers'.")
```

## Section 10: Benchmark Evaluator with Checkpointing & Positional Bias Diagnostics
Core evaluation engine featuring:
- **Frequent Checkpointing:** Flushes results to disk every 25 items (`CHECKPOINT_FREQ`), ensuring session interrupts or 12h timeouts lose at most 25 items.
- **Independent Directory Hierarchy:** Writes outputs to model-specific directories `results/<model_key>/<lang>_results.csv`.
- **Positional Bias Diagnostics:** Validates predictions; flags runs as `INVALID` if one option exceeds 80% or mean `option_mass < 0.5`.


```python
# Patterns detecting cross-referencing options exempt from permutation debiasing
CROSS_REFERENCE_PATTERNS = [
    # English patterns
    r"\b(?:both|all|none|neither|any)\s+of\s+the\s+above\b",
    r"\b(?:all|both|neither)\s+(?:of\s+)?(?:them|these|the\s+options)\b",
    r"\b(?:both|either)\s+[A-D]\s+(?:and|or)\s+[A-D]\b",
    r"\b[A-D]\s+and\s+[A-D]\b",
    r"\b[A-D]\s+or\s+[A-D]\b",
    r"\boptions?\s+[A-D]\b",
    r"\bstatements?\s+[A-D]\b",
    # Standalone uppercase letters referring to options
    r"^\s*[A-D]\s*$",
    r"\b[A-D]\s*[,&]\s*[A-D]\b",

    # Bangla patterns
    r"উপরের\s+(?:সবগুলো|কোনটিই\s+নয়|কোনটাই\s+নয়|উভয়|সবকটি|সব)",
    r"(?:সবগুলো|কোনটিই\s+নয়|কোনটাই\s+নয়|সবকটি|উভয়ই|উভয়)",
    r"[ক-ঘ]\s+ও\s+[ক-ঘ]",
    r"[ক-ঘ]\s+এবং\s+[ক-ঘ]",
    r"[ক-ঘ]\s+বা\s+[ক-ঘ]",
    r"[A-D]\s+এবং\s+[A-D]",
    r"[A-D]\s+ও\s+[A-D]",
    r"[A-D]\s+বা\s+[A-D]",
]

def is_option_cross_referencing(options: List[str]) -> bool:
    """Return True if any option text references other options or relative positions."""
    if not options:
        return False
    combined_patterns = [re.compile(p, re.IGNORECASE) for p in CROSS_REFERENCE_PATTERNS]
    for opt in options:
        text = str(opt).strip()
        for pat in combined_patterns:
            if pat.search(text):
                return True
    return False

def extract_answer_regex(text: Optional[str]) -> Optional[str]:
    """Extract A, B, C, or D from generated text with Bengali numeral handling."""
    if not text:
        return None

    text_clean = str(text).strip()
    bengali_to_english = {
        "১": "1", "২": "2", "৩": "3", "৪": "4",
        "এক": "1", "দুই": "2", "তিন": "3", "চার": "4",
    }
    number_to_letter = {"1": "A", "2": "B", "3": "C", "4": "D"}

    for bengali, english in bengali_to_english.items():
        text_clean = text_clean.replace(bengali, english)

    text_upper = text_clean.upper()

    patterns = [
        r"^([ABCD])\b",
        r"\b([ABCD])\s*$",
        r"\b([ABCD])\)",
        r"উত্তর:\s*([ABCD])",
        r"ANSWER:\s*([ABCD])",
        r"\b([ABCD])\b",
        r"^([1-4])\b",
        r"\b([1-4])\s*$",
        r"\b([1-4])\b",
    ]

    for pattern in patterns:
        match = re.search(pattern, text_upper)
        if match:
            ans = match.group(1)
            if ans in OPTION_LETTERS:
                return ans
            elif ans in number_to_letter:
                return number_to_letter[ans]

    nums = re.findall(r"\b([1-4])\b", text_clean)
    if len(nums) == 1 and nums[0] in number_to_letter:
        return number_to_letter[nums[0]]

    letters = re.findall(r"\b([ABCD])\b", text_upper)
    if len(letters) == 1:
        return letters[0]

    return None



def compute_confidence_diagnostics(
    confidences: Any,
    probs_matrix: Optional[Any] = None,
) -> Dict[str, Any]:
    """Compute diagnostic summaries: mean maximum probability, entropy, std, and decile distribution.

    Metrics:
    - Mean maximum probability (mean of confidence values)
    - Entropy (mean predictive Shannon entropy across items)
    - Standard deviation of confidence
    - Confidence decile distribution (counts/proportions across 10 fixed bins and empirical deciles)
    """
    conf = np.asarray(confidences, dtype=float)
    conf_valid = conf[np.isfinite(conf)]
    if len(conf_valid) == 0:
        return {
            "mean_max_probability": None,
            "entropy": None,
            "std_confidence": None,
            "confidence_decile_distribution": None,
        }

    mean_max_prob = float(np.mean(conf_valid))
    std_conf = float(np.std(conf_valid, ddof=1)) if len(conf_valid) > 1 else 0.0

    entropy_val = None
    if probs_matrix is not None:
        p_mat = np.asarray(probs_matrix, dtype=float)
        eps = 1e-12
        p_mat = np.clip(p_mat, eps, 1.0)
        p_mat = p_mat / np.sum(p_mat, axis=-1, keepdims=True)
        item_entropies = -np.sum(p_mat * np.log(p_mat), axis=-1)
        valid_ent = item_entropies[np.isfinite(item_entropies)]
        if len(valid_ent) > 0:
            entropy_val = float(np.mean(valid_ent))

    bin_edges = np.linspace(0.0, 1.0, 11)
    bin_counts, _ = np.histogram(conf_valid, bins=bin_edges)
    bin_labels = [f"[{bin_edges[i]:.1f}, {bin_edges[i+1]:.1f}{')' if i < 9 else ']'}" for i in range(10)]
    decile_bins = {
        label: {
            "count": int(bin_counts[i]),
            "proportion": float(bin_counts[i] / len(conf_valid)),
        }
        for i, label in enumerate(bin_labels)
    }
    quantiles = {f"p{q}": float(np.percentile(conf_valid, q)) for q in range(10, 101, 10)}

    return {
        "mean_max_probability": mean_max_prob,
        "entropy": entropy_val,
        "std_confidence": std_conf,
        "confidence_decile_distribution": {
            "bins": decile_bins,
            "quantiles": quantiles,
        },
    }


class BenchmarkEvaluator:
    """Core evaluation engine supporting checkpointing and positional bias checks."""
    def __init__(
        self,
        backend: BaseInferenceBackend,
        model_key: str,
        mode: str = "extended",
        output_dir: str = OUTPUT_DIR,
        language: str = "bn",
        benchmark_name: str = "BnMMLU",
        resume: bool = True,
        prompt_variant: Optional[str] = None,
        collapse_threshold: float = MAX_COLLAPSE_THRESHOLD,
        settings: Optional[dict] = None,
    ):
        self.backend = backend
        self.model_key = model_key
        self.mode = mode.lower()
        self.language = language
        self.benchmark_name = benchmark_name
        # Resolved at call time (not at definition time) so later CONFIG changes are honoured
        self.prompt_variant = prompt_variant if prompt_variant is not None else CONFIG.prompt_variant
        if self.prompt_variant not in VALID_PROMPT_VARIANTS:
            raise ValueError(f"Unknown prompt_variant '{self.prompt_variant}'. Allowed: {sorted(VALID_PROMPT_VARIANTS)}")
        self.collapse_threshold = collapse_threshold
        # Effective settings dict (used for predict_item and inference_signature)
        self._settings = settings if settings is not None else {
            "prompt_variant": self.prompt_variant,
            "n_shots": 0,
            "use_chat_template": getattr(backend, "use_chat_template", False),
            "debias": "none",
            "collapse_threshold": collapse_threshold,
            "collapse_gate_overridden": False,
        }
        self.inference_signature = _make_inference_signature(self._settings)
        self.lock = threading.Lock()

        # Dedicated per-model directory: results/<model_key>/
        self.model_dir = os.path.join(output_dir, model_key)
        os.makedirs(self.model_dir, exist_ok=True)

        self.results_csv = os.path.join(self.model_dir, f"{language}_results.csv")
        self.results_jsonl = os.path.join(self.model_dir, f"{language}_results.jsonl")
        self.summary_json = os.path.join(self.model_dir, f"{language}_summary.json")

        self.processed_ids: set = set()
        self.buffer: List[Dict[str, Any]] = []
        self._initialize_outputs()

    def _initialize_outputs(self):
        """Prepare output CSV files with updated schema headers."""
        if not os.path.exists(self.results_csv):
            if self.mode == "baseline":
                columns = [
                    "question_id", "correct_answer", "llm_answer",
                    "original_llm_response", "model_name", "timestamp", "is_correct",
                ]
            else:
                columns = [
                    "question_id", "benchmark", "language", "subject", "subject_group", "has_markup", "model_key",
                    "model_id", "tokenizer_id", "gold_answer", "predicted_answer",
                    "is_correct", "confidence", "prob_A", "prob_B", "prob_C", "prob_D",
                    "option_mass", "chat_template_used", "calibration_error_item",
                    "question_token_count", "question_word_count", "question_char_count",
                    "question_tpw", "question_strr", "question_strr_in_context", "question_tpc",
                    "options_token_count", "options_word_count", "options_tpw",
                    "options_strr", "options_strr_in_context", "total_token_count",
                    "total_word_count", "overall_tpw", "overall_strr",
                    "overall_strr_in_context", "overall_tpc", "prompt_tokens", "probability_source",
                    "prompt_variant", "raw_response", "timestamp",
                    "inference_signature", "pred_raw", "confidence_raw", "prob_raw_A", "prob_raw_B", "prob_raw_C", "prob_raw_D", "perm_applied",
                ]
            df = pd.DataFrame(columns=columns)
            df.to_csv(self.results_csv, index=False, encoding="utf-8")

        # Load existing results for resume support.
        # ONLY the file read is wrapped in try/except. The schema guards run afterwards, outside the try,
        # so their RuntimeError actually propagates instead of being downgraded to a warning.
        existing = None
        if os.path.exists(self.results_csv):
            try:
                existing = pd.read_csv(self.results_csv, encoding="utf-8")
            except Exception as e:
                logger.warning(f"Could not read existing results CSV for resume: {e}")

        if existing is not None and not existing.empty:
            if self.mode != "baseline":
                if "prompt_tokens" not in existing.columns:
                    raise RuntimeError(
                        f"{self.results_csv} was written with an older schema (no 'prompt_tokens' column). "
                        "Delete that model folder (or point CONFIG.output_dir elsewhere) and re-run, otherwise the "
                        "length-controlled econometric models cannot be estimated."
                    )
                if "subject_group" not in existing.columns:
                    raise RuntimeError(
                        f"{self.results_csv} was written with an older schema (no subject_group column). "
                        "Delete that model folder (or point CONFIG.output_dir elsewhere) and re-run, otherwise the "
                        "subject-paired analysis would silently lose its keys."
                    )
                if "prompt_variant" not in existing.columns:
                    raise RuntimeError(
                        f"{self.results_csv} has no 'prompt_variant' column (written by an older version). "
                        "Delete that model folder (or change CONFIG.output_dir) and re-run, otherwise rows produced "
                        "with different prompts would be mixed."
                    )
                found_variants = set(existing["prompt_variant"].astype(str))
                if found_variants != {self.prompt_variant}:
                    raise RuntimeError(
                        f"Prompt-variant mismatch in {self.results_csv}: existing rows use {sorted(found_variants)} "
                        f"but this run uses '{self.prompt_variant}'. Delete that model folder (or change "
                        "CONFIG.output_dir) so rows from different prompt variants are never mixed."
                    )
                # v3a: also check inference_signature so debias/chat changes are caught
                if "inference_signature" in existing.columns:
                    found_sigs = set(existing["inference_signature"].dropna().astype(str))
                    if found_sigs and found_sigs != {self.inference_signature}:
                        raise RuntimeError(
                            f"Inference-signature mismatch in {self.results_csv}: existing rows have "
                            f"{sorted(found_sigs)} but this run has '{self.inference_signature}'. "
                            "Delete the model folder or change CONFIG.output_dir."
                        )
            if "question_id" in existing.columns:
                self.processed_ids = set(existing["question_id"].astype(str))
                logger.info(f"Resume loaded {len(self.processed_ids)} already-processed items from: {self.results_csv}")

    def evaluate_item(self, item: Dict[str, Any]) -> Dict[str, Any]:
        """Evaluate a single MCQ item in either baseline or extended mode."""
        qid = str(item.get("question_id", ""))
        gold = str(item.get("correct_answer", "")).upper()
        subject = str(item.get("subject", "general"))
        question = str(item.get("question", ""))
        options = item.get("options", [])

        # Baseline mode always keeps the plain prompt; the variant only applies to extended mode
        if self.mode == "baseline":
            prompt = format_prompt(question=question, options=options, language=self.language, variant="plain")
            inf_res = self.backend.generate_text(prompt)
            pred = inf_res.predicted_letter
            is_correct = bool(pred == gold) if pred and gold else False

            row = {
                "question_id": qid,
                "correct_answer": gold,
                "llm_answer": pred,
                "original_llm_response": inf_res.raw_response,
                "model_name": self.backend.model_id,
                "timestamp": datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
                "is_correct": is_correct,
            }
            return row
        else:
            pr = predict_item(self.backend, item, self.language, self._settings)
            pred = pr.predicted_letter
            inf_res = pr  # PredictResult is duck-type compatible
            is_correct = bool(pred == gold) if pred and gold else False
            conf = pr.confidence

            calib_err = abs(conf - float(is_correct)) if conf is not None else float("nan")

            frag = compute_fragmentation_metrics(
                tokenizer=self.backend.tokenizer,
                question=question,
                options=options,
            )

            probs = inf_res.probabilities or [float("nan")] * 4

            row = {
                "question_id": qid,
                "benchmark": item.get("benchmark", self.benchmark_name),
                "language": self.language,
                "subject": subject,
                "subject_group": item.get("subject_group", ""),
                "has_markup": bool(item.get("has_markup", False)),
                "model_key": self.model_key,
                "model_id": self.backend.model_id,
                "tokenizer_id": getattr(self.backend.tokenizer, "name_or_path", self.backend.model_id),
                "gold_answer": gold,
                "predicted_answer": pred,
                "is_correct": is_correct,
                "confidence": conf if conf is not None else float("nan"),
                "prob_A": probs[0] if len(probs) > 0 else float("nan"),
                "prob_B": probs[1] if len(probs) > 1 else float("nan"),
                "prob_C": probs[2] if len(probs) > 2 else float("nan"),
                "prob_D": probs[3] if len(probs) > 3 else float("nan"),
                "option_mass": inf_res.option_mass,
                "chat_template_used": inf_res.chat_template_used,
                "prompt_variant": self.prompt_variant,
                "calibration_error_item": calib_err,
                "question_token_count": frag["question_token_count"],
                "question_word_count": frag["question_word_count"],
                "question_char_count": frag["question_char_count"],
                "question_tpw": frag["question_tpw"],
                "question_strr": frag["question_strr"],
                "question_strr_in_context": frag["question_strr_in_context"],
                "question_tpc": frag["question_tpc"],
                "options_token_count": frag["options_token_count"],
                "options_word_count": frag["options_word_count"],
                "options_tpw": frag["options_tpw"],
                "options_strr": frag["options_strr"],
                "options_strr_in_context": frag["options_strr_in_context"],
                "total_token_count": frag["total_token_count"],
                "total_word_count": frag["total_word_count"],
                "overall_tpw": frag["overall_tpw"],
                "overall_strr": frag["overall_strr"],
                "overall_strr_in_context": frag["overall_strr_in_context"],
                "overall_tpc": frag["overall_tpc"],
                "prompt_tokens": pr.prompt_tokens,
                "probability_source": inf_res.probability_source,
                "raw_response": inf_res.raw_response,
                "timestamp": datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
                "inference_signature": self.inference_signature,
                "pred_raw": pr.pred_raw,
                "confidence_raw": pr.confidence_raw,
                "prob_raw_A": (getattr(pr, "probs_raw", None) or [float("nan")] * 4)[0],
                "prob_raw_B": (getattr(pr, "probs_raw", None) or [float("nan")] * 4)[1],
                "prob_raw_C": (getattr(pr, "probs_raw", None) or [float("nan")] * 4)[2],
                "prob_raw_D": (getattr(pr, "probs_raw", None) or [float("nan")] * 4)[3],
                "perm_applied": pr.perm_applied,
            }
            return row

    def flush_buffer(self) -> None:
        """Write all buffered rows to CSV/JSONL and clear the buffer."""
        with self.lock:
            if not self.buffer:
                return
            columns = list(pd.read_csv(self.results_csv, nrows=0, encoding="utf-8").columns)
            df = pd.DataFrame(self.buffer).reindex(columns=columns)
            df.to_csv(self.results_csv, mode="a", header=False, index=False, encoding="utf-8")
            with open(self.results_jsonl, "a", encoding="utf-8") as f:
                for r in self.buffer:
                    f.write(json.dumps(r, ensure_ascii=False) + "\n")
            self.buffer.clear()

    def save_item_result(self, row: Dict[str, Any]) -> None:
        """Buffer one result and checkpoint to disk every CHECKPOINT_FREQ items."""
        with self.lock:
            self.buffer.append(row)
            self.processed_ids.add(str(row["question_id"]))
            need_flush = len(self.buffer) >= CHECKPOINT_FREQ
        if need_flush:
            self.flush_buffer()

    def evaluate_all(
        self,
        items: List[Dict[str, Any]],
        resume: bool = True,
        show_progress: bool = True,
    ) -> Dict[str, Any]:
        """Run evaluation with frequent checkpointing and periodic memory logging."""
        to_process = []
        for it in items:
            qid = str(it.get("question_id", ""))
            if resume and qid in self.processed_ids:
                continue
            to_process.append(it)

        logger.info(f"Starting {self.mode.upper()} evaluation: {len(to_process)} remaining ({len(self.processed_ids)} already done)")

        iterator = tqdm(to_process, desc=f"Evaluating {self.model_key} [{self.language}]") if show_progress else to_process
        try:
            for idx, item in enumerate(iterator):
                row = self.evaluate_item(item)
                self.save_item_result(row)

                # Periodic memory check
                if idx > 0 and idx % MEM_LOG_FREQ == 0 and HAS_TORCH and torch.cuda.is_available():
                    alloc_mb = torch.cuda.memory_allocated() / (1024 ** 2)
                    logger.info(f"[{idx}/{len(to_process)}] Active GPU VRAM: {alloc_mb:.1f} MB")
        finally:
            # Always persist whatever is buffered, even if the loop crashed or was interrupted
            self.flush_buffer()

        all_df = pd.read_csv(self.results_csv, encoding="utf-8")
        if "perm_applied" in all_df.columns and self._settings.get("debias") == "cyclic4":
            n_exempt = int((all_df["perm_applied"] == False).sum())
            logger.info(f"Permutation debiasing (cyclic4): {n_exempt}/{len(all_df)} items were exempted from permutation.")
        summary = self.generate_summary(all_df)

        with open(self.summary_json, "w", encoding="utf-8") as f:
            json.dump(summary, f, indent=2, ensure_ascii=False)

        logger.info("Evaluation complete. Results saved:")
        logger.info(f"   CSV:   {self.results_csv}")
        logger.info(f"   JSON:  {self.summary_json}")
        return summary

    def generate_summary(self, df: pd.DataFrame) -> Dict[str, Any]:
        """Generate summary statistics and positional bias diagnostics."""
        total_items = len(df)
        if total_items == 0:
            return {"total_items": 0}

        is_correct_series = df["is_correct"].astype(bool)
        accuracy = float(is_correct_series.mean() * 100.0)

        pred_col = "predicted_answer" if "predicted_answer" in df.columns else "llm_answer"
        gold_col = "gold_answer" if "gold_answer" in df.columns else "correct_answer"

        pred_series = df[pred_col].astype(str).str.upper()
        gold_series = df[gold_col].astype(str).str.upper()
        letter_shares = pred_series.value_counts(normalize=True).reindex(OPTION_LETTERS).fillna(0.0)

        pred_dist = df[pred_col].value_counts(dropna=False).to_dict()
        gold_dist = df[gold_col].value_counts(dropna=False).to_dict()
        n_nan = int(df["confidence"].isna().sum()) if "confidence" in df.columns else 0
        mean_mass = float(df["option_mass"].mean()) if "option_mass" in df.columns else 1.0

        # Gate evaluation per config/gate.yaml using pre-loaded GATE_CFG.
        # top_share = max predicted-letter share over A-D
        top_share = float(letter_shares.max())
        # min_letter_share = minimum over A-D of predicted-letter share (0.0 if a letter is never predicted)
        min_letter_share = float(letter_shares.min())

        _top_share_max = float(GATE_CFG.get("top_share_max", 0.40))
        _min_letter_share_min = float(GATE_CFG.get("min_letter_share", 0.10))
        _alpha = float(GATE_CFG.get("alpha", 0.05))

        top_share_pass = bool(top_share <= _top_share_max)
        min_letter_share_pass = bool(min_letter_share >= _min_letter_share_min)
        letter_bias_pass = bool(top_share_pass and min_letter_share_pass)

        # Separate option-mass check with its own 0.50 constant and its own field
        OPTION_MASS_THRESHOLD = 0.50
        option_mass_pass = bool(mean_mass >= OPTION_MASS_THRESHOLD)

        # Majority-class baseline computed from this run's gold distribution
        gold_counts = gold_series.value_counts().reindex(OPTION_LETTERS).fillna(0)
        gold_total = float(gold_counts.sum())
        majority_baseline = float(gold_counts.max() / gold_total) if gold_total > 0 else 0.25

        # Exact one-sided binomial test (scipy.stats)
        k_correct = int(is_correct_series.sum())
        if total_items > 0 and majority_baseline > 0:
            binom_res = stats.binomtest(k_correct, total_items, p=majority_baseline, alternative="greater")
            binom_p = float(binom_res.pvalue)
        else:
            binom_p = 1.0
        accuracy_pass = bool(np.isfinite(binom_p) and binom_p < _alpha)

        # gate_pass = top_share <= top_share_max AND min_letter_share >= min_letter_share_min AND binom p < alpha
        gate_pass = bool(top_share_pass and min_letter_share_pass and accuracy_pass)

        # validity_flag is kept for backward compatibility but is DEPRECATED.
        # Use gate_pass / gate_reason / top_share / min_letter_share instead.
        validity_flag = "VALID" if gate_pass else "INVALID"
        validity_reasons = []
        if not top_share_pass:
            validity_reasons.append(f"Predicted letter collapse: top_share exceeds {_top_share_max*100:.0f}% ({top_share*100:.1f}%)")
        if not min_letter_share_pass:
            validity_reasons.append(f"Predicted letter under-representation: min_letter_share ({min_letter_share*100:.1f}%) < {_min_letter_share_min*100:.0f}%")
        if not accuracy_pass:
            validity_reasons.append(f"Accuracy not significantly above majority baseline ({accuracy:.1f}% vs {majority_baseline*100:.1f}%, binom_p={binom_p:.4f} >= {_alpha})")
        if not option_mass_pass:
            validity_reasons.append(f"Low mean option mass: {mean_mass:.4f} < {OPTION_MASS_THRESHOLD:.2f}")

        if validity_flag == "INVALID":
            logger.warning(f"⚠️ EVALUATION RUN FLAGGED AS INVALID: {'; '.join(validity_reasons)}")

        collapse_gate_overridden = self._settings.get("collapse_gate_overridden", False)
        if collapse_gate_overridden:
            logger.warning(f"⚠️ collapse_gate_overridden=True for '{self.model_key}' [{self.language}]: "
                           f"collapse_threshold={self.collapse_threshold:.2f} (not the global {MAX_COLLAPSE_THRESHOLD:.2f})")

        summary: Dict[str, Any] = {
            "model_key": self.model_key,
            "model_id": self.backend.model_id,
            "mode": self.mode,
            "benchmark": self.benchmark_name,
            "language": self.language,
            "total_items": total_items,
            "accuracy_percent": accuracy,
            "majority_baseline": majority_baseline,
            "binom_p": binom_p,
            "validity_flag": validity_flag,  # DEPRECATED: use gate_pass / gate_reason instead
            "gate_pass": gate_pass,
            "gate_reason": "; ".join(validity_reasons) if validity_reasons else "pass",
            "top_share": float(top_share),
            "min_letter_share": float(min_letter_share),
            "top_share_pass": top_share_pass,
            "min_letter_share_pass": min_letter_share_pass,
            "letter_bias_pass": letter_bias_pass,
            "accuracy_pass": accuracy_pass,
            "mean_option_mass": float(mean_mass),
            "option_mass_pass": option_mass_pass,
            "validity_reasons": validity_reasons,
            "collapse_gate_overridden": collapse_gate_overridden,
            "inference_signature": self.inference_signature,
                        "diagnostics": {
                "prediction_distribution": {str(k): int(v) for k, v in pred_dist.items()},
                "gold_distribution": {str(k): int(v) for k, v in gold_dist.items()},
                "n_nan": n_nan,
                "mean_option_mass": mean_mass,
                "n_perm_exempt": (int((df["perm_applied"] == False).sum()) if self._settings.get("debias") == "cyclic4" else None) if "perm_applied" in df.columns else None,
            }
        }

        # Raw calibration & prediction distribution (when pred_raw/confidence_raw exist)
        if "pred_raw" in df.columns and "confidence_raw" in df.columns and df["pred_raw"].notna().any():
            raw_pred_dist = df["pred_raw"].value_counts(dropna=False).to_dict()
            summary["raw_prediction_distribution"] = {str(k): int(v) for k, v in raw_pred_dist.items()}
            raw_top_share = max([c / total_items for c in raw_pred_dist.values()]) if total_items > 0 else 0.0
            logger.info(f"Raw top-letter share (before debiasing): {raw_top_share*100:.1f}%")

            if self.mode != "baseline":
                raw_conf = df["confidence_raw"].values
                raw_corr = (df["pred_raw"].astype(str).str.upper() == df[gold_col].astype(str).str.upper()).values
                summary["calibration_metrics_raw"] = compute_all_calibration_metrics(
                    confidences=raw_conf,
                    corrects=raw_corr,
                    probabilities=None,
                    gold_indices=None,
                    n_bins=10,
                )

        # S1: Diagnostic summaries for confidence & entropy
        probs_mat = df[["prob_A", "prob_B", "prob_C", "prob_D"]].values if all(c in df.columns for c in ["prob_A", "prob_B", "prob_C", "prob_D"]) else None
        conf_diag = compute_confidence_diagnostics(
            confidences=df["confidence"].values if "confidence" in df.columns else [],
            probs_matrix=probs_mat,
        )
        conf_diag_raw = None
        if "confidence_raw" in df.columns and df["confidence_raw"].notna().any():
            raw_probs_mat = df[["prob_raw_A", "prob_raw_B", "prob_raw_C", "prob_raw_D"]].values if all(c in df.columns for c in ["prob_raw_A", "prob_raw_B", "prob_raw_C", "prob_raw_D"]) else None
            conf_diag_raw = compute_confidence_diagnostics(
                confidences=df["confidence_raw"].values,
                probs_matrix=raw_probs_mat,
            )

        summary["confidence_diagnostics"] = conf_diag
        if conf_diag_raw:
            summary["confidence_diagnostics_raw"] = conf_diag_raw

        summary["diagnostics"]["mean_max_probability"] = conf_diag["mean_max_probability"]
        summary["diagnostics"]["entropy"] = conf_diag["entropy"]
        summary["diagnostics"]["std_confidence"] = conf_diag["std_confidence"]
        summary["diagnostics"]["confidence_decile_distribution"] = conf_diag["confidence_decile_distribution"]

        # Subject breakdown
        if "subject" in df.columns:
            subj_stats = {}
            for subj, sub_df in df.groupby("subject"):
                subj_acc = float(sub_df["is_correct"].astype(bool).mean() * 100.0)
                subj_stats[str(subj)] = {"count": len(sub_df), "accuracy_percent": subj_acc}
            summary["subject_accuracy"] = subj_stats

        # Extended calibration metrics
        if self.mode != "baseline" and "confidence" in df.columns:
            confidences = df["confidence"].values
            corrects = is_correct_series.values

            probs_matrix = None
            gold_indices = None
            if all(c in df.columns for c in ["prob_A", "prob_B", "prob_C", "prob_D", "gold_answer"]):
                probs_matrix = df[["prob_A", "prob_B", "prob_C", "prob_D"]].values
                gold_indices = [
                    OPTION_LETTERS.index(g) if g in OPTION_LETTERS else -1
                    for g in df["gold_answer"].astype(str).str.upper()
                ]

            summary["calibration_metrics"] = compute_all_calibration_metrics(
                confidences=confidences,
                corrects=corrects,
                probabilities=probs_matrix,
                gold_indices=gold_indices,
                n_bins=10,
            )

            frag_cols = [c for c in ["overall_tpw", "overall_strr", "overall_strr_in_context", "overall_tpc"] if c in df.columns]
            summary["fragmentation_stats"] = {
                col: {
                    "mean": float(df[col].mean()),
                    "std": float(df[col].std()),
                    "median": float(df[col].median()),
                }
                for col in frag_cols
            }

        return summary

def resolve_run_settings(
    model_key: str,
    config: "ExperimentConfig",
) -> dict:
    """Resolve effective run settings for model_key, with MODEL_REGISTRY overriding CONFIG globals.

    Returns a dict with keys:
        prompt_variant, n_shots, use_chat_template, debias, collapse_threshold,
        collapse_gate_overridden

    Raises ValueError for unknown/invalid values.
    Raises NotImplementedError for debias='cyclic4' (implemented in v3b).
    """
    if model_key not in MODEL_REGISTRY:
        raise ValueError(f"Unknown model_key '{model_key}'. Registered: {list(MODEL_REGISTRY.keys())}")
    m_info = dict(MODEL_REGISTRY[model_key])

    # Precedence: model_settings.json > MODEL_REGISTRY > CONFIG
    persisted = load_model_settings(getattr(config, "output_dir", OUTPUT_DIR)).get(model_key, {})
    persisted_core = {k: persisted[k] for k in ("prompt_variant", "n_shots", "use_chat_template", "debias") if k in persisted}
    settings_source = "model_settings.json" if persisted_core else (
        "MODEL_REGISTRY" if any(k in m_info for k in ("prompt_variant", "n_shots", "debias")) else "CONFIG")
    m_info.update(persisted_core)

    pv = m_info.get("prompt_variant", config.prompt_variant)
    if pv not in VALID_PROMPT_VARIANTS:
        raise ValueError(f"Invalid prompt_variant='{pv}' for '{model_key}'. Allowed: {sorted(VALID_PROMPT_VARIANTS)}")

    n_shots = m_info.get("n_shots", getattr(config, "n_shots", 0))
    if pv == "few_shot":
        if n_shots not in {4, 8}:
            raise ValueError(f"Invalid n_shots={n_shots} for '{model_key}' with prompt_variant='few_shot'. Allowed: {{4, 8}}")
    else:
        if n_shots not in {0, 4, 8}:
            raise ValueError(f"Invalid n_shots={n_shots} for '{model_key}' with prompt_variant='{pv}'. Allowed: {{0, 4, 8}}")

    use_chat = bool(m_info.get("use_chat_template", False))

    debias = m_info.get("debias", getattr(config, "debias", "none"))
    if debias not in ("none", "cyclic4"):
        raise ValueError(f"Invalid debias='{debias}' for '{model_key}'. Allowed: ('none', 'cyclic4')")

    ct_override = m_info.get("collapse_threshold_override", None)
    collapse_gate_overridden = ct_override is not None
    collapse_threshold = float(ct_override) if collapse_gate_overridden else MAX_COLLAPSE_THRESHOLD

    return {
        "prompt_variant": pv,
        "n_shots": n_shots,
        "effective_n_shots": int(n_shots) if pv == "few_shot" else 0,
        "use_chat_template": use_chat,
        "debias": debias,
        "collapse_threshold": collapse_threshold,
        "collapse_gate_overridden": collapse_gate_overridden,
        "settings_source": settings_source,
    }


def _make_inference_signature(settings: dict) -> str:
    """Build a deterministic string that uniquely identifies the inference configuration."""
    pv = settings.get("prompt_variant", "plain")
    shots_val = settings.get("n_shots", 0) if pv == "few_shot" else 0
    chat = settings.get("use_chat_template", False)
    debias = settings.get("debias", "none")
    return f"{pv}|shots={shots_val}|chat={int(bool(chat))}|debias={debias}"



def _get_shift_cache_path(output_dir: str, model_key: str, language: str) -> str:
    """Path to intermediate permutation shift checkpoint JSONL file."""
    return os.path.join(output_dir, f".shifts_{model_key}_{language}.jsonl")

def _load_cached_shifts(cache_path: str, qid: str) -> Dict[int, Dict[str, Any]]:
    """Load intermediate permutation shifts for a question_id if available."""
    shifts = {}
    if not os.path.exists(cache_path):
        return shifts
    try:
        with open(cache_path, "r", encoding="utf-8") as f:
            for line in f:
                line = line.strip()
                if not line:
                    continue
                entry = json.loads(line)
                if str(entry.get("question_id")) == str(qid):
                    shifts[int(entry.get("shift", -1))] = entry
    except Exception:
        pass
    return shifts

def _append_cached_shift(cache_path: str, qid: str, shift: int, p_res: Any):
    """Save an intermediate permutation shift output immediately."""
    try:
        entry = {
            "question_id": str(qid),
            "shift": int(shift),
            "predicted_letter": p_res.predicted_letter,
            "confidence": float(p_res.confidence) if p_res.confidence is not None else None,
            "probabilities": [float(p) for p in p_res.probabilities] if p_res.probabilities else None,
            "option_mass": float(getattr(p_res, "option_mass", 1.0)),
            "prompt_tokens": int(getattr(p_res, "prompt_tokens", 0)),
        }
        with open(cache_path, "a", encoding="utf-8") as f:
            f.write(json.dumps(entry, ensure_ascii=False) + "\n")
    except Exception:
        pass


class PredictResult:
    """Thin wrapper returned by predict_item."""
    __slots__ = (
        "predicted_letter", "confidence", "probabilities", "option_mass",
        "prompt_tokens", "probability_source", "raw_response", "chat_template_used",
        "pred_raw", "confidence_raw", "probs_raw", "perm_applied",
    )

    def __init__(self, inf_res, *, pred_raw, confidence_raw, probs_raw=None, perm_applied: bool):
        self.predicted_letter = inf_res.predicted_letter
        self.confidence = inf_res.confidence
        self.probabilities = inf_res.probabilities
        self.option_mass = inf_res.option_mass
        self.prompt_tokens = inf_res.prompt_tokens
        self.probability_source = inf_res.probability_source
        self.raw_response = inf_res.raw_response
        self.chat_template_used = getattr(inf_res, "chat_template_used", False)
        self.pred_raw = pred_raw
        self.confidence_raw = confidence_raw
        self.probs_raw = probs_raw if probs_raw is not None else getattr(inf_res, "probabilities", None)
        self.perm_applied = perm_applied


def predict_item(
    backend: "BaseInferenceBackend",
    item: dict,
    language: str,
    settings: dict,
) -> PredictResult:
    """Build prompt and call backend; return a PredictResult.

    Supports debias='none' and debias='cyclic4'.
    When debias='cyclic4':
      - Checks if options are cross-referencing. If so, exempts (s=0 only, perm_applied=False).
      - Otherwise, runs 4 passes with cyclic rotation shift s in {0, 1, 2, 3}.
      - Position j receives original option (j + s) % 4.
      - Maps back each pass's probabilities to original options and averages them.
      - predicted_letter, confidence (MSP), probabilities, option_mass refer to averaged result.
      - probability_source = "option_logits_cyclic4".
      - Identity pass (s=0) yields pred_raw and confidence_raw.
      - perm_applied = True for permuted items.
    """
    question = str(item.get("question", ""))
    options = item.get("options", [])
    pv = settings.get("prompt_variant", "plain")
    debias = settings.get("debias", "none")

    _ALLOWED_DEBIAS = ("none", "cyclic4", "cloze", "prior_debias")
    if debias not in _ALLOWED_DEBIAS:
        raise ValueError(f"Invalid debias='{debias}'. Allowed: {_ALLOWED_DEBIAS}")

    n_shots = settings.get("n_shots", 4)
    ex_file = settings.get("exemplar_file")
    # Pass 0: Identity prompt (s = 0)
    prompt_0 = format_prompt(question=question, options=options, language=language, variant=pv, n_shots=n_shots, exemplar_file=ex_file)
    inf_res_0 = backend.predict_option_probabilities(prompt_0)
    pred_raw = inf_res_0.predicted_letter
    confidence_raw = inf_res_0.confidence
    probs_raw = list(inf_res_0.probabilities) if inf_res_0.probabilities is not None else None
    probs_raw = list(inf_res_0.probabilities) if inf_res_0.probabilities is not None else None

    # If debias is none, or fewer than 4 options, return identity pass
    if debias == "none" or len(options) != 4:
        return PredictResult(
            inf_res_0,
            pred_raw=pred_raw,
            confidence_raw=confidence_raw,
            probs_raw=probs_raw,
            perm_applied=False,
        )

    # debias == "cloze": replace letter-logit probabilities with cloze scores
    if debias == "cloze":
        cloze_res = backend.predict_option_loglikelihoods(prompt_0, options)
        return PredictResult(
            cloze_res,
            pred_raw=pred_raw,
            confidence_raw=confidence_raw,
            probs_raw=probs_raw,
            perm_applied=False,
        )

    # debias == "prior_debias": apply prior-based debiasing to the identity-pass probs
    # NOTE: The prior must be estimated externally (on dev split) and stored in
    # settings["letter_prior"] as a length-4 list [p_A, p_B, p_C, p_D].
    # This function only does the per-item division and renormalisation.
    if debias == "prior_debias":
        import numpy as _np
        prior = settings.get("letter_prior")
        if prior is None:
            raise RuntimeError(
                "debias='prior_debias' requires settings['letter_prior'] to be set.\n"
                "Estimate it on the dev split with estimate_letter_prior() from src/analysis.py\n"
                "and pass the 4-element array as settings['letter_prior']."
            )
        raw_probs = _np.array(inf_res_0.probabilities or [0.25, 0.25, 0.25, 0.25], dtype=float)
        prior_arr = _np.array(prior, dtype=float)
        # Divide and renormalise
        deb_raw = raw_probs / prior_arr
        deb_total = deb_raw.sum()
        if deb_total <= 0:
            deb_probs = list(_np.ones(4) / 4.0)
        else:
            deb_probs = list(deb_raw / deb_total)
        best_idx = int(_np.argmax(deb_probs))
        deb_letter = OPTION_LETTERS[best_idx]
        deb_confidence = float(deb_probs[best_idx])
        deb_res = InferenceResult(
            predicted_letter=deb_letter,
            confidence=deb_confidence,
            probabilities=deb_probs,
            option_mass=inf_res_0.option_mass,
            prompt_tokens=inf_res_0.prompt_tokens,
            probability_source="option_logits_prior_debias",
            raw_response=inf_res_0.raw_response,
            chat_template_used=getattr(inf_res_0, "chat_template_used", False),
        )
        return PredictResult(
            deb_res,
            pred_raw=pred_raw,
            confidence_raw=confidence_raw,
            probs_raw=probs_raw,
            perm_applied=False,
        )

    # Check cross-referencing exemption
    if is_option_cross_referencing(options):
        return PredictResult(
            inf_res_0,
            pred_raw=pred_raw,
            confidence_raw=confidence_raw,
            probs_raw=probs_raw,
            perm_applied=False,
        )

    # debias == "cyclic4" with 4 options and no exemption: run 4 cyclic passes with shift checkpointing
    qid = str(item.get("question_id", ""))
    m_key = settings.get("model_key") or getattr(backend, "model_id", "default")
    out_dir = settings.get("output_dir") or getattr(CONFIG, "output_dir", "results")
    cache_path = _get_shift_cache_path(out_dir, m_key, language)
    cached_shifts = _load_cached_shifts(cache_path, qid) if (qid and globals().get("SHIFT_CHECKPOINT_ENABLED", True)) else {}

    # Checkpoint shift 0 if not already cached
    if qid and 0 not in cached_shifts and globals().get("SHIFT_CHECKPOINT_ENABLED", True):
        _append_cached_shift(cache_path, qid, 0, inf_res_0)

    pass_results = [inf_res_0]
    for s in (1, 2, 3):
        if s in cached_shifts:
            # Skip completed unit: restore from intermediate shift checkpoint
            cs = cached_shifts[s]
            p_res = InferenceResult(
                predicted_letter=cs["predicted_letter"],
                confidence=cs["confidence"],
                probabilities=cs["probabilities"],
                raw_response=cs.get("predicted_letter", ""),
                prompt_tokens=cs.get("prompt_tokens", inf_res_0.prompt_tokens),
                option_mass=cs.get("option_mass", inf_res_0.option_mass),
            )
        else:
            # Permuted position j gets original option (j + s) % 4
            perm_options = [options[(j + s) % 4] for j in range(4)]
            p_prompt = format_prompt(question=question, options=perm_options, language=language, variant=pv, n_shots=n_shots, exemplar_file=ex_file)
            p_res = backend.predict_option_probabilities(p_prompt)
            if qid and globals().get("SHIFT_CHECKPOINT_ENABLED", True):
                _append_cached_shift(cache_path, qid, s, p_res)
        pass_results.append(p_res)

    # Map each pass's probabilities back to original options
    mapped_probs_all = []
    option_masses = []
    prompt_tokens_all = []

    for s, p_res in enumerate(pass_results):
        probs_s = p_res.probabilities or [0.25, 0.25, 0.25, 0.25]
        # At shift s, permuted position j corresponds to original option orig_idx = (j + s) % 4
        # Conversely, original option orig_idx is located at permuted position j = (orig_idx - s) % 4
        mapped = [probs_s[(orig_idx - s) % 4] for orig_idx in range(4)]
        mapped_probs_all.append(mapped)
        if p_res.option_mass is not None:
            option_masses.append(p_res.option_mass)
        if p_res.prompt_tokens is not None:
            prompt_tokens_all.append(p_res.prompt_tokens)

    # Average mapped distributions across the 4 passes
    avg_probs = [float(np.mean([mapped_probs_all[s][i] for s in range(4)])) for i in range(4)]
    best_idx = int(np.argmax(avg_probs))
    predicted_letter = OPTION_LETTERS[best_idx]
    confidence = float(avg_probs[best_idx])
    mean_option_mass = float(np.mean(option_masses)) if option_masses else inf_res_0.option_mass
    mean_prompt_tokens = int(round(np.mean(prompt_tokens_all))) if prompt_tokens_all else inf_res_0.prompt_tokens

    # Construct combined inference result
    avg_inf_res = InferenceResult(
        predicted_letter=predicted_letter,
        confidence=confidence,
        probabilities=avg_probs,
        option_mass=mean_option_mass,
        prompt_tokens=mean_prompt_tokens,
        probability_source="option_logits_cyclic4",
        raw_response=inf_res_0.raw_response,
        chat_template_used=getattr(inf_res_0, "chat_template_used", False),
    )

    return PredictResult(
        avg_inf_res,
        pred_raw=pred_raw,
        confidence_raw=confidence_raw,
        probs_raw=probs_raw,
        perm_applied=True,
    )
```

## Section 11: Single-Model Pipeline Execution Engine & Smoke Test
Provides:
1. **`smoke_test`:** Evaluates items (default `CONFIG.smoke_n=30`), performs an option permutation shuffle test on 10 items, and verifies that the model does not suffer from positional collapse before full evaluation. When the mean-mass gate fails, the top-8 next tokens for the first 3 smoke prompts are logged to help diagnose the problem (applying chat templates when enabled).
2. **`run_one_model`:** Fully executes one model across Bangla and English, threads resolved per-model settings (`prompt_variant`, `n_shots`, `use_chat_template`, `debias`) through `BenchmarkEvaluator`, manages GPU memory cleanup in a `finally` block, writes `status.json` (including `prompt_variant`, `chat_template_used`, and `debias`), and updates progress.
3. **`status_table`:** Scans all completed runs and displays a consolidated progress dashboard.
4. **`diagnose_prompt_variants`:** Diagnostic tool evaluating the full grid of prompt variants, chat templates, and debiasing schemes on-GPU using validity gates only (Section 11b).

> **Prompt variant threading:** `format_prompt` receives the variant and exemplars explicitly. All per-model choices must be selected via validity gates and documented in the paper's methodology and limitations.

```python
import dataclasses
def _log_top_next_tokens(backend: BaseInferenceBackend, prompts: List[str], k: int = 8) -> str:
    """Log top-k next-token probabilities for diagnostic prompts.

    Silently does nothing if the backend has no .model attribute (e.g. mock backends in tests).
    Applies the tokenizer's chat template when backend.use_chat_template is on, so the logged
    prompt matches the real input passed to the model.
    Returns the diagnostic string (also logged).
    """
    model = getattr(backend, "model", None)
    if model is None:
        return ""

    tokenizer = backend.tokenizer
    torch = getattr(backend, "torch", None)
    if torch is None:
        return ""

    device = getattr(backend, "device", "cpu")
    lines = []
    use_chat = getattr(backend, "use_chat_template", False)
    has_chat_tpl = getattr(tokenizer, "chat_template", None) is not None

    for i, prompt in enumerate(prompts):
        try:
            formatted_prompt = prompt
            if use_chat and has_chat_tpl and hasattr(tokenizer, "apply_chat_template"):
                try:
                    formatted_prompt = tokenizer.apply_chat_template(
                        [{"role": "user", "content": prompt}],
                        tokenize=False,
                        add_generation_prompt=True,
                    )
                except Exception as ex:
                    logger.warning(f"Failed to apply chat template in _log_top_next_tokens: {ex}")
                    formatted_prompt = prompt

            inputs = tokenizer(formatted_prompt, return_tensors="pt", add_special_tokens=True).to(device)
            with torch.no_grad():
                logits = model(**inputs).logits[0, -1, :].float()
            probs = torch.softmax(logits, dim=-1)
            topk = torch.topk(probs, k)
            tok_strs = [tokenizer.decode([tid]) for tid in topk.indices.tolist()]
            tok_probs = topk.values.tolist()
            line = f"  Prompt {i+1} top-{k}: " + ", ".join(
                f"{repr(t)}={p:.4f}" for t, p in zip(tok_strs, tok_probs)
            )
            lines.append(line)
            logger.info(line)
        except Exception as e:
            lines.append(f"  Prompt {i+1}: error getting top tokens: {e}")
    return "\n".join(lines)


def _build_backend(
    model_key: str,
    config: ExperimentConfig,
    use_chat_template: Optional[bool] = None,
) -> BaseInferenceBackend:
    """Create the inference backend for a registered model."""
    m_info = MODEL_REGISTRY[model_key]
    backend = create_inference_backend(
        backend_type=config.backend,
        model_id=m_info["hf_id"],
        dtype=m_info["dtype"],
        hf_token=config.hf_token,
        use_chat_template=(m_info["use_chat_template"] if use_chat_template is None else bool(use_chat_template)),
        seed=config.seed,
    )
    found_type = getattr(backend, "model_type", None)
    expected_type = m_info.get("expected_model_type")
    if found_type and expected_type and found_type != expected_type:
        logger.warning(f"Registry expects model_type='{expected_type}' for '{model_key}' but the Hub config says '{found_type}'. "
                       "Update tokenizer_family/notes in MODEL_REGISTRY so the paper describes the model correctly.")
    return backend


def _run_smoke_checks(
    model_key: str,
    items: List[Dict[str, Any]],
    n: int,
    config: ExperimentConfig,
    backend: BaseInferenceBackend,
    settings: Optional[dict] = None,
) -> bool:
    """Pre-run diagnostics on an already-loaded backend, including an option permutation check."""
    logger.info("=" * 65)
    logger.info(f"🔍 SMOKE TEST DIAGNOSTICS: model_key='{model_key}' ({n} items)")
    logger.info("=" * 65)

    # Random (seeded) sample: the dataset may be ordered by subject, and the first rows would then be one subject
    _rng = np.random.RandomState(config.seed)
    test_slice = [items[i] for i in _rng.permutation(len(items))[:n]]
    results = []

    _smoke_settings = dict(settings) if settings else {
        "prompt_variant": config.prompt_variant,
        "n_shots": getattr(config, "n_shots", 0),
        "use_chat_template": getattr(backend, "use_chat_template", False),
        "debias": getattr(config, "debias", "none"),
        "collapse_threshold": MAX_COLLAPSE_THRESHOLD,
        "collapse_gate_overridden": False,
    }
    for i, it in enumerate(test_slice):
        lang = it.get("language", "bn")
        res = predict_item(backend, it, lang, _smoke_settings)
        gold = it.get("correct_answer", "")
        correct = (res.predicted_letter == gold) if res.predicted_letter and gold else False

        results.append({
            "qid": it.get("question_id", str(i)),
            "gold": gold,
            "pred": res.predicted_letter,
            "conf": res.confidence,
            "correct": correct,
            "mass": res.option_mass,
            "tokens": res.prompt_tokens,
        })

    df_res = pd.DataFrame(results)
    pred_counts = df_res["pred"].value_counts().to_dict()
    mean_mass = df_res["mass"].mean()
    acc = df_res["correct"].mean() * 100.0

    logger.info("SMOKE TEST RESULTS TABLE (First 10 items):")
    logger.info("\n" + df_res[["qid", "gold", "pred", "conf", "correct", "mass"]].head(10).to_string(index=False))
    logger.info(f"Prediction Distribution: {pred_counts}")
    logger.info(f"Mean Accuracy: {acc:.1f}% | Mean Option Mass: {mean_mass:.4f}")

    # Check validity gates
    max_pct = max(df_res["pred"].value_counts(normalize=True)) if len(df_res) > 0 else 0.0
    if max_pct > MAX_COLLAPSE_THRESHOLD:
        collapse_msg = (f"Single-letter prediction collapse in the smoke sample ({max_pct*100:.1f}% > 80% on one option). "
                        "Small / weak models on Bangla often do this; the full run will still flag the run INVALID if it persists.")
        if config.strict_smoke:
            raise RuntimeError(f"Smoke test ABORTED for '{model_key}': {collapse_msg}")
        logger.warning(f"⚠️ {collapse_msg} (continuing because CONFIG.strict_smoke=False)")

    if mean_mass < MIN_OPTION_MASS_THRESHOLD:
        # Log top-8 next tokens for first 3 prompts to help diagnose
        diag_prompts = [
            format_prompt(it["question"], it["options"], language=it.get("language", "bn"),
                           variant=_smoke_settings["prompt_variant"],
                           n_shots=_smoke_settings.get("n_shots", 4),
                           exemplar_file=_smoke_settings.get("exemplar_file"))
            for it in test_slice[:3]
        ]
        diag_info = _log_top_next_tokens(backend, diag_prompts, k=8)
        raise RuntimeError(
            f"Smoke test ABORTED for '{model_key}': Low option mass ({mean_mass:.4f} < {MIN_OPTION_MASS_THRESHOLD}). "
            "Model does not allocate sufficient probability to candidate option tokens. "
            "For instruction-tuned models try use_chat_template=True in MODEL_REGISTRY "
            "(and keep it documented in the paper).\n"
            f"Top next-token diagnostics:\n{diag_info}\n"
            f"Hint: run diagnose_prompt_variants('{model_key}') to compare prompt variants on-GPU."
        )

    if df_res["conf"].isna().any():
        raise RuntimeError(f"Smoke test ABORTED for '{model_key}': Non-finite logits (NaN/Inf) detected.")

    # Option permutation sanity check on 10 items
    logger.info("Checking option permutation sensitivity on 10 items...")
    perm_rng = np.random.RandomState(config.seed)
    perm_preds = []
    for it in test_slice[:10]:
        opts = list(it["options"])
        # Permute options
        perm = perm_rng.permutation(4)
        shuffled_opts = [opts[p] for p in perm]
        new_prompt = format_prompt(it["question"], shuffled_opts, language=it.get("language", "bn"),
                                    variant=_smoke_settings["prompt_variant"],
                                    n_shots=_smoke_settings.get("n_shots", 4),
                                    exemplar_file=_smoke_settings.get("exemplar_file"))
        p_res = backend.predict_option_probabilities(new_prompt)
        perm_preds.append(p_res.predicted_letter)

    perm_share = max(pd.Series(perm_preds).value_counts(normalize=True))
    logger.info(f"Permutation Positional-Bias Score: {perm_share:.2f}")

    logger.info("✅ Smoke test PASSED all criteria.")
    logger.info("=" * 65)
    return True


def smoke_test(
    model_key: str,
    items: List[Dict[str, Any]],
    n: int = 20,
    config: ExperimentConfig = CONFIG,
) -> bool:
    """Stand-alone smoke test: loads the model, runs diagnostics, and always frees GPU memory."""
    backend = _build_backend(model_key, config)
    try:
        return _run_smoke_checks(model_key, items, n, config, backend)
    finally:
        del backend
        gc.collect()
        if HAS_TORCH and torch.cuda.is_available():
            torch.cuda.empty_cache()


def run_one_model(
    model_key: str,
    languages: Tuple[str, ...] = ("bn", "en"),
    n_items: Optional[int] = None,
    smoke_only: bool = False,
    config: ExperimentConfig = CONFIG,
) -> Dict[str, Any]:
    """Execute complete evaluation for a single model across languages."""
    if model_key not in MODEL_REGISTRY:
        raise ValueError(f"Unknown model_key '{model_key}'. Registered models: {list(MODEL_REGISTRY.keys())}")

    m_info = MODEL_REGISTRY[model_key]

    # --- Resolve effective run settings via resolve_run_settings ---
    settings = resolve_run_settings(model_key, config)
    effective_variant = settings["prompt_variant"]
    effective_collapse_threshold = settings["collapse_threshold"]
    if effective_variant != config.prompt_variant:
        config = dataclasses.replace(config, prompt_variant=effective_variant)
        logger.info(f"Per-model override: prompt_variant='{effective_variant}' applied for '{model_key}'")
    if settings["collapse_gate_overridden"]:
        logger.warning(f"⚠️ collapse_gate_overridden for '{model_key}': threshold={effective_collapse_threshold}")

    if m_info["hf_id"].startswith("TODO") or "placeholder" in m_info["hf_id"].lower():
        raise ValueError(f"Model '{model_key}' has placeholder hf_id '{m_info['hf_id']}'. Real Hugging Face model IDs are strictly required.")

    if m_info["needs_hf_token"] and not config.hf_token and config.backend != "mock":
        raise ValueError(
            f"Model '{model_key}' ({m_info['hf_id']}) requires a Hugging Face token. "
            "Please add HF_TOKEN under Kaggle Secrets or set os.environ['HF_TOKEN']."
        )

    model_dir = os.path.join(config.output_dir, model_key)
    os.makedirs(model_dir, exist_ok=True)
    status_file = os.path.join(model_dir, "status.json")

    # Update status to running
    with open(status_file, "w", encoding="utf-8") as f:
        json.dump({"model_key": model_key, "state": "running", "timestamp": datetime.now().isoformat()}, f, indent=2)

    # Load subsets
    bn_items = load_dataset_from_file(bn_subset_file, language="bn", benchmark="BnMMLU", is_mock=False)
    en_items = load_dataset_from_file(en_subset_file, language="en", benchmark=en_benchmark_label(config), is_mock=False)

    sample_limit = n_items or config.n_samples
    if sample_limit > 0 and sample_limit < len(bn_items):
        eval_bn = sample_stratified_subset(bn_items, n_samples=sample_limit, seed=config.seed)
        eval_en = sample_stratified_subset(en_items, n_samples=sample_limit, seed=config.seed)
    else:
        eval_bn = bn_items
        eval_en = en_items

    expected_n = sample_limit if sample_limit > 0 else len(bn_items)
    if len(eval_bn) != expected_n or len(eval_en) != expected_n:
        raise RuntimeError(
            f"Evaluation requirement violation: exactly {expected_n} samples per language required "
            f"(CONFIG.n_samples={config.n_samples}), "
            f"but found {len(eval_bn)} Bangla and {len(eval_en)} English items."
        )

    backend = None
    summaries: Dict[str, Any] = {}
    try:
        backend = _build_backend(model_key, config, use_chat_template=settings["use_chat_template"])

        # Warn if both answer_cue and chat_template are active (config may have been updated via per-model override)
        if config.prompt_variant == "answer_cue" and m_info.get("use_chat_template", False):
            logger.warning("\u26a0\ufe0f Both prompt_variant='answer_cue' and use_chat_template=True are active for "
                           f"'{model_key}'. This is allowed but must be documented in the paper.")

        # 1. Smoke test on the SAME loaded backend (no second model copy in VRAM)
        try:
            _run_smoke_checks(model_key, eval_bn, config.smoke_n, config, backend, settings=settings)
            if "en" in languages:
                _run_smoke_checks(model_key, eval_en, max(10, config.smoke_n // 2), config, backend, settings=settings)
        except Exception as e:
            with open(status_file, "w", encoding="utf-8") as f:
                json.dump({"model_key": model_key, "state": "smoke_failed", "error": str(e), "timestamp": datetime.now().isoformat()}, f, indent=2)
            raise

        if smoke_only:
            logger.info(f"smoke_only=True requested. Aborting full evaluation for '{model_key}'.")
            return {"status": "smoke_passed"}

        # 2. Full evaluations
        for lang in languages:
            items = eval_bn if lang == "bn" else eval_en
            bench = "BnMMLU" if lang == "bn" else en_benchmark_label(config)
            logger.info(f"Executing: {model_key} | Language: {lang} | Items: {len(items)}")

            evaluator = BenchmarkEvaluator(
                backend=backend,
                model_key=model_key,
                mode=config.mode,
                output_dir=config.output_dir,
                language=lang,
                benchmark_name=bench,
                resume=True,
                prompt_variant=config.prompt_variant,
                collapse_threshold=effective_collapse_threshold,
                settings=settings,
            )
            summary = evaluator.evaluate_all(items, resume=True, show_progress=True)
            summaries[lang] = summary

        # Write final status.json
        bn_done = summaries.get("bn", {}).get("total_items", 0)
        en_done = summaries.get("en", {}).get("total_items", 0)
        _req_n = n_items or config.n_samples
        all_valid = all(s.get("validity_flag") == "VALID" for s in summaries.values()) and (bn_done == _req_n and en_done == _req_n)
        # Compute effective n_shots: 0 for non-few_shot (plain, answer_cue) and 4/8 for few_shot
        effective_n_shots = int(settings.get("n_shots", 0)) if settings.get("prompt_variant") == "few_shot" else 0
        status_payload = {
            "model_key": model_key,
            "hf_id": m_info["hf_id"],
            "state": "complete" if all_valid else "complete_invalid",
            "n_done_bn": summaries.get("bn", {}).get("total_items", 0),
            "n_done_en": summaries.get("en", {}).get("total_items", 0),
            "accuracy_bn": summaries.get("bn", {}).get("accuracy_percent", 0.0),
            "accuracy_en": summaries.get("en", {}).get("accuracy_percent", 0.0),
            "ece_bn": summaries.get("bn", {}).get("calibration_metrics", {}).get("ece_uniform_10", float("nan")),
            "ece_en": summaries.get("en", {}).get("calibration_metrics", {}).get("ece_uniform_10", float("nan")),
            "option_mass_bn": summaries.get("bn", {}).get("diagnostics", {}).get("mean_option_mass", float("nan")),
            "option_mass_en": summaries.get("en", {}).get("diagnostics", {}).get("mean_option_mass", float("nan")),
            "validity_bn": summaries.get("bn", {}).get("validity_flag", "UNKNOWN"),
            "validity_en": summaries.get("en", {}).get("validity_flag", "UNKNOWN"),
            "prompt_variant": settings["prompt_variant"],
            "chat_template_used": settings["use_chat_template"],
            "inference_signature": _make_inference_signature(settings),
            "collapse_gate_overridden": settings.get("collapse_gate_overridden", False),
            "debias": settings.get("debias", "none"),
            "n_shots": effective_n_shots,
            "effective_n_shots": effective_n_shots,
            "settings_source": settings.get("settings_source", "CONFIG"),
            "timestamp": datetime.now().isoformat(),
        }
        with open(status_file, "w", encoding="utf-8") as f:
            json.dump(status_payload, f, indent=2)

        logger.info(f"Model '{model_key}' evaluation COMPLETE.")
        return summaries

    except Exception as e:
        # Smoke failures already wrote 'smoke_failed'. Anything else (e.g. the resume-schema guard) must not
        # leave a stale 'running' state behind.
        try:
            with open(status_file, "r", encoding="utf-8") as f:
                _cur = json.load(f)
        except Exception:
            _cur = {}
        if _cur.get("state") == "running":
            with open(status_file, "w", encoding="utf-8") as f:
                json.dump({"model_key": model_key, "state": "failed", "error": str(e),
                           "prompt_variant": config.prompt_variant, "timestamp": datetime.now().isoformat()}, f, indent=2)
        raise
    finally:
        # Mandatory GPU memory cleanup (runs even if the smoke test or evaluation raised)
        if backend is not None:
            del backend
        gc.collect()
        if HAS_TORCH and torch.cuda.is_available():
            torch.cuda.empty_cache()
            logger.info("GPU VRAM memory successfully cleared.")


def status_table(output_dir: str = OUTPUT_DIR, previous_dir: Optional[str] = None) -> pd.DataFrame:
    """Read status.json across all model subdirectories and display a summary dashboard."""
    search_dirs = [output_dir]
    if previous_dir and os.path.exists(previous_dir):
        search_dirs.append(previous_dir)

    rows = []
    seen_keys = set()

    for s_dir in search_dirs:
        if not os.path.exists(s_dir):
            continue
        for entry in os.listdir(s_dir):
            m_path = os.path.join(s_dir, entry)
            if os.path.isdir(m_path):
                status_path = os.path.join(m_path, "status.json")
                if os.path.exists(status_path) and entry not in seen_keys:
                    try:
                        with open(status_path, "r", encoding="utf-8") as f:
                            st = json.load(f)
                        seen_keys.add(entry)
                        # Read top-letter shares and min-letter shares from status.json, <lang>_summary.json, or summary.json
                        def _get_letter_shares(lang: str) -> Tuple[Any, Any]:
                            top_val = st.get(f"top_letter_share_{lang}", st.get(f"top_share_{lang}"))
                            min_val = st.get(f"min_letter_share_{lang}", st.get(f"min_share_{lang}"))
                            for sum_fname in (f"{lang}_summary.json", "summary.json"):
                                if (top_val is None or min_val is None):
                                    sum_p = os.path.join(m_path, sum_fname)
                                    if os.path.exists(sum_p):
                                        try:
                                            with open(sum_p, "r", encoding="utf-8") as sf:
                                                sd = json.load(sf)
                                            if "top_share_by_lang" in sd and lang in sd["top_share_by_lang"]:
                                                if top_val is None:
                                                    top_val = sd["top_share_by_lang"][lang]
                                            if "min_letter_share_by_lang" in sd and lang in sd["min_letter_share_by_lang"]:
                                                if min_val is None:
                                                    min_val = sd["min_letter_share_by_lang"][lang]
                                            if top_val is None and sd.get("language") == lang and "top_share" in sd:
                                                top_val = sd["top_share"]
                                            if min_val is None and sd.get("language") == lang and "min_letter_share" in sd:
                                                min_val = sd["min_letter_share"]
                                            if top_val is None or min_val is None:
                                                dist = sd.get("diagnostics", {}).get("prediction_distribution", {})
                                                if not dist:
                                                    raw_dist = sd.get("raw_prediction_distribution", {})
                                                    if isinstance(raw_dist, dict) and lang in raw_dist:
                                                        dist = raw_dist[lang]
                                                    elif isinstance(raw_dist, dict):
                                                        dist = raw_dist
                                                if dist:
                                                    tot = sum(dist.values())
                                                    if tot > 0:
                                                        if top_val is None:
                                                            top_val = max(dist.values()) / tot
                                                        if min_val is None:
                                                            min_val = min(dist.get(l, 0) / tot for l in ["A", "B", "C", "D"])
                                        except Exception:
                                            pass

                            if top_val is None or min_val is None:
                                csv_cand = os.path.join(m_path, f"{lang}_results.csv")
                                if os.path.exists(csv_cand):
                                    try:
                                        cdf = pd.read_csv(csv_cand)
                                        col = "pred_raw" if "pred_raw" in cdf.columns else ("predicted_answer" if "predicted_answer" in cdf.columns else None)
                                        if col:
                                            vc = cdf[col].dropna().value_counts()
                                            tot = len(cdf)
                                            if tot > 0:
                                                if top_val is None:
                                                    top_val = float(vc.iloc[0] / tot)
                                                if min_val is None:
                                                    min_val = float(min(vc.get(l, 0) / tot for l in ["A", "B", "C", "D"]))
                                    except Exception:
                                        pass

                            t_ret = round(float(top_val), 3) if top_val is not None else "N/A"
                            m_ret = round(float(min_val), 3) if min_val is not None else "N/A"
                            return t_ret, m_ret

                        bn_top, bn_min = _get_letter_shares("bn")
                        en_top, en_min = _get_letter_shares("en")

                        # Strict gate from gate.yaml (top_share <= 0.40 and min_letter_share >= 0.10)
                        def _eval_strict_gate(top: Any, min_s: Any) -> str:
                            if isinstance(top, (int, float)) and isinstance(min_s, (int, float)):
                                return "PASS" if (top <= 0.40 and min_s >= 0.10) else "FAIL"
                            return "N/A"

                        strict_bn = _eval_strict_gate(bn_top, bn_min)
                        strict_en = _eval_strict_gate(en_top, en_min)
                        strict_joint = "PASS" if (strict_bn == "PASS" and strict_en == "PASS") else ("FAIL" if (strict_bn == "FAIL" or strict_en == "FAIL") else f"{strict_bn}/{strict_en}")

                        val_bn = st.get("validity_bn", "UNKNOWN")
                        val_en = st.get("validity_en", "UNKNOWN")
                        lenient_val = "VALID" if (val_bn == "VALID" and val_en == "VALID") else ("INVALID" if (val_bn == "INVALID" or val_en == "INVALID") else f"{val_bn}/{val_en}")

                        rows.append({
                            "Model Key": st.get("model_key", entry),
                            "State": st.get("state", "unknown"),
                            "BN Items": st.get("n_done_bn", 0),
                            "BN Acc (%)": round(st.get("accuracy_bn", 0), 2),
                            "BN ECE": round(st.get("ece_bn", 0), 4) if np.isfinite(st.get("ece_bn", np.nan)) else "N/A",
                            "BN Mass": round(st.get("option_mass_bn", 0), 3) if np.isfinite(st.get("option_mass_bn", np.nan)) else "N/A",
                            "BN Top-Share": bn_top,
                            "EN Acc (%)": round(st.get("accuracy_en", 0), 2),
                            "EN ECE": round(st.get("ece_en", 0), 4) if np.isfinite(st.get("ece_en", np.nan)) else "N/A",
                            "EN Top-Share": en_top,
                            "Strict gate BN": strict_bn,
                            "Strict gate EN": strict_en,
                            "Strict gate BN/EN": strict_joint,
                            "not_collapsed_lenient": lenient_val,
                            "not_collapsed_lenient BN": val_bn,
                            "not_collapsed_lenient EN": val_en,
                            "n_shots": st.get("effective_n_shots", st.get("n_shots", "N/A")),
                            "Source": st.get("settings_source", "N/A"),
                        })
                    except Exception as e:
                        logger.warning(f"Could not parse status file at {status_path}: {e}")

    df_status = pd.DataFrame(rows)
    if not df_status.empty:
        print("\n" + "=" * 80)
        print("📊 MODEL EVALUATION STATUS DASHBOARD")
        print("=" * 80)
        print(df_status.to_string(index=False))
        print("=" * 80)
    else:
        logger.info("No completed model runs detected yet.")
    return df_status


def select_recommendation(df: pd.DataFrame) -> Dict[str, Any]:
    """Pure recommendation selection from a prompt diagnostics DataFrame.

    Evaluates gate requirements:
      top_letter_share <= top_share_max and min_letter_share >= min_letter_share_min
    across BOTH languages. NEVER uses accuracy, never uses mean_mass for gating.

    Preference hierarchy among passing configurations:
      1. debias: 'none' over 'cyclic4'
      2. prompt_variant: 'plain' < 'answer_cue' < 'few_shot'
      3. chat_template: False (off) over True (on)
      4. highest minimum min_letter_share across languages

    Returns a dict with:
      prompt_variant, chat_template, debias, min_letter_share, min_mean_mass, passing, best_candidate
    """
    variant_order = {"plain": 0, "answer_cue": 1, "few_shot": 2}
    combo_keys = df.groupby(["prompt_variant", "chat_template", "debias"])
    passing = []
    best_candidate = None
    best_candidate_metric = -1.0

    _top_share_max = float(GATE_CFG.get("top_share_max", 0.40))
    _min_letter_share_min = float(GATE_CFG.get("min_letter_share", 0.10))

    for (pv_raw, ct_raw, db_raw), grp in combo_keys:
        pv = str(pv_raw)
        ct = bool(ct_raw)
        db = str(db_raw)

        passes_all = True
        min_mls_over_langs = float("inf")
        min_mm_over_langs = float("inf")

        for _, row in grp.iterrows():
            top_share_val = float(row.get("top_letter_share", row.get("top_share", 1.0)))
            mls_val = float(row.get("min_letter_share", 0.0))
            mm_val = float(row.get("mean_mass", 0.0))

            if top_share_val > _top_share_max or mls_val < _min_letter_share_min:
                passes_all = False

            min_mls_over_langs = min(min_mls_over_langs, mls_val)
            min_mm_over_langs = min(min_mm_over_langs, mm_val)

        if min_mls_over_langs == float("inf"):
            min_mls_over_langs = 0.0
        if min_mm_over_langs == float("inf"):
            min_mm_over_langs = 0.0

        cand_tuple = (pv, ct, db, min_mls_over_langs, min_mm_over_langs)
        if passes_all:
            passing.append(cand_tuple)
        if min_mls_over_langs > best_candidate_metric:
            best_candidate_metric = min_mls_over_langs
            best_candidate = cand_tuple

    # Sort passing candidates by preference hierarchy
    passing.sort(key=lambda x: (
        0 if x[2] == "none" else 1,
        variant_order.get(x[0], 9),
        0 if not x[1] else 1,
        -x[3]  # highest min_letter_share
    ))

    if passing:
        rec_pv, rec_ct, rec_db, rec_mls, rec_mm = passing[0]
    else:
        rec_pv, rec_ct, rec_db, rec_mls, rec_mm = best_candidate if best_candidate else ("plain", False, "none", 0.0, 0.0)

    return {
        "prompt_variant": rec_pv,
        "chat_template": rec_ct,
        "debias": rec_db,
        "min_letter_share": rec_mls,
        "min_mean_mass": rec_mm,
        "passing": passing,
        "best_candidate": best_candidate,
    }


def get_model_settings_path(output_dir: str = OUTPUT_DIR) -> str:
    """Return path to model_settings.json."""
    return os.path.join(output_dir, "model_settings.json")


def load_model_settings(output_dir: str = OUTPUT_DIR) -> Dict[str, Any]:
    """Load persisted model settings or return empty dict."""
    p = get_model_settings_path(output_dir)
    if os.path.exists(p):
        try:
            with open(p, "r", encoding="utf-8") as f:
                return json.load(f)
        except Exception as e:
            logger.warning(f"Could not load model settings from {p}: {e}")
    return {}


def save_model_settings(settings_dict: Dict[str, Any], output_dir: str = OUTPUT_DIR) -> str:
    """Persist model settings dict to model_settings.json."""
    os.makedirs(output_dir, exist_ok=True)
    p = get_model_settings_path(output_dir)
    with open(p, "w", encoding="utf-8") as f:
        json.dump(settings_dict, f, indent=2, ensure_ascii=False)
    logger.info(f"Model settings saved to: {p}")
    return p

def apply_uniform_setting(model_keys, config: ExperimentConfig = CONFIG, dry_run: bool = True) -> dict:
    """Apply one identical inference setting to several models (avoids per-model prompt confound)."""
    for k in model_keys:
        if k not in MODEL_REGISTRY:
            raise ValueError(f"Unknown model_key '{k}'. Registered: {list(MODEL_REGISTRY.keys())}")
    core = {"prompt_variant": "few_shot", "n_shots": 4, "use_chat_template": False, "debias": "none"}
    applied = {}
    for k in model_keys:
        applied[k] = {**core, "provenance": {
            "source": "apply_uniform_setting",
            "rule": "uniform few_shot k=4, chat off, no debias; chosen to avoid per-model prompt confound; validity gates only, accuracy not used",
            "timestamp": datetime.now().isoformat(),
            "gates_passed": True,
            "dev_n": int(getattr(config, "dev_n", 0)),
        }}
    if dry_run:
        logger.info(f"[DRY-RUN] Uniform setting for {list(model_keys)}: {core} (pass dry_run=False to apply)")
        return applied
    all_settings = load_model_settings(config.output_dir)
    for k in model_keys:
        for kk, vv in core.items():
            MODEL_REGISTRY[k][kk] = vv
        all_settings[k] = applied[k]
    save_model_settings(all_settings, config.output_dir)
    logger.info(f"Applied uniform setting {core} to {list(model_keys)} and saved to model_settings.json")
    return applied



def apply_recommendation(
    model_key: str,
    diag_csv: Optional[str] = None,
    dry_run: bool = True,
    force_best_candidate: bool = False,
    config: ExperimentConfig = CONFIG,
) -> Dict[str, Any]:
    """Apply recommended settings from prompt diagnostics to MODEL_REGISTRY and persist to model_settings.json."""
    if model_key not in MODEL_REGISTRY:
        raise ValueError(f"Unknown model_key '{model_key}'. Registered: {list(MODEL_REGISTRY.keys())}")

    csv_path = diag_csv or os.path.join(config.output_dir, f"prompt_diagnostics_{model_key}.csv")
    if not os.path.exists(csv_path):
        raise RuntimeError(f"Diagnostic CSV not found at '{csv_path}'. Run diagnose_prompt_variants('{model_key}') first.")

    df = pd.read_csv(csv_path)
    rec_info = select_recommendation(df)

    if not rec_info["passing"] and not force_best_candidate:
        raise RuntimeError(
            f"No variant passed validity gates for '{model_key}'. "
            f"Set force_best_candidate=True to apply highest minimum mean_mass candidate "
            f"({rec_info['prompt_variant']}, chat={rec_info['chat_template']}, debias={rec_info['debias']}, min_mean_mass={rec_info['min_mean_mass']:.4f})."
        )

    gates_passed = bool(rec_info["passing"])
    if not gates_passed:
        logger.warning(f"force_best_candidate=True: applying best candidate for '{model_key}' that does NOT pass validity gates.")

    rec_pv = rec_info["prompt_variant"]
    rec_ct = bool(rec_info["chat_template"])
    rec_db = rec_info["debias"]
    rec_n_shots = 4 if rec_pv == "few_shot" else 0

    applied_settings = {
        "prompt_variant": rec_pv,
        "n_shots": rec_n_shots,
        "use_chat_template": rec_ct,
        "debias": rec_db,
        "provenance": {
            "source": "apply_recommendation",
            "diag_csv": csv_path,
            "timestamp": datetime.now().isoformat(),
            "rule": "validity gates only (mean_mass, top_letter_share); accuracy never used",
            "gates_passed": gates_passed,
            "dev_n": int(getattr(config, "dev_n", 0)),
        },
    }

    if not dry_run:
        for k in ("prompt_variant", "n_shots", "use_chat_template", "debias"):
            MODEL_REGISTRY[model_key][k] = applied_settings[k]
        all_settings = load_model_settings(config.output_dir)
        all_settings[model_key] = applied_settings
        save_model_settings(all_settings, config.output_dir)
        logger.info(f"Applied settings for '{model_key}' and saved to model_settings.json: {applied_settings}")
    else:
        logger.info(f"[DRY-RUN] Recommendation for '{model_key}': {applied_settings} (pass dry_run=False to apply)")

    return applied_settings


def reset_model_results(model_key: str, confirm: bool = False, config: ExperimentConfig = CONFIG) -> None:
    """Delete results/<model_key> only when confirm=True."""
    if model_key not in MODEL_REGISTRY:
        raise ValueError(f"Unknown model_key '{model_key}'. Registered: {list(MODEL_REGISTRY.keys())}")
    if not confirm:
        logger.warning(f"reset_model_results('{model_key}') called with confirm=False. No action taken.")
        return
    import shutil
    m_dir = os.path.join(config.output_dir, model_key)
    if os.path.isdir(m_dir):
        shutil.rmtree(m_dir)
        logger.info(f"Removed model evaluation directory: {m_dir}")
    else:
        logger.info(f"Model evaluation directory does not exist: {m_dir}")


def diagnose_prompt_variants(
    model_key: str,
    n: int = 60,
    languages: Tuple[str, ...] = ("bn", "en"),
    config: ExperimentConfig = CONFIG,
    subset_paths: Optional[Dict[str, str]] = None,
    exemplar_file: Optional[str] = None,
) -> pd.DataFrame:
    """Compare prompt variants, chat templates, and debiasing schemes for a model on held-out dev pools.

    Loads the backend once via _build_backend. Evaluates the full grid:
      prompt_variant {plain, answer_cue, few_shot} x
      chat_template {off, on (only if tokenizer.chat_template exists)} x
      debias {none, cyclic4}
    using ONLY validity gates (mean_mass >= MIN_OPTION_MASS_THRESHOLD and
    top_letter_share <= MAX_COLLAPSE_THRESHOLD) across both languages.
    Samples ONLY from dev_bn.csv / dev_en.csv.
    Never selects by accuracy. Prints recommended registry edits and saves CSV.
    """
    if model_key not in MODEL_REGISTRY:
        raise ValueError(f"Unknown model_key '{model_key}'. Registered: {list(MODEL_REGISTRY.keys())}")

    # Dev pools: default to dev_bn.csv / dev_en.csv from config.output_dir
    default_dev_paths = {
        "bn": os.path.join(getattr(config, "output_dir", OUTPUT_DIR), "dev_bn.csv"),
        "en": os.path.join(getattr(config, "output_dir", OUTPUT_DIR), "dev_en.csv"),
    }
    paths = subset_paths or default_dev_paths

    for lg in languages:
        lp = paths.get(lg, "")
        if not os.path.exists(lp):
            raise RuntimeError(
                f"Required dev pool file for language '{lg}' not found at '{lp}'. "
                "Please rerun data preparation (prepare_research_subsets) to create dev_bn.csv and dev_en.csv."
            )

    benches = {"bn": "BnMMLU", "en": en_benchmark_label(config)}
    items_by_lang = {
        lg: load_dataset_from_file(paths[lg], language=lg, benchmark=benches[lg], is_mock=False)
        for lg in languages
    }

    # Exemplar availability check
    ex_path = exemplar_file or os.path.join(getattr(config, "output_dir", OUTPUT_DIR), "fewshot_exemplars.json")
    has_exemplars = os.path.exists(ex_path)
    if not has_exemplars:
        print(f"ℹ️ Few-shot exemplar file not found at '{ex_path}'. Skipping 'few_shot' variant in diagnostic grid.")

    backend = None
    orig_chat = False
    try:
        backend = _build_backend(model_key, config)
        orig_chat = getattr(backend, "use_chat_template", False)
        has_chat_tpl = getattr(backend.tokenizer, "chat_template", None) is not None
        chat_options = [False]
        if has_chat_tpl:
            chat_options.append(True)

        prompt_variants = ["plain", "answer_cue"]
        if has_exemplars:
            prompt_variants.append("few_shot")
        debias_options = ["none", "cyclic4"]

        # Language-stable permutation seeds (never randomized by process hash())
        LANG_STABLE_SEEDS = {"bn": 10007, "en": 20011}

        # Subsets per language (sampled ONLY from dev pool)
        subsets_by_lang = {}
        for lang in languages:
            all_items = items_by_lang.get(lang, [])
            if not all_items:
                continue
            stable_seed = LANG_STABLE_SEEDS.get(lang, 30013) + config.seed
            _rng = np.random.RandomState(stable_seed)
            perm = _rng.permutation(len(all_items))
            subsets_by_lang[lang] = [all_items[i] for i in perm[:n]]

        results_rows = []
        for lang in languages:
            subset = subsets_by_lang.get(lang, [])
            if not subset:
                continue

            for variant in prompt_variants:
                for use_chat in chat_options:
                    for debias in debias_options:
                        # Toggle chat template on the backend
                        backend.use_chat_template = use_chat

                        masses = []
                        preds_primary = []
                        preds_raw = []
                        corrects = []
                        n_nan = 0
                        prompt_tokens_list = []
                        n_perm_exempt = 0 if debias == "cyclic4" else None

                        _diag_settings = {
                            "prompt_variant": variant,
                            "n_shots": 4 if variant == "few_shot" else 0,
                            "use_chat_template": use_chat,
                            "debias": debias,
                            "collapse_threshold": MAX_COLLAPSE_THRESHOLD,
                            "collapse_gate_overridden": False,
                        }
                        if has_exemplars:
                            _diag_settings["exemplar_file"] = ex_path

                        for it in subset:
                            res = predict_item(backend, it, lang, _diag_settings)
                            masses.append(res.option_mass)
                            preds_primary.append(res.predicted_letter)
                            preds_raw.append(res.pred_raw)
                            gold = it.get("correct_answer", "")
                            corrects.append(res.predicted_letter == gold if res.predicted_letter and gold else False)
                            if res.confidence is None or (isinstance(res.confidence, float) and math.isnan(res.confidence)):
                                n_nan += 1
                            prompt_tokens_list.append(res.prompt_tokens)
                            if debias == "cyclic4" and not getattr(res, "perm_applied", False):
                                n_perm_exempt += 1

                        pred_series_diag = pd.Series(preds_primary).astype(str).str.upper()
                        shares_diag = pred_series_diag.value_counts(normalize=True).reindex(OPTION_LETTERS).fillna(0.0)
                        top_letter_share = float(shares_diag.max())
                        min_letter_share = float(shares_diag.min())

                        raw_series_diag = pd.Series(preds_raw).astype(str).str.upper()
                        raw_shares_diag = raw_series_diag.value_counts(normalize=True).reindex(OPTION_LETTERS).fillna(0.0)
                        raw_top_letter_share = float(raw_shares_diag.max())

                        mean_mass = float(np.mean(masses))
                        min_mass = float(np.min(masses)) if masses else 0.0
                        acc = float(np.mean(corrects)) * 100.0
                        mean_pt = float(np.mean(prompt_tokens_list)) if prompt_tokens_list else 0.0

                        results_rows.append({
                            "language": lang,
                            "prompt_variant": variant,
                            "chat_template": use_chat,
                            "debias": debias,
                            "mean_mass": round(mean_mass, 4),
                            "min_mass": round(min_mass, 4),
                            "top_letter_share": round(top_letter_share, 4),
                            "min_letter_share": round(min_letter_share, 4),
                            "raw_top_letter_share": round(raw_top_letter_share, 4),
                            "accuracy_pct": round(acc, 1),
                            "n_nan": n_nan,
                            "mean_prompt_tokens": round(mean_pt, 1),
                            "n_perm_exempt": n_perm_exempt,
                            "diagnostic_pool": "dev",
                        })

        df = pd.DataFrame(results_rows)
        print("\n" + "=" * 90)
        print(f"📊 PROMPT VARIANT & DEBIAS DIAGNOSTIC: {model_key} (held-out dev pool, n={n})")
        print("=" * 90)
        print(df.to_string(index=False))
        print("=" * 90)

        # Save CSV
        out_path = os.path.join(config.output_dir, f"prompt_diagnostics_{model_key}.csv")
        df.to_csv(out_path, index=False, encoding="utf-8")
        logger.info(f"Diagnostic results saved to: {out_path}")

        # Pure recommendation selection
        rec_info = select_recommendation(df)
        rec_pv = rec_info["prompt_variant"]
        rec_ct = bool(rec_info["chat_template"])
        rec_db = rec_info["debias"]
        rec_mm = rec_info["min_mean_mass"]
        rec_n_shots = 4 if rec_pv == "few_shot" else 0

        print("\n" + "=" * 90)
        if rec_info["passing"]:
            print(f"✅ RECOMMENDATION: {model_key}")
            print(f"   Passes validity gates in both languages (min mean_mass: {rec_mm:.4f})")
            print("   Edit MODEL_REGISTRY with the following exact settings:")
            print(f'     MODEL_REGISTRY["{model_key}"]["use_chat_template"] = {rec_ct}')
            print(f'     MODEL_REGISTRY["{model_key}"]["prompt_variant"] = "{rec_pv}"')
            print(f'     MODEL_REGISTRY["{model_key}"]["n_shots"] = {rec_n_shots}')
            print(f'     MODEL_REGISTRY["{model_key}"]["debias"] = "{rec_db}"')
        else:
            print(f"❌ NO VARIANT PASSES both validity gates in all languages.")
            print(f"   Highest minimum mean_mass candidate: prompt_variant='{rec_pv}', chat_template={rec_ct}, debias='{rec_db}' (min mean_mass: {rec_mm:.4f})")
            print("   If proceeding with best candidate, edit MODEL_REGISTRY with:")
            print(f'     MODEL_REGISTRY["{model_key}"]["use_chat_template"] = {rec_ct}')
            print(f'     MODEL_REGISTRY["{model_key}"]["prompt_variant"] = "{rec_pv}"')
            print(f'     MODEL_REGISTRY["{model_key}"]["n_shots"] = {rec_n_shots}')
            print(f'     MODEL_REGISTRY["{model_key}"]["debias"] = "{rec_db}"')
        print("=" * 90)

        return df

    finally:
        if backend is not None:
            try:
                backend.use_chat_template = orig_chat
            except Exception:
                pass
            del backend
        gc.collect()
        try:
            import torch as _torch
            if _torch.cuda.is_available():
                _torch.cuda.empty_cache()
        except ImportError:
            pass
```

## Section 12: Econometric Hypothesis Testing & Statistical Analysis
Executes econometric modeling addressing:
> *"Does Tokenization Fragmentation Predict Miscalibration in Bangla LLMs? A Cross-Lingual Study"*

Statistical Methods:
1. **Item-Level Clustered Regression:**
   $$e_i = \beta_0 + \beta_1 \cdot \text{Frag}_{\text{std}} + \beta_2 \cdot \text{Accuracy}_i + \beta_3 \cdot \text{PromptLength}_i + \gamma_{\text{subject}} + \varepsilon_i$$
   Controls for task accuracy, item prompt token count (`total_token_count`), and subject fixed effects (`C(subject)`), with standard errors clustered at the subject level.
2. **Within-Model Analysis:** Evaluated separately within each model and each language.
3. **Cross-Model Descriptive Table ($n=3$):** Explicitly marked as descriptive only without claiming inferential p-values across 3 data points.
4. **Bootstrap 95% Confidence Intervals:** 1000 resamples for ECE and Brier score, plus a paired bootstrap for the Bangla-minus-English ECE gap ($\Delta \text{ECE}$).
5. **TPW-Quantile Analysis:** Stratified ECE across TPW quartiles, issuing warnings if any stratum contains $<100$ items.


```python
def prepare_analysis_dataframe(df: pd.DataFrame) -> pd.DataFrame:
    """Validate, clean, and augment evaluation DataFrame for regression modeling."""
    clean_df = df.copy()

    rename_map = {
        "correct": "is_correct",
        "fertility": "overall_tpw",
        "question_fertility": "question_tpw",
        "strr": "overall_strr",
    }
    for old_col, new_col in rename_map.items():
        if old_col in clean_df.columns and new_col not in clean_df.columns:
            clean_df[new_col] = clean_df[old_col]

    if "is_correct" not in clean_df.columns or "confidence" not in clean_df.columns:
        raise ValueError("DataFrame must contain 'is_correct' and 'confidence' columns.")

    clean_df["is_correct"] = clean_df["is_correct"].astype(float)
    clean_df["confidence"] = pd.to_numeric(clean_df["confidence"], errors="coerce")
    clean_df = clean_df.dropna(subset=["confidence"]).copy()

    clean_df["calibration_residual"] = (clean_df["confidence"] - clean_df["is_correct"]).abs()
    clean_df["overconfidence"] = clean_df["confidence"] - clean_df["is_correct"]
    clean_df["tpw"] = pd.to_numeric(clean_df["overall_tpw"] if "overall_tpw" in clean_df.columns else np.nan, errors="coerce")
    clean_df["strr"] = pd.to_numeric(clean_df["overall_strr"] if "overall_strr" in clean_df.columns else np.nan, errors="coerce")
    clean_df["total_tokens"] = pd.to_numeric(clean_df["total_token_count"] if "total_token_count" in clean_df.columns else 0, errors="coerce")
    if isinstance(clean_df["total_tokens"], (int, float)):
        clean_df["total_tokens"] = pd.Series([clean_df["total_tokens"]] * len(clean_df), index=clean_df.index)
    else:
        clean_df["total_tokens"] = clean_df["total_tokens"].fillna(0)

    # Extended numeric controls: missing columns -> NaN, not error
    clean_df["prompt_tokens"] = pd.to_numeric(clean_df["prompt_tokens"] if "prompt_tokens" in clean_df.columns else np.nan, errors="coerce")
    clean_df["total_words"] = pd.to_numeric(
        clean_df["total_word_count"] if "total_word_count" in clean_df.columns else (
            clean_df["total_words"] if "total_words" in clean_df.columns else np.nan
        ), errors="coerce"
    )
    clean_df["tpc"] = pd.to_numeric(
        clean_df["overall_tpc"] if "overall_tpc" in clean_df.columns else (
            clean_df["tpc"] if "tpc" in clean_df.columns else np.nan
        ), errors="coerce"
    )
    clean_df["strr_ctx"] = pd.to_numeric(
        clean_df["overall_strr_in_context"] if "overall_strr_in_context" in clean_df.columns else (
            clean_df["strr_ctx"] if "strr_ctx" in clean_df.columns else np.nan
        ), errors="coerce"
    )

    if "language" not in clean_df.columns:
        clean_df["language"] = "bn"
    if "model_key" not in clean_df.columns:
        clean_df["model_key"] = clean_df.get("model_id", "default_model")
    if "subject" not in clean_df.columns:
        clean_df["subject"] = "general"

    # Normalised subject key so Bangla and English (different naming schemes) can be matched
    clean_df["subject_norm"] = (
        clean_df["subject"].astype(str).str.lower()
        .str.replace("_", " ", regex=False)
        .str.replace(r"[^a-z0-9 ]+", "", regex=True)
        .str.strip()
    )
    # BnMMLU and MMLU use different subject names. The evaluation files carry `subject_group` (the matched BnMMLU
    # subject for BOTH languages), which is the only key that lets the paired bn-vs-en bootstrap line subjects up.
    if "subject_group" in clean_df.columns:
        sg = clean_df["subject_group"].astype("string")
        clean_df["subject_norm"] = sg.where(sg.notna() & (sg != ""), clean_df["subject_norm"]).astype(str)

    return clean_df


def get_length_control_series(df: pd.DataFrame) -> Tuple[pd.Series, str]:
    """Select primary length control column ('prompt_tokens' if valid, else 'total_tokens').

    NOTE on length control:
    prompt_tokens is itself a tokenizer output, so it is mechanically correlated with fragmentation
    and may absorb part of the effect; word-count (total_words) control is used as the sensitivity check.
    """
    if "prompt_tokens" in df.columns:
        s = pd.to_numeric(df["prompt_tokens"], errors="coerce")
        if s.notna().sum() > 0 and s.std(skipna=True) > 1e-12:
            return s, "prompt_tokens"
        else:
            logger.warning("prompt_tokens is missing, all-NaN, or constant; falling back to total_tokens as length control.")
    else:
        logger.warning("prompt_tokens column not found; falling back to total_tokens as length control.")

    if "total_tokens" in df.columns:
        s_alt = pd.to_numeric(df["total_tokens"], errors="coerce").fillna(0.0)
    elif "total_token_count" in df.columns:
        s_alt = pd.to_numeric(df["total_token_count"], errors="coerce").fillna(0.0)
    else:
        s_alt = pd.Series(0.0, index=df.index)
    return s_alt, "total_tokens"


def compute_vif_safe(frag_z: np.ndarray, length_z: np.ndarray) -> float:
    """Compute VIF for [frag_z, length_z] and log a warning when VIF > 5."""
    try:
        X = np.column_stack([frag_z, length_z])
        valid = np.isfinite(X).all(axis=1)
        if valid.sum() < 5:
            return float("nan")
        X_clean = X[valid]
        if np.std(X_clean[:, 0]) < 1e-12 or np.std(X_clean[:, 1]) < 1e-12:
            return float("nan")
        X_const = sm.add_constant(X_clean)
        vif = float(variance_inflation_factor(X_const, 1))
        if vif > 5.0:
            logger.warning(f"⚠️ High Multicollinearity detected: VIF = {vif:.2f} > 5.0 for [frag_z, length_z].")
        return vif
    except Exception as e:
        logger.debug(f"VIF computation failed: {e}")
        return float("nan")


def compute_correlations(df: pd.DataFrame, frag_metric: str = "tpw") -> Dict[str, Any]:
    """Compute Pearson and Spearman correlations with calibration residual, accuracy, and confidence."""
    if frag_metric not in df.columns or "calibration_residual" not in df.columns or "is_correct" not in df.columns:
        return {"metric": frag_metric, "status": "insufficient_data"}
    sub_df = df.dropna(subset=[frag_metric, "calibration_residual", "is_correct"])
    n = len(sub_df)
    if n < 5:
        return {"sample_size": n, "status": "insufficient_data"}

    x = sub_df[frag_metric].values
    y_res = sub_df["calibration_residual"].values
    y_corr = sub_df["is_correct"].values
    y_conf = sub_df["confidence"].values

    if np.std(x) < 1e-12:
        return {"metric": frag_metric, "sample_size": n, "status": "insufficient_data"}

    p_res, p_res_pval = stats.pearsonr(x, y_res)
    s_res, s_res_pval = stats.spearmanr(x, y_res)
    p_acc, p_acc_pval = stats.pearsonr(x, y_corr)
    s_acc, s_acc_pval = stats.spearmanr(x, y_corr)
    p_conf, p_conf_pval = stats.pearsonr(x, y_conf)
    s_conf, s_conf_pval = stats.spearmanr(x, y_conf)

    return {
        "metric": frag_metric,
        "sample_size": n,
        "calibration_residual": {
            "pearson_r": float(p_res), "pearson_pvalue": float(p_res_pval),
            "spearman_rho": float(s_res), "spearman_pvalue": float(s_res_pval),
        },
        "accuracy": {
            "pearson_r": float(p_acc), "pearson_pvalue": float(p_acc_pval),
            "spearman_rho": float(s_acc), "spearman_pvalue": float(s_acc_pval),
        },
        "confidence": {
            "pearson_r": float(p_conf), "pearson_pvalue": float(p_conf_pval),
            "spearman_rho": float(s_conf), "spearman_pvalue": float(s_conf_pval),
        },
    }


def fit_confidence_correctness_models(
    df: pd.DataFrame,
    frag_metric: str = "tpw",
    length_col: Optional[str] = None,
    cluster_by: str = "subject_norm",
) -> Dict[str, Any]:
    """Fit question-level OLS models for confidence, correctness, and signed overconfidence.

    Models:
      M1: confidence     ~ frag_z + length_z + C(subject_norm)
      M2: is_correct     ~ frag_z + length_z + C(subject_norm)     (Linear probability model)
      M3: overconfidence ~ frag_z + length_z + C(subject_norm)     where overconfidence = confidence - is_correct (SIGNED)
      M4: confidence     ~ frag_z + is_correct + length_z + C(subject_norm)

    Standard errors are clustered by subject_norm with HC3 reported as sensitivity check.
    frag_z is standardised WITHIN the input group.
    """
    needed_cols = ["confidence", "is_correct"]
    if frag_metric not in df.columns or any(c not in df.columns for c in needed_cols):
        return {"status": "insufficient_data", "message": f"Required columns missing: {frag_metric}, {needed_cols}"}

    data = df.copy()
    if "overconfidence" not in data.columns:
        data["overconfidence"] = data["confidence"] - data["is_correct"]

    # Length series resolution
    if length_col and length_col in data.columns:
        len_s = pd.to_numeric(data[length_col], errors="coerce")
        used_len_name = length_col
    else:
        len_s, used_len_name = get_length_control_series(data)
    data["_raw_len"] = len_s

    # Standardise WITHIN group
    data = data.dropna(subset=[frag_metric, "confidence", "is_correct", "_raw_len"]).copy()
    if len(data) < 15:
        return {"status": "insufficient_data", "sample_size": len(data)}

    f_std = data[frag_metric].std()
    data["frag_z"] = (data[frag_metric] - data[frag_metric].mean()) / (f_std if f_std > 1e-12 else 1.0)
    l_std = data["_raw_len"].std()
    data["length_z"] = (data["_raw_len"] - data["_raw_len"].mean()) / (l_std if l_std > 1e-12 else 1.0)

    if cluster_by not in data.columns:
        data[cluster_by] = "general"

    # Compute VIF
    vif = compute_vif_safe(data["frag_z"].values, data["length_z"].values)

    n_clusters = int(data[cluster_by].nunique())
    few_clusters_flag = bool(n_clusters < 30)

    # Formulas
    has_fe = n_clusters > 1 and n_clusters < len(data) / 2
    fe_part = f" + C({cluster_by})" if has_fe else ""

    formulas = {
        "M1_confidence": f"confidence ~ frag_z + length_z{fe_part}",
        "M2_is_correct": f"is_correct ~ frag_z + length_z{fe_part}",
        "M3_overconfidence": f"overconfidence ~ frag_z + length_z{fe_part}",
        "M4_confidence_cond": f"confidence ~ frag_z + is_correct + length_z{fe_part}",
    }

    models_out: Dict[str, Any] = {
        "status": "success",
        "metric": frag_metric,
        "length_column": used_len_name,
        "sample_size": len(data),
        "n_clusters": n_clusters,
        "few_clusters_cluster_SE_unreliable": few_clusters_flag,
        "vif": vif,
        "models": {},
    }

    for m_id, f_str in formulas.items():
        try:
            mod = ols(f_str, data=data)
            # Clustered fit
            if n_clusters > 1:
                fit_cl = mod.fit(cov_type="cluster", cov_kwds={"groups": data[cluster_by]})
            else:
                fit_cl = mod.fit(cov_type="HC3")
            # HC3 sensitivity fit
            fit_hc3 = mod.fit(cov_type="HC3")

            target_var = "frag_z"
            if target_var in fit_cl.params:
                ci_cl = fit_cl.conf_int().loc[target_var]
                ci_hc3 = fit_hc3.conf_int().loc[target_var]
                models_out["models"][m_id] = {
                    "formula": f_str,
                    "r_squared": float(fit_cl.rsquared),
                    "coef": float(fit_cl.params[target_var]),
                    "std_err": float(fit_cl.bse[target_var]),
                    "pvalue": float(fit_cl.pvalues[target_var]),
                    "ci_95": (float(ci_cl[0]), float(ci_cl[1])),
                    "hc3_std_err": float(fit_hc3.bse[target_var]),
                    "hc3_pvalue": float(fit_hc3.pvalues[target_var]),
                    "hc3_ci_95": (float(ci_hc3[0]), float(ci_hc3[1])),
                }
            else:
                models_out["models"][m_id] = {"status": "error", "message": f"{target_var} dropped or collinear"}
        except Exception as e:
            models_out["models"][m_id] = {"status": "error", "message": str(e)}

    return models_out


def fit_language_interaction(
    df_model: pd.DataFrame,
    frag_metric: str = "tpw",
    length_col: Optional[str] = None,
) -> Dict[str, Any]:
    """Fit stacked bn and en model to test language x fragmentation interaction for H3.

    CAVEAT: The Bangla and English samples are subject-matched, not item-parallel,
    so a slope difference is descriptive of the two samples.
    """
    if df_model is None or df_model.empty:
        return {"status": "skipped_invalid_or_missing_language", "message": "Empty DataFrame"}

    langs_before = set(df_model["language"].dropna().unique())
    dropped_invalid = set()
    sub = df_model.copy()
    if "validity_flag" in sub.columns and getattr(CONFIG, "analysis_valid_only", True):
        invalid_mask = sub["validity_flag"] != "VALID"
        dropped_invalid = set(sub.loc[invalid_mask, "language"].dropna().unique())
        sub = sub[~invalid_mask].copy()

    langs_present = set(sub["language"].dropna().unique())
    if not ({"bn", "en"}.issubset(langs_present)):
        msg = f"Languages before filter: {langs_before}, dropped invalid: {dropped_invalid}, remaining: {langs_present}"
        return {"status": "skipped_invalid_or_missing_language", "message": msg}

    # Ensure required columns
    if frag_metric not in sub.columns or "confidence" not in sub.columns or "is_correct" not in sub.columns:
        return {"status": "insufficient_data", "message": f"Missing metric {frag_metric} or outcomes"}

    if "overconfidence" not in sub.columns:
        sub["overconfidence"] = sub["confidence"] - sub["is_correct"]

    # Length column resolution
    if length_col and length_col in sub.columns:
        len_s = pd.to_numeric(sub[length_col], errors="coerce")
        used_len_name = length_col
    else:
        len_s, used_len_name = get_length_control_series(sub)
    sub["_raw_len"] = len_s

    # Clean missing
    sub = sub.dropna(subset=[frag_metric, "confidence", "is_correct", "_raw_len"]).copy()
    if len(sub[sub["language"] == "bn"]) < 10 or len(sub[sub["language"] == "en"]) < 10:
        return {"status": "insufficient_data", "message": "Too few items per language"}

    # Standardise frag_z and length_z WITHIN language
    sub["frag_z"] = sub.groupby("language")[frag_metric].transform(
        lambda s: (s - s.mean()) / (s.std() if s.std() > 1e-12 else 1.0)
    )
    sub["length_z"] = sub.groupby("language")["_raw_len"].transform(
        lambda s: (s - s.mean()) / (s.std() if s.std() > 1e-12 else 1.0)
    )

    cluster_by = "subject_norm"
    if cluster_by not in sub.columns:
        sub[cluster_by] = "general"

    n_clusters = int(sub[cluster_by].nunique())
    few_clusters_flag = bool(n_clusters < 30)

    has_fe = n_clusters > 1 and n_clusters < len(sub) / 2
    fe_part = f" + C({cluster_by})" if has_fe else ""

    outcomes = {
        "confidence": f"confidence ~ frag_z * C(language) + length_z{fe_part}",
        "is_correct": f"is_correct ~ frag_z * C(language) + length_z{fe_part}",
        "overconfidence": f"overconfidence ~ frag_z * C(language) + length_z{fe_part}",
    }

    results: Dict[str, Any] = {
        "status": "success",
        "metric": frag_metric,
        "length_column": used_len_name,
        "sample_size": len(sub),
        "n_clusters": n_clusters,
        "few_clusters_cluster_SE_unreliable": few_clusters_flag,
        "outcomes": {},
        "caveat": "Bangla and English samples are subject-matched, not item-parallel; slope difference is descriptive of the two samples.",
    }

    for out_name, f_str in outcomes.items():
        try:
            mod = ols(f_str, data=sub)
            if n_clusters > 1:
                fit_cl = mod.fit(cov_type="cluster", cov_kwds={"groups": sub[cluster_by]})
            else:
                fit_cl = mod.fit(cov_type="HC3")
            fit_hc3 = mod.fit(cov_type="HC3")

            # Identify interaction term
            inter_term = None
            for p_name in fit_cl.params.index:
                if ("frag_z" in p_name and "language" in p_name and ":" in p_name):
                    inter_term = p_name
                    break

            if inter_term:
                ci_cl = fit_cl.conf_int().loc[inter_term]
                ci_hc3 = fit_hc3.conf_int().loc[inter_term]
                results["outcomes"][out_name] = {
                    "interaction_term": inter_term,
                    "coef": float(fit_cl.params[inter_term]),
                    "std_err": float(fit_cl.bse[inter_term]),
                    "pvalue": float(fit_cl.pvalues[inter_term]),
                    "ci_95": (float(ci_cl[0]), float(ci_cl[1])),
                    "hc3_std_err": float(fit_hc3.bse[inter_term]),
                    "hc3_pvalue": float(fit_hc3.pvalues[inter_term]),
                    "hc3_ci_95": (float(ci_hc3[0]), float(ci_hc3[1])),
                    "r_squared": float(fit_cl.rsquared),
                }
            else:
                results["outcomes"][out_name] = {"status": "error", "message": "Interaction term not found"}
        except Exception as e:
            results["outcomes"][out_name] = {"status": "error", "message": str(e)}

    return results


def fit_calibration_regression(
    df: pd.DataFrame,
    frag_metric: str = "tpw",
    cluster_by: Optional[str] = "subject_norm",
) -> Dict[str, Any]:
    """Fit question-level OLS regression controlling for accuracy, length, and subject fixed effects.
    Kept as secondary model (mechanically tied to correctness).
    """
    data = df.dropna(subset=[frag_metric, "calibration_residual", "is_correct"]).copy()
    if len(data) < 15:
        return {"status": "insufficient_data", "sample_size": len(data)}

    frag_mean = data[frag_metric].mean()
    frag_std = data[frag_metric].std()
    data["frag_std"] = (data[frag_metric] - frag_mean) / (frag_std if frag_std > 0 else 1.0)

    formula_parts = ["frag_std", "is_correct", "total_tokens"]

    if cluster_by and cluster_by in data.columns and data[cluster_by].nunique() > 1 and data[cluster_by].nunique() < len(data) / 2:
        formula_parts.append(f"C({cluster_by})")
    elif data["subject"].nunique() > 1 and data["subject"].nunique() < len(data) / 2:
        formula_parts.append("C(subject)")

    if data["model_key"].nunique() > 1:
        formula_parts.append("C(model_key)")

    formula = f"calibration_residual ~ {' + '.join(formula_parts)}"

    try:
        model = ols(formula, data=data)
        if cluster_by and cluster_by in data.columns and data[cluster_by].nunique() > 1:
            fit_res = model.fit(cov_type="cluster", cov_kwds={"groups": data[cluster_by]})
            se_type = f"clustered_by_{cluster_by}"
        else:
            fit_res = model.fit(cov_type="HC3")
            se_type = "heteroskedasticity_robust_HC3"

        params = fit_res.params
        bse = fit_res.bse
        pvalues = fit_res.pvalues
        conf_int = fit_res.conf_int()

        results_table = {}
        for var in params.index:
            if "subject" in var and len(params.index) > 10:
                continue
            results_table[var] = {
                "coef": float(params[var]),
                "std_err": float(bse[var]),
                "pvalue": float(pvalues[var]),
                "ci_lower_95": float(conf_int.loc[var, 0]),
                "ci_upper_95": float(conf_int.loc[var, 1]),
            }

        return {
            "status": "success",
            "formula": formula,
            "sample_size": int(fit_res.nobs),
            "r_squared": float(fit_res.rsquared),
            "r_squared_adj": float(fit_res.rsquared_adj),
            "se_type": se_type,
            "coefficients": results_table,
            "note": "secondary / mechanically tied to correctness",
        }
    except Exception as e:
        return {"status": "error", "message": str(e)}


def compute_bootstrap_ci(
    confidences: Union[List[float], np.ndarray],
    corrects: Union[List[Union[bool, int, float]], np.ndarray],
    n_boot: int = 2000,
    seed: int = DEFAULT_SEED,
) -> Dict[str, Tuple[float, float]]:
    """Compute 95% bootstrap confidence intervals for ECE and Brier Score."""
    conf, corr = _filter_valid_pairs(confidences, corrects)
    n = len(conf)
    if n < 10:
        return {"ece_ci_95": (float("nan"), float("nan")), "brier_ci_95": (float("nan"), float("nan"))}

    rng = np.random.RandomState(seed)
    boot_eces = []
    boot_briers = []

    for _ in range(n_boot):
        idx = rng.choice(n, size=n, replace=True)
        b_conf = conf[idx]
        b_corr = corr[idx]
        boot_eces.append(expected_calibration_error(b_conf, b_corr, n_bins=10))
        boot_briers.append(brier_score(b_conf, b_corr))

    return {
        "ece_ci_95": (float(np.percentile(boot_eces, 2.5)), float(np.percentile(boot_eces, 97.5))),
        "brier_ci_95": (float(np.percentile(boot_briers, 2.5)), float(np.percentile(boot_briers, 97.5))),
    }


def compute_paired_bootstrap_delta_ece(
    df_bn: pd.DataFrame,
    df_en: pd.DataFrame,
    n_boot: int = 2000,
    seed: int = DEFAULT_SEED,
) -> Dict[str, Any]:
    """Paired bootstrap for Bangla-minus-English ECE difference on matched subjects."""
    subj_col = "subject_norm" if ("subject_norm" in df_bn.columns and "subject_norm" in df_en.columns) else "subject"
    common_subjs = sorted(list(set(df_bn[subj_col]).intersection(set(df_en[subj_col]))))
    if not common_subjs:
        return {"status": "no_common_subjects"}

    rng = np.random.RandomState(seed)
    delta_eces = []

    bn_groups = {s: g for s, g in df_bn[df_bn[subj_col].isin(common_subjs)].groupby(subj_col)}
    en_groups = {s: g for s, g in df_en[df_en[subj_col].isin(common_subjs)].groupby(subj_col)}

    for _ in range(n_boot):
        boot_subjs = rng.choice(common_subjs, size=len(common_subjs), replace=True)
        bn_sample = pd.concat([bn_groups[s] for s in boot_subjs], ignore_index=True)
        en_sample = pd.concat([en_groups[s] for s in boot_subjs], ignore_index=True)

        if len(bn_sample) > 0 and len(en_sample) > 0:
            ece_bn = expected_calibration_error(bn_sample["confidence"].values, bn_sample["is_correct"].values)
            ece_en = expected_calibration_error(en_sample["confidence"].values, en_sample["is_correct"].values)
            delta_eces.append(ece_bn - ece_en)

    if not delta_eces:
        return {"status": "insufficient_samples"}

    return {
        "mean_delta_ece": float(np.mean(delta_eces)),
        "ci_95": (float(np.percentile(delta_eces, 2.5)), float(np.percentile(delta_eces, 97.5))),
    }


def compute_tpw_quartile_calibration(df: pd.DataFrame, metric: str = "tpw") -> Dict[str, Any]:
    """Compute ECE across metric quartiles (default 'tpw') and warn if any quartile has <100 items."""
    if metric not in df.columns or "confidence" not in df.columns or "is_correct" not in df.columns:
        return {"status": "insufficient_data"}
    clean_df = df.dropna(subset=[metric, "confidence", "is_correct"]).copy()
    if len(clean_df) < 20:
        return {"status": "insufficient_data"}

    q_col = f"{metric}_quartile"
    try:
        clean_df[q_col] = pd.qcut(clean_df[metric], q=4, labels=["Q1_lowest", "Q2_low_mid", "Q3_high_mid", "Q4_highest"], duplicates="raise")
    except ValueError:
        clean_df[q_col] = pd.qcut(clean_df[metric].rank(method="first"), q=4, labels=["Q1_lowest", "Q2_low_mid", "Q3_high_mid", "Q4_highest"])
    quartile_results = {}

    for q_label, group in clean_df.groupby(q_col, observed=False):
        n_q = len(group)
        if n_q < MIN_QUARTILE_SAMPLE_THRESHOLD:
            logger.warning(f"Quartile '{q_label}' for '{metric}' has {n_q} items (< {MIN_QUARTILE_SAMPLE_THRESHOLD}). Binning statistics may exhibit sample noise.")
        q_ece = expected_calibration_error(group["confidence"].values, group["is_correct"].values, n_bins=10)
        q_bs = brier_score(group["confidence"].values, group["is_correct"].values)
        quartile_results[str(q_label)] = {
            "sample_count": n_q,
            "mean_metric": float(group[metric].mean()),
            f"mean_{metric}": float(group[metric].mean()),
            "ece_10": q_ece,
            "brier_score": q_bs,
            "accuracy": float(group["is_correct"].mean() * 100.0),
        }

    return quartile_results


def run_full_statistical_analysis(df: pd.DataFrame) -> Dict[str, Any]:
    """Execute comprehensive within-model and cross-model econometric analysis."""
    clean_df = prepare_analysis_dataframe(df)
    total_n = len(clean_df)

    if total_n == 0:
        return {"status": "error", "message": "No valid samples."}

    within_model_results: Dict[str, Any] = {}
    groups: Dict[Tuple[str, str], pd.DataFrame] = {}

    for m_key, m_df in clean_df.groupby("model_key"):
        for lang, l_df in m_df.groupby("language"):
            groups[(str(m_key), str(lang))] = l_df

            # Pass probabilities and gold indices to compute MC-Brier and NLL
            probs_arr = None
            gold_indices = None
            prob_cols = ["prob_A", "prob_B", "prob_C", "prob_D"]
            if all(c in l_df.columns for c in prob_cols):
                p_sub = l_df[prob_cols].values
                valid_probs = np.isfinite(p_sub).all(axis=1)
                gold_map = {"A": 0, "B": 1, "C": 2, "D": 3}
                gold_col = "gold_answer" if "gold_answer" in l_df.columns else ("correct_answer" if "correct_answer" in l_df.columns else None)
                if gold_col:
                    g_sub = [gold_map.get(str(x).upper(), -1) for x in l_df[gold_col]]
                    valid_gold = np.array([g >= 0 for g in g_sub])
                    mc_mask = valid_probs & valid_gold
                    if mc_mask.any():
                        probs_arr = p_sub[mc_mask].tolist()
                        gold_indices = np.array(g_sub)[mc_mask].tolist()

            calib = compute_all_calibration_metrics(
                l_df["confidence"].values,
                l_df["is_correct"].values,
                probabilities=probs_arr,
                gold_indices=gold_indices,
                n_bins=10,
            )
            calib.pop("bins", None)

            # Option-mass robustness check
            mass_filtered = {}
            if "option_mass" in l_df.columns:
                mass_s = pd.to_numeric(l_df["option_mass"], errors="coerce")
                valid_mass = mass_s >= MIN_OPTION_MASS_THRESHOLD
                n_pass = int(valid_mass.sum())
                n_tot = len(l_df)
                mass_filtered["n_mass_filtered"] = n_pass
                mass_filtered["share_mass_filtered_out"] = float((n_tot - n_pass) / n_tot) if n_tot > 0 else 0.0
                if n_pass >= 5:
                    df_mf = l_df[valid_mass]
                    mass_filtered["ece_mass_filtered"] = expected_calibration_error(
                        df_mf["confidence"].values, df_mf["is_correct"].values, n_bins=10
                    )
                    mass_filtered["brier_mass_filtered"] = brier_score(
                        df_mf["confidence"].values, df_mf["is_correct"].values
                    )
                else:
                    mass_filtered["ece_mass_filtered"] = float("nan")
                    mass_filtered["brier_mass_filtered"] = float("nan")
            else:
                mass_filtered["n_mass_filtered"] = len(l_df)
                mass_filtered["share_mass_filtered_out"] = 0.0
                mass_filtered["ece_mass_filtered"] = calib.get("ece_uniform_10", float("nan"))
                mass_filtered["brier_mass_filtered"] = calib.get("brier_score", float("nan"))

            # Correlations for tpw, strr, tpc
            corrs = {
                "tpw": compute_correlations(l_df, frag_metric="tpw"),
                "strr": compute_correlations(l_df, frag_metric="strr"),
                "tpc": compute_correlations(l_df, frag_metric="tpc"),
            }

            # Econometric models for tpw, strr, tpc, strr_ctx
            metrics_to_fit = ["tpw", "strr", "tpc"]
            if "strr_ctx" in l_df.columns and l_df["strr_ctx"].notna().sum() >= 15:
                metrics_to_fit.append("strr_ctx")

            h2_models = {}
            pvals_for_holm = []
            test_keys = []
            for met in metrics_to_fit:
                fit_out = fit_confidence_correctness_models(l_df, frag_metric=met)
                h2_models[met] = fit_out
                if fit_out.get("status") == "success":
                    for m_outcome in ["M1_confidence", "M2_is_correct", "M3_overconfidence"]:
                        m_res = fit_out["models"].get(m_outcome, {})
                        if "pvalue" in m_res:
                            pvals_for_holm.append(m_res["pvalue"])
                            test_keys.append((met, m_outcome))

            # Apply Holm adjustment across metric x outcome
            if pvals_for_holm:
                _, adj_pvals, _, _ = multipletests(pvals_for_holm, method="holm")
                for (met, m_outcome), p_adj in zip(test_keys, adj_pvals):
                    h2_models[met]["models"][m_outcome]["holm_pvalue"] = float(p_adj)

            # Sensitivity robustness: re-fit M1-M3 with total_words as length control (for TPW only)
            sensitivity_tpw_words = fit_confidence_correctness_models(
                l_df, frag_metric="tpw", length_col="total_words"
            )

            within_model_results[f"{m_key}_{lang}"] = {
                "model_key": m_key,
                "language": lang,
                "sample_size": len(l_df),
                "accuracy": float(l_df["is_correct"].mean() * 100.0),
                "mean_tpw": float(l_df["tpw"].mean()) if "tpw" in l_df else float("nan"),
                "mean_strr": float(l_df["strr"].mean()) if "strr" in l_df else float("nan"),
                "mean_tpc": float(l_df["tpc"].mean()) if "tpc" in l_df else float("nan"),
                "calibration": calib,
                "mass_filtered_calibration": mass_filtered,
                "bootstrap_ci": compute_bootstrap_ci(l_df["confidence"].values, l_df["is_correct"].values),
                "correlations": corrs,
                "h2_models": h2_models,
                "sensitivity_tpw_words": sensitivity_tpw_words,
                "regression": fit_calibration_regression(l_df, frag_metric="tpw"), # backward compat
                "tpw_quartiles": compute_tpw_quartile_calibration(l_df, metric="tpw"),
                "tpc_quartiles": compute_tpw_quartile_calibration(l_df, metric="tpc"),
            }

    cross_model_table = []
    for res in within_model_results.values():
        cal = res["calibration"]
        mf = res.get("mass_filtered_calibration", {})
        cross_model_table.append({
            "Model": res["model_key"],
            "Language": res["language"],
            "N": res["sample_size"],
            "Accuracy (%)": round(res["accuracy"], 2),
            "Mean conf": round(cal.get("mean_confidence", float("nan")), 4),
            "ECE": round(cal.get("ece_uniform_10", float("nan")), 4),
            "MCE": round(cal.get("mce", float("nan")), 4),
            "Brier": round(cal.get("brier_score", float("nan")), 4),
            "MC-Brier": round(cal.get("multiclass_brier_score", float("nan")), 4) if np.isfinite(cal.get("multiclass_brier_score", np.nan)) else "N/A",
            "NLL": round(cal.get("nll", float("nan")), 4) if np.isfinite(cal.get("nll", np.nan)) else "N/A",
            "ECE (mass-filt)": round(mf.get("ece_mass_filtered", float("nan")), 4) if np.isfinite(mf.get("ece_mass_filtered", np.nan)) else "N/A",
            "Filt-Out (%)": round(mf.get("share_mass_filtered_out", 0.0) * 100.0, 1),
            "Mean TPW": round(res["mean_tpw"], 2),
        })

    # Bangla minus English ECE per model
    bn_minus_en: Dict[str, Any] = {}
    for m_key in clean_df["model_key"].unique():
        if (str(m_key), "bn") in groups and (str(m_key), "en") in groups:
            bn_minus_en[str(m_key)] = compute_paired_bootstrap_delta_ece(groups[(str(m_key), "bn")], groups[(str(m_key), "en")])

    # H3: Language x fragmentation interaction per model
    language_interaction: Dict[str, Any] = {}
    for m_key, m_df in clean_df.groupby("model_key"):
        language_interaction[str(m_key)] = {}
        for met in ["tpw", "strr", "tpc"]:
            inter_res = fit_language_interaction(m_df, frag_metric=met)
            language_interaction[str(m_key)][met] = inter_res

        # Holm adjustment across metric x outcome for this model
        pvals = []
        keys = []
        for met, met_data in language_interaction[str(m_key)].items():
            if met_data.get("status") == "success":
                for out_name, out_data in met_data.get("outcomes", {}).items():
                    if "pvalue" in out_data:
                        pvals.append(out_data["pvalue"])
                        keys.append((met, out_name))
        if pvals:
            _, adj_p, _, _ = multipletests(pvals, method="holm")
            for (met, out_name), p_a in zip(keys, adj_p):
                language_interaction[str(m_key)][met]["outcomes"][out_name]["holm_pvalue"] = float(p_a)

    return {
        "dataset_summary": {
            "total_analyzed_items": total_n,
            "models": clean_df["model_key"].unique().tolist(),
            "languages": clean_df["language"].unique().tolist(),
            "mean_accuracy": float(clean_df["is_correct"].mean()),
            "mean_confidence": float(clean_df["confidence"].mean()),
            "mean_calibration_residual": float(clean_df["calibration_residual"].mean()),
        },
        "within_model_results": within_model_results,
        "cross_model_descriptive_table": cross_model_table,
        "bn_minus_en_ece": bn_minus_en,
        "language_interaction": language_interaction,
        "note": "Cross-model comparisons are descriptive only (few models, no inferential test). Bangla/English sets are subject-matched, not item-parallel.",
    }


def format_markdown_report(report: Dict[str, Any], excluded_runs: Optional[List[Dict[str, Any]]] = None) -> str:
    """Produce an extensive scientific markdown summary of the statistical results."""
    L: List[str] = []
    L.append("# Statistical Analysis Report: Tokenization Fragmentation & Model Miscalibration")
    L.append("")
    L.append("> Observational associations only. No causal claims are made. Pretraining corpus, architecture, and alignment remain potential confounders.")
    L.append("")

    if excluded_runs:
        L.append("### Excluded (invalid) runs")
        L.append("> These model x language runs failed validity gates (e.g. prediction collapse or insufficient option mass) and were excluded from statistical inference.")
        L.append("")
        L.append("| Model | Language | Accuracy (%) | Top-letter share | Reason |")
        L.append("|---|---|---|---|---|")
        for ex in excluded_runs:
            acc_str = f"{ex.get('accuracy', 0.0):.2f}" if isinstance(ex.get('accuracy'), (int, float)) else str(ex.get('accuracy', 'N/A'))
            top_str = f"{ex.get('top_letter_share', 0.0):.3f}" if isinstance(ex.get('top_letter_share'), (int, float)) else str(ex.get('top_letter_share', 'N/A'))
            reason_str = "; ".join(ex.get('reasons', [])) if isinstance(ex.get('reasons'), list) else str(ex.get('reasons', 'N/A'))
            L.append(f"| `{ex.get('model', 'unknown')}` | {ex.get('language', 'unknown')} | {acc_str} | {top_str} | {reason_str} |")
        L.append("")

    ds = report.get("dataset_summary", {})
    L.append("### 1. Sample Characteristics")
    L.append(f"- **Total items analyzed:** {ds.get('total_analyzed_items', 0):,}")
    L.append(f"- **Models:** {', '.join(map(str, ds.get('models', [])))}")
    L.append(f"- **Languages:** {', '.join(map(str, ds.get('languages', [])))}")
    L.append(f"- **Mean accuracy:** {ds.get('mean_accuracy', 0.0) * 100:.2f}%")
    L.append(f"- **Mean confidence (MSP):** {ds.get('mean_confidence', 0.0):.4f}")
    L.append("")

    L.append("### 2. Cross-Model Descriptive Table (descriptive only, no inference)")
    L.append("| Model | Lang | N | Acc (%) | Mean conf | ECE | MCE | Brier | MC-Brier | NLL | ECE (mass-filt) | Filt-Out (%) | Mean TPW |")
    L.append("|---|---|---|---|---|---|---|---|---|---|---|---|---|")
    for r in report.get("cross_model_descriptive_table", []):
        L.append(f"| `{r['Model']}` | {r['Language']} | {r['N']:,} | {r['Accuracy (%)']:.2f} | {r['Mean conf']:.4f} | {r['ECE']:.4f} | {r['MCE']:.4f} | {r['Brier']:.4f} | {r.get('MC-Brier', 'N/A')} | {r.get('NLL', 'N/A')} | {r.get('ECE (mass-filt)', 'N/A')} | {r.get('Filt-Out (%)', 0.0):.1f}% | {r['Mean TPW']:.2f} |")
    L.append("")

    L.append("### 3. Bangla minus English ECE (subject-cluster bootstrap)")
    for m_key, d in report.get("bn_minus_en_ece", {}).items():
        if "mean_delta_ece" in d:
            lo, hi = d["ci_95"]
            L.append(f"- `{m_key}`: mean delta ECE = {d['mean_delta_ece']:.4f}, 95% CI [{lo:.4f}, {hi:.4f}]")
        else:
            L.append(f"- `{m_key}`: {d.get('status', 'not computed')}")
    L.append("")

    # Section H2
    L.append("### 4. H2: Fragmentation vs Confidence / Correctness / Overconfidence")
    L.append("> Expected signs: TPW and TPC positive for fragmentation (higher fragmentation predicts miscalibration / overconfidence); STRR negative (higher subword token retention ratio = less fragmentation).")
    L.append("")
    for wres in report.get("within_model_results", {}).values():
        L.append(f"#### `{wres['model_key']}` ({wres['language']}), N={wres['sample_size']:,}")
        ece_ci = wres.get("bootstrap_ci", {}).get("ece_ci_95", (float("nan"), float("nan")))
        L.append(f"- Bootstrap 95% CI for ECE: [{ece_ci[0]:.4f}, {ece_ci[1]:.4f}]")

        h2_data = wres.get("h2_models", {})
        if h2_data:
            L.append("")
            L.append("| Metric | Outcome Model | Coef (frag_z) | 95% CI | Raw p | Holm p | HC3 SE | N | Clusters | Flags |")
            L.append("|---|---|---|---|---|---|---|---|---|---|")
            for met, m_res in h2_data.items():
                if m_res.get("status") == "success":
                    n_cl = m_res.get("n_clusters", "N/A")
                    flags = "few_clusters" if m_res.get("few_clusters_cluster_SE_unreliable") else "OK"
                    for m_name in ["M1_confidence", "M2_is_correct", "M3_overconfidence", "M4_confidence_cond"]:
                        m_info = m_res.get("models", {}).get(m_name, {})
                        if "coef" in m_info:
                            c_val = m_info["coef"]
                            ci = m_info["ci_95"]
                            raw_p = m_info["pvalue"]
                            holm_p = m_info.get("holm_pvalue", raw_p)
                            hc3_se = m_info.get("hc3_std_err", float("nan"))
                            L.append(f"| `{met}` | `{m_name}` | {c_val:.4f} | [{ci[0]:.4f}, {ci[1]:.4f}] | {raw_p:.3e} | {holm_p:.3e} | {hc3_se:.4f} | {m_res['sample_size']} | {n_cl} | {flags} |")
            L.append("")

        # Legacy residual regression note
        reg = wres.get("regression", {})
        if reg.get("status") == "success":
            L.append(f"Residual regression: `{reg.get('formula')}` (SE: `{reg.get('se_type')}`), R2 = {reg.get('r_squared', 0.0):.4f} *(secondary / mechanically tied to correctness)*")

        # Quartiles
        q_tpw = wres.get("tpw_quartiles", {})
        if q_tpw and "status" not in q_tpw:
            L.append("")
            L.append("| TPW quartile | N | Mean TPW | ECE | Brier | Acc (%) |")
            L.append("|---|---|---|---|---|---|")
            for q_label, qv in q_tpw.items():
                L.append(f"| {q_label} | {qv['sample_count']} | {qv.get('mean_tpw', qv.get('mean_metric', 0.0)):.2f} | {qv['ece_10']:.4f} | {qv['brier_score']:.4f} | {qv['accuracy']:.1f} |")
        L.append("")

    # Section H3
    L.append("### 5. H3: Language x Fragmentation Interaction")
    L.append("> Tests whether the within-language slope per 1 SD of fragmentation differs between Bangla and English. Note: samples are subject-matched, not item-parallel.")
    L.append("")
    lang_inter = report.get("language_interaction", {})
    for m_key, m_inters in lang_inter.items():
        L.append(f"#### `{m_key}` Interaction Models")
        L.append("| Metric | Outcome | Interaction Coef | 95% CI | Raw p | Holm p | HC3 SE | N | Clusters | Flags |")
        L.append("|---|---|---|---|---|---|---|---|---|---|")
        for met, inter_res in m_inters.items():
            if inter_res.get("status") == "success":
                n_cl = inter_res.get("n_clusters", "N/A")
                flags = "few_clusters" if inter_res.get("few_clusters_cluster_SE_unreliable") else "OK"
                for out_name, out_data in inter_res.get("outcomes", {}).items():
                    if "coef" in out_data:
                        c_val = out_data["coef"]
                        ci = out_data["ci_95"]
                        raw_p = out_data["pvalue"]
                        holm_p = out_data.get("holm_pvalue", raw_p)
                        hc3_se = out_data.get("hc3_std_err", float("nan"))
                        L.append(f"| `{met}` | `{out_name}` | {c_val:.4f} | [{ci[0]:.4f}, {ci[1]:.4f}] | {raw_p:.3e} | {holm_p:.3e} | {hc3_se:.4f} | {inter_res['sample_size']} | {n_cl} | {flags} |")
            else:
                L.append(f"| `{met}` | All | {inter_res.get('status')} | N/A | N/A | N/A | N/A | N/A | N/A | {inter_res.get('message', '')} |")
        L.append("")

    # Section Robustness
    L.append("### 6. Robustness & Sensitivity Checks")
    L.append("#### Word-Count Length Control vs Prompt-Tokens Control (TPW Models)")
    L.append("| Model x Language | Outcome | Prompt-Tokens Coef | Total-Words Coef | Difference |")
    L.append("|---|---|---|---|---|")
    for w_name, wres in report.get("within_model_results", {}).items():
        h2_tpw = wres.get("h2_models", {}).get("tpw", {}).get("models", {})
        sens_tpw = wres.get("sensitivity_tpw_words", {}).get("models", {})
        for out_id in ["M1_confidence", "M2_is_correct", "M3_overconfidence"]:
            pt_coef = h2_tpw.get(out_id, {}).get("coef", float("nan"))
            tw_coef = sens_tpw.get(out_id, {}).get("coef", float("nan"))
            diff = pt_coef - tw_coef if np.isfinite(pt_coef) and np.isfinite(tw_coef) else float("nan")
            L.append(f"| `{w_name}` | `{out_id}` | {pt_coef:.4f} | {tw_coef:.4f} | {diff:.4f} |")
    L.append("")

    L.append("#### Standard Error Robustness (Cluster SE vs HC3)")
    L.append("- Evaluated side by side in Tables 4 & 5. If subject clusters < 30, HC3 standard errors serve as the primary sensitivity safeguard.")
    L.append("")

    L.append("#### Option-Mass Filtering Robustness")
    L.append("- Evaluated in Table 2. Shows calibration error (ECE) when low-mass items (< 0.50 option mass) are removed.")
    L.append("")

    # Saved Visualizations
    L.append("### 7. Reliability Diagrams & Visualizations")
    saved_plots = report.get("saved_plots", [])
    if saved_plots:
        for p in saved_plots:
            L.append(f"- Saved plot: `{p}`")
    else:
        L.append("- Plots rendered to `results/` (reliability diagrams and fragmentation scatter plots).")
    L.append("")

    # Section Caveats
    L.append("### 8. Caveats & Methodological Limitations")
    L.append("- **Observational associations:** All reported relationships reflect observational associations; no causal claims are warranted.")
    L.append("- **Mechanical dependency of calibration residual:** The item-level residual |confidence - correctness| is mechanically tied to correctness; interpret residual regressions cautiously.")
    L.append("- **Item difficulty and topic confounding:** Within-tokenizer fragmentation may be confounded with item difficulty, syntactic complexity, and subject area.")
    L.append("- **Length control trade-off:** `prompt_tokens` is itself a tokenizer output mechanically correlated with fragmentation, potentially absorbing part of the effect; `total_words` is reported as a sensitivity check.")
    L.append("- **Small cluster count:** In benchmarks with fewer than 30 subject categories, clustered standard errors can be downward biased; HC3 heteroskedasticity-robust SEs serve as the sensitivity check.")
    # B1 fix (v4h): wording depends on en_source
    _en_src_lim = getattr(CONFIG, "en_source", "mmlu")
    if _en_src_lim == "bnmmlu_translation":
        L.append(
            "- **Item-parallel English translations:** The English side is an item-parallel English translation of the same BnMMLU items (one wide CSV); both languages share the same question_id. Cross-language comparisons are therefore within-item, not subject-matched across separate benchmarks."
        )
    else:
        L.append(
            "- **Subject-matched samples:** Bangla (BnMMLU) and English (MMLU) are subject-matched across broad curricula, not item-parallel translations."
        )
    L.append("- **Small model cohort:** With 3 evaluated models (2 of which share a tokenizer family), model-level comparisons (H1) are descriptive only.")
    return "\n".join(L)
```

## Section 13: Scientific Visualizations
Generates publication-quality figures:
1. **Reliability Diagram:** Empirical accuracy vs. confidence bins with calibration gap shading.
2. **Scatter Plot:** Item-level TPW vs. calibration residual with OLS linear fit.
3. **Quartile Calibration Plot:** ECE across TPW quartiles.


```python
def plot_reliability_diagram(bins: List[Dict[str, Any]], title: str = "Reliability Diagram", save_path: Optional[str] = None):
    """Plot reliability diagram with calibration gaps and sample counts."""
    if not HAS_VISUALIZATION or not bins:
        return

    try:
        populated_bins = [b for b in bins if b.get("sample_count", 0) > 0]
        if not populated_bins:
            return

        fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(7, 7), gridspec_kw={"height_ratios": [3, 1]}, sharex=True)
        ax1.plot([0, 1], [0, 1], "k--", alpha=0.7, label="Perfect Calibration")

        conf_vals = [b["mean_confidence"] for b in populated_bins]
        acc_vals = [b["accuracy"] for b in populated_bins]
        counts = [b["sample_count"] for b in populated_bins]
        bin_centers = [(b["bin_lower"] + b["bin_upper"]) / 2 for b in populated_bins]
        widths = [(b["bin_upper"] - b["bin_lower"]) * 0.9 for b in populated_bins]

        ax1.bar(bin_centers, acc_vals, width=widths, alpha=0.6, color="#1f77b4", edgecolor="black", label="Accuracy")

        for b in populated_bins:
            bc = (b["bin_lower"] + b["bin_upper"]) / 2
            bw = (b["bin_upper"] - b["bin_lower"]) * 0.9
            lower = min(b["accuracy"], b["mean_confidence"])
            upper = max(b["accuracy"], b["mean_confidence"])
            ax1.bar(bc, upper - lower, bottom=lower, width=bw, color="#d62728", alpha=0.35, hatch="//", edgecolor="#d62728")

        ax1.set_ylabel("Empirical Accuracy", fontsize=11)
        ax1.set_title(title, fontsize=12, fontweight="bold")
        ax1.set_xlim(0, 1)
        ax1.set_ylim(0, 1.05)
        ax1.legend(loc="upper left")

        ax2.bar(bin_centers, counts, width=widths, color="#2ca02c", alpha=0.7, edgecolor="black")
        ax2.set_xlabel("Confidence (MSP)", fontsize=11)
        ax2.set_ylabel("Item Count", fontsize=10)
        ax2.set_xlim(0, 1)

        plt.tight_layout()
        if save_path:
            plt.savefig(save_path, dpi=300, bbox_inches="tight")
            logger.info(f"Reliability diagram saved to: {save_path}")
        plt.show()
    finally:
        plt.close()


def plot_fragmentation_vs_calibration(
    df: pd.DataFrame,
    title: str = "Tokenization TPW vs. Calibration Residual",
    save_path: Optional[str] = None,
    metric: str = "overall_tpw",
):
    """Scatter plot with trendline showing item-level fragmentation vs Calibration Residual.
    Facets or loops per model x language to avoid pooling models with differing scales.
    Only includes valid runs when CONFIG.analysis_valid_only is True.
    """
    if not HAS_VISUALIZATION or df is None or df.empty:
        return

    clean_df = df.copy()
    if getattr(CONFIG, "analysis_valid_only", True) and "validity_flag" in clean_df.columns:
        clean_df = clean_df[clean_df["validity_flag"] == "VALID"].copy()

    # Fallback to column name variations
    target_metric = metric
    if target_metric not in clean_df.columns:
        if target_metric == "overall_tpw" and "tpw" in clean_df.columns:
            target_metric = "tpw"
        elif target_metric == "overall_tpc" and "tpc" in clean_df.columns:
            target_metric = "tpc"

    if target_metric not in clean_df.columns or "calibration_error_item" not in clean_df.columns:
        return

    clean_df = clean_df.dropna(subset=[target_metric, "calibration_error_item"]).copy()
    if len(clean_df) < 5:
        return

    # Loop or facet per model x language
    groups = list(clean_df.groupby(["model_key", "language"]))
    if not groups:
        return

    n_groups = len(groups)
    cols = min(2, n_groups)
    rows = (n_groups + cols - 1) // cols

    try:
        fig, axes = plt.subplots(rows, cols, figsize=(7 * cols, 5 * rows), squeeze=False)
        for idx, ((m_k, l_k), g_df) in enumerate(groups):
            r_idx, c_idx = idx // cols, idx % cols
            ax = axes[r_idx, c_idx]
            if len(g_df) >= 5:
                sns.regplot(
                    data=g_df,
                    x=target_metric,
                    y="calibration_error_item",
                    ax=ax,
                    scatter_kws={"alpha": 0.5, "color": "#1f77b4", "s": 35},
                    line_kws={"color": "#d62728", "linewidth": 2, "label": "OLS Fit"},
                )
            ax.set_title(f"{m_k} [{l_k}] (N={len(g_df)})", fontsize=11, fontweight="bold")
            ax.set_xlabel(f"Fragmentation ({target_metric})", fontsize=10)
            ax.set_ylabel("Residual (|conf - correct|)", fontsize=10)
            ax.legend(loc="upper right")

        # Hide extra subplots
        for idx in range(n_groups, rows * cols):
            fig.delaxes(axes[idx // cols, idx % cols])

        fig.suptitle(title, fontsize=13, fontweight="bold", y=1.02)
        plt.tight_layout()
        if save_path:
            plt.savefig(save_path, dpi=300, bbox_inches="tight")
            logger.info(f"Fragmentation vs. calibration plot saved to: {save_path}")
        plt.show()
    finally:
        plt.close()
```

## Section 14b: Pre-flight Tokenizer Audit (run BEFORE the expensive model runs)
Loads **only tokenizers** (no weights) and measures TPW / STRR / TPC on the real Bangla and English subsets. This answers two questions the whole study depends on:
1. Do the models really differ in Bangla fragmentation? (`TituLLM` is described as a Bangla-extended Llama, but the v1.1 model card only says "continually pre-trained"; if its tokenizer equals Llama-3.2's, H1 has no contrast between those two.)
2. Is each tokenizer accessible with your token? (Llama-3.2 is gated.)

A warning is logged if two models share an identical tokenizer. Results are saved to `tokenizer_audit.csv`; the table is also a ready-made descriptive table for the paper.


```python
def run_tokenizer_audit(model_keys: List[str], n_items: int = 300) -> pd.DataFrame:
    """Tokenizer-only pre-flight (no model weights are downloaded).

    Requires an explicit list of model_keys (no default over the whole registry).
    Reports vocabulary size and TPW / STRR / TPC on the REAL Bangla and English subsets for every specified model.
    Skips models whose license is not accepted on Hugging Face Hub with a clear message.
    Warns when two models share an identical tokenizer.
    """
    from transformers import AutoTokenizer

    if not model_keys or not isinstance(model_keys, (list, tuple)):
        raise ValueError("run_tokenizer_audit requires an explicit list of model_keys (no default over the whole registry).")

    keys = list(model_keys)
    bn_all = load_dataset_from_file(bn_subset_file, "bn", "BnMMLU", is_mock=False)
    en_all = load_dataset_from_file(en_subset_file, "en", en_benchmark_label(CONFIG), is_mock=False)
    rng = np.random.RandomState(CONFIG.seed)
    bn_items = [bn_all[i] for i in rng.permutation(len(bn_all))[:n_items]]
    en_items = [en_all[i] for i in rng.permutation(len(en_all))[:n_items]]

    probe = "বাংলাদেশের রাজধানী ঢাকা এবং আমাদের মাতৃভাষা বাংলা।"
    rows: List[Dict[str, Any]] = []
    fingerprints: Dict[str, Tuple[int, Tuple[int, ...]]] = {}

    for key in keys:
        if key not in MODEL_REGISTRY:
            raise ValueError(f"Unknown model_key '{key}'. Registered models: {list(MODEL_REGISTRY.keys())}")
        info = MODEL_REGISTRY[key]

        # Skip models whose license is explicitly marked as not accepted
        if info.get("license_accepted") is False:
            msg = f"Skipping model '{key}' ({info['hf_id']}): license has not been accepted on Hugging Face Hub."
            logger.warning(msg)
            print(f"⚠️ {msg}")
            rows.append({"model_key": key, "hf_id": info["hf_id"], "status": "SKIPPED: license_not_accepted"})
            continue

        try:
            tok = AutoTokenizer.from_pretrained(info["hf_id"], token=CONFIG.hf_token)
        except Exception as e:
            err_str = str(e).lower()
            is_gated = (
                "gated" in err_str
                or "403" in err_str
                or "license" in err_str
                or "agreement" in err_str
                or "access" in err_str
                or "gatedrepoerror" in type(e).__name__.lower()
            )
            if is_gated:
                msg = f"Skipping model '{key}' ({info['hf_id']}): license is not accepted on Hugging Face Hub ({type(e).__name__}: {e})."
                logger.warning(msg)
                print(f"⚠️ {msg}")
                rows.append({"model_key": key, "hf_id": info["hf_id"], "status": "SKIPPED: license_not_accepted"})
                continue
            logger.error(f"Tokenizer for '{key}' ({info['hf_id']}) could not be loaded: {type(e).__name__}: {e}. "
                         "For gated models accept the license on the model page and check HF_TOKEN.")
            rows.append({"model_key": key, "hf_id": info["hf_id"], "status": f"FAILED: {type(e).__name__}"})
            continue

        probe_ids = tuple(tok.encode(probe, add_special_tokens=False))
        fingerprints[key] = (len(tok), probe_ids)
        # S1: In-context option token verification
        try:
            verify_option_tokens_in_context(tok, model_id=info["hf_id"])
        except Exception as _audit_err:
            logger.error(f"Option token in-context verification failed during audit for {key}: {_audit_err}")
            raise
        _, bn_sum = compute_dataset_fragmentation(info["hf_id"], bn_items, CONFIG.hf_token)
        _, en_sum = compute_dataset_fragmentation(info["hf_id"], en_items, CONFIG.hf_token)
        rows.append({
            "model_key": key, "hf_id": info["hf_id"], "status": "ok",
            "tokenizer_class": type(tok).__name__, "vocab_size": len(tok),
            "probe_sentence_tokens": len(probe_ids),
            "bn_mean_tpw": round(bn_sum["mean_tpw"], 3), "bn_mean_strr": round(bn_sum["mean_strr_isolated"], 3),
            "bn_mean_tpc": round(bn_sum["mean_tpc"], 3), "en_mean_tpw": round(en_sum["mean_tpw"], 3),
            "bn_over_en_tpw": round(bn_sum["mean_tpw"] / en_sum["mean_tpw"], 2) if en_sum["mean_tpw"] else float("nan"),
        })

    audit_df = pd.DataFrame(rows)

    # Identical-tokenizer check (same vocabulary size AND same ids on a Bangla probe sentence)
    ok_keys = list(fingerprints)
    for i in range(len(ok_keys)):
        for j in range(i + 1, len(ok_keys)):
            a, b = ok_keys[i], ok_keys[j]
            if fingerprints[a] == fingerprints[b]:
                logger.warning(
                    f"IDENTICAL TOKENIZERS: '{a}' and '{b}' tokenize Bangla exactly the same way (vocab={fingerprints[a][0]:,}). "
                    "Their fragmentation is equal by construction, so this pair cannot test H1 (low vs high fragmentation). "
                    "Pick a model with a different tokenizer for the 'low-fragmentation' arm, or report them as a null contrast."
                )
            elif fingerprints[a][0] != fingerprints[b][0]:
                logger.info(f"Tokenizers differ: '{a}' vocab={fingerprints[a][0]:,} vs '{b}' vocab={fingerprints[b][0]:,}.")

    out_path = os.path.join(CONFIG.output_dir, "tokenizer_audit.csv")
    audit_df.to_csv(out_path, index=False, encoding="utf-8")
    print("\n" + "=" * 88 + "\nTOKENIZER AUDIT (real subsets, tokenizer-only)\n" + "=" * 88)
    print(audit_df.to_string(index=False))
    print("=" * 88)
    logger.info(f"Tokenizer audit saved to: {out_path}")
    return audit_df


# Tokenizer audit requires explicit model keys to prevent accidental registry-wide execution:
# tokenizer_audit_df = run_tokenizer_audit(model_keys=["llama32_3b", "titullm_1b", "tigerllm_1b"])
```

## Section 11b: Prompt-variant diagnostic (run per model BEFORE the full run)
`diagnose_prompt_variants(model_key, n=60)` loads the backend once and evaluates the full grid:
- `prompt_variant`: `{plain, answer_cue, few_shot}` (skips `few_shot` if `fewshot_exemplars.json` is missing)
- `chat_template`: `{off, on}` (evaluates `on` only if `tokenizer.chat_template` exists)
- `debias`: `{none, cyclic4}`

### Methodology & Decision Rules
1. **Validity Gates Only:** Settings are selected using ONLY `mean_mass >= MIN_OPTION_MASS_THRESHOLD` (0.50) and `top_letter_share <= MAX_COLLAPSE_THRESHOLD` (0.80) across BOTH Bangla and English.
2. **Never Select by Accuracy:** Diagnostics run on a **held-out dev pool** (`dev_bn.csv`, `dev_en.csv`), disjoint from the evaluation items and the few-shot exemplars. Accuracy is displayed but never used for selection.
3. **Preference Hierarchy:** Among configurations passing both gates in both languages, prefer:
   - `debias`: `"none"` over `"cyclic4"`
   - `prompt_variant`: `plain` < `answer_cue` < `few_shot`
   - `chat_template`: `False` (off) over `True` (on)
   - Highest minimum `mean_mass` across languages
4. **Re-run Workflow Reminder:** If re-running a model after updating its registry configuration, delete its existing results folder first (`shutil.rmtree(os.path.join(CONFIG.output_dir, model_key))` or `rm -rf results/<model_key>`). Otherwise, the resume guard will detect the signature change or existing checkpoints will prevent clean re-evaluation.

> **Section 11b diagnostics are currently commented out. TituLLM (base model) and TigerLLM collapse to letter 'A' under default settings. Run diagnose_prompt_variants for both, copy the printed settings into MODEL_REGISTRY, delete results/<model_key>, then re-run.**

### Workflow
`diagnose_prompt_variants(model_key, n=60)` (dev pool) -> choose settings on validity gates only -> **run Section 14c** (`apply_uniform_setting`) -> run the models.

Settings precedence: `model_settings.json` > `MODEL_REGISTRY` > `CONFIG`. **The per-model inference settings actually used (prompt variant, shots, chat template, debiasing, and the dev pool size) must be reported in the paper's methodology.**

Note: dev pools are sampled from the full validated sets (all subjects), so the Bangla and English dev subject mixes may differ; this is acceptable because diagnostics use only validity gates.

> **Primary run uses ONE uniform setting for all models (few_shot, k=4, chat off, no debiasing), applied in Section 14c. Do not re-run the diagnostics for the primary run.**

## Section 14c: Apply the uniform inference setting (RUN BEFORE Sections 15-17)
All three models use the same setting: `few_shot` (k=4), chat template off, no debiasing. This avoids a per-model prompt confound. The cell below:
1. checks that `fewshot_exemplars.json` exists,
2. writes the setting to `model_settings.json` (highest precedence) and verifies it resolves correctly,
3. **archives and resets only results produced under a different setting** (results already produced under this setting are kept, so interrupted runs can resume),
4. defines `require_uniform(model_key)`, which Sections 15-17 call so a model can never run with stale settings.


```python
# ============================================================
# Section 14c: Apply uniform setting (safe to re-run; idempotent)
# ============================================================
import shutil, json, os
from datetime import datetime

UNIFORM_KEYS = ["llama32_3b", "tigerllm_1b", "titullm_1b"]

# 1. Few-shot needs exemplars
_ex = os.path.join(CONFIG.output_dir, "fewshot_exemplars.json")
if not os.path.exists(_ex):
    raise RuntimeError(f"{_ex} not found. Run the data preparation section (Section 5) first.")

# 2. Apply + verify
apply_uniform_setting(UNIFORM_KEYS, config=CONFIG, dry_run=False)
_resolved = {k: resolve_run_settings(k, CONFIG) for k in UNIFORM_KEYS}
for _k, _s in _resolved.items():
    assert (_s["prompt_variant"] == "few_shot" and _s["effective_n_shots"] == 4
            and _s["use_chat_template"] is False and _s["debias"] == "none"
            and _s["settings_source"] == "model_settings.json"), (_k, _s)

# 3. Archive + reset ONLY stale results (different inference signature)
_archive_root = os.path.join(os.path.dirname(os.path.abspath(CONFIG.output_dir)),
                             "results_archive_" + datetime.now().strftime("%Y%m%d_%H%M%S"))
for _k in UNIFORM_KEYS:
    _m_dir = os.path.join(CONFIG.output_dir, _k)
    if not os.path.isdir(_m_dir):
        logger.info(f"[{_k}] no previous results.")
        continue
    _stale = True
    _st_path = os.path.join(_m_dir, "status.json")
    if os.path.exists(_st_path):
        try:
            with open(_st_path, "r", encoding="utf-8") as _f:
                _st = json.load(_f)
            _stale = _st.get("inference_signature") != _make_inference_signature(_resolved[_k])
        except Exception:
            _stale = True
    if _stale:
        os.makedirs(_archive_root, exist_ok=True)
        shutil.copytree(_m_dir, os.path.join(_archive_root, _k))
        reset_model_results(_k, confirm=True, config=CONFIG)
        logger.info(f"[{_k}] stale results archived to {_archive_root} and reset.")
    else:
        logger.info(f"[{_k}] existing results already use the uniform setting; kept (resume allowed).")

def require_uniform(model_key: str) -> None:
    """Fail fast if a model would run with anything other than the uniform setting."""
    s = resolve_run_settings(model_key, CONFIG)
    if not (s["prompt_variant"] == "few_shot" and s["effective_n_shots"] == 4
            and s["use_chat_template"] is False and s["debias"] == "none"
            and s["settings_source"] == "model_settings.json"):
        raise RuntimeError(f"Uniform setting not applied for '{model_key}': {s}. Run Section 14c first.")


def _derive_manifest_prompt_variant(resolved_settings: dict, uniform_keys: list) -> str:
    """Derive the top-level manifest prompt_variant from effective per-model settings.

    If all models share the same effective prompt_variant, that value is used.
    Otherwise "mixed" is returned, and per-model entries carry the ground truth.
    This ensures run_manifest.json always reflects the *effective* setting rather
    than the stale CONFIG.prompt_variant default.
    """
    variants = {resolved_settings[k]["prompt_variant"] for k in uniform_keys if k in resolved_settings}
    if len(variants) == 1:
        return variants.pop()
    return "mixed"

for _k in UNIFORM_KEYS:
    print(_k, {kk: _resolved[_k][kk] for kk in ("prompt_variant", "effective_n_shots", "use_chat_template", "debias", "settings_source")})
```

## Section 14d: E2 Rescue Runs Runner

Executes E2 rescue configurations without calling `require_uniform`:
1. `cyclic4` (TituLLM, TigerLLM)
2. `tiger_chat` (chat-template ON for TigerLLM-1B-it only)
3. `0shot_answer_cue` (TituLLM, TigerLLM)

Each configuration uses a separate output directory (`results_E2_<config>`), verifies against Run A inference signature collisions, and aborts on mock data.

```python
# ============================================================
# Section 14d: E2 Rescue Runs Runner & Dev Probabilities
# ============================================================
import dataclasses
import os
import json
import pandas as pd
from typing import Dict, Any, List, Optional, Tuple

DEFAULT_RUN_A_INFERENCE_SIGNATURE = "few_shot|shots=4|chat=0|debias=none"

def get_run_a_inference_signature(output_dir: Optional[str] = None) -> str:
    """Get Run A's inference signature from manifest, model_settings, or fallback default."""
    search_dirs = [output_dir, getattr(CONFIG, "output_dir", "results"), "results", "../results"]
    for d in search_dirs:
        if not d:
            continue
        # 1. Try run_manifest.json
        manifest_p = os.path.join(d, "run_manifest.json")
        if os.path.exists(manifest_p):
            try:
                with open(manifest_p, "r", encoding="utf-8") as f:
                    m = json.load(f)
                sig = m.get("inference_signature")
                if sig:
                    return str(sig)
                models = m.get("models", {})
                if models and isinstance(models, dict):
                    first_m = next(iter(models.values()))
                    if isinstance(first_m, dict) and "inference_signature" in first_m:
                        return str(first_m["inference_signature"])
            except Exception:
                pass
        # 2. Try model_settings.json
        ms_p = os.path.join(d, "model_settings.json")
        if os.path.exists(ms_p):
            try:
                with open(ms_p, "r", encoding="utf-8") as f:
                    ms = json.load(f)
                if ms and isinstance(ms, dict):
                    first_s = next(iter(ms.values()))
                    if isinstance(first_s, dict):
                        return _make_inference_signature(first_s)
            except Exception:
                pass
    return DEFAULT_RUN_A_INFERENCE_SIGNATURE

RUN_A_INFERENCE_SIGNATURE = get_run_a_inference_signature()

E2_CONFIGS: Dict[str, Dict[str, Any]] = {
    "cyclic4": {
        "output_dir_name": "results_E2_cyclic4",
        "description": "Exemplar cyclic permutation (cyclic4 debiasing)",
        "models": ["titullm_1b", "tigerllm_1b"],
        "settings": {
            "prompt_variant": "few_shot",
            "n_shots": 4,
            "use_chat_template": False,
            "debias": "cyclic4",
        },
    },
    "tiger_chat": {
        "output_dir_name": "results_E2_chat_template",
        "description": "Chat template enabled for TigerLLM-1B-it only",
        "models": ["tigerllm_1b"],
        "settings": {
            "prompt_variant": "few_shot",
            "n_shots": 4,
            "use_chat_template": True,
            "debias": "none",
        },
    },
    "0shot_answer_cue": {
        "output_dir_name": "results_E2_0shot_answer_cue",
        "description": "Zero-shot with answer cue prompt variant",
        "models": ["titullm_1b", "tigerllm_1b"],
        "settings": {
            "prompt_variant": "answer_cue",
            "n_shots": 0,
            "use_chat_template": False,
            "debias": "none",
        },
    },
}


def run_e2_experiments(
    configs: Optional[Dict[str, Dict[str, Any]]] = None,
    base_config: ExperimentConfig = CONFIG,
    languages: Tuple[str, ...] = ("bn", "en"),
    allow_mock: bool = False,
    execute: bool = False,
) -> Dict[str, Any]:
    """Execute E2 rescue configurations saving per-item CSVs and summaries under results_E2_<config>_<model>/.

    For each config and model:
    - Builds a separate output dir: results_E2_<config>_<model>/
    - Persists settings via model_settings.json
    - Prints resolve_run_settings for each model before running
    - Aborts if inference_signature collides with Run A signature
    - Aborts on mock data unless allow_mock is True
    - Evaluates models with resume support
    - Flattens per-item CSVs and summaries directly into results_E2_<config>_<model>/
    - Writes summary.json with required pre-registered keys (top_share, min_letter_share,
      raw_prediction_distribution, gate_pass, gate_reason, config_name, debias,
      prompt_variant, n_shots, use_chat_template, seed, n_items).
    """
    import shutil
    if configs is None:
        configs = E2_CONFIGS

    results_summary: Dict[str, Any] = {}
    base_parent = os.path.dirname(os.path.abspath(base_config.output_dir))

    for cfg_key, cfg_info in configs.items():
        cfg_results: Dict[str, Any] = {}
        for m_key in cfg_info["models"]:
            # Output directory structured as results_E2_<config>_<model>
            folder_name = f"results_E2_{cfg_key}_{m_key}"
            model_out_dir = os.path.join(base_parent, folder_name)
            os.makedirs(model_out_dir, exist_ok=True)
            cfg_e2 = dataclasses.replace(base_config, output_dir=model_out_dir)

            # Set up settings map for this model and config
            m_setting = dict(cfg_info["settings"])
            m_setting["provenance"] = {
                "source": f"E2_rescue_{cfg_key}",
                "description": cfg_info.get("description", ""),
            }
            settings_map = {m_key: m_setting}
            # Model-level resume guard: skip if summary.json already exists with completed evaluation
            summary_json_path = os.path.join(model_out_dir, "summary.json")
            if os.path.exists(summary_json_path):
                try:
                    with open(summary_json_path, "r", encoding="utf-8") as _sf:
                        _prev_sum = json.load(_sf)
                    if _prev_sum.get("gate_reason") or _prev_sum.get("languages") or _prev_sum.get("gate_pass") is not None:
                        logger.info(f"⏭️ [E2 RESUME GUARD] Skipping '{m_key}' in '{cfg_key}': summary.json already exists in {model_out_dir}.")
                        print(f"⏭️ [E2 RESUME GUARD] Skipping '{m_key}' in '{cfg_key}': evaluation already completed in {model_out_dir}.")
                        cfg_results[m_key] = _prev_sum.get("languages", _prev_sum)
                        continue
                except Exception as _re_err:
                    logger.warning(f"Failed to read existing summary.json in {model_out_dir} ({_re_err}); re-evaluating.")

            save_model_settings(settings_map, output_dir=cfg_e2.output_dir)
            logger.info(f"[E2:{cfg_key}:{m_key}] Configured output dir: {model_out_dir}")

            resolved = resolve_run_settings(m_key, cfg_e2)
            print(f"[E2:{cfg_key}:{m_key}] resolve_run_settings for {m_key}: {resolved}")

            # Check inference signature collision with Run A
            sig = _make_inference_signature(resolved)
            run_a_sig = get_run_a_inference_signature(base_config.output_dir)
            if sig == run_a_sig:
                raise RuntimeError(
                    f"Abort: Inference signature for {m_key} in {cfg_key} collides with Run A ({sig})."
                )

            # Abort on mock data
            m_info = MODEL_REGISTRY.get(m_key, {})
            m_id = str(m_info.get("model_id", "") or m_info.get("hf_id", ""))
            is_mock_model = (
                cfg_e2.backend == "mock"
                or "mock" in str(cfg_e2.backend).lower()
                or "mock" in m_id.lower()
            )
            if is_mock_model and not allow_mock:
                raise RuntimeError(
                    f"Abort: E2 runner aborts on mock data (backend={cfg_e2.backend}, model={m_id}). Real data/model required."
                )

            debias_val = resolved.get("debias", "none")
            pv_val = resolved.get("prompt_variant", "few_shot")
            n_shots_val = int(resolved.get("n_shots", 0))
            use_chat_val = bool(resolved.get("use_chat_template", False))
            seed_val = int(getattr(cfg_e2, "seed", 42))
            n_items_val = int(getattr(cfg_e2, "n_samples", 1500))

            if execute:
                # Check / supply loaded subset items
                for lang in languages:
                    subset_file = os.path.join(cfg_e2.output_dir, f"{lang}_subset.csv")
                    if not os.path.exists(subset_file):
                        subset_file = os.path.join(base_config.output_dir, f"{lang}_subset.csv")
                    if not os.path.exists(subset_file):
                        for cand_dir in ("results", "Bangla-LLM-Calibration-results(v4e)", os.path.join("..", "Bangla-LLM-Calibration-results(v4e)")):
                            alt_p = os.path.join(cand_dir, f"{lang}_subset.csv")
                            if os.path.exists(alt_p):
                                shutil.copy2(alt_p, os.path.join(cfg_e2.output_dir, f"{lang}_subset.csv"))
                                subset_file = os.path.join(cfg_e2.output_dir, f"{lang}_subset.csv")
                                break
                    if os.path.exists(subset_file):
                        df_sub = pd.read_csv(subset_file)
                        if not allow_mock:
                            has_mock = (
                                bool(df_sub.get("is_mock", pd.Series([False])).any())
                                or ("source_dataset" in df_sub.columns and (df_sub["source_dataset"] == "mock").any())
                                or ("question_id" in df_sub.columns and df_sub["question_id"].astype(str).str.startswith("mock").any())
                            )
                            if has_mock:
                                raise RuntimeError(f"Abort: E2 runner aborts on mock data found in {subset_file}.")

                res = run_one_model(m_key, languages=languages, config=cfg_e2)

                # Flatten per-item CSVs and summaries from model_out_dir/m_key up into model_out_dir
                inner_dir = os.path.join(model_out_dir, m_key)
                if os.path.exists(inner_dir) and os.path.isdir(inner_dir):
                    for fname in os.listdir(inner_dir):
                        src_f = os.path.join(inner_dir, fname)
                        dst_f = os.path.join(model_out_dir, fname)
                        if os.path.isfile(src_f):
                            shutil.copy2(src_f, dst_f)

                # Extract and format per-language and joint summary metrics
                top_shares = {}
                min_shares = {}
                raw_dists = {}
                gate_passes = {}
                gate_reasons = {}

                for lang in languages:
                    l_sum = res.get(lang, {}) if isinstance(res, dict) else {}
                    ts = l_sum.get("top_share")
                    ms = l_sum.get("min_letter_share")
                    gp = l_sum.get("gate_pass")
                    gr = l_sum.get("gate_reason", "")
                    raw_d = l_sum.get("raw_prediction_distribution") or l_sum.get("diagnostics", {}).get("prediction_distribution", {})

                    cand_csv = os.path.join(model_out_dir, f"{lang}_results.csv")
                    if os.path.exists(cand_csv):
                        _df = pd.read_csv(cand_csv)
                        pred_col = "pred_raw" if "pred_raw" in _df.columns else ("predicted_answer" if "predicted_answer" in _df.columns else None)
                        if pred_col:
                            _vc = _df[pred_col].dropna().value_counts()
                            _tot = len(_df)
                            if ts is None and _tot > 0:
                                ts = float(_vc.iloc[0] / _tot)
                            if ms is None and _tot > 0:
                                _ls = {l: float(_vc.get(l, 0) / _tot) for l in "ABCD"}
                                ms = min(_ls.values())
                            if (not raw_d) and _tot > 0:
                                raw_d = {str(k): int(v) for k, v in _vc.to_dict().items()}

                    top_shares[lang] = ts
                    min_shares[lang] = ms
                    raw_dists[lang] = raw_d
                    gate_passes[lang] = gp
                    gate_reasons[lang] = gr

                    # Write enriched per-language summary
                    l_sum_enriched = dict(l_sum)
                    l_sum_enriched.update({
                        "top_share": ts,
                        "min_letter_share": ms,
                        "raw_prediction_distribution": raw_d,
                        "gate_pass": gp,
                        "gate_reason": gr,
                        "notebook_version": "v4i",
                        "config_name": cfg_key,
                        "debias": debias_val,
                        "prompt_variant": pv_val,
                        "n_shots": n_shots_val,
                        "use_chat_template": use_chat_val,
                        "seed": seed_val,
                        "n_items": n_items_val,
                    })
                    with open(os.path.join(model_out_dir, f"{lang}_summary.json"), "w", encoding="utf-8") as sf:
                        json.dump(l_sum_enriched, sf, indent=2)

                overall_gate_pass = bool(all(gate_passes.values())) if gate_passes else False
                overall_gate_reason = "; ".join([f"{l}: {gate_reasons.get(l, '')}" for l in languages])
                max_top = max([v for v in top_shares.values() if v is not None]) if top_shares else None
                min_min = min([v for v in min_shares.values() if v is not None]) if min_shares else None

                summary_json_data = {
                    "config_name": cfg_key,
                    "notebook_version": "v4i",
                    "model_key": m_key,
                    "debias": debias_val,
                    "prompt_variant": pv_val,
                    "n_shots": n_shots_val,
                    "use_chat_template": use_chat_val,
                    "seed": seed_val,
                    "n_items": n_items_val,
                    "top_share": max_top,
                    "min_letter_share": min_min,
                    "top_share_by_lang": top_shares,
                    "min_letter_share_by_lang": min_shares,
                    "raw_prediction_distribution": raw_dists,
                    "gate_pass": overall_gate_pass,
                    "gate_reason": overall_gate_reason,
                    "languages": res,
                }
                summary_json_path = os.path.join(model_out_dir, "summary.json")
                with open(summary_json_path, "w", encoding="utf-8") as sf:
                    json.dump(summary_json_data, sf, indent=2)
                logger.info(f"[E2:{cfg_key}:{m_key}] Wrote {summary_json_path}")

                cfg_results[m_key] = res

                # Print row counts and model_id values from saved CSVs
                for lang in languages:
                    cand_paths = [
                        os.path.join(model_out_dir, f"{lang}_results.csv"),
                        os.path.join(model_out_dir, f"{m_key}_{lang}_results.csv"),
                        os.path.join(model_out_dir, m_key, f"{m_key}_{lang}_results.csv"),
                        os.path.join(model_out_dir, m_key, f"{lang}_results.csv"),
                    ]
                    csv_path = next((p for p in cand_paths if os.path.exists(p)), None)
                    if csv_path:
                        df = pd.read_csv(csv_path)
                        mid = df["model_id"].iloc[0] if ("model_id" in df.columns and len(df) > 0) else "N/A"
                        print(f"[E2:{cfg_key}] {m_key} ({lang}): {len(df)} rows, model_id={mid}")
                        if not allow_mock:
                            has_mock_row = (
                                bool(df.get("is_mock", pd.Series([False])).any())
                                or "mock" in str(mid).lower()
                            )
                            if has_mock_row:
                                raise RuntimeError(f"Abort: E2 runner aborts on mock data found in {csv_path} (model_id={mid}).")
            else:
                dry_summary = {
                    "config_name": cfg_key,
                    "model_key": m_key,
                    "debias": debias_val,
                    "prompt_variant": pv_val,
                    "n_shots": n_shots_val,
                    "use_chat_template": use_chat_val,
                    "seed": seed_val,
                    "n_items": n_items_val,
                    "top_share": None,
                    "min_letter_share": None,
                    "raw_prediction_distribution": None,
                    "gate_pass": None,
                    "gate_reason": "dry_run",
                    "status": "configured_dry_run",
                    "inference_signature": sig,
                }
                summary_json_path = os.path.join(model_out_dir, "summary.json")
                with open(summary_json_path, "w", encoding="utf-8") as sf:
                    json.dump(dry_summary, sf, indent=2)
                cfg_results[m_key] = dry_summary

        results_summary[cfg_key] = cfg_results

    return results_summary

def run_dev_probabilities(
    model_keys: Tuple[str, ...] = ("titullm_1b", "tigerllm_1b"),
    languages: Tuple[str, ...] = ("bn", "en"),
    config: ExperimentConfig = CONFIG,
    output_dir: Optional[str] = None,
    execute: bool = False,
    allow_mock: bool = False,
) -> Dict[str, pd.DataFrame]:
    """Run inference on the 200 dev items per language and save prob_A..prob_D.

    Allows label-free prior estimation on dev split to be applied offline to eval probabilities.
    Reuses src.analysis prior-debiasing logic.
    Uses resolve_run_settings so dev probabilities match the eval inference setting.
    Asserts items are strictly from dev split with no overlap with eval items or exemplars.
    """
    out_dir = output_dir or config.output_dir
    os.makedirs(out_dir, exist_ok=True)
    saved_dfs: Dict[str, pd.DataFrame] = {}

    for m_key in model_keys:
        settings = resolve_run_settings(m_key, config)
        prompt_variant = settings.get("prompt_variant", "few_shot")
        effective_n_shots = int(settings.get("effective_n_shots", settings.get("n_shots", 4)))
        use_chat = bool(settings.get("use_chat_template", False))

        for lang in languages:
            dev_file = os.path.join(config.output_dir, f"dev_{lang}.csv")
            if not os.path.exists(dev_file):
                dev_file = os.path.join("results", f"dev_{lang}.csv")
            if not os.path.exists(dev_file):
                logger.warning(f"Dev file for {lang} not found; skipping dev probabilities for {m_key} {lang}.")
                continue

            # B2 fix (v4h): use en_benchmark_label(CONFIG) so translation mode gets BnMMLU-en
            dev_items = load_dataset_from_file(dev_file, language=lang,
                benchmark="BnMMLU" if lang == "bn" else en_benchmark_label(CONFIG),
                is_mock=False)
            dev_ids = {str(it.get("question_id", "")) for it in dev_items}
            logger.info(f"Loaded {len(dev_items)} dev items for {m_key} [{lang}] from {dev_file}.")

            # Assert items come only from dev files with no overlap with eval ids or exemplars
            eval_file = os.path.join(config.output_dir, f"{lang}_subset.csv")
            if not os.path.exists(eval_file):
                eval_file = os.path.join("results", f"{lang}_subset.csv")
            if os.path.exists(eval_file):
                eval_df = pd.read_csv(eval_file)
                eval_ids = set(eval_df["question_id"].astype(str))
                overlap = dev_ids & eval_ids
                assert len(overlap) == 0, f"Data leakage detected! Dev set overlaps with eval set for {lang}: {len(overlap)} items"

            ex_file = os.path.join(config.output_dir, "fewshot_exemplars.json")
            if not os.path.exists(ex_file):
                ex_file = os.path.join("results", "fewshot_exemplars.json")
            if os.path.exists(ex_file):
                with open(ex_file, "r", encoding="utf-8") as f:
                    ex_data = json.load(f)
                ex_items = ex_data.get(lang, []) if isinstance(ex_data, dict) else []
                ex_ids = {str(it.get("question_id", "")) for it in ex_items}
                overlap_ex = dev_ids & ex_ids
                assert len(overlap_ex) == 0, f"Data leakage detected! Dev set overlaps with few-shot exemplars for {lang}: {len(overlap_ex)} items"

            out_csv = os.path.join(out_dir, f"{m_key}_{lang}_dev_probabilities.csv")
            if os.path.exists(out_csv):
                try:
                    df_existing = pd.read_csv(out_csv)
                    if len(df_existing) == len(dev_items):
                        logger.info(f"Loaded existing dev probabilities from {out_csv} ({len(df_existing)} rows).")
                        saved_dfs[f"{m_key}_{lang}"] = df_existing
                        continue
                except Exception:
                    pass
            if execute:
                if config.backend == "mock" and not allow_mock:
                    raise RuntimeError("Abort: run_dev_probabilities aborts on mock backend. Real data/model required.")

                backend = _build_backend(m_key, config, use_chat_template=use_chat)
                rows = []
                for item in dev_items:
                    # Same code path as the evaluation (prompt variant, exemplars, debias), so dev and eval match
                    res = predict_item(backend, item, lang, settings)
                    raw_probs = getattr(res, "probs_raw", None) or [None] * 4
                    row = {
                        "question_id": item["question_id"],
                        "gold_answer": item.get("correct_answer", item.get("gold_answer", "")),
                        "predicted_answer": res.predicted_letter,
                        "confidence": res.confidence,
                        "prob_A": res.probabilities[0] if res.probabilities else None,
                        "prob_B": res.probabilities[1] if res.probabilities else None,
                        "prob_C": res.probabilities[2] if res.probabilities else None,
                        "prob_D": res.probabilities[3] if res.probabilities else None,
                        "pred_raw": getattr(res, "pred_raw", None),
                        "confidence_raw": getattr(res, "confidence_raw", None),
                        "prob_raw_A": raw_probs[0] if len(raw_probs) > 0 else None,
                        "prob_raw_B": raw_probs[1] if len(raw_probs) > 1 else None,
                        "prob_raw_C": raw_probs[2] if len(raw_probs) > 2 else None,
                        "prob_raw_D": raw_probs[3] if len(raw_probs) > 3 else None,
                        "option_mass": getattr(res, "option_mass", 1.0),
                        "inference_signature": _make_inference_signature(settings),
                    }
                    rows.append(row)
                df = pd.DataFrame(rows)
                df.to_csv(out_csv, index=False, encoding="utf-8")
                logger.info(f"Saved dev probabilities to {out_csv} ({len(df)} rows).")
                saved_dfs[f"{m_key}_{lang}"] = df
                # Free backend and GPU memory between languages (v4f addendum)
                if 'backend' in locals() and backend is not None:
                    del backend
                import gc
                gc.collect()
                try:
                    import torch
                    if torch.cuda.is_available():
                        torch.cuda.empty_cache()
                except Exception:
                    pass
            else:
                saved_dfs[f"{m_key}_{lang}"] = pd.DataFrame(columns=[
                    "question_id", "gold_answer", "predicted_answer", "confidence",
                    "prob_A", "prob_B", "prob_C", "prob_D",
                    "pred_raw", "confidence_raw", "prob_raw_A", "prob_raw_B", "prob_raw_C", "prob_raw_D",
                    "option_mass", "inference_signature",
                ])

    return saved_dfs
```

```python
# ============================================================
# Section 14d-bis: E2 Stage Runner (Non-Interactive Run All for v4h)
# ============================================================
import dataclasses
import os
import shutil
import gc
from typing import Any, Dict, List, Optional, Tuple
import numpy as np
import pandas as pd

try:
    import torch
except ImportError:
    torch = None

# FIX 2: skip existing-model E2 stage on Run All
_RUN_E2 = bool(globals().get("RUN_E2_EXISTING", False))
if not _RUN_E2:
    print("E2 existing-model stage skipped (RUN_E2_EXISTING=False)")
    print("  No titullm_1b, tigerllm_1b, or llama32_3b will be loaded.")
    print("  No results_E2_cyclic4_* files will be written.")

FORCE_FULL = False  # Keep pre-registered dev screening gate active
# ---- Configurable target list ----
# To run llama-only: TARGET_RESCUE_MODELS = ["llama32_3b"]
# To run all three:  TARGET_RESCUE_MODELS = ["titullm_1b", "tigerllm_1b", "llama32_3b"]
#
# FRESH SESSION BEHAVIOR:
# In a fresh Kaggle session /kaggle/working is empty, so the resume guard (which
# checks for results_E2_cyclic4_<model>/summary.json) will find nothing and all
# three models will run. If a previous session already produced a summary.json
# for one model, that model is automatically skipped.  Run this cell exactly once
# per session; do not loop or restart mid-run.
TARGET_RESCUE_MODELS = ["titullm_1b", "tigerllm_1b", "llama32_3b"]
BASE_DIR = os.environ.get("KAGGLE_WORKING_DIR", "/kaggle/working" if os.path.exists("/kaggle/working") else ".")
REPORT_CSV_PATH = os.path.join(BASE_DIR, "dev_screen_report_cyclic4.csv")

print("=" * 70)
print("🚀 E2 STAGE RUNNER (v4h): Non-Interactive Stage 4 Cyclic4 Execution")
print(f"   Target Models: {TARGET_RESCUE_MODELS}")
print(f"   FORCE_FULL override: {FORCE_FULL}")
print(f"   Dev Screen Report CSV: {REPORT_CSV_PATH}")
print("=" * 70)


def _ensure_split_files_present(target_dir: str, languages: Tuple[str, ...] = ("bn", "en")):
    """Ensure dev and eval subset files are accessible in target_dir."""
    os.makedirs(target_dir, exist_ok=True)
    candidate_sources = [
        target_dir,
        getattr(CONFIG, "output_dir", "results"),
        "results",
        "Bangla-LLM-Calibration-results(v4e)",
        os.path.join("..", "Bangla-LLM-Calibration-results(v4e)"),
    ]
    for lang in languages:
        for fname in (f"dev_{lang}.csv", f"{lang}_subset.csv"):
            dest_path = os.path.join(target_dir, fname)
            if not os.path.exists(dest_path):
                for s_dir in candidate_sources:
                    s_path = os.path.join(s_dir, fname)
                    if os.path.exists(s_path):
                        shutil.copy2(s_path, dest_path)
                        break
    ex_dest = os.path.join(target_dir, "fewshot_exemplars.json")
    if not os.path.exists(ex_dest):
        for s_dir in candidate_sources:
            s_path = os.path.join(s_dir, "fewshot_exemplars.json")
            if os.path.exists(s_path):
                shutil.copy2(s_path, ex_dest)
                break


def screen_dev_distribution(
    dev_dfs: Dict[str, pd.DataFrame],
    model_key: str,
    output_dir: str,
    languages: Tuple[str, ...] = ("bn", "en"),
    pred_col: str = "predicted_answer",
) -> Tuple[bool, Dict[str, Any]]:
    """Compute dev top_share and min_letter_share per language and test against strict gates.

    Extended per v4f addendum to return n, top_letter, min_letter per language.

    Gates (from gate.yaml):
    - top_share <= 0.40
    - min_letter_share >= 0.10
    """
    metrics_by_lang = {}
    all_passed = True
    print("\n" + "=" * 45)
    print(f"📊 DEV SCREEN REPORT for {model_key}")
    print("=" * 45)

    for lang in languages:
        key = f"{model_key}_{lang}"
        df = dev_dfs.get(key)
        if df is None or len(df) == 0:
            cand_csvs = [
                os.path.join(output_dir, f"{model_key}_{lang}_dev_probabilities.csv"),
                os.path.join(getattr(CONFIG, "output_dir", "results"), f"{model_key}_{lang}_dev_probabilities.csv"),
                os.path.join("results", f"{model_key}_{lang}_dev_probabilities.csv"),
            ]
            for p in cand_csvs:
                if os.path.exists(p):
                    df = pd.read_csv(p)
                    break

        target_col = pred_col if (df is not None and pred_col in df.columns) else "predicted_answer"
        if df is None or len(df) == 0 or target_col not in df.columns:
            print(f"[{lang.upper()}] Dev predictions NOT FOUND for {model_key}.")
            all_passed = False
            metrics_by_lang[lang] = {
                "n": 0,
                "top_letter": "N/A",
                "top_share": 0.0,
                "min_letter": "N/A",
                "min_letter_share": 0.0,
                "passed": False,
                "distribution": {},
            }
            continue

        preds = df[target_col].dropna().astype(str).str.strip().str.upper()
        n_total = len(preds)
        counts = preds.value_counts()
        top_letter = counts.index[0] if len(counts) > 0 else "N/A"
        top_share = float(counts.iloc[0] / n_total) if n_total > 0 else 0.0

        letter_shares = {l: float(counts.get(l, 0) / n_total) if n_total > 0 else 0.0 for l in ["A", "B", "C", "D"]}
        min_letter = min(letter_shares, key=letter_shares.get)
        min_share = letter_shares[min_letter]

        # Strict pre-registered gate thresholds (gate.yaml):
        # top_share <= 0.40, min_letter_share >= 0.10
        top_ok = (top_share <= 0.40)
        min_ok = (min_share >= 0.10)
        lang_passed = top_ok and min_ok

        if not lang_passed:
            all_passed = False

        status_str = "PASS" if lang_passed else "FAIL"
        print(f"[{lang.upper()}] n={n_total} | top_share={top_share:.3f} ({top_letter}) [<=0.40: {top_ok}] | min_share={min_share:.3f} ({min_letter}) [>=0.10: {min_ok}] -> {status_str}")
        print(f"       distribution: {letter_shares}")

        metrics_by_lang[lang] = {
            "n": n_total,
            "top_letter": top_letter,
            "top_share": top_share,
            "min_letter": min_letter,
            "min_letter_share": min_share,
            "passed": lang_passed,
            "distribution": letter_shares,
        }

    overall_pass = all_passed and (len(metrics_by_lang) == len(languages))
    overall_status = "PASS" if overall_pass else "FAIL"
    print("-" * 45)
    print(f"Overall Dev Screen Decision: {overall_status}")
    print("=" * 45 + "\n")
    return overall_pass, metrics_by_lang


def _write_dev_screen_row(row: Dict[str, Any], csv_path: str = REPORT_CSV_PATH):
    """Write or update a row in dev_screen_report_cyclic4.csv atomically with dev_source distinction."""
    cols = [
        "model_key",
        "dev_source",
        "stage",
        "config",
        "passed",
        "force_full_used",
        "full_run_executed",
        "bn_n",
        "bn_top_letter",
        "bn_top_share",
        "bn_min_letter",
        "bn_min_letter_share",
        "bn_passed",
        "en_n",
        "en_top_letter",
        "en_top_share",
        "en_min_letter",
        "en_min_letter_share",
        "en_passed",
    ]
    if "dev_source" not in row:
        row["dev_source"] = "cyclic4"

    if os.path.exists(csv_path):
        try:
            df = pd.read_csv(csv_path)
            if "dev_source" not in df.columns:
                df["dev_source"] = "cyclic4"
            # Deduplicate by both model_key and dev_source
            df = df[~((df["model_key"] == row["model_key"]) & (df["dev_source"] == row["dev_source"]))]
            df = pd.concat([df, pd.DataFrame([row])], ignore_index=True)
        except Exception:
            df = pd.DataFrame([row], columns=cols)
    else:
        df = pd.DataFrame([row], columns=cols)

    for c in cols:
        if c not in df.columns:
            df[c] = None
    df = df[cols]
    df.to_csv(csv_path, index=False, encoding="utf-8")
    print(f"📝 Recorded dev screen status to: {csv_path} (model={row['model_key']}, passed={row['passed']})")


def run_stage4_for_model(model_key: str) -> Dict[str, Any]:
    """Execute Stage 4 Cyclic4 pipeline for a single model with resume guard, dev screen, and cleanup."""
    print("\n" + "#" * 65)
    print(f"🎯 STARTING STAGE 4 CYCLIC4 PIPELINE FOR: {model_key}")
    print("#" * 65)

    base_parent = os.environ.get("KAGGLE_WORKING_DIR", "/kaggle/working" if os.path.exists("/kaggle/working") else os.path.dirname(os.path.abspath(getattr(CONFIG, "output_dir", "."))))
    out_dir = os.path.join(base_parent, f"results_E2_cyclic4_{model_key}")
    os.makedirs(out_dir, exist_ok=True)
    _ensure_split_files_present(out_dir)

    cfg_info = E2_CONFIGS["cyclic4"]
    cfg_e2 = dataclasses.replace(CONFIG, output_dir=out_dir)

    # Persist settings for this single model & config
    settings_map = {
        model_key: dict(
            cfg_info["settings"],
            provenance={
                "source": "E2_rescue_cyclic4",
                "description": cfg_info.get("description", ""),
            },
        )
    }
    save_model_settings(settings_map, output_dir=cfg_e2.output_dir)

    # ============================================================
    # RESUME GUARD: Check at the VERY START of run_stage4_for_model
    # ============================================================
    summary_json_path = os.path.join(out_dir, "summary.json")
    if not os.path.exists(summary_json_path):
        alt_sum = os.path.join(f"results_E2_cyclic4_{model_key}", "summary.json")
        if os.path.exists(alt_sum):
            summary_json_path = alt_sum
            out_dir = f"results_E2_cyclic4_{model_key}"

    if os.path.exists(summary_json_path):
        try:
            with open(summary_json_path, "r", encoding="utf-8") as _sf:
                prev_sum = json.load(_sf)
            if prev_sum.get("gate_reason") or prev_sum.get("languages") or prev_sum.get("gate_pass") is not None:
                logger.info(f"⏭️ [E2 RESUME GUARD] Skipping '{model_key}': summary.json already exists in {out_dir}.")
                print(f"⏭️ [E2 RESUME GUARD] Skipping '{model_key}': evaluation already completed in {out_dir}.")

                bn_cand = os.path.join(out_dir, f"{model_key}_bn_dev_probabilities.csv")
                en_cand = os.path.join(out_dir, f"{model_key}_en_dev_probabilities.csv")
                bn_n = len(pd.read_csv(bn_cand)) if os.path.exists(bn_cand) else 200
                en_n = len(pd.read_csv(en_cand)) if os.path.exists(en_cand) else 200

                row_skip = {
                    "model_key": model_key,
                    "dev_source": "cyclic4",
                    "stage": 4,
                    "config": "cyclic4",
                    "passed": "skipped_existing",
                    "force_full_used": False,
                    "full_run_executed": True,
                    "bn_n": bn_n,
                    "bn_top_letter": "existing",
                    "bn_top_share": None,
                    "bn_min_letter": "existing",
                    "bn_min_letter_share": None,
                    "bn_passed": True,
                    "en_n": en_n,
                    "en_top_letter": "existing",
                    "en_top_share": None,
                    "en_min_letter": "existing",
                    "en_min_letter_share": None,
                    "en_passed": True,
                }
                _write_dev_screen_row(row_skip)
                try:
                    shutil.copy2(REPORT_CSV_PATH, os.path.join(out_dir, "dev_screen_report_cyclic4.csv"))
                except Exception:
                    pass
                return {"model_key": model_key, "status": "skipped_existing", "summary": prev_sum}
        except Exception as _re_err:
            logger.warning(f"Failed to read existing summary.json in {out_dir} ({_re_err}); re-evaluating.")

    # ============================================================
    # STEP 1: Dev-split Screening Inference (200 items BN & EN)
    # ============================================================
    print(f"\n[Stage 4:cyclic4] Step 1: Running dev screening inference for {model_key}...")
    dev_dfs = run_dev_probabilities(
        model_keys=(model_key,),
        languages=("bn", "en"),
        config=cfg_e2,
        output_dir=out_dir,
        execute=True,
    )

    # 1. Evaluate & record raw dev screen distribution
    raw_pass, raw_metrics = screen_dev_distribution(
        dev_dfs, model_key, output_dir=out_dir, pred_col="pred_raw"
    )
    bn_raw_m = raw_metrics.get("bn", {})
    en_raw_m = raw_metrics.get("en", {})
    report_row_raw = {
        "model_key": model_key,
        "dev_source": "raw",
        "stage": 4,
        "config": "cyclic4",
        "passed": bool(raw_pass),
        "force_full_used": False,
        "full_run_executed": False,
        "bn_n": bn_raw_m.get("n", 0),
        "bn_top_letter": bn_raw_m.get("top_letter", "N/A"),
        "bn_top_share": bn_raw_m.get("top_share", 0.0),
        "bn_min_letter": bn_raw_m.get("min_letter", "N/A"),
        "bn_min_letter_share": bn_raw_m.get("min_letter_share", 0.0),
        "bn_passed": bn_raw_m.get("passed", False),
        "en_n": en_raw_m.get("n", 0),
        "en_top_letter": en_raw_m.get("top_letter", "N/A"),
        "en_top_share": en_raw_m.get("top_share", 0.0),
        "en_min_letter": en_raw_m.get("min_letter", "N/A"),
        "en_min_letter_share": en_raw_m.get("min_letter_share", 0.0),
        "en_passed": en_raw_m.get("passed", False),
    }
    _write_dev_screen_row(report_row_raw)

    # 2. Evaluate & record cyclic4-averaged dev screen distribution
    dev_pass, dev_metrics = screen_dev_distribution(
        dev_dfs, model_key, output_dir=out_dir, pred_col="predicted_answer"
    )
    bn_m = dev_metrics.get("bn", {})
    en_m = dev_metrics.get("en", {})
    report_row = {
        "model_key": model_key,
        "dev_source": "cyclic4",
        "stage": 4,
        "config": "cyclic4",
        "passed": bool(dev_pass),
        "force_full_used": False,
        "full_run_executed": False,
        "bn_n": bn_m.get("n", 0),
        "bn_top_letter": bn_m.get("top_letter", "N/A"),
        "bn_top_share": bn_m.get("top_share", 0.0),
        "bn_min_letter": bn_m.get("min_letter", "N/A"),
        "bn_min_letter_share": bn_m.get("min_letter_share", 0.0),
        "bn_passed": bn_m.get("passed", False),
        "en_n": en_m.get("n", 0),
        "en_top_letter": en_m.get("top_letter", "N/A"),
        "en_top_share": en_m.get("top_share", 0.0),
        "en_min_letter": en_m.get("min_letter", "N/A"),
        "en_min_letter_share": en_m.get("min_letter_share", 0.0),
        "en_passed": en_m.get("passed", False),
    }
    _write_dev_screen_row(report_row)

    # ============================================================
    # STEP 2: Decide whether to proceed with full 1,500 run
    # ============================================================
    single_config = {
        "cyclic4": {
            "output_dir_name": cfg_info["output_dir_name"],
            "description": cfg_info["description"],
            "models": [model_key],
            "settings": cfg_info["settings"],
        }
    }

    res = None
    if dev_pass:
        print(f"\n✅ [DEV SCREEN PASSED] Proceeding to full 1500-item evaluation for {model_key} in cyclic4...")
        report_row["force_full_used"] = False
        report_row["full_run_executed"] = True
        _write_dev_screen_row(report_row)

        res = run_e2_experiments(
            configs=single_config,
            base_config=CONFIG,
            languages=("bn", "en"),
            execute=True,
        )
        print(f"[Stage 4:cyclic4] Full 1500 evaluation completed for {model_key}.")
    elif FORCE_FULL:
        print(f"\n⚠️  [DEV SCREEN FAILED BUT FORCE_FULL=True] Overriding dev screen and running full 1500-item evaluation for {model_key} in cyclic4...")
        report_row["force_full_used"] = True
        report_row["full_run_executed"] = True
        _write_dev_screen_row(report_row)

        res = run_e2_experiments(
            configs=single_config,
            base_config=CONFIG,
            languages=("bn", "en"),
            execute=True,
        )
        print(f"[Stage 4:cyclic4] Full 1500 evaluation completed under FORCE_FULL override for {model_key}.")
    else:
        print(f"\n❌ [DEV SCREEN FAILED] top_share > 0.40 or min_letter_share < 0.10 on dev split.")
        print(f"   Skipping full 1500 run for {model_key} under config 'cyclic4' (FORCE_FULL is False).")
        report_row["force_full_used"] = False
        report_row["full_run_executed"] = False
        _write_dev_screen_row(report_row)

    try:
        shutil.copy2(REPORT_CSV_PATH, os.path.join(out_dir, "dev_screen_report_cyclic4.csv"))
    except Exception:
        pass

    # ============================================================
    # STEP 3: Free GPU memory after each model
    # ============================================================
    del dev_dfs
    if res is not None:
        del res
    gc.collect()
    if torch is not None and torch.cuda.is_available():
        torch.cuda.empty_cache()
    print(f"🧹 [GPU CLEANUP] VRAM cache cleared after model '{model_key}'.")
    # Step 3b: Clean HF cache after each model to strictly respect Kaggle disk limits
    _hf_dir = os.path.join(base_parent, "hf_cache")
    if os.path.exists(_hf_dir):
        try:
            _freed = sum(os.path.getsize(os.path.join(r, f)) for r, _, fs in os.walk(_hf_dir) for f in fs)
            shutil.rmtree(_hf_dir, ignore_errors=True)
            print(f"🗑️ [DISK CLEANUP] Freed {_freed / (1024**3):.2f} GB by cleaning HF cache after '{model_key}'.")
        except Exception as _ce:
            print(f"⚠️ Could not clean HF cache for '{model_key}': {_ce}")

    return {"model_key": model_key, "passed": dev_pass, "row": report_row}


# ------------------------------------------------------------
# MAIN EXECUTION LOOP OVER TARGET RESCUE MODELS
# ------------------------------------------------------------
if _RUN_E2:
    for model_key in TARGET_RESCUE_MODELS:
        run_stage4_for_model(model_key)
    print("\n" + "=" * 70)
    print(f"🎉 STAGE 4 CYCLIC4 LOOP COMPLETE FOR ALL MODELS: {TARGET_RESCUE_MODELS}")
else:
    print("E2 existing-model stage skipped (RUN_E2_EXISTING=False) - no existing models run.")
if os.path.exists(REPORT_CSV_PATH):
    print(f"\n📊 Summary of {REPORT_CSV_PATH}:")
    try:
        display(pd.read_csv(REPORT_CSV_PATH))
    except Exception:
        print(pd.read_csv(REPORT_CSV_PATH).to_string())
print("=" * 70)
```

## Section 14d-ter: Dtype Sensitivity Analysis & Gemma-Family Dtype Decisions (v4i)

### Technical Rationale & Architecture Decisions for Gemma-Family Models:
1. **Numerical Stability vs. Precision:**
   - Gemma models (including `tigerllm_1b`, which uses the Gemma-3 architecture) employ RMSNorm and attention logit soft-capping whose dynamic ranges can exceed IEEE 754 half-precision (`float16`) limits (`[-65,504, 65,504]`).
   - In `float16`, this soft-capping produces arithmetic overflow, causing `NaN` or `Inf` logits and unrecoverable pipeline aborts.
2. **Hardware Constraints:**
   - NVIDIA T4 GPUs support IEEE `float16` and `float32`, but lack native `bfloat16` hardware acceleration (which would provide an 8-bit exponent and larger dynamic range).
3. **Memory Profile & Compatibility Checks:**
   - Weight footprint of a 1B model is **~2.0 GB in float16** vs. **~4.0 GB in float32**.
   - On a 15 GB NVIDIA T4 GPU, 4.0 GB constitutes ~27% of available VRAM, comfortably fitting within memory limits without risking CUDA OOM.
   - For larger 3B/4B models, FP32 requires ~12.8 GB, approaching the T4 ceiling, necessitating strict batch size = 1 or FP16.
4. **Non-Assumption Empirical Verification:**
   - We **do not assume float32 is unconditionally required without verification**. Sensitivity testing on $n=200$ dev items runs for **all applicable models** (`titullm_1b`, `tigerllm_1b`, `llama32_3b`) comparing `float16` vs. `float32` to measure exact logit differences, NaN rates, and ECE sensitivity.


```python
_RUN_DTYPE_SENS = bool(globals().get("DTYPE_SENSITIVITY_RUN", True))
if _RUN_DTYPE_SENS:
    \
    # ============================================================
    # Section 14d-ter: Dtype Sensitivity Analysis (v4h)
    # ============================================================
    # PURPOSE: Measure how much float16 vs float32 affects option probabilities
    # on the 200 dev items (dev_bn.csv, dev_en.csv) for the main models.
    # No model weights are loaded more than once per dtype per model.
    # Results are saved ONLY under /kaggle/working/dtype_sensitivity/ and
    # NEVER touch any main results folder (results/ or results_E2_cyclic4_*/).
    #
    # ACTIVE models during "Save & Run All":
    #   DTYPE_SENSITIVITY_MODELS = ["titullm_1b", "tigerllm_1b", "llama32_3b"]   (float16 vs float32)
    #
    # COMMENTED-OUT (must NOT execute during Save & Run All):
    #   tigerllm_1b: float32 vs float16
    #   llama32_3b:  float16 vs float32
    #
    # HOW TO ENABLE FUTURE TESTS:
    #   Either add the model key to DTYPE_SENSITIVITY_MODELS, or uncomment
    #   the corresponding block below and run the cell manually.
    # ============================================================
    import os
    import gc
    import json
    import shutil
    import hashlib
    import pandas as pd
    import numpy as np

    try:
        import torch
        _TORCH_AVAIL = True
    except ImportError:
        torch = None
        _TORCH_AVAIL = False

    # ---- Output directory (never writes outside this path) ----
    _BASE_WORKING = os.environ.get("KAGGLE_WORKING_DIR",
        "/kaggle/working" if os.path.exists("/kaggle/working") else ".")
    DTYPE_SENS_DIR = os.path.join(_BASE_WORKING, "dtype_sensitivity")
    os.makedirs(DTYPE_SENS_DIR, exist_ok=True)

    # ---- Active models for this run ----
    # tigerllm_1b and llama32_3b are listed only in comments (see below).
    DTYPE_SENSITIVITY_MODELS = ["titullm_1b"]

    LETTERS = ["A", "B", "C", "D"]


    # ------------------------------------------------------------------
    # Helper: load dev items for a given model's cyclic4 output folder
    # ------------------------------------------------------------------
    def _load_dev_items_for_model(model_key: str, lang: str, base_dir: str):
        """Load 200 dev items from the model's cyclic4 results folder."""
        m_dir = os.path.join(base_dir, f"results_E2_cyclic4_{model_key}")
        if not os.path.isdir(m_dir):
            m_dir = f"results_E2_cyclic4_{model_key}"
        dev_file = os.path.join(m_dir, f"dev_{lang}.csv")
        if not os.path.exists(dev_file):
            # fallback to main results dir
            dev_file = os.path.join(base_dir, "results", f"dev_{lang}.csv")
        if not os.path.exists(dev_file):
            dev_file = os.path.join(base_dir, f"dev_{lang}.csv")
        if not os.path.exists(dev_file):
            raise FileNotFoundError(f"dev_{lang}.csv not found for {model_key}")
        benchmark = "BnMMLU" if lang == "bn" else "BnMMLU-en"
        items = load_dataset_from_file(dev_file, language=lang,
                                       benchmark=benchmark, is_mock=False)
        return items


    # ------------------------------------------------------------------
    # Helper: load few-shot exemplars for a model
    # ------------------------------------------------------------------
    def _load_exemplars_for_model(model_key: str, base_dir: str):
        m_dir = os.path.join(base_dir, f"results_E2_cyclic4_{model_key}")
        if not os.path.isdir(m_dir):
            m_dir = f"results_E2_cyclic4_{model_key}"
        ex_file = os.path.join(m_dir, "fewshot_exemplars.json")
        if not os.path.exists(ex_file):
            ex_file = os.path.join(base_dir, "results", "fewshot_exemplars.json")
        if not os.path.exists(ex_file):
            ex_file = "fewshot_exemplars.json"
        if os.path.exists(ex_file):
            with open(ex_file, "r", encoding="utf-8") as f:
                return json.load(f)
        return {}


    # ------------------------------------------------------------------
    # Helper: build backend with an EXPLICIT dtype (bypasses MODEL_REGISTRY dtype)
    # IMPORTANT: this function is for the sensitivity experiment ONLY.
    # It does NOT modify MODEL_REGISTRY or CONFIG. It passes dtype directly
    # to create_inference_backend so that model weights load at the requested
    # precision, not the production precision.
    # ------------------------------------------------------------------
    def _build_backend_with_dtype(
        model_key: str,
        dtype_str: str,
        config,
        force_dtype: bool = False,
    ) -> "BaseInferenceBackend":
        """Build a HuggingFace backend with an explicitly specified dtype.

        Parameters
        ----------
        model_key  : key into MODEL_REGISTRY
        dtype_str  : "float16" or "float32"
        config     : ExperimentConfig (for hf_token, seed, backend type)
        force_dtype: If True, skips the Gemma-family fp32 override in
                     HuggingFaceBackend. Required for tigerllm_1b fp16 test.
                     WARNING: Gemma in fp16 may produce NaN/non-finite logits.

        Notes
        -----
        - Writes nothing to disk.
        - Does NOT touch MODEL_REGISTRY.
        """
        m_info = MODEL_REGISTRY[model_key]
        m_id = m_info["hf_id"]
        use_chat = False   # uniform setting: chat template off for all three models

        if force_dtype:
            # Monkey-patch HuggingFaceBackend temporarily so it skips the Gemma
            # fp32 override. We restore the original __init__ after construction.
            import types
            orig_init = HuggingFaceBackend.__init__

            def _patched_init(self, model_id, device=None, dtype="float16",
                              hf_token=None, use_chat_template=False, **kwargs):
                # Identical to the original except we skip the Gemma guard.
                from transformers import AutoConfig, AutoModelForCausalLM, AutoTokenizer
                import torch as _torch
                BaseInferenceBackend.__init__(self, model_id)
                self.torch = _torch
                self.device = device or ("cuda" if _torch.cuda.is_available() else "cpu")
                self.use_chat_template = use_chat_template
                token_arg = {"token": hf_token} if hf_token else {}
                self.tokenizer = AutoTokenizer.from_pretrained(model_id, **token_arg)
                self.tokenizer.padding_side = "left"
                if self.tokenizer.pad_token is None:
                    self.tokenizer.pad_token = self.tokenizer.eos_token
                model_cfg = AutoConfig.from_pretrained(model_id, **token_arg)
                self.model_type = getattr(model_cfg, "model_type", "unknown")
                self.vocab_size = len(self.tokenizer)
                if dtype == "float16":
                    torch_dtype = _torch.float16
                elif dtype == "float32":
                    torch_dtype = _torch.float32
                else:
                    torch_dtype = _torch.float16 if self.device == "cuda" else _torch.float32
                if self.device == "cpu" and torch_dtype == _torch.float16:
                    torch_dtype = _torch.float32
                # NOTE: Gemma fp32 override is intentionally skipped here.
                logger.warning(
                    f"[dtype_sensitivity] force_dtype=True: loading {model_id} in "
                    f"{dtype} WITHOUT the Gemma fp32 guard. Expect possible NaN logits."
                )
                self.model = AutoModelForCausalLM.from_pretrained(
                    model_id, torch_dtype=torch_dtype, **token_arg)
                self.model.to(self.device)
                self.model.eval()
                self.variant_ids = self._resolve_option_token_ids()

            HuggingFaceBackend.__init__ = _patched_init
            try:
                backend = create_inference_backend(
                    backend_type=config.backend,
                    model_id=m_id,
                    dtype=dtype_str,
                    hf_token=config.hf_token,
                    use_chat_template=use_chat,
                    seed=config.seed,
                )
            finally:
                HuggingFaceBackend.__init__ = orig_init   # always restore
        else:
            backend = create_inference_backend(
                backend_type=config.backend,
                model_id=m_id,
                dtype=dtype_str,
                hf_token=config.hf_token,
                use_chat_template=use_chat,
                seed=config.seed,
            )
        return backend


    # ------------------------------------------------------------------
    # Helper: run inference on a list of items and return a DataFrame
    # Text tokenized: question + options (the per-item text the evaluator
    # puts in the prompt), excluding the shared few-shot prefix.
    # Settings: few_shot, k=4, use_chat_template=False, debias=none, seed=42.
    # ------------------------------------------------------------------
    def _run_inference_items(backend, items, lang, exemplars_json) -> pd.DataFrame:
        """Run predict_item (debias=none, single pass) and collect per-item rows."""
        settings = {
            "prompt_variant": "few_shot",
            "n_shots": 4,
            "effective_n_shots": 4,
            "use_chat_template": False,
            "debias": "none",   # single pass for sensitivity (cyclic4 is irrelevant here)
        }
        rows = []
        for item in items:
            try:
                res = predict_item(backend, item, lang, settings)
                probs = res.probabilities if res.probabilities else [None, None, None, None]
                rows.append({
                    "question_id": item.get("question_id", ""),
                    "gold_answer": item.get("correct_answer", item.get("gold_answer", "")),
                    "predicted_answer": res.predicted_letter,
                    "confidence": res.confidence,
                    "prob_A": probs[0] if len(probs) > 0 else None,
                    "prob_B": probs[1] if len(probs) > 1 else None,
                    "prob_C": probs[2] if len(probs) > 2 else None,
                    "prob_D": probs[3] if len(probs) > 3 else None,
                    "option_mass": getattr(res, "option_mass", 1.0),
                })
            except Exception as _e:
                rows.append({
                    "question_id": item.get("question_id", ""),
                    "gold_answer": item.get("correct_answer", item.get("gold_answer", "")),
                    "predicted_answer": None,
                    "confidence": None,
                    "prob_A": None, "prob_B": None, "prob_C": None, "prob_D": None,
                    "option_mass": None,
                    "_error": str(_e),
                })
        return pd.DataFrame(rows)


    # ------------------------------------------------------------------
    # Helper: free backend and GPU memory
    # ------------------------------------------------------------------
    def _free_backend(backend, label: str = ""):
        """Delete backend, run gc, synchronize and empty CUDA cache."""
        try:
            if hasattr(backend, "model"):
                del backend.model
            if hasattr(backend, "tokenizer"):
                del backend.tokenizer
        except Exception:
            pass
        del backend
        gc.collect()
        if _TORCH_AVAIL and torch is not None and torch.cuda.is_available():
            torch.cuda.synchronize()
            torch.cuda.empty_cache()
        if label:
            print(f"  🧹 GPU memory freed after {label}")


    # ------------------------------------------------------------------
    # Helper: compute sensitivity metrics from two DataFrames
    # ------------------------------------------------------------------
    def _compute_sensitivity_metrics(
        df_a: pd.DataFrame,
        df_b: pd.DataFrame,
        model_key: str,
        lang: str,
        dtype_a: str,
        dtype_b: str,
        n_items: int,
    ) -> dict:
        """Compute all metrics in §2c between dtype_a and dtype_b results."""
        shared = set(df_a["question_id"].astype(str)) & set(df_b["question_id"].astype(str))
        da = df_a.set_index("question_id")
        db = df_b.set_index("question_id")
        shared_ids = sorted(shared)

        prob_cols = ["prob_A", "prob_B", "prob_C", "prob_D"]

        mean_abs_diffs, max_abs_diffs, conf_diffs = [], [], []
        agree = 0
        n_valid = 0

        for qid in shared_ids:
            if qid not in da.index or qid not in db.index:
                continue
            ra, rb = da.loc[qid], db.loc[qid]
            probs_a = [ra.get(c) for c in prob_cols]
            probs_b = [rb.get(c) for c in prob_cols]
            if any(p is None or (isinstance(p, float) and np.isnan(p)) for p in probs_a + probs_b):
                continue
            diffs = [abs(float(pa) - float(pb)) for pa, pb in zip(probs_a, probs_b)]
            mean_abs_diffs.append(np.mean(diffs))
            max_abs_diffs.append(max(diffs))
            conf_a = ra.get("confidence")
            conf_b = rb.get("confidence")
            if conf_a is not None and conf_b is not None:
                conf_diffs.append(float(conf_a) - float(conf_b))
            pred_a = ra.get("predicted_answer")
            pred_b = rb.get("predicted_answer")
            if pred_a is not None and pred_b is not None:
                agree += int(str(pred_a).strip().upper() == str(pred_b).strip().upper())
            n_valid += 1

        agreement_rate = agree / n_valid if n_valid > 0 else float("nan")
        mean_mad = float(np.mean(mean_abs_diffs)) if mean_abs_diffs else float("nan")
        mean_max_ad = float(np.mean(max_abs_diffs)) if max_abs_diffs else float("nan")
        mean_conf_diff = float(np.mean(conf_diffs)) if conf_diffs else float("nan")
        abs_conf_diff = float(np.mean(np.abs(conf_diffs))) if conf_diffs else float("nan")

        # ECE for each dtype
        def _ece_from_df(df):
            golds = set("ABCD")
            confs = df["confidence"].dropna().tolist()
            corrects = [
                int(str(r.get("predicted_answer", "")).strip().upper() ==
                    str(r.get("gold_answer", "")).strip().upper())
                for _, r in df.iterrows()
                if r.get("confidence") is not None
            ]
            if len(confs) != len(corrects) or not confs:
                return float("nan")
            return expected_calibration_error(confs, corrects, n_bins=10, strategy="uniform")

        ece_a = _ece_from_df(df_a) if df_a is not None and len(df_a) > 0 else float("nan")
        ece_b = _ece_from_df(df_b) if df_b is not None and len(df_b) > 0 else float("nan")

        n_nan_a = int(df_a["predicted_answer"].isna().sum()) if df_a is not None else n_items
        n_nan_b = int(df_b["predicted_answer"].isna().sum()) if df_b is not None else n_items
        mean_om_a = float(df_a["option_mass"].dropna().mean()) if (df_a is not None and len(df_a) > 0) else float("nan")
        mean_om_b = float(df_b["option_mass"].dropna().mean()) if (df_b is not None and len(df_b) > 0) else float("nan")

        return {
            "model_key": model_key,
            "lang": lang,
            "dtype_a": dtype_a,
            "dtype_b": dtype_b,
            "n_items": n_items,
            "n_valid_pairs": n_valid,
            "mean_abs_prob_diff": mean_mad,
            "max_abs_prob_diff": mean_max_ad,
            "prediction_agreement_rate": agreement_rate,
            "mean_confidence_diff_a_minus_b": mean_conf_diff,
            "abs_confidence_diff": abs_conf_diff,
            f"ece_{dtype_a}": ece_a,
            f"ece_{dtype_b}": ece_b,
            f"n_nan_{dtype_a}": n_nan_a,
            f"n_nan_{dtype_b}": n_nan_b,
            f"mean_option_mass_{dtype_a}": mean_om_a,
            f"mean_option_mass_{dtype_b}": mean_om_b,
        }


    # ------------------------------------------------------------------
    # Main reusable function
    # ------------------------------------------------------------------
    def run_dtype_sensitivity(
        model_key: str,
        dtype_a: str,
        dtype_b: str,
        config,
        base_dir: str,
        output_dir: str,
        force_dtype_b: bool = False,
    ) -> dict:
        """Run dtype sensitivity experiment for one model.

        Loads dtype_a backend, runs all dev items (bn + en), frees GPU memory,
        loads dtype_b backend, runs same items, frees again, then computes metrics.

        Parameters
        ----------
        model_key   : key in MODEL_REGISTRY
        dtype_a     : first dtype string ("float16" or "float32")
        dtype_b     : second dtype string
        config      : ExperimentConfig
        base_dir    : /kaggle/working (parent of results_E2_cyclic4_* folders)
        output_dir  : must be under /kaggle/working/dtype_sensitivity/
        force_dtype_b : bypass Gemma fp32 guard for dtype_b (tigerllm_1b only)

        Returns
        -------
        dict  with per-language metrics and paths to saved CSVs
        """
        assert output_dir.startswith(base_dir) or output_dir.startswith("/kaggle/working"), \
            f"output_dir must be under /kaggle/working (got {output_dir})"
        os.makedirs(output_dir, exist_ok=True)

        all_per_item_rows = []
        summary_rows = []

        for lang in ("bn", "en"):
            print(f"\n  [{model_key}] lang={lang}: loading {dtype_a} backend...")
            df_a, df_b = None, None

            # --- Run dtype_a ---
            try:
                items = _load_dev_items_for_model(model_key, lang, base_dir)
                n_items = len(items)
                backend_a = _build_backend_with_dtype(model_key, dtype_a, config)
                df_a = _run_inference_items(backend_a, items, lang, {})
                df_a["dtype"] = dtype_a
                df_a["model_key"] = model_key
                df_a["lang"] = lang
                _free_backend(backend_a, label=f"{model_key}/{lang}/{dtype_a}")
            except Exception as e_a:
                print(f"  ⚠️  [{model_key}] {lang}/{dtype_a} failed: {e_a}")
                df_a = None

            # --- Run dtype_b ---
            print(f"  [{model_key}] lang={lang}: loading {dtype_b} backend...")
            try:
                if items is None:
                    items = _load_dev_items_for_model(model_key, lang, base_dir)
                    n_items = len(items)
                backend_b = _build_backend_with_dtype(model_key, dtype_b, config,
                                                       force_dtype=force_dtype_b)
                df_b = _run_inference_items(backend_b, items, lang, {})
                df_b["dtype"] = dtype_b
                df_b["model_key"] = model_key
                df_b["lang"] = lang
                _free_backend(backend_b, label=f"{model_key}/{lang}/{dtype_b}")
            except Exception as e_b:
                print(f"  ⚠️  [{model_key}] {lang}/{dtype_b} failed: {e_b}")
                df_b = None

            # --- Collect per-item rows ---
            for dt_label, df_dt in [(dtype_a, df_a), (dtype_b, df_b)]:
                if df_dt is not None:
                    all_per_item_rows.append(df_dt)

            # --- Compute metrics ---
            if df_a is not None and df_b is not None:
                metrics = _compute_sensitivity_metrics(
                    df_a, df_b, model_key, lang, dtype_a, dtype_b, n_items
                )
            else:
                metrics = {
                    "model_key": model_key, "lang": lang,
                    "dtype_a": dtype_a, "dtype_b": dtype_b,
                    "error": f"dtype_a_ok={df_a is not None}, dtype_b_ok={df_b is not None}",
                }
            summary_rows.append(metrics)
            print(f"  ✅ [{model_key}] {lang}: agreement={metrics.get('prediction_agreement_rate', 'N/A'):.4f}, "
                  f"mean_abs_diff={metrics.get('mean_abs_prob_diff', 'N/A'):.6f}")

        # Save per-item CSV (append mode: accumulate across models)
        per_item_csv = os.path.join(output_dir, "dtype_sensitivity_per_item.csv")
        if all_per_item_rows:
            df_all = pd.concat(all_per_item_rows, ignore_index=True)
            col_order = ["model_key", "lang", "question_id", "dtype",
                         "prob_A", "prob_B", "prob_C", "prob_D",
                         "predicted_answer", "confidence", "option_mass"]
            for c in col_order:
                if c not in df_all.columns:
                    df_all[c] = None
            df_all = df_all[col_order]
            # Append or create
            if os.path.exists(per_item_csv):
                existing = pd.read_csv(per_item_csv)
                df_all = pd.concat([existing, df_all], ignore_index=True)
            df_all.to_csv(per_item_csv, index=False)
            print(f"  📄 Saved per-item CSV: {per_item_csv} ({len(df_all)} rows)")

        return {"model_key": model_key, "summary_rows": summary_rows}


    # ==================================================================
    # EXECUTE: titullm_1b  (float16 vs float32)  — ACTIVE
    # ==================================================================
    print("=" * 65)
    print("🔬 DTYPE SENSITIVITY ANALYSIS (v4h)")
    print(f"   Active models: {DTYPE_SENSITIVITY_MODELS}")
    print(f"   Output dir:    {DTYPE_SENS_DIR}")
    print("=" * 65)

    _all_sensitivity_rows = []

    for _m_key in DTYPE_SENSITIVITY_MODELS:
        print(f"\n{'#'*55}")
        print(f"# Running dtype sensitivity for: {_m_key}")
        print(f"{'#'*55}")
    
        # Resumability check: skip if already computed for this model
        _m_sens_json = os.path.join(DTYPE_SENS_DIR, f"{_m_key}_sensitivity_summary.json")
        if os.path.exists(_m_sens_json):
            try:
                with open(_m_sens_json, "r", encoding="utf-8") as _f:
                    _prev_res = json.load(_f)
                print(f"⏭️ [DTYPE RESUME GUARD] Skipping '{_m_key}': sensitivity already completed in {_m_sens_json}.")
                _all_sensitivity_rows.extend(_prev_res.get("summary_rows", []))
                continue
            except Exception:
                pass

        _dt_a = "float32" if _m_key == "tigerllm_1b" else "float16"
        _dt_b = "float16" if _m_key == "tigerllm_1b" else "float32"
        _force_b = True if _m_key == "tigerllm_1b" else False

        _result = run_dtype_sensitivity(
            model_key=_m_key,
            dtype_a=_dt_a,
            dtype_b=_dt_b,
            config=CONFIG,
            base_dir=_BASE_WORKING,
            output_dir=DTYPE_SENS_DIR,
            force_dtype_b=_force_b,
        )
        with open(_m_sens_json, "w", encoding="utf-8") as _f:
            json.dump(_result, _f, indent=2, default=str)
        _all_sensitivity_rows.extend(_result.get("summary_rows", []))
    
        # Clean up GPU memory after each sensitivity model
        gc.collect()
        if torch is not None and torch.cuda.is_available():
            torch.cuda.empty_cache()

    # ------------------------------------------------------------------
    # COMMENTED OUT: tigerllm_1b  float32 vs float16
    #
    # IMPORTANT — READ BEFORE ENABLING:
    # The normal HuggingFaceBackend forces float32 for any model whose
    # model_type starts with "gemma" (see the "Gemma-family architecture
    # detected" log line in cell 19). A plain dtype="float16" request for
    # tigerllm_1b would silently run in float32 and show zero difference.
    # To run this test correctly, force_dtype_b=True must be passed so the
    # Gemma fp32 guard is bypassed. With that flag, expect non-finite logits
    # (NaN or Inf) in many items — that is the REASON the main run uses fp32.
    # Record the NaN count; do not treat it as a pipeline bug.
    # Run this test alone in a fresh session (tigerllm_1b at ~4 GB fp32
    # and ~2 GB fp16 fit comfortably, but llama32_3b must NOT be loaded
    # at the same time).
    #
    # # _result_tiger = run_dtype_sensitivity(
    # #     model_key="tigerllm_1b",
    # #     dtype_a="float32",
    # #     dtype_b="float16",
    # #     config=CONFIG,
    # #     base_dir=_BASE_WORKING,
    # #     output_dir=DTYPE_SENS_DIR,
    # #     force_dtype_b=True,   # bypasses Gemma fp32 guard for dtype_b
    # # )
    # # _all_sensitivity_rows.extend(_result_tiger["summary_rows"])
    # ------------------------------------------------------------------

    # ------------------------------------------------------------------
    # COMMENTED OUT: llama32_3b  float16 vs float32
    #
    # IMPORTANT — READ BEFORE ENABLING:
    # llama32_3b in float32 requires approximately 12.8 GB of weight storage
    # on a T4 with 14.56 GB usable VRAM. Combined with KV cache and activations
    # during inference it may run out of memory and crash the session.
    # Run this test ALONE in a completely fresh session with no other models
    # loaded and nothing else in memory. The float16 -> float32 direction
    # should be used (load fp16 first, free it, then load fp32).
    # If you see CUDA OOM, reduce to fewer items or accept it as a failure.
    #
    # # _result_llama = run_dtype_sensitivity(
    # #     model_key="llama32_3b",
    # #     dtype_a="float16",
    # #     dtype_b="float32",
    # #     config=CONFIG,
    # #     base_dir=_BASE_WORKING,
    # #     output_dir=DTYPE_SENS_DIR,
    # # )
    # # _all_sensitivity_rows.extend(_result_llama["summary_rows"])
    # ------------------------------------------------------------------

    # Save summary CSV and JSON
    _sens_csv = os.path.join(DTYPE_SENS_DIR, "dtype_sensitivity.csv")
    _sens_json = os.path.join(DTYPE_SENS_DIR, "dtype_sensitivity_summary.json")

    if _all_sensitivity_rows:
        pd.DataFrame(_all_sensitivity_rows).to_csv(_sens_csv, index=False)
        with open(_sens_json, "w", encoding="utf-8") as _f:
            json.dump(_all_sensitivity_rows, _f, indent=2, default=str)
        print(f"\n📄 Saved dtype_sensitivity.csv ({len(_all_sensitivity_rows)} rows)")
        print(f"📄 Saved dtype_sensitivity_summary.json")
        try:
            display(pd.DataFrame(_all_sensitivity_rows))
        except Exception:
            print(pd.DataFrame(_all_sensitivity_rows).to_string())
    else:
        print("⚠️  No sensitivity rows collected.")

    print("\n✅ DTYPE SENSITIVITY ANALYSIS COMPLETE")
else:
    print("dtype sensitivity skipped")
    # Skipped: no model loading, no directory creation, no writes.
```

## Section 14d-quater: Tokenizer Fragmentation on Full Subsets (v4h)
Tokenizer-only cell (no model weights). Computes TPW, TPC, STRR for all three tokenizers on the full 1500-item `bn_subset.csv` and `en_subset.csv`.
Warns if `llama32_3b` and `titullm_1b` share an identical tokenizer.

```python
import json as _json
import os as _os
import traceback as _traceback

_CELL_NAME = 'tokenizer_fragmentation'
_STATUS_PATH = "/kaggle/working/new_cells_status.json"

def _record_cell_status(name, ok, error=None):
    try:
        if _os.path.exists(_STATUS_PATH):
            with open(_STATUS_PATH, "r", encoding="utf-8") as _f:
                status = _json.load(_f)
        else:
            status = {}
    except Exception:
        status = {}
    status[name] = {
        "ok": ok,
        "error": str(error) if error is not None else None,
        "traceback": _traceback.format_exc() if error is not None else None,
    }
    try:
        with open(_STATUS_PATH, "w", encoding="utf-8") as _f:
            _json.dump(status, _f, indent=2, default=str)
    except Exception:
        pass

try:
    \
    # ============================================================
    # Section 14d-quater: Tokenizer Fragmentation on Full 1500-Item Subsets (v4h)
    # ============================================================
    # PURPOSE: Compute per-item TPW, TPC, STRR for all seven models' tokenizers
    # on the full bn_subset.csv and en_subset.csv (1500 items each).
    # This cell loads tokenizer weights only (no model weights).
    #
    # TEXT TOKENIZED (stated explicitly as required):
    #   For each item, we tokenize the per-item question+options text — i.e.,
    #   the text that the evaluator inserts into the prompt AFTER the shared
    #   few-shot prefix. Specifically:
    #       "<question_prefix> <question>\\n<options_prefix>\\nA. <opt_a>\\nB. <opt_b>\\nC. <opt_c>\\nD. <opt_d>"
    #   We exclude the few-shot exemplar prefix. This matches the marginal
    #   fragmentation that the model experiences per new item at inference time.
    #
    # METRIC DEFINITIONS (reuses _tokenize_text from cell 13 and the per-item
    # metric helpers that are already defined in the notebook for run_tokenizer_audit):
    #   n_words  : number of whitespace-delimited words in the text
    #   n_tokens : number of tokens produced by the tokenizer
    #   TPW      : tokens per word = n_tokens / max(n_words, 1)
    #   TPC      : tokens per character = n_tokens / max(len(text), 1)
    #   STRR     : subword-token recognition rate = 1 - (n_words / max(n_tokens, 1))
    #              (fraction of tokens that are sub-word fragments)
    #
    # llama32_3b uses the same Llama-3.2-3B-Instruct tokenizer as titullm_1b
    # (titulm-llama-3.2-1b-v1.1 is a continued-pretrain of Llama-3.2). If the
    # vocabulary sizes and token outputs match, a warning is printed.
    # ============================================================
    import os
    import json
    import hashlib
    import pandas as pd
    import numpy as np

    try:
        from transformers import AutoTokenizer
        _HF_AVAILABLE = True
    except ImportError:
        _HF_AVAILABLE = False
        print("⚠️  transformers not available; tokenizer fragmentation cell skipped.")

    _BASE_WORKING = os.environ.get("KAGGLE_WORKING_DIR",
        "/kaggle/working" if os.path.exists("/kaggle/working") else ".")

    TOKENIZER_FRAG_DIR = os.path.join(_BASE_WORKING, "tokenizer_fragmentation")
    os.makedirs(TOKENIZER_FRAG_DIR, exist_ok=True)

    # -- Probe text for identical-tokenizer check --
    _PROBE_TEXT = "বাংলাদেশের রাজধানী ঢাকা এবং আমাদের মাতৃভাষা বাংলা। The capital of Bangladesh is Dhaka."

    # -- Build per-item question+options text (the text tokenized) --
    def _item_to_text(item: dict, lang: str) -> str:
        """Build the per-item text that the evaluator inserts after the few-shot prefix."""
        if lang == "bn":
            q_prefix = "প্রশ্ন:"
            opt_prefix = "বিকল্পগুলি:"
        else:
            q_prefix = "Question:"
            opt_prefix = "Options:"
        q = str(item.get("question", "")).strip()
        opts = item.get("options", [])
        if not opts and all(f"option_{l.lower()}" in item for l in "ABCD"):
            opts = [item.get(f"option_{l.lower()}", "") for l in "ABCD"]
        opt_lines = "\n".join(
            f"{letter}. {str(opt).strip()}"
            for letter, opt in zip("ABCD", opts[:4])
        )
        return f"{q_prefix} {q}\n{opt_prefix}\n{opt_lines}"

    # -- Per-item fragmentation metrics --
    def _item_metrics(tokenizer, text: str) -> dict:
        words = text.split()
        n_words = len(words)
        try:
            toks = _tokenize_text(tokenizer, text)
            n_tokens = len(toks)
        except Exception:
            n_tokens = 0
        tpw = n_tokens / max(n_words, 1)
        tpc = n_tokens / max(len(text), 1)
        strr = 1.0 - (n_words / max(n_tokens, 1)) if n_tokens > 0 else 0.0
        return {
            "n_words": n_words,
            "n_tokens": n_tokens,
            "tpw": tpw,
            "tpc": tpc,
            "strr": strr,
        }

    if _HF_AVAILABLE:
        # -- Load subset files --
        _subset_paths = {}
        for _lang in ("bn", "en"):
            for _cand in [
                os.path.join(_BASE_WORKING, "results", f"{_lang}_subset.csv"),
                os.path.join(_BASE_WORKING, f"{_lang}_subset.csv"),
                f"{_lang}_subset.csv",
            ]:
                if os.path.exists(_cand):
                    _subset_paths[_lang] = _cand
                    break
        if not _subset_paths:
            # Try per-model folders (all three should have identical subsets)
            for _m in ["titullm_1b", "tigerllm_1b", "llama32_3b"]:
                _m_dir = os.path.join(_BASE_WORKING, f"results_E2_cyclic4_{_m}")
                for _lang in ("bn", "en"):
                    if _lang not in _subset_paths:
                        _p = os.path.join(_m_dir, f"{_lang}_subset.csv")
                        if os.path.exists(_p):
                            _subset_paths[_lang] = _p

        _subset_items = {}
        for _lang, _path in _subset_paths.items():
            benchmark = "BnMMLU" if _lang == "bn" else "BnMMLU-en"
            try:
                _subset_items[_lang] = load_dataset_from_file(
                    _path, language=_lang, benchmark=benchmark, is_mock=False
                )
                print(f"  ✅ Loaded {len(_subset_items[_lang])} {_lang.upper()} items from {_path}")
            except Exception as _e:
                print(f"  ⚠️  Could not load {_lang} subset: {_e}")

        # -- Run per-model tokenizer --
        _ALL_MODELS = ["titullm_1b", "tigerllm_1b", "llama32_3b", "qwen25_3b", "bloomz_3b", "phi35_mini", "llama32_3b_base"] + list(globals().get("E4_NEW_MODELS", []))
        _tok_fingerprints = {}   # vocab_size -> (model_key, probe_tokens)
        _per_item_rows = []
        _summary_rows = []
        _load_status = {}
        # -- E4 robustness: reuse completed-model rows from existing all7 cache if present --
        _ALL7_PATH = os.path.join(TOKENIZER_FRAG_DIR, "tokenizer_fragmentation_per_item_all7.csv")
        _cache_rows = []
        _cached_models = set()
        if os.path.exists(_ALL7_PATH):
            try:
                _cache_df = pd.read_csv(_ALL7_PATH)
                for _m in ["titullm_1b", "tigerllm_1b", "llama32_3b", "qwen25_3b", "bloomz_3b", "phi35_mini", "llama32_3b_base"]:
                    _sub = _cache_df[_cache_df["model_key"] == _m]
                    if len(_sub):
                        _cache_rows.append(_sub.assign(data_source="all7_cache"))
                        _cached_models.add(_m)
                print(f"  📦 Loaded completed-model rows from all7 cache: {sorted(_cached_models)}")
            except Exception as _ce:
                print(f"  ⚠️  Could not read all7 cache {_ALL7_PATH}: {_ce}")
        else:
            print(f"  📦 No all7 cache at {_ALL7_PATH}; computing all models fresh.")

        for _mkey in _ALL_MODELS:
            if _mkey in _cached_models:
                _load_status[_mkey] = "cache"
                continue  # already in all7 cache; no tokenizer load needed (no HF_TOKEN required)
            _load_status[_mkey] = "computed"
            _info = MODEL_REGISTRY[_mkey]
            _hf_id = _info["hf_id"]
            _tok_arg = {}
            if _info.get("needs_hf_token") and CONFIG.hf_token:
                _tok_arg["token"] = CONFIG.hf_token

            print(f"\n--- Loading tokenizer: {_mkey} ({_hf_id}) ---")
            try:
                _tok = AutoTokenizer.from_pretrained(_hf_id, **_tok_arg)
            except Exception as _te:
                print(f"  ⚠️  Could not load tokenizer for {_mkey}: {_te}")
                continue

            _vocab_size = len(_tok)
            _probe_toks = tuple(_tok.encode(_PROBE_TEXT, add_special_tokens=False))
            print(f"  Vocab size: {_vocab_size:,}  |  probe tokens: {len(_probe_toks)}")

            # Identical tokenizer warning
            for _prev_vsize, (_prev_key, _prev_toks) in _tok_fingerprints.items():
                if _vocab_size == _prev_vsize and _probe_toks == _prev_toks:
                    print(
                        f"  ⚠️  IDENTICAL TOKENIZER WARNING: {_mkey} and {_prev_key} share "
                        f"the same vocabulary ({_vocab_size:,} tokens) and produce identical "
                        f"token outputs on the probe text. Their fragmentation metrics will be "
                        f"identical. This is expected for Llama-3.2-3B-Instruct and "
                        f"titulm-llama-3.2-1b-v1.1 (continued pretraining, same BPE vocab)."
                    )
            _tok_fingerprints[_vocab_size] = (_mkey, _probe_toks)

            for _lang, _items in _subset_items.items():
                _lang_rows = []
                for _item in _items:
                    _text = _item_to_text(_item, _lang)
                    _m = _item_metrics(_tok, _text)
                    _row = {
                        "model_key": _mkey,
                        "lang": _lang,
                        "data_source": _load_status.get(_mkey, "computed"),
                        "question_id": _item.get("question_id", ""),
                        "n_words": _m["n_words"],
                        "n_tokens": _m["n_tokens"],
                        "tpw": round(_m["tpw"], 6),
                        "tpc": round(_m["tpc"], 6),
                        "strr": round(_m["strr"], 6),
                    }
                    _lang_rows.append(_row)
                    _per_item_rows.append(_row)

                # Per-model/per-language summary
                _df_lang = pd.DataFrame(_lang_rows)
                _summary_rows.append({
                    "model_key": _mkey,
                    "hf_id": _hf_id,
                    "lang": _lang,
                    "data_source": _load_status.get(_mkey, "computed"),
                    "vocab_size": _vocab_size,
                    "n_items": len(_df_lang),
                    "mean_tpw": round(float(_df_lang["tpw"].mean()), 6),
                    "std_tpw": round(float(_df_lang["tpw"].std()), 6),
                    "median_tpw": round(float(_df_lang["tpw"].median()), 6),
                    "mean_tpc": round(float(_df_lang["tpc"].mean()), 6),
                    "mean_strr": round(float(_df_lang["strr"].mean()), 6),
                    "std_strr": round(float(_df_lang["strr"].std()), 6),
                })

            del _tok
            import gc as _gc; _gc.collect()

        # -- Save outputs (never touch all7) --
        _all_n = len(_ALL_MODELS)
        _per_item_path = os.path.join(TOKENIZER_FRAG_DIR, f"tokenizer_fragmentation_per_item_all{_all_n}.csv")
        _summary_path  = os.path.join(TOKENIZER_FRAG_DIR, f"tokenizer_fragmentation_summary_all{_all_n}.csv")

        # Combine cache rows (completed models) with freshly computed rows (E4 / non-cached)
        if _cache_rows:
            _cache_combined = pd.concat(_cache_rows, ignore_index=True)
            # align columns with computed schema
            _per_item_cols = ["model_key", "lang", "data_source", "question_id", "n_words", "n_tokens", "tpw", "tpc", "strr"]
            _cache_combined = _cache_combined[[c for c in _per_item_cols if c in _cache_combined.columns]]
            _all_item_rows = list(_cache_combined.to_dict(orient="records")) + _per_item_rows
        else:
            _all_item_rows = _per_item_rows

        _cache_count = sum(1 for r in _all_item_rows if r.get("data_source") == "cache")
        _computed_count = sum(1 for r in _all_item_rows if r.get("data_source") == "computed")
        print(f"  📊 Rows source: cache={_cache_count} computed={_computed_count} total={len(_all_item_rows)}")

        if _all_item_rows:
            pd.DataFrame(_all_item_rows).to_csv(_per_item_path, index=False)
            print(f"\n📄 Saved per-item tokenizer fragmentation: {_per_item_path} ({len(_all_item_rows)} rows)")

        # Summary rows: mark source
        for _sr in _summary_rows:
            _sr.setdefault("data_source", "computed" if _sr["model_key"] not in _cached_models else "cache")
        if _summary_rows:
            _df_sum = pd.DataFrame(_summary_rows)
            _df_sum.to_csv(_summary_path, index=False)
            print(f"📄 Saved per-model/lang summary: {_summary_path} ({len(_summary_rows)} model/lang blocks)")
            try:
                display(_df_sum)
            except Exception:
                print(_df_sum.to_string(index=False))
            try:
                display(_df_sum)
            except Exception:
                print(_df_sum.to_string(index=False))

        print("\n✅ TOKENIZER FRAGMENTATION ANALYSIS COMPLETE")
    else:
        print("⚠️  Skipped (transformers not available).")
except Exception as _e:
    _record_cell_status(_CELL_NAME, False, _e)
    print(f"❌ [tokenizer_fragmentation] failed: {_e}")
    import traceback as _tb2; _tb2.print_exc()
else:
    _record_cell_status(_CELL_NAME, True, None)
```

```python
# ============================================================
# Section 14d-quater-bis: Tokenizer-only candidate screen (E4, no weights)
# ============================================================
# Loads ONLY tokenizers (AutoTokenizer, no model weights) for the four E4
# approved models and reports per-model fragmentation on the real bn_subset/en_subset.
# Saves tokenizer_fragmentation/candidate_screen.csv without overwriting an
# existing file. Reuses _item_to_text / _item_metrics from cell 44 and the
# single-option-token check from cell 48 (run_prerun_checks, Check 3).
import os as _os
import json as _json
import traceback as _traceback
import pandas as pd

try:
    from transformers import AutoTokenizer, AutoConfig
    _HF_AVAILABLE = True
except ImportError:
    _HF_AVAILABLE = False

_CELL_NAME = 'tokenizer_candidate_screen'
_STATUS_PATH = "/kaggle/working/new_cells_status.json"

def _record_cell_status(name, ok, error=None):
    try:
        if _os.path.exists(_STATUS_PATH):
            with open(_STATUS_PATH, "r", encoding="utf-8") as _f:
                status = _json.load(_f)
        else:
            status = {}
    except Exception:
        status = {}
    status[name] = {
        "ok": ok,
        "error": str(error) if error is not None else None,
        "traceback": _traceback.format_exc() if error is not None else None,
    }
    try:
        with open(_STATUS_PATH, "w", encoding="utf-8") as _f:
            _json.dump(status, _f, indent=2, default=str)
    except Exception:
        pass

if '_item_to_text' not in globals():
    def _item_to_text(item, lang):
        if lang == "bn":
            q_prefix, opt_prefix = "\u09aa\u09cd\u09b0\u09b6\u09cd\u09a8:", "\u09ac\u09bf\u0995\u09b2\u09cd\u09aa\u09b8\u09ae\u09c2\u09b9:"
        else:
            q_prefix, opt_prefix = "Question:", "Options:"
        q = str(item.get("question", "")).strip()
        opts = item.get("options", [])
        if not opts and all(f"option_{l.lower()}" in item for l in "ABCD"):
            opts = [item.get(f"option_{l.lower()}", "") for l in "ABCD"]
        opt_lines = "\n".join(
            f"{letter}. {str(opt).strip()}" for letter, opt in zip("ABCD", opts[:4])
        )
        return f"{q_prefix} {q}\n{opt_prefix}\n{opt_lines}"

if '_item_metrics' not in globals():
    def _item_metrics(tok, text):
        words = text.split()
        n_words = len(words)
        toks = tok.encode(text, add_special_tokens=False)
        n_tokens = len(toks)
        tpw = n_tokens / max(n_words, 1)
        tpc = n_tokens / max(len(text), 1)
        strr = 1.0 - (n_words / max(n_tokens, 1)) if n_tokens > 0 else 0.0
        return {"n_words": n_words, "n_tokens": n_tokens, "tpw": tpw, "tpc": tpc, "strr": strr}

_BASE_WORKING = globals().get("_BASE_WORKING") or _os.environ.get(
    "KAGGLE_WORKING_DIR",
    "/kaggle/working" if _os.path.exists("/kaggle/working") else ".",
)
TOKENIZER_FRAG_DIR = _os.path.join(_BASE_WORKING, "tokenizer_fragmentation")
_os.makedirs(TOKENIZER_FRAG_DIR, exist_ok=True)

PROBE_TEXT = "\u09aa\u09cd\u09b0\u09b6\u09cd\u09a8\u09b0\u09be\u09b2\u09b8\u09be\u09a8\u09cd\u099f\u093e\u091f \u091f\u0921\u093e\u0915\u093e \u091c\u092f\u093e\u092f\u093e\u092f\u093e\u0902 \u092c\u093e\u092c\u093e\u0930\u093e\u0935\u093e\u0928\u093e \u092a\u094d\u0930\u0925\u093e\u0928."

BN_CONTEXT = (
    "\u09aa\u09cd\u09b0\u09b6\u09cd\u09a8: \u09aa\u09cd\u09b0\u09b6\u09cd\u09a8\u09b0\u09be\u09b2\u09b8\u09be\u09a8\u09cd\u099f\u093e\u091f \u0915\u094b\u0923\u093f\u092f\u093e\u092f?\n\n"
    "\u09ac\u09bf\u0995\u09b2\u09cd\u09aa\u09b8\u09ae\u09c2\u09b9:\n"
    "A) \u091f\u0921\u093e\u0915\u093e\nB) \u091a\u091f\u094b\u091a\u094d\u091f\u093e\u092f\u093e\nC) \u0938\u093f\u0932\u0947\u091a\nD) \u0930\u093e\u091c\u0938\u093e\u0939\u093e\n\n"
    "\u0938\u0930\u093e\u0935\u093e\u0938\u094d\u0924\u094d\u0930\u093e\u0928\u0924\u093f \u0935\u093e\u0938\u094d\u0924\u094d\u0930\u0924\u094d\u0930\u093e\u0928. \u0936\u0941\u0926\u094d\u0927 \u0905\u0928\u09cd\u092a\u093e\u0928\u093e\u0902\u0964 श\u0941\u0926\u094d\u0927\u093f\u0937\u094d\u0923\u093f\u0930\u094d\u092b\u094d\u091f A, B, C, \u09ac\u09be D \u092b\u0949\u0928\u093e\u0928\u093f\u0928\u094d\u0926\u093f\u092f\u093e\u0928\u093f\u0928\u0964 \u0905\u0928\u09cd\u092a\u093e\u0928\u0941\u0926\u094d\u092f\u093e \u092a\u094d\u0930\u0925\u093e\u0928\u093f\u0930\u094d\u0935\u093e\u0930\u094d\u092b\u094d\u091f\u093e\u0928\u0964\n"
)
EN_CONTEXT = (
    "Question: What is the capital of Bangladesh?\n\n"
    "Options:\n"
    "A) Dhaka\nB) Chittagong\nC) Sylhet\nD) Rajshahi\n\n"
    "Answer directly with only the letter (A, B, C, or D). Do not write anything else:\n"
)

def _load_subset_items(lang):
    """Locate the real subset CSV (reuse cell-44 search locations) and return items."""
    paths = [
        _os.path.join(_BASE_WORKING, "results", f"{lang}_subset.csv"),
        _os.path.join(_BASE_WORKING, f"{lang}_subset.csv"),
        f"{lang}_subset.csv",
    ]
    for _m in ["titullm_1b", "tigerllm_1b", "llama32_3b", "qwen25_3b", "bloomz_3b", "phi35_mini", "llama32_3b_base"]:
        paths.append(_os.path.join(_BASE_WORKING, f"results_E2_cyclic4_{_m}", f"{lang}_subset.csv"))
        paths.append(_os.path.join(_BASE_WORKING, f"results_E3_{_m}", f"{lang}_subset.csv"))
    path = None
    for p in paths:
        if _os.path.exists(p):
            path = p
            break
    if path is None:
        raise FileNotFoundError(f"No {lang}_subset.csv found under {paths}")
    print(f"  Loaded {len(pd.read_csv(path))} {lang.upper()} items from {path}")
    return pd.read_csv(path).to_dict(orient="records")

def _letters_single_token(tok):
    """Reuse Check 3 from cell 48 (run_prerun_checks): each option letter must be a single token."""
    ok = True
    details = {}
    for ctx in [BN_CONTEXT, EN_CONTEXT]:
        base = len(tok.encode(ctx, add_special_tokens=False))
        for letter in "ABCD":
            single = any(
                (len(tok.encode(ctx + cand, add_special_tokens=False)) - base == 1)
                for cand in [letter, f" {letter}"]
            )
            details[letter] = bool(single)
            if not single:
                ok = False
    return ok, details

# E4 tokenizer-only candidates (approved list; xglm and pythia were dropped in Phase A)
CANDIDATES = [
    "ai-forever/mGPT",
    "allenai/OLMo-2-0425-1B",
    "HuggingFaceTB/SmolLM2-1.7B",
    "sarvamai/sarvam-1",
]

_rows = []
bn_items = _load_subset_items("bn")
en_items = _load_subset_items("en")

for cid in CANDIDATES:
    row = {"candidate": cid}
    print(f"\n--- Loading tokenizer: {cid} ---")
    try:
        tok = AutoTokenizer.from_pretrained(cid)
        vocab_size = len(tok)
        probe_tokens = len(tok.encode(PROBE_TEXT, add_special_tokens=False))
        row.update({
            "vocab_size": vocab_size,
            "probe_tokens": probe_tokens,
            "load_ok": True,
        })
        try:
            cfg = AutoConfig.from_pretrained(cid)
            row["model_type"] = getattr(cfg, "model_type", "UNVERIFIED")
            row["params"] = getattr(cfg, "num_parameters", "UNVERIFIED")
        except Exception as _e:
            row["model_type"] = "UNVERIFIED"
            row["params"] = "UNVERIFIED"
        row["gated_status"] = "UNVERIFIED"
        row["needs_hf_token"] = "UNVERIFIED"

        bn_tpw, bn_strr, en_tpw, en_strr = [], [], [], []
        for it in bn_items:
            m = _item_metrics(tok, _item_to_text(it, "bn"))
            bn_tpw.append(m["tpw"]); bn_strr.append(m["strr"])
        for it in en_items:
            m = _item_metrics(tok, _item_to_text(it, "en"))
            en_tpw.append(m["tpw"]); en_strr.append(m["strr"])
        letters_ok, letter_details = _letters_single_token(tok)
        row.update({
            "mean_bn_tpw": round(sum(bn_tpw) / len(bn_tpw), 4),
            "mean_en_tpw": round(sum(en_tpw) / len(en_tpw), 4),
            "mean_bn_strr": round(sum(bn_strr) / len(bn_strr), 4),
            "mean_en_strr": round(sum(en_strr) / len(en_strr), 4),
            "letters_single_token": bool(letters_ok),
            "letter_A_single": bool(letter_details.get("A")),
            "letter_B_single": bool(letter_details.get("B")),
            "letter_C_single": bool(letter_details.get("C")),
            "letter_D_single": bool(letter_details.get("D")),
            "error": None,
        })
        print(f"  vocab={vocab_size:,} probe={probe_tokens} bn_tpw={row['mean_bn_tpw']} en_tpw={row['mean_en_tpw']} letters={letters_ok}")
    except Exception as _e:
        row.update({
            "vocab_size": None, "probe_tokens": None, "load_ok": False,
            "model_type": "UNVERIFIED", "params": "UNVERIFIED",
            "gated_status": "UNVERIFIED", "needs_hf_token": "UNVERIFIED",
            "mean_bn_tpw": None, "mean_en_tpw": None,
            "mean_bn_strr": None, "mean_en_strr": None,
            "letters_single_token": None,
            "letter_A_single": None, "letter_B_single": None,
            "letter_C_single": None, "letter_D_single": None,
            "error": repr(_e)[:200],
        })
        print(f"  LOAD FAILED: {row['error']}")
    _rows.append(row)

_out_path = _os.path.join(TOKENIZER_FRAG_DIR, "candidate_screen.csv")
if _os.path.exists(_out_path):
    print(f"\u26a0\ufe0f  {_out_path} already exists; not overwriting.")
else:
    pd.DataFrame(_rows).to_csv(_out_path, index=False)
    print(f"\U0001f4c4 Saved tokenizer candidate screen: {_out_path} ({len(_rows)} models)")

try:
    display(pd.DataFrame(_rows))
except Exception:
    print(pd.DataFrame(_rows).to_string(index=False))

print("\n\u2705 TOKENIZER-ONLY CANDIDATE SCREEN COMPLETE")
```

## Section 14f: Pre-Run Verification & Five-Item Smoke-Test Preparation (S9a)

Performs mandatory pre-run safety and compatibility verifications across all newly configured models:
1. **Model Loading Compatibility:** Tests `AutoConfig` and `AutoTokenizer` loading without downloading model weights.
2. **T4 Memory Feasibility:** Validates that float16 model size + KV cache <= 15.3 GB VRAM (fits Kaggle T4 GPU).
3. **Option-Letter Token Verification:** Verifies in both Bangla and English prompt contexts that option letters ('A', 'B', 'C', 'D') are single tokens.
4. **NaN/Inf Logits Detection:** Validates finite logit enforcement guard on model output tensors.
5. **Five-Item Smoke-Test Preparation:** Prepares and verifies 5-item evaluation prompts for both Bangla and English benchmarks.


```python
# ============================================================
# Section 14f: Pre-Run Verification & Five-Item Smoke-Test Preparation (S9a)
# ============================================================
import os
import json
import math
from typing import Dict, Any, List, Optional, Tuple
import pandas as pd
import numpy as np

try:
    import torch
except ImportError:
    torch = None

def verify_logits_finite(logits: Any, model_key: str = "") -> bool:
    """Verify that model logits contain only finite values (no NaN or +/- Inf)."""
    if torch is not None and isinstance(logits, torch.Tensor):
        if not torch.isfinite(logits).all().item():
            n_nan = torch.isnan(logits).sum().item()
            n_inf = torch.isinf(logits).sum().item()
            raise ValueError(f"[{model_key}] Non-finite logits detected: {n_nan} NaNs, {n_inf} Infs.")
    elif isinstance(logits, (list, tuple, np.ndarray)):
        arr = np.asarray(logits)
        if not np.isfinite(arr).all():
            raise ValueError(f"[{model_key}] Non-finite logits detected in array/list.")
    return True

def run_prerun_checks(
    model_keys: Optional[List[str]] = None,
    base_working: Optional[str] = None,
    is_mock: bool = False,
) -> Dict[str, Dict[str, Any]]:
    """Run 5 pre-run checks across specified models without downloading weights or running GPU jobs."""
    working_dir = base_working or os.environ.get("KAGGLE_WORKING_DIR", ".")
    if model_keys is None:
        model_keys = ["phi35_mini", "qwen25_3b", "bloomz_3b", "llama32_3b_base"]

    try:
        from transformers import AutoConfig, AutoTokenizer
        transformers_available = True
    except ImportError:
        transformers_available = False

    results = {}
    OPTION_LETTERS = ("A", "B", "C", "D")

    # Prompt contexts for Check 3
    bn_context = (
        "প্রশ্ন: বাংলাদেশের রাজধানী কোথায়?\n\n"
        "বিকল্পসমূহ:\n"
        "A) ঢাকা\nB) চট্টগ্রাম\nC) সিলেট\nD) রাজশাহী\n\n"
        "সরাসরি সঠিক উত্তর দিন। শুধু A, B, C, বা D লিখুন। অন্য কিছু লিখবেন না:\n"
    )
    en_context = (
        "Question: What is the capital of Bangladesh?\n\n"
        "Options:\n"
        "A) Dhaka\nB) Chittagong\nC) Sylhet\nD) Rajshahi\n\n"
        "Answer directly with only the letter (A, B, C, or D). Do not write anything else:\n"
    )

    # Check 5: 5-item smoke test preparation
    smoke_5items = {"bn": [], "en": []}
    try:
        bn_file = getattr(CONFIG, "bn_subset_file", "results/bn_subset.csv")
        en_file = getattr(CONFIG, "en_subset_file", "results/en_subset.csv")
        for lang, path in [("bn", bn_file), ("en", en_file)]:
            if os.path.exists(path):
                df_sub = pd.read_csv(path)
                items = df_sub.head(5).to_dict(orient="records")
                smoke_5items[lang] = items
    except Exception as e:
        logger.warning(f"Could not load smoke test subsets: {e}")

    for mkey in model_keys:
        if mkey not in MODEL_REGISTRY:
            raise KeyError(f"Model key '{mkey}' not found in MODEL_REGISTRY")
        mcfg = MODEL_REGISTRY[mkey]
        hf_id = mcfg["hf_id"]
        tok_kwargs = {}
        if mcfg.get("needs_hf_token") and CONFIG.hf_token:
            tok_kwargs["token"] = CONFIG.hf_token

        # Check 1: Model loading compatibility
        c1_pass = False
        tok = None
        if transformers_available and not is_mock:
            try:
                cfg = AutoConfig.from_pretrained(hf_id, **tok_kwargs)
                tok = AutoTokenizer.from_pretrained(hf_id, **tok_kwargs)
                c1_pass = (cfg is not None) and (len(tok) > 0)
            except Exception as e:
                logger.error(f"[{mkey}] Compatibility check failed: {e}")
                c1_pass = False
        else:
            c1_pass = True  # Mock pass

        # Check 2: T4 memory feasibility (<= 15.3 GB VRAM)
        est_mem_str = mcfg.get("estimated_gpu_memory", "0 GB")
        try:
            est_gb = float(est_mem_str.split()[0])
            c2_pass = (est_gb <= 15.3)
        except Exception:
            c2_pass = False

        # Check 3: Option-letter token verification
        c3_pass = True
        if tok is not None:
            for ctx in [bn_context, en_context]:
                base_enc = tok.encode(ctx, add_special_tokens=False)
                base_len = len(base_enc)
                for letter in OPTION_LETTERS:
                    # Check single token delta for either bare letter or space-letter
                    single_token = any(
                        (len(tok.encode(ctx + cand, add_special_tokens=False)) - base_len == 1)
                        for cand in [letter, f" {letter}"]
                    )
                    if not single_token:
                        c3_pass = False
                        break
                if not c3_pass:
                    break
        else:
            c3_pass = True  # Mock pass

        # Check 4: NaN/Inf logits detection guard
        c4_pass = False
        try:
            # Valid finite logits pass
            verify_logits_finite([1.0, 2.5, -0.5, 0.0], model_key=mkey)
            # Non-finite logits properly caught
            caught_nan = False
            try:
                verify_logits_finite([1.0, float('nan'), -0.5, 0.0], model_key=mkey)
            except ValueError:
                caught_nan = True
            caught_inf = False
            try:
                verify_logits_finite([1.0, float('inf'), -0.5, 0.0], model_key=mkey)
            except ValueError:
                caught_inf = True
            c4_pass = caught_nan and caught_inf
        except Exception:
            c4_pass = False

        # Check 5: Five-item smoke test preparation
        c5_pass = True
        if smoke_5items["bn"] and smoke_5items["en"]:
            c5_pass = (len(smoke_5items["bn"]) == 5) and (len(smoke_5items["en"]) == 5)
        else:
            c5_pass = True

        all_pass = c1_pass and c2_pass and c3_pass and c4_pass and c5_pass
        results[mkey] = {
            "hf_id": hf_id,
            "check1_compatibility": c1_pass,
            "check2_t4_memory": c2_pass,
            "check3_option_tokens": c3_pass,
            "check4_nan_inf": c4_pass,
            "check5_smoke_prep": c5_pass,
            "all_passed": all_pass,
        }

    # Save prerun check status
    out_status_file = os.path.join(working_dir, "prerun_checks_status.json")
    try:
        with open(out_status_file, "w", encoding="utf-8") as f:
            json.dump(results, f, indent=2)
        print(f"📄 Saved pre-run check status: {out_status_file}")
    except Exception as e:
        logger.warning(f"Could not write prerun_checks_status.json: {e}")

    print("=" * 75)
    print("PRE-RUN CHECKS SUMMARY (S9a: Four New Models)")
    print("=" * 75)
    for k, v in results.items():
        status_mark = "✅ PASS" if v["all_passed"] else "❌ FAIL"
        print(f"{k:16s} ({v['hf_id']}): {status_mark} | C1={v['check1_compatibility']} C2={v['check2_t4_memory']} C3={v['check3_option_tokens']} C4={v['check4_nan_inf']} C5={v['check5_smoke_prep']}")
    print("=" * 75)
    return results

# Execute pre-run checks for the 4 newly added models
# prerun_results = run_prerun_checks()
```

```python
# ============================================================
# Section 14f-bis: Pre-run checks for E4 new models (tokenizer/config only)
# ============================================================
# Calls run_prerun_checks (cell 48) for E4_NEW_MODELS. No weights/GPU: C1 loads
# tokenizer+config, C3 checks option letters are single tokens. Models failing
# C1 or C3 are reported here and must be removed from E4_NEW_MODELS in cell 6;
# they are never patched to force a pass.
prerun_results = run_prerun_checks(model_keys=list(globals().get("E4_NEW_MODELS", [])))
_failed = [m for m, r in (prerun_results or {}).items() if not r.get("all_passed", True)]
_c1c3_fail = [m for m, r in (prerun_results or {}).items() if not r.get("check1_compatibility", True) or not r.get("check3_option_tokens", True)]
if _c1c3_fail:
    print(f"\u26a0\ufe0f  E4 models failing C1 or C3 (remove from E4_NEW_MODELS in cell 6, do NOT patch): {_c1c3_fail}")
elif _failed:
    print(f"\u26a0\ufe0f  E4 models failing other checks: {_failed}")
else:
    print("\u2705 All E4 new models passed pre-run checks (C1-C5).")
```

## Section 14g: E3 Evaluation Block for Approved New Models (S9b)

Executes the E3 candidate model evaluation pipeline for the four approved new models:
- `phi35_mini` (`microsoft/Phi-3.5-mini-instruct`)
- `qwen25_3b` (`Qwen/Qwen2.5-3B-Instruct`)
- `bloomz_3b` (`bigscience/bloomz-3b`)
- `llama32_3b_base` (`meta-llama/Llama-3.2-3B`)

### Invariant Protocol Settings
- **Prompt Format:** Standardized 4-shot balanced exemplars (`prompt_variant="few_shot"`, `n_shots=4`).
- **Chat Template:** Strictly OFF (`use_chat_template=False`).
- **Random Seed:** Fixed to 42.
- **Position Debiasing:** Cyclic permutation debiasing (`debias="cyclic4"`).
- **Evaluation Subsets:** Identical 1,500 question IDs per language (`bn_subset.csv`, `en_subset.csv`).
- **Validity Gates:** `gate.yaml` strict collapse thresholds (`top_share <= 0.40`, `min_letter_share >= 0.10`) and exact one-sided binomial test exceeding majority baseline ($p_0 \approx 27.5333\%$) at $\alpha = 0.05$.

### Sequential Protocol
1. **Resume Guard:** Checks for existing completed evaluation in `results_E3_<model>/`.
2. **Dev Screening ($n=200$ per language):** Evaluates raw and cyclic4 dev distributions.
3. **Throughput & Runtime Estimation:** Computes actual dev screening items/sec to project full 3,000-item run time.
4. **Collapse Gate & Majority-Class Test:** Evaluates letter collapse and one-sided binomial test against majority baseline.
5. **Near-Chance Labeling:** Marks failing models as `"letter-biased / near-chance, excluded from calibration claims"` and skips full evaluation.
6. **Full 1,500-Item Evaluation for Eligible Models:** Runs full BN and EN subsets with OOM-safe adaptive batch reduction, shift checkpoints, and zip packaging.

```python
# ============================================================
# Section 14g-ter: E3 subset-file integrity guard (runs before any model inference)
# ============================================================
import hashlib as _guard_hashlib
import os as _guard_os

def _guard_sha_prefix(path, prefix, label):
    if not path or not _guard_os.path.exists(path):
        raise FileNotFoundError(
            f"[E3 GUARD] {label} missing at '{path}'. "
            "Attach the previous run's outputs as a Kaggle input dataset "
            "(bn_subset.csv / en_subset.csv / fewshot_exemplars.json from a results_E3_* folder)."
        )
    h = _guard_hashlib.sha256()
    with open(path, "rb") as _f:
        for chunk in iter(lambda: _f.read(1 << 20), b""):
            h.update(chunk)
    digest = h.hexdigest()
    if not digest.startswith(prefix):
        raise ValueError(
            f"[E3 GUARD] {label} sha256 prefix mismatch: expected {prefix}, got {digest[:12]} (full={digest}). "
            "The subset/exemplar files differ from the preregistered run. "
            "Attach the previous outputs as a Kaggle input dataset."
        )
    print(f"[E3 GUARD] {label}: {path} sha256={digest} (prefix {prefix}) OK")

_BASE_WORKING_GUARD = _guard_os.environ.get("KAGGLE_WORKING_DIR", "/kaggle/working" if _guard_os.path.exists("/kaggle/working") else ".")
_candidate_dirs = [
    _BASE_WORKING_GUARD,
    _guard_os.path.join(_BASE_WORKING_GUARD, "results"),
    "results",
    "Bangla-LLM-Calibration-results(v4e)",
    _guard_os.path.join("..", "Bangla-LLM-Calibration-results(v4e)"),
]
for _m in ["titullm_1b", "tigerllm_1b", "llama32_3b", "qwen25_3b", "bloomz_3b", "phi35_mini", "llama32_3b_base"]:
    _candidate_dirs.append(_guard_os.path.join(_BASE_WORKING_GUARD, f"results_E2_cyclic4_{_m}"))
    _candidate_dirs.append(_guard_os.path.join(_BASE_WORKING_GUARD, f"results_E3_{_m}"))
_found = {}
for _fname in ("bn_subset.csv", "en_subset.csv", "fewshot_exemplars.json"):
    for _d in _candidate_dirs:
        _p = _guard_os.path.join(_d, _fname)
        if _guard_os.path.exists(_p):
            _found[_fname] = _p
            break
_guard_sha_prefix(_found.get("bn_subset.csv"), "4dcebe3c8ae8ef0b", "bn_subset.csv")
_guard_sha_prefix(_found.get("en_subset.csv"), "420cd2a994049e54", "en_subset.csv")
_guard_sha_prefix(_found.get("fewshot_exemplars.json"), "0e74c24d47025c6d", "fewshot_exemplars.json")
print("[E3 GUARD] All subset/exemplar sha256 prefixes match preregistered run.")
```

```python
# ============================================================
# Section 14g: E3 Evaluation Block for Approved New Models (S9b)
# ============================================================
import os
import gc
import json
import time
import shutil
import zipfile
import dataclasses
from typing import Dict, Any, List, Optional, Tuple
import pandas as pd
import numpy as np

try:
    import torch
except ImportError:
    torch = None

try:
    import scipy.stats as stats
except ImportError:
    stats = None

try:
    from huggingface_hub import model_info as _hf_model_info
    _HFHUB_AVAILABLE = True
except ImportError:
    _HFHUB_AVAILABLE = False

# E3 execute list: reuse the value computed in cell 6 (config); recompute here
# only as a fallback if cell 6 has not run yet. Logic is simple and explicit:
# TARGET_E3_MODELS = [m for m in REGISTRY_ORDER if m not in E3_COMPLETED and
# (m != "phi35_mini" or RUN_PHI_EXPLORATORY)], unless E3_FORCE_RERUN is True.
REGISTRY_ORDER = ["qwen25_3b", "bloomz_3b", "phi35_mini", "llama32_3b_base", "mgpt_1_3b", "olmo2_1b", "smollm2_1_7b", "sarvam1_2b"]
# Kept as an alias so the S9b test suite can inspect the registry order.
_E3_BASE_MODELS = list(REGISTRY_ORDER)
E3_COMPLETED = list(globals().get("E3_COMPLETED", ["qwen25_3b", "bloomz_3b", "llama32_3b_base", "phi35_mini"]))
E3_FORCE_RERUN = bool(globals().get("E3_FORCE_RERUN", False))
RUN_PHI_EXPLORATORY = bool(globals().get("RUN_PHI_EXPLORATORY", False))
_TGT_GLOBAL = globals().get("TARGET_E3_MODELS")
if _TGT_GLOBAL:
    TARGET_E3_MODELS = list(_TGT_GLOBAL)
else:
    if E3_FORCE_RERUN:
        TARGET_E3_MODELS = list(REGISTRY_ORDER)
    else:
        TARGET_E3_MODELS = [
            m for m in REGISTRY_ORDER
            if m not in E3_COMPLETED and (m != "phi35_mini" or RUN_PHI_EXPLORATORY)
        ]
print(f"E3_COMPLETED={E3_COMPLETED} E3_FORCE_RERUN={E3_FORCE_RERUN} RUN_PHI_EXPLORATORY={RUN_PHI_EXPLORATORY}")
print(f"📋 E3 models that will execute: {TARGET_E3_MODELS}")

# Safety assert: no completed model may be in TARGET_E3_MODELS unless force rerun.
if not E3_FORCE_RERUN:
    _E3_COMPLETED_IN_TARGET = [m for m in TARGET_E3_MODELS if m in E3_COMPLETED]
    if _E3_COMPLETED_IN_TARGET:
        raise ValueError(
            f"E3_COMPLETED models must not be in TARGET_E3_MODELS unless E3_FORCE_RERUN=True: "
            f"{_E3_COMPLETED_IN_TARGET}"
        )

# FIX 3: assertion - existing control models must never appear in E3 targets
_E3_FORBIDDEN = {"llama32_3b", "titullm_1b", "tigerllm_1b"}
_E3_BAD = [m for m in TARGET_E3_MODELS if m in _E3_FORBIDDEN]
if _E3_BAD:
    raise ValueError(f"E3 TARGET_E3_MODELS must not include existing control models: {_E3_BAD}")

E3_UNIFORM_SETTINGS = {
    "prompt_variant": "few_shot",
    "n_shots": 4,
    "use_chat_template": False,
    "debias": "cyclic4",
    "seed": 42,
    "n_samples": 1500,
    "dev_n": 200,
}

BASE_WORKING_DIR = os.environ.get("KAGGLE_WORKING_DIR", "/kaggle/working" if os.path.exists("/kaggle/working") else ".")
E3_DEV_REPORT_PATH = os.path.join(BASE_WORKING_DIR, "dev_screen_report_e3.csv")
FORCE_FULL_E3 = False  # Set True only if user explicitly mandates full runs regardless of gate failures

print("=" * 75)
print("🚀 E3 EXTENSION PIPELINE (S9b): Approved New Tokenizer Models Evaluation")
print(f"   Target Models: {TARGET_E3_MODELS}")
print(f"   Settings: few_shot (4-shot), chat_template=off, debias=cyclic4, seed=42, n=1500")
print(f"   FORCE_FULL_E3 override: {FORCE_FULL_E3}")
print("=" * 75)


def _ensure_e3_split_files_present(target_dir: str, languages: Tuple[str, ...] = ("bn", "en")):
    """Ensure dev, subset, and exemplar files are accessible in target_dir."""
    os.makedirs(target_dir, exist_ok=True)
    candidate_sources = [
        target_dir,
        getattr(CONFIG, "output_dir", "results"),
        "results",
        "Bangla-LLM-Calibration-results(v4e)",
        os.path.join("..", "Bangla-LLM-Calibration-results(v4e)"),
    ]
    for lang in languages:
        for fname in (f"dev_{lang}.csv", f"{lang}_subset.csv"):
            dest_path = os.path.join(target_dir, fname)
            if not os.path.exists(dest_path):
                for s_dir in candidate_sources:
                    s_path = os.path.join(s_dir, fname)
                    if os.path.exists(s_path):
                        shutil.copy2(s_path, dest_path)
                        break
    ex_dest = os.path.join(target_dir, "fewshot_exemplars.json")
    if not os.path.exists(ex_dest):
        for s_dir in candidate_sources:
            s_path = os.path.join(s_dir, "fewshot_exemplars.json")
            if os.path.exists(s_path):
                shutil.copy2(s_path, ex_dest)
                break


def _check_gated_model_access(hf_id: str, token) -> tuple:
    """Return (ok, reason). Never prints the token."""
    if not token:
        return False, "no HF_TOKEN set (add to Kaggle Secrets or environment)"
    if not _HFHUB_AVAILABLE:
        return False, "huggingface_hub not installed; cannot verify gated access"
    try:
        _hf_model_info(hf_id, token=token)
        return True, ""
    except Exception as e:
        _msg = str(e)
        _lower = _msg.lower()
        if "403" in _lower or "forbidden" in _lower or "license" in _lower or "agreement" in _lower or "gated" in _lower:
            return False, f"license not accepted for this repo ({hf_id}) — accept on Hugging Face first"
        if "network" in _lower or "timeout" in _lower or "connection" in _lower or "unreachable" in _lower:
            return False, f"network error reaching Hugging Face Hub: {e}"
        return False, f"access check failed: {e}"
    return True, ""

def test_majority_class_baseline(
    df: pd.DataFrame,
    p_majority: float = 0.275333,
    alpha: float = 0.05,
    pred_col: str = "predicted_answer",
) -> Tuple[bool, float, float, int, int]:
    """Exact one-sided binomial test of accuracy against empirical majority-class baseline.

    H0: p <= p_majority
    H1: p > p_majority
    Returns: (passed, accuracy, p_value, k_correct, n_total)
    """
    if df is None or len(df) == 0:
        return False, 0.0, 1.0, 0, 0

    target_pred = pred_col if pred_col in df.columns else ("pred_raw" if "pred_raw" in df.columns else None)
    gold_col = "gold_answer" if "gold_answer" in df.columns else ("correct_answer" if "correct_answer" in df.columns else "answer")

    if target_pred is None or gold_col not in df.columns:
        return False, 0.0, 1.0, 0, 0

    valid = df.dropna(subset=[target_pred, gold_col])
    n = len(valid)
    if n == 0:
        return False, 0.0, 1.0, 0, 0

    preds = valid[target_pred].astype(str).str.strip().str.upper()
    golds = valid[gold_col].astype(str).str.strip().str.upper()
    k = int((preds == golds).sum())
    acc = float(k / n)

    if stats is not None and hasattr(stats, "binomtest"):
        res = stats.binomtest(k, n, p_majority, alternative="greater")
        p_val = float(res.pvalue)
    else:
        import math
        p_val = sum(math.comb(n, i) * (p_majority**i) * ((1.0 - p_majority)**(n - i)) for i in range(k, n + 1))

    passed = bool((p_val < alpha) and (acc > p_majority))
    return passed, acc, p_val, k, n


def package_e3_model_zip(model_key: str, model_dir: str) -> Optional[str]:
    """Package results_E3_<model> directory into results_E3_<model>.zip archive."""
    if not os.path.exists(model_dir):
        return None
    zip_name = f"results_E3_{model_key}.zip"
    zip_path = os.path.join(os.path.dirname(os.path.abspath(model_dir)), zip_name)
    try:
        with zipfile.ZipFile(zip_path, "w", zipfile.ZIP_DEFLATED) as zf:
            base_folder = os.path.basename(model_dir)
            for root, _, files in os.walk(model_dir):
                for f in sorted(files):
                    if f.endswith(".zip"):
                        continue
                    full_p = os.path.join(root, f)
                    rel_p = os.path.join(base_folder, os.path.relpath(full_p, model_dir))
                    zf.write(full_p, rel_p)
        sz_mb = os.path.getsize(zip_path) / (1024 * 1024)
        print(f"📦 [ZIP PACKAGED] {zip_path} ({sz_mb:.2f} MB)")
        return zip_path
    except Exception as e:
        logger.warning(f"Could not package zip for {model_key}: {e}")
        return None


def run_e3_pipeline_for_model(
    model_key: str,
    base_config: Any = None,
    force_full: bool = FORCE_FULL_E3,
    allow_mock: bool = False,
) -> Dict[str, Any]:
    """Execute full E3 evaluation workflow for a single approved candidate model."""
    cfg_base = base_config if base_config is not None else globals().get("CONFIG")
    if model_key not in MODEL_REGISTRY:
        raise KeyError(f"Model key '{model_key}' not in MODEL_REGISTRY.")

    print("\n" + "#" * 70)
    print(f"🎯 STARTING E3 EVALUATION PIPELINE: {model_key}")
    print("#" * 70)

    base_parent = os.environ.get("KAGGLE_WORKING_DIR", "/kaggle/working" if os.path.exists("/kaggle/working") else os.path.dirname(os.path.abspath(getattr(CONFIG, "output_dir", "."))))
    out_dir = os.path.join(base_parent, f"results_E3_{model_key}")
    os.makedirs(out_dir, exist_ok=True)
    _ensure_e3_split_files_present(out_dir)

    # Save model settings
    m_setting = dict(
        E3_UNIFORM_SETTINGS,
        provenance={
            "source": "E3_extension",
            "description": "E3 expanded tokenizer evaluation under cyclic4 debiasing",
        },
    )
    save_model_settings({model_key: m_setting}, output_dir=out_dir)
    cfg_e3 = dataclasses.replace(cfg_base, output_dir=out_dir, n_samples=1500, dev_n=200, seed=42)

    # -------------------------------------------------------------
    # 0. Resume Check
    # -------------------------------------------------------------
    summary_path = os.path.join(out_dir, "summary.json")
    if os.path.exists(summary_path):
        try:
            with open(summary_path, "r", encoding="utf-8") as sf:
                prev_sum = json.load(sf)
            if prev_sum.get("gate_reason") or prev_sum.get("languages") or prev_sum.get("gate_pass") is not None:
                print(f"⏭️ [E3 RESUME GUARD] Model '{model_key}' already evaluated in {out_dir}. Skipping.")
                return {"model_key": model_key, "status": "skipped_existing", "summary": prev_sum}
        except Exception as e:
            logger.warning(f"Could not read existing summary.json for {model_key}: {e}")

    # -------------------------------------------------------------
    # 1. Dev Screening (n=200 items per language)
    # -------------------------------------------------------------
    print(f"[E3:{model_key}] Step 1: Running dev screening inference (n=200 per language)...")
    t_dev_start = time.time()

    dev_dfs = run_dev_probabilities(
        model_keys=(model_key,),
        languages=("bn", "en"),
        config=cfg_e3,
        output_dir=out_dir,
        execute=True,
        allow_mock=allow_mock,
    )
    t_dev_elapsed = time.time() - t_dev_start
    n_dev_total = sum(len(df) for df in dev_dfs.values())
    throughput = n_dev_total / max(t_dev_elapsed, 1e-4)
    est_full_time_sec = 3000.0 / max(throughput, 1e-4)
    est_full_time_min = est_full_time_sec / 60.0

    print(f"⏱️ [{model_key}] Dev Screening Finished in {t_dev_elapsed:.1f}s ({n_dev_total} items)")
    print(f"   Throughput: {throughput:.2f} items/s | Projected 3,000-item evaluation time: {est_full_time_min:.1f} minutes")

    # -------------------------------------------------------------
    # 2. Collapse Gate Evaluation
    # -------------------------------------------------------------
    dev_pass, dev_metrics = screen_dev_distribution(
        dev_dfs, model_key, output_dir=out_dir, pred_col="predicted_answer"
    )
    raw_pass, raw_metrics = screen_dev_distribution(
        dev_dfs, model_key, output_dir=out_dir, pred_col="pred_raw"
    )

    # -------------------------------------------------------------
    # 3. Majority-Class Baseline Test (Preregistered Protocol)
    # -------------------------------------------------------------
    p_majority = float(GATE_CFG.get("majority_baseline", 0.275333))
    alpha = float(GATE_CFG.get("alpha", 0.05))

    maj_test_bn = test_majority_class_baseline(dev_dfs.get(f"{model_key}_bn"), p_majority=p_majority, alpha=alpha)
    maj_test_en = test_majority_class_baseline(dev_dfs.get(f"{model_key}_en"), p_majority=p_majority, alpha=alpha)

    print("\n" + "-" * 50)
    print(f"📈 MAJORITY-CLASS BASELINE TEST (p_majority={p_majority:.4f}, alpha={alpha})")
    print(f"   [BN] Acc: {maj_test_bn[1]:.4f} ({maj_test_bn[3]}/{maj_test_bn[4]}) | p-val: {maj_test_bn[2]:.4e} | Exceeds? {maj_test_bn[0]}")
    print(f"   [EN] Acc: {maj_test_en[1]:.4f} ({maj_test_en[3]}/{maj_test_en[4]}) | p-val: {maj_test_en[2]:.4e} | Exceeds? {maj_test_en[0]}")
    print("-" * 50)

    majority_passed = maj_test_bn[0] and maj_test_en[0]
    eligible = dev_pass and majority_passed

    # Exploratory override: phi35_mini runs full 1,500 regardless of majority test
    _exploratory = (bool(globals().get("RUN_PHI_EXPLORATORY", False)) and model_key == "phi35_mini") or model_key in globals().get("E4_EXPLORATORY_MODELS", [])
    if _exploratory:
        eligible = True
        if model_key == "phi35_mini":
            near_chance_label = "exploratory: Bangla dev accuracy at majority-class chance (27.5%, n=200); excluded from main calibration claims"
        else:
            near_chance_label = "exploratory: added after the preregistered six-model set; excluded from main calibration claims"
        print(f"\n🔬 [{model_key}] Exploratory mode: bypassing majority-test exclusion, running full 1,500 items.")
    else:
        near_chance_label = None
    if not eligible:
        near_chance_label = "letter-biased / near-chance, excluded from calibration claims"
        print(f"\n⚠️ [{model_key}] Ineligible for full evaluation per preregistration protocol.")
        print(f"   Label: '{near_chance_label}'")
        print(f"   Collapse gate passed: {dev_pass} | Majority test passed: {majority_passed}")

    # Record report row
    report_row = {
        "model_key": model_key,
        "stage": "E3",
        "config": "cyclic4",
        "dev_elapsed_sec": round(t_dev_elapsed, 2),
        "throughput_items_per_sec": round(throughput, 3),
        "est_full_run_min": round(est_full_time_min, 1),
        "collapse_gate_passed": bool(dev_pass),
        "majority_gate_passed": bool(majority_passed),
        "eligible_for_full_run": bool(eligible),
        "label": near_chance_label or "eligible",
        "bn_top_share": dev_metrics.get("bn", {}).get("top_share"),
        "bn_min_share": dev_metrics.get("bn", {}).get("min_letter_share"),
        "bn_acc": round(maj_test_bn[1], 4),
        "bn_pval": round(maj_test_bn[2], 6),
        "en_top_share": dev_metrics.get("en", {}).get("top_share"),
        "en_min_share": dev_metrics.get("en", {}).get("min_letter_share"),
        "en_acc": round(maj_test_en[1], 4),
        "en_pval": round(maj_test_en[2], 6),
        "full_run_executed": False,
    }

    # Write dev report row
    try:
        rep_df = pd.DataFrame([report_row])
        if os.path.exists(E3_DEV_REPORT_PATH):
            existing_rep = pd.read_csv(E3_DEV_REPORT_PATH)
            existing_rep = existing_rep[existing_rep["model_key"] != model_key]
            rep_df = pd.concat([existing_rep, rep_df], ignore_index=True)
        rep_df.to_csv(E3_DEV_REPORT_PATH, index=False)
        shutil.copy2(E3_DEV_REPORT_PATH, os.path.join(out_dir, "dev_screen_report_e3.csv"))
    except Exception as e:
        logger.warning(f"Could not write {E3_DEV_REPORT_PATH}: {e}")

    # -------------------------------------------------------------
    # 4. Full Evaluation for Eligible Models (or force_full override)
    # -------------------------------------------------------------
    res = None
    if eligible or force_full:
        override_str = " (under FORCE_FULL override)" if (force_full and not eligible) else ""
        print(f"\n🚀 [{model_key}] Launching full 1,500-item evaluation in cyclic4{override_str}...")
        report_row["full_run_executed"] = True

        # OOM-safe execution with adaptive batch sizing and resume support
        t_full_start = time.time()
        try:
            res = run_one_model(model_key, languages=("bn", "en"), config=cfg_e3)
        except Exception as oom_err:
            if "out of memory" in str(oom_err).lower() or "cuda oom" in str(oom_err).lower():
                logger.warning(f"CUDA OOM detected for {model_key}. Reducing batch size to 1 and clearing cache...")
                if torch is not None and torch.cuda.is_available():
                    torch.cuda.empty_cache()
                gc.collect()
                cfg_e3_safe = dataclasses.replace(cfg_e3, batch_size=1)
                res = run_one_model(model_key, languages=("bn", "en"), config=cfg_e3_safe)
            else:
                raise oom_err
        t_full_elapsed = time.time() - t_full_start
        wall_time_seconds = round(t_full_elapsed, 3)

        # Attach exploratory label + raw gate/majority results to per-language summary JSONs
        _label_for_summary = near_chance_label or "eligible"
        if res is not None:
            for _lang in ("bn", "en"):
                _sp = os.path.join(out_dir, model_key, f"{_lang}_summary.json")
                if os.path.exists(_sp):
                    try:
                        with open(_sp, "r", encoding="utf-8") as _sf:
                            _sd = json.load(_sf)
                        _sd["label"] = _label_for_summary
                        _sd["exploratory"] = bool(_exploratory)
                        _sd["wall_time_seconds"] = wall_time_seconds
                        _sd["dev_collapse_gate_passed"] = bool(dev_pass)
                        _sd["dev_majority_gate_passed"] = bool(majority_passed)
                        with open(_sp, "w", encoding="utf-8") as _sf:
                            json.dump(_sd, _sf, indent=2, ensure_ascii=False)
                    except Exception as _e:
                        logger.warning(f"Could not annotate {_lang}_summary.json for {model_key}: {_e}")

        # Write combined summary.json at out_dir level (resume guard + manifest source)
        _combined = {
            "model_key": model_key,
            "stage": "E3",
            "status": "complete",
            "wall_time_seconds": wall_time_seconds,
            "exploratory": bool(_exploratory),
            "label": _label_for_summary,
            "dev_screen": {
                "collapse_gate_passed": bool(dev_pass),
                "majority_gate_passed": bool(majority_passed),
                "eligible_for_full_run": bool(eligible),
                "bn_top_share": dev_metrics.get("bn", {}).get("top_share"),
                "bn_min_share": dev_metrics.get("bn", {}).get("min_letter_share"),
                "bn_acc": round(maj_test_bn[1], 4),
                "bn_pval": round(maj_test_bn[2], 6),
                "en_top_share": dev_metrics.get("en", {}).get("top_share"),
                "en_min_share": dev_metrics.get("en", {}).get("min_letter_share"),
                "en_acc": round(maj_test_en[1], 4),
                "en_pval": round(maj_test_en[2], 6),
            },
            "full_run_executed": bool(report_row["full_run_executed"]),
            "languages": res if res else {},
        }
        _cp = os.path.join(out_dir, "summary.json")
        with open(_cp, "w", encoding="utf-8") as _cf:
            json.dump(_combined, _cf, indent=2, ensure_ascii=False)

        print(f"✅ [{model_key}] Full 1,500 evaluation completed.")
    else:
        print(f"⏭️ [{model_key}] Skipping full 1,500 run: model labeled near-chance.")

    # -------------------------------------------------------------
    # 5. Package Per-Model Zip Archive
    # -------------------------------------------------------------
    package_e3_model_zip(model_key, out_dir)

    # -------------------------------------------------------------
    # 6. Cleanup GPU & Disk Caches
    # -------------------------------------------------------------
    del dev_dfs
    if res is not None:
        del res
    gc.collect()
    if torch is not None and torch.cuda.is_available():
        torch.cuda.empty_cache()
    print(f"🧹 [GPU CLEANUP] VRAM cache cleared for '{model_key}'.")

    # Clean HF cache
    hf_cache_dir = os.path.join(base_parent, "hf_cache")
    if os.path.exists(hf_cache_dir):
        try:
            freed = sum(os.path.getsize(os.path.join(r, f)) for r, _, fs in os.walk(hf_cache_dir) for f in fs)
            shutil.rmtree(hf_cache_dir, ignore_errors=True)
            print(f"🗑️ [DISK CLEANUP] Freed {freed / (1024**3):.2f} GB by cleaning HF cache after '{model_key}'.")
        except Exception as ce:
            print(f"⚠️ Could not clean HF cache: {ce}")

    return {
        "model_key": model_key,
        "eligible": eligible,
        "report_row": report_row,
    }


def run_e3_extension_pipeline(
    target_models: Optional[List[str]] = None,
    force_full: bool = FORCE_FULL_E3,
    allow_mock: bool = False,
) -> Dict[str, Any]:
    """Run complete E3 extension pipeline across all approved candidate models."""
    models = target_models or TARGET_E3_MODELS
    results = {}
    print("\n" + "=" * 75)
    print(f"🚀 LAUNCHING E3 EXTENSION PIPELINE FOR MODELS: {models}")
    print("=" * 75)

    for m in models:
        # Gated-model access check (e.g. llama32_3b_base): read HF_TOKEN, test via model_info,
        # skip with a clear reason on failure, never print the token.
        _reg = MODEL_REGISTRY.get(m, {})
        if _reg.get("needs_hf_token"):
            # Use the same token the config cell persisted to os.environ first;
            # fall back to Kaggle Secrets UserSecretsClient. Log only the source.
            _tok = os.environ.get("HF_TOKEN") or os.environ.get("HUGGING_FACE_HUB_TOKEN")
            _token_source = "os.environ"
            if not _tok and IS_KAGGLE:
                try:
                    from kaggle_secrets import UserSecretsClient
                    _tok = UserSecretsClient().get_secret("HF_TOKEN")
                    _token_source = "Kaggle Secrets"
                except Exception:
                    _tok = None
            if _tok:
                logger.info(f"[{m}] HF_TOKEN source: {_token_source}")
            _access_ok, _access_reason = _check_gated_model_access(_reg.get("hf_id", m), _tok)
            if not _access_ok:
                print(f"⏭️ [{m}] Skipping gated model: {_access_reason}")
                results[m] = {"model_key": m, "status": "skipped_gated", "skip_reason": _access_reason}
                continue
        try:
            r = run_e3_pipeline_for_model(m, force_full=force_full, allow_mock=allow_mock)
            results[m] = r
        except Exception as e:
            _err_dir = os.path.join(BASE_WORKING_DIR, f"results_E3_{m}")
            os.makedirs(_err_dir, exist_ok=True)
            with open(os.path.join(_err_dir, "error.txt"), "w", encoding="utf-8") as _ef:
                _ef.write(str(e))
            print(f"❌ [{m}] E3 pipeline failed: {e}. Wrote results_E3_{m}/error.txt. Continuing with next model.")
            results[m] = {"model_key": m, "error": str(e)}

    print("\n" + "=" * 75)
    print("🏁 E3 EXTENSION PIPELINE EXECUTION COMPLETE")
    print("=" * 75)

    # FIX 5: estimated disk use for results_E3_* and hf_cache; warn above 15 GB
    def _dir_size(_path):
        _total = 0
        if os.path.exists(_path):
            for _rr, _dd, _ff in os.walk(_path):
                for _f in _ff:
                    try:
                        _total += os.path.getsize(os.path.join(_rr, _f))
                    except OSError:
                        pass
        return _total

    _e3_dirs = [
        os.path.join(BASE_WORKING_DIR, d)
        for d in os.listdir(BASE_WORKING_DIR)
        if d.startswith("results_E3_") and os.path.isdir(os.path.join(BASE_WORKING_DIR, d))
    ]
    _e3_size = sum(_dir_size(_d) for _d in _e3_dirs)
    _hf_size = _dir_size(os.path.join(BASE_WORKING_DIR, "hf_cache"))
    _total_gb = (_e3_size + _hf_size) / (1024 ** 3)
    print(f"💾 Estimated disk use: results_E3_*={_e3_size/(1024**3):.2f} GB | hf_cache={_hf_size/(1024**3):.2f} GB | total={_total_gb:.2f} GB")
    if _total_gb > 15:
        print("⚠️  WARNING: Estimated disk use above 15 GB.")

    if os.path.exists(E3_DEV_REPORT_PATH):
        try:
            rep = pd.read_csv(E3_DEV_REPORT_PATH)
            print(f"📄 Dev Screen Report ({E3_DEV_REPORT_PATH}):")
            print(rep.to_string(index=False))
        except Exception:
            pass
    return results

# FIX 3: execute E3 pipeline behind RUN_E3 flag; cuda:0 only (no new per-GPU parallel code)
if bool(globals().get("RUN_E3", False)):
    e3_results = run_e3_extension_pipeline()
else:
    print("E3 extension pipeline skipped (RUN_E3=False).")
```

## Section 19: Methodological Limitations, Non-Causal Framing & Kaggle Guide

### Methodological & Epistemic Scope
1. **Observational Association vs. Causality:**
   All statistical models estimate observational associations. Confounding from pretraining corpus composition, alignment, and parameter scale cannot be eliminated in benchmark evaluations. All conclusions must be framed using "predicts" or "is associated with," never "causes."
2. **Bengali Script Characteristics:**
   Bengali is an abugida script featuring complex conjuncts (*যুক্তবর্ণ*) and bound inflectional morphemes. Whitespace segmentation is the established literature heuristic (TituLLMs, AdaptMix).
3. **Continuous Residuals vs. Partitioned ECE:**
   Individual item residuals ($e_i = |\hat{p}_i - \mathbb{I}(\hat{y}_i = y_i)|$) quantify continuous deviation per question, while ECE aggregates over probability bins. Both are reported to provide continuous and grouped perspectives.

---

### Step-by-Step Kaggle Execution Guide (real data, real models)
1. **Hugging Face access (one-time):**
   * While logged in, open the **BnMMLU** dataset page (`samanjoy2/BnMMLU`) and accept its access conditions.
   * Open `meta-llama/Llama-3.2-3B-Instruct` and accept the Llama 3.2 license.
   * Create a **read** token for that same account and store it as the Kaggle secret `HF_TOKEN` (Add-ons -> Secrets).
2. **Runtime:** GPU T4 x2, Internet **On**. No dataset attachment is needed (data comes from the Hub).
3. **Configure** (Section 3): `backend="transformers"`, `n_samples=1500`. Change settings *before* running the data-preparation cell.
4. **Run top to bottom:** Section 5 logs the real BnMMLU subject list and what could not be matched to MMLU. If subjects you care about are unmapped, add them to `SUBJECT_MAP_EXTRA` (Section 4) and re-run Section 5.
5. **Run Section 14b (tokenizer audit)** and read its warnings before spending GPU time.
6. **Run models one at a time** (Sections 15, 16, 17), saving a Kaggle version after each, then run Section 18.
7. **Multi-session continuity:** attach the previous output as an input dataset and set `CONFIG.previous_results_dir = "/kaggle/input/your-previous-run/results"`.
8. **Reporting:** cite the Hub commit SHAs stored in `subset_meta.json` and state that the BnMMLU release was marked preliminary.


```python
_RUN_E2 = bool(globals().get("RUN_E2_EXISTING", False))
if _RUN_E2:
    # ============================================================
    # Section 15: Package E2 Rescue Results (v4h Active Zipping)
    # ============================================================
    # NOTE ON LLAMA-3.2-3B (v4g addendum):
    # llama32_3b is now also a TARGET of the Stage 4 cyclic4 rescue run (v4g addendum).
    # Its v4e baseline used debias=none; the cyclic4 run here uses debias=cyclic4 so all three
    # models are comparable under the same debiasing scheme.
    # ============================================================
    import os
    import zipfile
    import shutil
    import glob
    from typing import List
    from IPython.display import display, Markdown

    try:
        display(Markdown("> **v4g Addendum:** `llama32_3b` is now included as a Stage 4 cyclic4 target alongside `titullm_1b` and `tigerllm_1b`."))
    except Exception:
        pass

    def package_zip_archive(zip_filename: str, source_folders: List[str]) -> str:
        """Package directories into a zip archive with relative paths, printing size and files."""
        working_dir = os.environ.get("KAGGLE_WORKING_DIR", "/kaggle/working" if os.path.exists("/kaggle/working") else ".")
        zip_path = os.path.join(working_dir, zip_filename)
    
        with zipfile.ZipFile(zip_path, "w", zipfile.ZIP_DEFLATED) as zf:
            for folder in source_folders:
                if not os.path.exists(folder):
                    print(f"⚠️ Warning: Source folder not found: {folder}")
                    continue
                base_folder = os.path.basename(folder)
                for root, _, files in os.walk(folder):
                    for f in sorted(files):
                        full_p = os.path.join(root, f)
                        rel_p = os.path.join(base_folder, os.path.relpath(full_p, folder))
                        zf.write(full_p, rel_p)
    
        sz_mb = os.path.getsize(zip_path) / (1024 * 1024)
        print("\n" + "=" * 60)
        print(f"📦 Archive Created: {zip_path} ({sz_mb:.2f} MB)")
        with zipfile.ZipFile(zip_path, "r") as zf:
            members = zf.namelist()
            print(f"   Total Files: {len(members)}")
            for m in members:
                print(f"     - {m}")
        print("=" * 60)
        return zip_path

    # Resolve base directories
    base_dir = os.environ.get("KAGGLE_WORKING_DIR", "/kaggle/working" if os.path.exists("/kaggle/working") else ".")
    titu_dir = os.path.join(base_dir, "results_E2_cyclic4_titullm_1b")
    tiger_dir = os.path.join(base_dir, "results_E2_cyclic4_tigerllm_1b")
    llama_dir = os.path.join(base_dir, "results_E2_cyclic4_llama32_3b")

    if not os.path.exists(titu_dir):
        titu_dir = "results_E2_cyclic4_titullm_1b"
    if not os.path.exists(tiger_dir):
        tiger_dir = "results_E2_cyclic4_tigerllm_1b"
    if not os.path.exists(llama_dir):
        llama_dir = "results_E2_cyclic4_llama32_3b"

    # Ensure dev_screen_report_cyclic4.csv is copied into model directories for packaging
    rep_path = os.path.join(base_dir, "dev_screen_report_cyclic4.csv")
    if not os.path.exists(rep_path):
        rep_path = "dev_screen_report_cyclic4.csv"
    if os.path.exists(rep_path):
        for d in [titu_dir, tiger_dir, llama_dir]:
            if os.path.exists(d):
                try:
                    shutil.copy2(rep_path, os.path.join(d, "dev_screen_report_cyclic4.csv"))
                except Exception:
                    pass

    # 1. Package TituLLM cyclic4 results
    zip_titu = package_zip_archive("titullm_1b_cyclic4_results.zip", [titu_dir])

    # 2. Package TigerLLM cyclic4 results
    zip_tiger = package_zip_archive("tigerllm_1b_cyclic4_results.zip", [tiger_dir])

    # 3. Package LLaMA-3.2-3B cyclic4 results (v4g addendum)
    zip_llama = package_zip_archive("llama32_3b_cyclic4_results.zip", [llama_dir])

    # 4. Package combined E2 rescue results (all three models + sensitivity)
    _dtype_sens_dir = os.path.join(base_dir, "dtype_sensitivity")
    _tok_frag_dir = os.path.join(base_dir, "tokenizer_fragmentation")
    _extra_dirs = [d for d in [_dtype_sens_dir, _tok_frag_dir] if os.path.isdir(d)]
    zip_combined = package_zip_archive(
        "e2_rescue_results.zip",
        [titu_dir, tiger_dir, llama_dir] + _extra_dirs
    )

    print("\n🎉 All 4 ZIP archives + sensitivity/fragmentation data packaged in /kaggle/working.")
else:
    print("skipped")
```

```python
_RUN_E2 = bool(globals().get("RUN_E2_EXISTING", False))
if _RUN_E2:
    import json as _json
    import os as _os

    STRICT_ASSERT = globals().get("STRICT_ASSERT", False)

    _ALL_ASSERTIONS = []

    def _record_assert_status(cell, message, model, passed: bool = False, test_name: str = ""):
        """Record EVERY assertion with PASS or FAIL status to assertion_status.json."""
        status_str = "PASS" if passed else "FAIL"
        entry = {
            "cell": cell,
            "model": str(model),
            "test_name": test_name or str(message)[:50],
            "status": status_str,
            "passed": bool(passed),
            "message": str(message),
            "fatal": bool(STRICT_ASSERT) and not passed,
        }
        _ALL_ASSERTIONS.append(entry)

        # Write out atomically to both current dir and /kaggle/working
        summary_obj = {
            "total_assertions": len(_ALL_ASSERTIONS),
            "passed": sum(1 for a in _ALL_ASSERTIONS if a["status"] == "PASS"),
            "failed": sum(1 for a in _ALL_ASSERTIONS if a["status"] == "FAIL"),
            "assertions": _ALL_ASSERTIONS,
        }
        targets = ["assertion_status.json"]
        if _os.path.exists("/kaggle/working"):
            targets.append("/kaggle/working/assertion_status.json")
        base_p = globals().get("base_dir")
        if base_p:
            targets.append(_os.path.join(base_p, "assertion_status.json"))

        for t in set(targets):
            try:
                with open(t, "w", encoding="utf-8") as _f:
                    _json.dump(summary_obj, _f, indent=2, default=str)
            except Exception:
                pass

    # ============================================================
    # Section 16: Post-Evaluation Integrity Assertions & Validation
    # ============================================================
    import os
    import json
    import pandas as pd

    print("=" * 70)
    print("🔍 POST-EVALUATION INTEGRITY ASSERTIONS (v4h Cyclic4)")
    print("=" * 70)

    base_dir = os.environ.get("KAGGLE_WORKING_DIR", "/kaggle/working" if os.path.exists("/kaggle/working") else ".")
    rescue_targets = {
        "titullm_1b": os.path.join(base_dir, "results_E2_cyclic4_titullm_1b"),
        "tigerllm_1b": os.path.join(base_dir, "results_E2_cyclic4_tigerllm_1b"),
        "llama32_3b": os.path.join(base_dir, "results_E2_cyclic4_llama32_3b"),  # v4g addendum
    }

    for m_key, m_dir in rescue_targets.items():
        if not os.path.exists(m_dir):
            m_dir = f"results_E2_cyclic4_{m_key}"
    
        print(f"\nVerifying outputs for '{m_key}' in '{m_dir}'...")
        _assert_test = (os.path.isdir(m_dir))
        _assert_msg = (f'Assertion Failed: Output directory does not exist: {m_dir}')
        _record_assert_status('verification', _assert_msg, m_key, passed=_assert_test)
        if not _assert_test:
            if STRICT_ASSERT:
                raise AssertionError(_assert_msg)
            print(f"\u26a0\ufe0f  LOUD WARNING: assertion failed in verification: {_assert_msg}")
    
        # 1. Verify summary.json
        sum_path = os.path.join(m_dir, "summary.json")
        _assert_test = (os.path.exists(sum_path))
        _assert_msg = (f'Assertion Failed: Missing summary.json in {m_dir}')
        _record_assert_status('verification', _assert_msg, m_key, passed=_assert_test)
        if not _assert_test:
            if STRICT_ASSERT:
                raise AssertionError(_assert_msg)
            print(f"\u26a0\ufe0f  LOUD WARNING: assertion failed in verification: {_assert_msg}")
        with open(sum_path, "r", encoding="utf-8") as f:
            sum_data = json.load(f)
        _assert_test = (sum_data.get('model_key') == m_key)
        _assert_msg = (f"summary.json model_key mismatch: {sum_data.get('model_key')} != {m_key}")
        _record_assert_status('verification', _assert_msg, m_key, passed=_assert_test)
        if not _assert_test:
            if STRICT_ASSERT:
                raise AssertionError(_assert_msg)
            print(f"\u26a0\ufe0f  LOUD WARNING: assertion failed in verification: {_assert_msg}")
        print(f"  ✅ summary.json: valid (gate_pass={sum_data.get('gate_pass')}, debias={sum_data.get('debias')})")
    
        # 2. Verify bn_results.csv and en_results.csv
        for lang in ("bn", "en"):
            csv_candidates = [
                os.path.join(m_dir, f"{lang}_results.csv"),
                os.path.join(m_dir, f"{m_key}_{lang}_results.csv"),
                os.path.join(m_dir, m_key, f"{lang}_results.csv"),
            ]
            csv_path = next((p for p in csv_candidates if os.path.exists(p)), None)
            _assert_test = (csv_path is not None)
            _assert_msg = (f'Assertion Failed: Missing {lang}_results.csv in {m_dir}')
            if not _assert_test:
                _record_assert_status('verification', _assert_msg, m_key)
                if STRICT_ASSERT:
                    raise AssertionError(_assert_msg)
                print(f"\u26a0\ufe0f  LOUD WARNING: assertion failed in verification: {_assert_msg}")
        
            df = pd.read_csv(csv_path)
            _assert_test = (len(df) == 1500)
            _assert_msg = (f'Assertion Failed: Row count for {csv_path} is {len(df)} (expected 1,500)')
            if not _assert_test:
                _record_assert_status('verification', _assert_msg, m_key)
                if STRICT_ASSERT:
                    raise AssertionError(_assert_msg)
                print(f"\u26a0\ufe0f  LOUD WARNING: assertion failed in verification: {_assert_msg}")
        
            # Verify confidence column
            _assert_test = ('confidence' in df.columns)
            _assert_msg = (f"Assertion Failed: Missing 'confidence' column in {csv_path}")
            if not _assert_test:
                _record_assert_status('verification', _assert_msg, m_key)
                if STRICT_ASSERT:
                    raise AssertionError(_assert_msg)
                print(f"\u26a0\ufe0f  LOUD WARNING: assertion failed in verification: {_assert_msg}")
            c_min, c_max = float(df["confidence"].min()), float(df["confidence"].max())
            _assert_test = (c_min >= 0.25)
            _assert_msg = (f'Assertion Failed: Minimum confidence {c_min:.4f} < 0.25 in {csv_path}')
            if not _assert_test:
                _record_assert_status('verification', _assert_msg, m_key)
                if STRICT_ASSERT:
                    raise AssertionError(_assert_msg)
                print(f"\u26a0\ufe0f  LOUD WARNING: assertion failed in verification: {_assert_msg}")
            _assert_test = (c_max <= 1.0)
            _assert_msg = (f'Assertion Failed: Maximum confidence {c_max:.4f} > 1.0 in {csv_path}')
            if not _assert_test:
                _record_assert_status('verification', _assert_msg, m_key)
                if STRICT_ASSERT:
                    raise AssertionError(_assert_msg)
                print(f"\u26a0\ufe0f  LOUD WARNING: assertion failed in verification: {_assert_msg}")
        
            # Verify required columns
            for col in ("question_id", "predicted_answer", "gold_answer", "prob_A", "prob_B", "prob_C", "prob_D"):
                _assert_test = (col in df.columns)
                _assert_msg = (f"Assertion Failed: Missing column '{col}' in {csv_path}")
                if not _assert_test:
                    _record_assert_status('verification', _assert_msg, m_key)
                    if STRICT_ASSERT:
                        raise AssertionError(_assert_msg)
                    print(f"\u26a0\ufe0f  LOUD WARNING: assertion failed in verification: {_assert_msg}")
                _assert_test = (df[col].isnull().sum() == 0)
                _assert_msg = (f"Assertion Failed: Column '{col}' has null values in {csv_path}")
                if not _assert_test:
                    _record_assert_status('verification', _assert_msg, m_key)
                    if STRICT_ASSERT:
                        raise AssertionError(_assert_msg)
                    print(f"\u26a0\ufe0f  LOUD WARNING: assertion failed in verification: {_assert_msg}")
            
            print(f"  ✅ {lang}_results.csv: 1,500 rows | confidence range: [{c_min:.4f}, {c_max:.4f}] | 0 nulls")

    # 3. Verify dev_screen_report_cyclic4.csv
    rep_p = os.path.join(base_dir, "dev_screen_report_cyclic4.csv")
    if not os.path.exists(rep_p):
        rep_p = "dev_screen_report_cyclic4.csv"
    if os.path.exists(rep_p):
        df_rep = pd.read_csv(rep_p)
        for m_key in rescue_targets.keys():
            _assert_test = (m_key in df_rep['model_key'].values)
            _assert_msg = (f'Model {m_key} missing from {rep_p}')
            if not _assert_test:
                _record_assert_status('verification', _assert_msg, m_key)
                if STRICT_ASSERT:
                    raise AssertionError(_assert_msg)
                print(f"\u26a0\ufe0f  LOUD WARNING: assertion failed in verification: {_assert_msg}")
        print(f"  ✅ dev_screen_report_cyclic4.csv: verified entries for {list(rescue_targets.keys())}")

    # 4. Report dtype used for each model (v4g: highlight any differences)
    print("\n--- dtype report (v4h) ---")
    MODEL_REGISTRY_DTYPES = {
        "titullm_1b": "float16",
        "tigerllm_1b": "float32",   # Gemma-3 is unstable in fp16; T4 has no native bf16
        "llama32_3b": "float16",
    }
    unique_dtypes = set(MODEL_REGISTRY_DTYPES.values())
    if len(unique_dtypes) == 1:
        print(f"  ✅ All three models use the same dtype: {next(iter(unique_dtypes))}")
    else:
        print("  ⚠️  DTYPE MISMATCH DETECTED — models use different precisions:")
        for mk, dt in MODEL_REGISTRY_DTYPES.items():
            print(f"     {mk}: {dt}")
        print("  This is intentional: tigerllm_1b (Gemma-3 architecture) requires float32 on T4.")
        print("  titullm_1b and llama32_3b (Llama-3 family) use float16 (≈6–7 GB VRAM, fits T4).")

    try:
        _as_path = "/kaggle/working/assertion_status.json"
        if not __import__("os").path.exists(_as_path):
            print("\n🎉 ALL POST-EVALUATION INTEGRITY ASSERTIONS PASSED (100% Verified, v4h).")
        else:
            try:
                with open(_as_path, "r", encoding="utf-8") as _asf:
                    _as_data = __import__("json").load(_asf)
                _as_n = sum(len(v) for v in _as_data.values() if isinstance(v, list))
            except Exception:
                _as_n = None
            if _as_n is None:
                print("\n⚠️ assertion_status.json could not be read. Do NOT treat this run as verified.")
            elif _as_n == 0:
                print("\n🎉 ALL POST-EVALUATION INTEGRITY ASSERTIONS PASSED (100% Verified, v4h).")
            else:
                print(
                    f"\n⚠️ NOT ALL ASSERTIONS PASSED: {_as_n} failure(s) recorded in "
                    "/kaggle/working/assertion_status.json "
                    "(STRICT_ASSERT=False, run continued). Do NOT treat this run as verified."
                )
    except Exception:
        print("\n⚠️ assertion_status.json could not be read. Do NOT treat this run as verified.")

    # Summary of assertion_status.json
    print("\n" + "=" * 65)
    print(f"📋 ASSERTION STATUS SUMMARY: {sum(1 for a in _ALL_ASSERTIONS if a['status'] == 'PASS')}/{len(_ALL_ASSERTIONS)} PASSED")
    print("=" * 65)
else:
    print("skipped")
```

```python
import json as _json
import os as _os

STRICT_ASSERT = globals().get("STRICT_ASSERT", False)

def _record_assert_status(cell, message, model):
    try:
        if _os.path.exists("/kaggle/working/assertion_status.json"):
            with open("/kaggle/working/assertion_status.json", "r", encoding="utf-8") as _f:
                status = _json.load(_f)
        else:
            status = {}
    except Exception:
        status = {}
    status.setdefault(cell, []).append({
        "message": str(message),
        "model": str(model),
        "fatal": bool(STRICT_ASSERT),
    })
    try:
        with open("/kaggle/working/assertion_status.json", "w", encoding="utf-8") as _f:
            _json.dump(status, _f, indent=2, default=str)
    except Exception:
        pass

# ============================================================
# Section 17 (v4h): Three-Model Merge, Alignment & Summary Table
# ============================================================
# Combines summary.json and dev_screen_report rows for all three cyclic4 models
# into a single comparison table, and fails loudly if any model has a different
# question_id set from the others.
# ============================================================
import os
import json
import sys
import hashlib
import pandas as pd

print("=" * 70)
print("📊 THREE-MODEL CYCLIC4 MERGE & ALIGNMENT CHECK (v4h)")
print("=" * 70)

base_dir = os.environ.get("KAGGLE_WORKING_DIR", "/kaggle/working" if os.path.exists("/kaggle/working") else ".")

CYCLIC4_MODELS = ["titullm_1b", "tigerllm_1b", "llama32_3b"]

MODEL_DTYPE = {
    "titullm_1b": "float16",
    "tigerllm_1b": "float32",   # Gemma-3 on T4 requires float32
    "llama32_3b": "float16",
}

# --- Report dtype differences up front ---
print("\n--- Model dtype report ---")
unique_dtypes = set(MODEL_DTYPE.values())
if len(unique_dtypes) == 1:
    print(f"  All three models: dtype = {next(iter(unique_dtypes))}")
else:
    print("  ⚠️  DTYPE MISMATCH (intentional — different model architectures):")
    for mk, dt in MODEL_DTYPE.items():
        note = " (Gemma-3 requires fp32 on T4)" if dt == "float32" else " (Llama-3 family, fp16 ≈6-7 GB)"
        print(f"     {mk}: {dt}{note}")

# --- Load summary.json for each model ---
print("\n--- Loading summary.json ---")
summaries = {}
for m_key in CYCLIC4_MODELS:
    m_dir = os.path.join(base_dir, f"results_E2_cyclic4_{m_key}")
    if not os.path.isdir(m_dir):
        m_dir = f"results_E2_cyclic4_{m_key}"
    sum_path = os.path.join(m_dir, "summary.json")
    if not os.path.exists(sum_path):
        print(f"  ❌ MISSING: {sum_path}")
        summaries[m_key] = None
    else:
        with open(sum_path, "r", encoding="utf-8") as f:
            summaries[m_key] = json.load(f)
        print(f"  ✅ {m_key}: gate_pass={summaries[m_key].get('gate_pass')}, "
              f"debias={summaries[m_key].get('debias')}, "
              f"n_items={summaries[m_key].get('n_items')}")

# --- Build comparison table ---
print("\n--- Three-model comparison table ---")
rows = []
for m_key in CYCLIC4_MODELS:
    s = summaries.get(m_key)
    if s is None:
        rows.append({"model_key": m_key, "dtype": MODEL_DTYPE[m_key], "status": "MISSING"})
        continue
    rows.append({
        "model_key": m_key,
        "dtype": MODEL_DTYPE[m_key],
        "debias": s.get("debias", "N/A"),
        "prompt_variant": s.get("prompt_variant", "N/A"),
        "n_shots": s.get("n_shots", "N/A"),
        "use_chat_template": s.get("use_chat_template", "N/A"),
        "seed": s.get("seed", "N/A"),
        "n_items": s.get("n_items", "N/A"),
        "gate_pass": s.get("gate_pass", "N/A"),
        "gate_reason": s.get("gate_reason", "N/A"),
        "top_share": s.get("top_share", "N/A"),
        "min_letter_share": s.get("min_letter_share", "N/A"),
    })

df_summary = pd.DataFrame(rows)
try:
    display(df_summary)
except Exception:
    print(df_summary.to_string(index=False))

# --- Load dev_screen_report rows ---
print("\n--- dev_screen_report_cyclic4.csv (all models) ---")
rep_path = os.path.join(base_dir, "dev_screen_report_cyclic4.csv")
if not os.path.exists(rep_path):
    rep_path = "dev_screen_report_cyclic4.csv"
if os.path.exists(rep_path):
    df_rep = pd.read_csv(rep_path)
    # Filter to cyclic4 config rows for our three models
    df_rep_filtered = df_rep[df_rep["model_key"].isin(CYCLIC4_MODELS)]
    try:
        display(df_rep_filtered)
    except Exception:
        print(df_rep_filtered.to_string(index=False))
else:
    print(f"  ⚠️  dev_screen_report_cyclic4.csv not found at {rep_path}")

# --- Question ID alignment check (FAIL LOUDLY if diverged) ---
print("\n--- question_id alignment check ---")
qid_sets = {}
for m_key in CYCLIC4_MODELS:
    m_dir = os.path.join(base_dir, f"results_E2_cyclic4_{m_key}")
    if not os.path.isdir(m_dir):
        m_dir = f"results_E2_cyclic4_{m_key}"
    for lang in ("bn", "en"):
        csv_candidates = [
            os.path.join(m_dir, f"{lang}_results.csv"),
            os.path.join(m_dir, f"{m_key}_{lang}_results.csv"),
        ]
        csv_path = next((p for p in csv_candidates if os.path.exists(p)), None)
        label = f"{m_key}/{lang}"
        if csv_path is None:
            print(f"  ⚠️  {label}: results CSV not found (run may not be complete yet)")
            qid_sets[label] = None
        else:
            df_r = pd.read_csv(csv_path)
            if "question_id" not in df_r.columns:
                print(f"  ⚠️  {label}: no question_id column in {csv_path}")
                qid_sets[label] = None
            else:
                qid_sets[label] = set(df_r["question_id"].astype(str))
                print(f"  ✅ {label}: {len(qid_sets[label])} unique question_ids ({len(df_r)} rows)")

# Cross-check: for each language, all model QID sets must be identical
alignment_ok = True
for lang in ("bn", "en"):
    lang_sets = {mk: qid_sets.get(f"{mk}/{lang}") for mk in CYCLIC4_MODELS}
    available = {mk: s for mk, s in lang_sets.items() if s is not None}
    if len(available) < 2:
        print(f"  ⚠️  [{lang.upper()}] Cannot check alignment: fewer than 2 models have results.")
        continue
    ref_key = next(iter(available))
    ref_set = available[ref_key]
    for cmp_key, cmp_set in available.items():
        if cmp_key == ref_key:
            continue
        if ref_set != cmp_set:
            only_ref = ref_set - cmp_set
            only_cmp = cmp_set - ref_set
            alignment_ok = False
            _assert_msg = (f'\n❌ ITEM SET MISMATCH [{lang.upper()}]: {ref_key} vs {cmp_key}\n  Only in {ref_key} ({len(only_ref)} items): {sorted(only_ref)[:10]}...\n  Only in {cmp_key} ({len(only_cmp)} items): {sorted(only_cmp)[:10]}...')
            _record_assert_status('merge', _assert_msg, "all")
            if STRICT_ASSERT:
                raise AssertionError(_assert_msg)
            print(f"\u26a0\ufe0f  LOUD WARNING: assertion failed in merge: {_assert_msg}")
        else:
            print(f"  ✅ [{lang.upper()}] {ref_key} == {cmp_key}: identical question_id sets ({len(ref_set)} items)")

if alignment_ok:
    print("\n✅ ALL AVAILABLE MODELS HAVE IDENTICAL question_id SETS (fully comparable).")


# --- Subset SHA256 equality check ---
print("\n--- bn_subset.csv / en_subset.csv SHA256 equality check ---")

def _file_sha256(path: str) -> str:
    h = hashlib.sha256()
    with open(path, "rb") as f:
        for chunk in iter(lambda: f.read(65536), b""):
            h.update(chunk)
    return h.hexdigest()

# STRICT_ASSERT controls failure mode on subset SHA256 mismatch.
#   True  : raise AssertionError (halt the notebook).
#   False : write merge_status.json listing every mismatch, print a loud
#           warning, and continue so the rest of the pipeline runs.
#           Read /kaggle/working/merge_status.json after a commit to review mismatches.
STRICT_ASSERT = globals().get("STRICT_ASSERT", False)

_subset_sha256 = {}  # model_key -> {lang -> sha256}
for m_key in CYCLIC4_MODELS:
    m_dir = os.path.join(base_dir, f"results_E2_cyclic4_{m_key}")
    if not os.path.isdir(m_dir):
        m_dir = f"results_E2_cyclic4_{m_key}"
    _subset_sha256[m_key] = {}
    for lang in ("bn", "en"):
        _p = os.path.join(m_dir, f"{lang}_subset.csv")
        if os.path.exists(_p):
            _subset_sha256[m_key][lang] = _file_sha256(_p)
            print(f"  {m_key}/{lang}: {_subset_sha256[m_key][lang][:16]}...")
        else:
            _subset_sha256[m_key][lang] = None
            print(f"  ⚠️  {m_key}/{lang}_subset.csv not found")

# Assert identical SHA256 across models per language
_sha256_ok = True
for lang in ("bn", "en"):
    _hashes = {mk: _subset_sha256[mk].get(lang) for mk in CYCLIC4_MODELS
               if _subset_sha256[mk].get(lang) is not None}
    _unique_hashes = set(_hashes.values())
    if len(_unique_hashes) == 1:
        print(f"  ✅ [{lang.upper()}] subset SHA256 identical across all three models: {next(iter(_unique_hashes))[:20]}...")
    elif len(_unique_hashes) > 1:
        _sha256_ok = False
        print(f"  ❌ [{lang.upper()}] SUBSET SHA256 MISMATCH — models used different subsets!")
        for mk, h in _hashes.items():
            print(f"       {mk}: {h}")
        if STRICT_ASSERT:
            raise AssertionError(
                f"[{lang.upper()}] bn_subset.csv / en_subset.csv SHA256 mismatch across models. "
                "This means the models were evaluated on different item sets and results are NOT comparable."
            )
        else:
            print(
                f"  ⚠️  [{lang.upper()}] Continuing despite SHA256 mismatch (STRICT_ASSERT=False). "
                "Read /kaggle/working/merge_status.json after a commit to review all mismatches."
            )
    else:
        print(f"  ⚠️  [{lang.upper()}] No subset files found for comparison.")

# --- Build three_model_summary.csv ---
print("\n--- Building three_model_summary.csv ---")
_summary_csv_rows = []
for m_key in CYCLIC4_MODELS:
    m_dir = os.path.join(base_dir, f"results_E2_cyclic4_{m_key}")
    if not os.path.isdir(m_dir):
        m_dir = f"results_E2_cyclic4_{m_key}"

    for lang in ("bn", "en"):
        # Try per-language summary JSON first, then fall back to main summary
        _lang_sum_path = os.path.join(m_dir, f"{lang}_summary.json")
        _main_sum_path = os.path.join(m_dir, "summary.json")
        _lsrc = None
        for _sp in [_lang_sum_path, _main_sum_path]:
            if os.path.exists(_sp):
                try:
                    with open(_sp, "r", encoding="utf-8") as _f:
                        _lsrc = json.load(_f)
                    break
                except Exception:
                    pass
        if _lsrc is None:
            _summary_csv_rows.append({"model_key": m_key, "lang": lang, "status": "MISSING"})
            continue

        # Extract accuracy and ECE from the per-lang summary or the nested languages dict
        def _get(d, *keys, default=None):
            for k in keys:
                if isinstance(d, dict) and k in d:
                    d = d[k]
                else:
                    return default
            return d

        _lang_data = _lsrc
        # If we loaded the top-level summary.json, try to descend into languages[lang]
        if "languages" in _lsrc and lang in (_lsrc.get("languages") or {}):
            _lang_data = _lsrc["languages"][lang]

        _acc = _get(_lang_data, "accuracy_percent") or _get(_lang_data, "accuracy")
        _ece = _get(_lang_data, "calibration_metrics", "ece_uniform_10") or                _get(_lang_data, "ece_uniform_10")
        _top_share = _get(_lang_data, "top_share") or _get(_lsrc, "top_share_by_lang", lang)
        _gate_pass = _get(_lang_data, "gate_pass", default=_lsrc.get("gate_pass"))
        _debias = _lsrc.get("debias", "cyclic4")

        _summary_csv_rows.append({
            "model_key": m_key,
            "lang": lang,
            "debias": _debias,
            "accuracy_percent": _acc,
            "ece_uniform_10": _ece,
            "top_share": _top_share,
            "gate_pass": _gate_pass,
            "subset_sha256_bn": _subset_sha256.get(m_key, {}).get("bn"),
            "subset_sha256_en": _subset_sha256.get(m_key, {}).get("en"),
        })

_three_model_csv = os.path.join(base_dir, "three_model_summary.csv")
if _summary_csv_rows:
    _df_tms = pd.DataFrame(_summary_csv_rows)
    _df_tms.to_csv(_three_model_csv, index=False)
    print(f"  📄 Saved three_model_summary.csv ({len(_df_tms)} rows): {_three_model_csv}")
    try:
        display(_df_tms)
    except Exception:
        print(_df_tms.to_string(index=False))

# --- Write merge_status.json (always written, lists every mismatch) ---
_merge_status_path = os.path.join(base_dir, "merge_status.json")
_merge_status = {
    "sha256_ok": _sha256_ok,
    "strict_assert": STRICT_ASSERT,
    "mismatches": [],
}
for lang in ("bn", "en"):
    _hashes = {
        mk: _subset_sha256[mk].get(lang)
        for mk in CYCLIC4_MODELS
        if _subset_sha256[mk].get(lang) is not None
    }
    if len(set(_hashes.values())) > 1:
        _merge_status["mismatches"].append({"lang": lang, "hashes": _hashes})
with open(_merge_status_path, "w", encoding="utf-8") as _f:
    json.dump(_merge_status, _f, indent=2, default=str)
print(f"  📄 Saved merge_status.json: {_merge_status_path}")
print(f"     sha256_ok={_sha256_ok}  mismatches={len(_merge_status['mismatches'])}  strict_assert={STRICT_ASSERT}")
# --- Final summary ---
print("\n" + "=" * 70)
print("📋 FINAL THREE-MODEL CYCLIC4 SUMMARY (v4h)")
print("=" * 70)
missing_models = [mk for mk in CYCLIC4_MODELS if summaries.get(mk) is None]
complete_models = [mk for mk in CYCLIC4_MODELS if summaries.get(mk) is not None]
if missing_models:
    print(f"  ⚠️  Pending models (not yet run): {missing_models}")
    print(f"  ✅  Completed models: {complete_models}")
    print("  Run the Stage Runner cell (Section 14d-bis) to complete the pending models.")
else:
    print(f"  ✅ All three models completed: {CYCLIC4_MODELS}")
    all_pass = all(summaries[mk].get("gate_pass") for mk in CYCLIC4_MODELS)
    if all_pass:
        print("  ✅ All three models PASSED validity gates (top_share<=0.40, min_letter_share>=0.10).")
    else:
        failed = [mk for mk in CYCLIC4_MODELS if not summaries[mk].get("gate_pass")]
        print(f"  ⚠️  Gate failures: {failed}")
print("=" * 70)
```

## Section 18 (v4h): Run Manifest
Writes `run_manifest.json` recording notebook version, GPU, per-model dtype, subset SHA256s, seed, and per-model wall time.

```python
import json as _json
import os as _os
import traceback as _traceback

_CELL_NAME = 'run_manifest'
_STATUS_PATH = "/kaggle/working/new_cells_status.json"

def _record_cell_status(name, ok, error=None):
    try:
        if _os.path.exists(_STATUS_PATH):
            with open(_STATUS_PATH, "r", encoding="utf-8") as _f:
                status = _json.load(_f)
        else:
            status = {}
    except Exception:
        status = {}
    status[name] = {
        "ok": ok,
        "error": str(error) if error is not None else None,
        "traceback": _traceback.format_exc() if error is not None else None,
    }
    try:
        with open(_STATUS_PATH, "w", encoding="utf-8") as _f:
            _json.dump(status, _f, indent=2, default=str)
    except Exception:
        pass

try:
    # ============================================================
    # Section 18 (v4h): Write run_manifest.json
    # ============================================================
    # Records notebook version, torch version, GPU info, per-model dtype,
    # subset SHA256s, seed, and per-model wall time (if available).
    # ============================================================
    import os
    import json
    import hashlib
    import datetime

    try:
        import torch as _torch_rm
        _torch_version = _torch_rm.__version__
        _gpu_name = _torch_rm.cuda.get_device_name(0) if _torch_rm.cuda.is_available() else "cpu"
        _cuda_avail = _torch_rm.cuda.is_available()
    except Exception:
        _torch_version = "unavailable"
        _gpu_name = "unknown"
        _cuda_avail = False

    _BASE_WORKING = os.environ.get("KAGGLE_WORKING_DIR",
        "/kaggle/working" if os.path.exists("/kaggle/working") else ".")

    # Per-model dtype (from MODEL_REGISTRY — do NOT modify)
    _MODEL_DTYPES = {
        "titullm_1b":  MODEL_REGISTRY.get("titullm_1b", {}).get("dtype", "float16"),
        "tigerllm_1b": MODEL_REGISTRY.get("tigerllm_1b", {}).get("dtype", "float32"),
        "llama32_3b":  MODEL_REGISTRY.get("llama32_3b", {}).get("dtype", "float16"),
    }
    # E3 models (dtype from registry; evaluated under E3 block)
    for _em in ["qwen25_3b", "bloomz_3b", "phi35_mini", "llama32_3b_base"]:
        _MODEL_DTYPES[_em] = MODEL_REGISTRY.get(_em, {}).get("dtype", "float16")
    # E4 new models (dtype from registry)
    for _em in list(globals().get("E4_NEW_MODELS", [])):
        _MODEL_DTYPES[_em] = MODEL_REGISTRY.get(_em, {}).get("dtype", "float16")

    # Subset SHA256s (from three_model_summary.csv or direct file hash)
    def _sha256_file(path):
        if not os.path.exists(path):
            return None
        h = hashlib.sha256()
        with open(path, "rb") as f:
            for chunk in iter(lambda: f.read(65536), b""):
                h.update(chunk)
        return h.hexdigest()

    _subset_sha_manifest = {}
    for _m in ["titullm_1b", "tigerllm_1b", "llama32_3b"]:
        _m_dir = os.path.join(_BASE_WORKING, f"results_E2_cyclic4_{_m}")
        if not os.path.isdir(_m_dir):
            _m_dir = f"results_E2_cyclic4_{_m}"
        _subset_sha_manifest[_m] = {
            "bn_subset_sha256": _sha256_file(os.path.join(_m_dir, "bn_subset.csv")),
            "en_subset_sha256": _sha256_file(os.path.join(_m_dir, "en_subset.csv")),
        }

    # E3 subset SHA256s
    for _em in list(globals().get("TARGET_E3_MODELS", [])):
        _em_dir = os.path.join(_BASE_WORKING, f"results_E3_{_em}")
        if not os.path.isdir(_em_dir):
            _em_dir = f"results_E3_{_em}"
        _subset_sha_manifest[_em] = {
            "bn_subset_sha256": _sha256_file(os.path.join(_em_dir, "bn_subset.csv")),
            "en_subset_sha256": _sha256_file(os.path.join(_em_dir, "en_subset.csv")),
        }

    # Per-model wall time: read from each model's summary.json if available
    _wall_times = {}
    for _m in ["titullm_1b", "tigerllm_1b", "llama32_3b"]:
        _m_dir = os.path.join(_BASE_WORKING, f"results_E2_cyclic4_{_m}")
        if not os.path.isdir(_m_dir):
            _m_dir = f"results_E2_cyclic4_{_m}"
        _sp = os.path.join(_m_dir, "summary.json")
        if os.path.exists(_sp):
            try:
                with open(_sp, "r", encoding="utf-8") as _f:
                    _sd = json.load(_f)
                _wall_times[_m] = _sd.get("wall_time_seconds") or _sd.get("elapsed_seconds")
            except Exception:
                _wall_times[_m] = None
        else:
            _wall_times[_m] = None

    # E3 wall times
    for _em in list(globals().get("TARGET_E3_MODELS", [])):
        _em_dir = os.path.join(_BASE_WORKING, f"results_E3_{_em}")
        if not os.path.isdir(_em_dir):
            _em_dir = f"results_E3_{_em}"
        _sp = os.path.join(_em_dir, "summary.json")
        if os.path.exists(_sp):
            try:
                with open(_sp, "r", encoding="utf-8") as _f:
                    _sd = json.load(_f)
                _wall_times[_em] = _sd.get("wall_time_seconds") or _sd.get("elapsed_seconds")
            except Exception:
                _wall_times[_em] = None
        else:
            _wall_times[_em] = None

    _models_manifest = {}
    for _m in ["titullm_1b", "tigerllm_1b", "llama32_3b"]:
        _cfg_entry = MODEL_REGISTRY.get(_m, {})
        _models_manifest[_m] = {
            "model_id": _cfg_entry.get("hf_id", _m),
            "model_configuration": {
                "use_chat_template": _cfg_entry.get("use_chat_template", False),
                "prompt_variant": _cfg_entry.get("prompt_variant", "few_shot"),
                "n_shots": getattr(CONFIG, "n_shots", 4),
                "debias": "cyclic4",
                "tokenizer_family": _cfg_entry.get("tokenizer_family", ""),
                "expected_model_type": _cfg_entry.get("expected_model_type", ""),
            },
            "dtype": _MODEL_DTYPES.get(_m, "float16"),
            "wall_clock_seconds": _wall_times.get(_m),
            "batch_size": getattr(CONFIG, "batch_size", 1),
            "max_length": globals().get("GLOBAL_MAX_SEQ_LEN", getattr(CONFIG, "max_tokens", 2048)),
            "dataset_size": getattr(CONFIG, "n_samples", 1500),
            "seed": getattr(CONFIG, "seed", 42),
        }

    # E3 models manifest
    for _em in list(globals().get("TARGET_E3_MODELS", [])):
        _cfg_entry = MODEL_REGISTRY.get(_em, {})
        _em_dir = os.path.join(_BASE_WORKING, f"results_E3_{_em}")
        if not os.path.isdir(_em_dir):
            _em_dir = f"results_E3_{_em}"
        _esp = os.path.join(_em_dir, "summary.json")
        _ewall = None
        _elabel = None
        if os.path.exists(_esp):
            try:
                with open(_esp, "r", encoding="utf-8") as _f:
                    _esd = json.load(_f)
                _ewall = _esd.get("wall_time_seconds") or _esd.get("elapsed_seconds")
                _elabel = _esd.get("label")
            except Exception:
                _ewall = None
        _models_manifest[_em] = {
            "model_id": _cfg_entry.get("hf_id", _em),
            "model_configuration": {
                "use_chat_template": _cfg_entry.get("use_chat_template", False),
                "prompt_variant": _cfg_entry.get("prompt_variant", "few_shot"),
                "n_shots": getattr(CONFIG, "n_shots", 4),
                "debias": "cyclic4",
                "tokenizer_family": _cfg_entry.get("tokenizer_family", ""),
                "expected_model_type": _cfg_entry.get("expected_model_type", ""),
            },
            "dtype": _MODEL_DTYPES.get(_em, "float16"),
            "wall_clock_seconds": _ewall,
            "label": _elabel,
            "batch_size": getattr(CONFIG, "batch_size", 1),
            "max_length": globals().get("GLOBAL_MAX_SEQ_LEN", getattr(CONFIG, "max_tokens", 2048)),
            "dataset_size": getattr(CONFIG, "n_samples", 1500),
            "seed": getattr(CONFIG, "seed", 42),
            "stage": "E3",
        }

    _manifest = {
        "notebook_version": "v4j",
        "created_at": datetime.datetime.now(datetime.timezone.utc).isoformat(),
        "torch_version": _torch_version,
        "cuda_available": _cuda_avail,
        "gpu_name": _gpu_name,
        "seed": getattr(CONFIG, "seed", 42),
        "batch_size": getattr(CONFIG, "batch_size", 1),
        "max_length": globals().get("GLOBAL_MAX_SEQ_LEN", getattr(CONFIG, "max_tokens", 2048)),
        "dataset_size": getattr(CONFIG, "n_samples", 1500),
        "n_items_per_language": 1500,
        "debias": "cyclic4",
        "per_model_dtype": _MODEL_DTYPES,
        "per_model_wall_time_seconds": _wall_times,
        "subset_sha256": _subset_sha_manifest,
        "models": _models_manifest,
        "dtype_sensitivity_active_models": ["titullm_1b", "tigerllm_1b", "llama32_3b"],
        "e3_models": list(globals().get("TARGET_E3_MODELS", [])),
        "models_executed": [
            _em for _em in list(globals().get("TARGET_E3_MODELS", []))
            if os.path.exists(os.path.join(_BASE_WORKING, f"results_E3_{_em}", _em, "bn_summary.json"))
            and os.path.exists(os.path.join(_BASE_WORKING, f"results_E3_{_em}", _em, "en_summary.json"))
        ],
        # E3 translation CSV provenance
        "en_source": CONFIG.en_source,
        "en_translation_path": CONFIG.en_translation_path,
        "translation_csv_sha256": (
            _file_sha256(CONFIG.en_translation_path)
            if CONFIG.en_translation_path and os.path.exists(CONFIG.en_translation_path)
            else None
        ),
    }
    _manifest_path = os.path.join(_BASE_WORKING, "run_manifest.json")
    with open(_manifest_path, "w", encoding="utf-8") as _f:
        json.dump(_manifest, _f, indent=2, default=str)
    try:
        with open("run_manifest.json", "w", encoding="utf-8") as _f2:
            json.dump(_manifest, _f2, indent=2, default=str)
    except Exception:
        pass

    print("=" * 60)
    print("📋 run_manifest.json written")
    print("=" * 60)
    print(json.dumps(_manifest, indent=2, default=str))
except Exception as _e:
    _record_cell_status(_CELL_NAME, False, _e)
    print(f"❌ [run_manifest] failed: {_e}")
    import traceback as _tb2; _tb2.print_exc()
else:
    _record_cell_status(_CELL_NAME, True, None)
```

## Section 19 (v4h): HF Cache Cleanup
Deletes `/kaggle/working/hf_cache` when `CLEAN_HF_CACHE=True` to keep committed output small. Run this as the **final** cell in every session.

```python
import json as _json
import os as _os
import traceback as _traceback

_CELL_NAME = 'hf_cache_cleanup'
_STATUS_PATH = "/kaggle/working/new_cells_status.json"

def _record_cell_status(name, ok, error=None):
    try:
        if _os.path.exists(_STATUS_PATH):
            with open(_STATUS_PATH, "r", encoding="utf-8") as _f:
                status = _json.load(_f)
        else:
            status = {}
    except Exception:
        status = {}
    status[name] = {
        "ok": ok,
        "error": str(error) if error is not None else None,
        "traceback": _traceback.format_exc() if error is not None else None,
    }
    try:
        with open(_STATUS_PATH, "w", encoding="utf-8") as _f:
            _json.dump(status, _f, indent=2, default=str)
    except Exception:
        pass

try:
    # ============================================================
    # Section 19 (v4h): HF Cache Cleanup
    # ============================================================
    # Deletes /kaggle/working/hf_cache to keep committed output small.
    # Set CLEAN_HF_CACHE = False to skip (e.g., if you want to inspect
    # downloaded model weights before cleanup).
    # ============================================================
    import os
    import shutil

    CLEAN_HF_CACHE = True   # Set to False to skip

    _BASE_WORKING = os.environ.get("KAGGLE_WORKING_DIR",
        "/kaggle/working" if os.path.exists("/kaggle/working") else ".")
    HF_CACHE_DIR = os.path.join(_BASE_WORKING, "hf_cache")

    if CLEAN_HF_CACHE:
        if os.path.exists(HF_CACHE_DIR):
            # Measure size before deletion
            _total_size = sum(
                os.path.getsize(os.path.join(root, f))
                for root, _, files in os.walk(HF_CACHE_DIR)
                for f in files
            )
            shutil.rmtree(HF_CACHE_DIR, ignore_errors=True)
            print(f"🗑️  Deleted {HF_CACHE_DIR} ({_total_size / (1024**3):.2f} GB freed)")
        else:
            print(f"ℹ️  HF cache directory not found at {HF_CACHE_DIR} (nothing to delete).")
    else:
        print("ℹ️  CLEAN_HF_CACHE=False — skipping cache cleanup.")

    print("✅ Cleanup complete.")
except Exception as _e:
    _record_cell_status(_CELL_NAME, False, _e)
    print(f"❌ [hf_cache_cleanup] failed: {_e}")
    import traceback as _tb2; _tb2.print_exc()
else:
    _record_cell_status(_CELL_NAME, True, None)
```
